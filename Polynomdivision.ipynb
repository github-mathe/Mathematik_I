# Polynomdivision

Die **Polynomdivision** ist ein Verfahren zur Division zweier Polynome.

Sie funktioniert ähnlich wie die schriftliche Division von Zahlen.

Gegeben seien zwei Polynome

$$
p(x)
$$

und

$$
q(x),
$$

wobei

$$
q(x)\neq 0.
$$

Dann lässt sich die Division in der Form

$$
p(x)=q(x)\cdot s(x)+r(x)
$$

darstellen.

Dabei bezeichnet

- $p(x)$ das zu dividierende Polynom,
- $q(x)$ den Divisor,
- $s(x)$ den Quotienten,
- $r(x)$ den Rest.

Für den Rest gilt

$$
\deg(r)<\deg(q).
$$

Ist

$$
r(x)=0,
$$

so ist $p(x)$ ohne Rest durch $q(x)$ teilbar.

<a id="vorgehen-polynomdivision"></a>

## Vorgehen bei der Polynomdivision

Die Polynomdivision wird schrittweise durchgeführt:

1. Dividiere den Term mit der höchsten Potenz des Polynoms $p(x)$ durch den Term mit der höchsten Potenz des Divisors $q(x)$.

2. Multipliziere das erhaltene Ergebnis mit dem Divisor $q(x)$.

3. Subtrahiere das erhaltene Polynom von $p(x)$.

4. Wiederhole das Verfahren mit dem verbleibenden Polynom.

5. Die Division endet, wenn der Grad des Restpolynoms kleiner als der Grad des Divisors ist.
