## Lineare Gleichungssysteme mit komplexen Zahlen

- Lineare Gleichungssysteme können auch komplexe Koeffizienten enthalten.
- Das Gaußsche Eliminationsverfahren kann genauso wie im reellen Fall angewendet werden, um Lösungen zu finden.

**Beispiel:**

$$
\begin{aligned}
x+y+z &= 6 \\
x+2y+3z &= 14 \\
2x+(3+i)y+(4-i)z &= 18+2i
\end{aligned}
$$

Die erweiterte Matrix ist:

$$
\left[
\begin{array}{ccc|c}
1 & 1 & 1 & 6 \\
1 & 2 & 3 & 14 \\
2 & 3+i & 4-i & 18+2i
\end{array}
\right]
$$

### Vorwärtselimination: Schritt 1

Ausgangsmatrix:

$$
\begin{array}{l|ccc|c}
\text{I}   & 1 & 1 & 1 & 6 \\
\text{II}  & 1 & 2 & 3 & 14 \\
\text{III} & 2 & 3+i & 4-i & 18+2i
\end{array}
$$

Subtrahiere Zeile I von Zeile II und subtrahiere $2\times\text{Zeile I}$ von Zeile III:

$$
\begin{array}{l|ccc|c|l}
\text{I}    & 1 & 1 & 1 & 6 & \\
\text{II'}  & 0 & 1 & 2 & 8 & \text{II' = II - I} \\
\text{III'} & 0 & 1+i & 2-i & 6+2i & \text{III' = III - 2I}
\end{array}
$$

### Vorwärtselimination: Schritt 2

Subtrahiere $(1+i)\times\text{Zeile II'}$ von Zeile III':

$$
\begin{array}{l|ccc|c|l}
\text{I}     & 1 & 1 & 1 & 6 & \\
\text{II'}   & 0 & 1 & 2 & 8 & \\
\text{III''} & 0 & 0 & -3i & -2-6i &
\text{III'' = III' - (1+i)II'}
\end{array}
$$

### Rückwärtseinsetzung

Jetzt führen wir die Rückwärtseinsetzung durch:

- Aus Zeile III'' ergibt sich:

$$
-3i\,z=-2-6i
\quad\Rightarrow\quad
z=2-\frac{2i}{3}.
$$

- Setze

$$
z=2-\frac{2i}{3}
$$

in Zeile II' ein:

$$
y+2z=8
\quad\Rightarrow\quad
y=4+\frac{4i}{3}.
$$

- Setze

$$
y=4+\frac{4i}{3}
$$

und

$$
z=2-\frac{2i}{3}
$$

in Zeile I ein:

$$
x+y+z=6
\quad\Rightarrow\quad
x=-\frac{2i}{3}.
$$

Somit ist die Lösung:

$$
\boxed{
x=-\frac{2i}{3},
\qquad
y=4+\frac{4i}{3},
\qquad
z=2-\frac{2i}{3}
}
$$
