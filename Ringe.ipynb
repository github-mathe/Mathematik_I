## Von Gruppen zu Ringen und Körpern

Während Gruppen eine einzelne Verknüpfung betrachten, beinhalten **Ringe** und **Körper** zwei Verknüpfungen: Addition und Multiplikation. Diese Strukturen sind in der Algebra ebenso wichtig und erweitern die Konzepte der Gruppen.

<a id="ringe"></a>

## Ringe

### Definition eines Rings

Ein **Ring** $R$ ist eine Menge zusammen mit zwei Operationen:

- Addition $+$
- Multiplikation $\cdot$

Diese Operationen erfüllen die folgenden Axiome:

1. $(R,+)$ ist eine abelsche Gruppe.
2. Die Multiplikation ist assoziativ.
3. Es gilt das Distributivgesetz:

$$
a\cdot(b+c)
=
a\cdot b+a\cdot c.
$$

### Beispiele für Ringe

- Die ganzen Zahlen $\mathbb Z$ mit Addition und Multiplikation.
- Der Ring der $n\times n$-Matrizen

$$
M_n(\mathbb R).
$$
