## Basis eines Vektorraums

### Definition

Eine **Basis** eines Vektorraums $V$ ist eine Menge von Vektoren, die den Vektorraum aufspannen und linear unabhängig sind.

**Eigenschaften einer Basis:**

- Jeder Vektor in $V$ kann als eindeutige Linearkombination der Basisvektoren geschrieben werden.
- Die Anzahl der Vektoren in der Basis ist die **Dimension** des Vektorraums.
- Wenn ein Vektorraum $V$ $n$-dimensional ist, enthält jede Basis genau $n$ Vektoren.
- Die Basis eines Vektorraums ist nicht eindeutig: Ein Vektorraum kann verschiedene Basen haben, aber die Anzahl der Basisvektoren bleibt immer gleich (Dimension).

<a id="vektor-als-linearkombination"></a>

### Darstellung eines Vektors als Linearkombination

Jeder Vektor

$$
\mathbf{v}\in\mathbb{R}^n
$$

kann als Linearkombination von Basisvektoren

$$
\mathbf{b}_1,\mathbf{b}_2,\dots,\mathbf{b}_n
$$

dargestellt werden:

$$
\mathbf{v}
=
c_1\mathbf{b}_1
+
c_2\mathbf{b}_2
+
\dots
+
c_n\mathbf{b}_n.
$$

Die Koeffizienten

$$
c_1,c_2,\dots,c_n
$$

sind eindeutig bestimmt, wenn die Basisvektoren linear unabhängig sind.

<a id="beispiele-basen-r2"></a>

### Beispiele von Basen in $\mathbb{R}^2$

**Beispiel 1 (Standardbasis):**

Die Vektoren

$$
\vec{v}_1=
\begin{pmatrix}
1\\
0
\end{pmatrix}
$$

und

$$
\vec{v}_2=
\begin{pmatrix}
0\\
1
\end{pmatrix}
$$

bilden die Standardbasis von $\mathbb{R}^2$.

Jede Linearkombination

$$
\lambda_1
\begin{pmatrix}
1\\
0
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
0\\
1
\end{pmatrix}
=
\begin{pmatrix}
\lambda_1\\
\lambda_2
\end{pmatrix}
$$

beschreibt jeden Vektor im $\mathbb{R}^2$.

**Beispiel 2 (Nicht-Standardbasis):**

Die Vektoren

$$
\vec{v}_1=
\begin{pmatrix}
1\\
2
\end{pmatrix}
$$

und

$$
\vec{v}_2=
\begin{pmatrix}
3\\
1
\end{pmatrix}
$$

bilden ebenfalls eine Basis von $\mathbb{R}^2$, da sie linear unabhängig sind.

Jede Linearkombination

$$
\lambda_1
\begin{pmatrix}
1\\
2
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
3\\
1
\end{pmatrix}
=
\begin{pmatrix}
\lambda_1+3\lambda_2\\
2\lambda_1+\lambda_2
\end{pmatrix}
$$

beschreibt jeden Vektor im $\mathbb{R}^2$.

<a id="beziehung-m-n"></a>

### Beziehung zwischen $m$ und $n$

- Der Raum $\mathbb{R}^m$ besteht aus allen Vektoren mit $m$ Komponenten.

- Wenn $n$ Vektoren

$$
\vec{v}_1,\vec{v}_2,\dots,\vec{v}_n
$$

in $\mathbb{R}^m$ gegeben sind, dann ist der aufgespannte Raum der kleinste Unterraum von $\mathbb{R}^m$, der diese Vektoren enthält.

- **Fall 1:** Wenn $n=m$ und die Vektoren

$$
\{\vec{v}_1,\vec{v}_2,\dots,\vec{v}_n\}
$$

linear unabhängig sind, dann spannen sie den gesamten Raum $\mathbb{R}^m$ auf. In diesem Fall bilden sie eine Basis von $\mathbb{R}^m$.

- **Fall 2:** Wenn $n<m$, können die Vektoren

$$
\vec{v}_1,\vec{v}_2,\dots,\vec{v}_n
$$

nur einen Unterraum von $\mathbb{R}^m$ aufspannen, da weniger als $m$ Vektoren den gesamten Raum nicht vollständig aufspannen können.

- **Fall 3:** Wenn $n>m$, sind die Vektoren

$$
\vec{v}_1,\vec{v}_2,\dots,\vec{v}_n
$$

linear abhängig, da es nicht mehr als $m$ linear unabhängige Vektoren in $\mathbb{R}^m$ geben kann.

Der aufgespannte Raum bleibt $\mathbb{R}^m$, aber einige der Vektoren in der Menge

$$
\{\vec{v}_1,\vec{v}_2,\dots,\vec{v}_n\}
$$

sind linear abhängig.
