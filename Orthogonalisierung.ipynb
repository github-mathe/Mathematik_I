## Gram-Schmidt-Algorithmus

**Ziel:**

- Umwandlung einer gegebenen Basis eines Vektorraums in eine orthonormale Basis.

**Prinzip:**

- Iterative Konstruktion: Von jedem Vektor wird die Projektion auf die zuvor konstruierten Basisvektoren subtrahiert, sodass der verbleibende Teil orthogonal zu diesen ist.
- Das Resultat ist eine Menge von orthogonalen Vektoren, die den gleichen Raum aufspannen wie die ursprüngliche Basis.

**Anwendung:**

- Der Algorithmus ist nützlich bei der Konstruktion von orthonormalen Basen für Unterräume, z.B. in der linearen Regression, QR-Zerlegung oder Fourier-Analyse.
- Erleichtert Projektionen und Transformationen, da orthonormale Basen numerisch stabiler sind.

### Verbindung zu Projektionen

- Für eine gegebene Basis

$$
\{\mathbf{v}_1,\mathbf{v}_2,\ldots,\mathbf{v}_n\}
$$

möchten wir eine orthonormale Basis

$$
\{\mathbf{u}_1,\mathbf{u}_2,\ldots,\mathbf{u}_n\}
$$

finden.

- Jeder Vektor $\mathbf{u}_k$ wird konstruiert, indem der entsprechende Basisvektor $\mathbf{v}_k$ um den Anteil modifiziert wird, der seiner Projektion auf die bereits berechneten orthonormalen Vektoren entspricht:

$$
\mathbf{u}_k
=
\mathbf{v}_k
-
\sum_{j=1}^{k-1}
P_{\mathbf{u}_j}(\mathbf{v}_k),
$$

wobei die Projektion durch

$$
P_{\mathbf{u}_j}(\mathbf{v}_k)
=
\langle \mathbf{v}_k,\mathbf{u}_j\rangle\,\mathbf{u}_j
$$

definiert ist.

- Anschließend wird $\mathbf{u}_k$ normalisiert:

$$
\mathbf{u}_k
=
\frac{\mathbf{u}_k}{\|\mathbf{u}_k\|}.
$$

### Gram-Schmidt-Algorithmus

**Eingabe:** Eine Basis

$$
\{\mathbf{v}_1,\mathbf{v}_2,\ldots,\mathbf{v}_n\}
$$

des Vektorraums.

**Ausgabe:** Eine orthonormale Basis

$$
\{\mathbf{u}_1,\mathbf{u}_2,\ldots,\mathbf{u}_n\}.
$$

**Schritte:**

1. Setze

   $$
   \mathbf{u}_1
   =
   \frac{\mathbf{v}_1}{\|\mathbf{v}_1\|}.
   $$

2. Für $k=2,\ldots,n$:

   1. Berechne den orthogonalen Anteil:

      $$
      \mathbf{w}_k
      =
      \mathbf{v}_k
      -
      \sum_{j=1}^{k-1}
      P_{\mathbf{u}_j}(\mathbf{v}_k).
      $$

   2. Normalisiere:

      $$
      \mathbf{u}_k
      =
      \frac{\mathbf{w}_k}{\|\mathbf{w}_k\|}.
      $$

### Beispiel: Gram-Schmidt in $\mathbb{R}^3$

**Gegeben:** Basis

$$
\left\{
\mathbf{v}_1=(1,1,0)^T,\;
\mathbf{v}_2=(1,0,1)^T,\;
\mathbf{v}_3=(0,1,1)^T
\right\}.
$$

**Berechnung:**

#### Schritt 1

$$
\mathbf{u}_1
=
\frac{\mathbf{v}_1}{\|\mathbf{v}_1\|}
=
\frac{(1,1,0)^T}{\sqrt{2}}
=
\left(
\frac{1}{\sqrt{2}},
\frac{1}{\sqrt{2}},
0
\right)^T.
$$

#### Schritt 2: Berechnung von $\mathbf{u}_2$

$$
P_{\mathbf{u}_1}(\mathbf{v}_2)
=
\langle\mathbf{v}_2,\mathbf{u}_1\rangle\mathbf{u}_1,
$$

$$
\langle\mathbf{v}_2,\mathbf{u}_1\rangle
=
\frac{1}{\sqrt{2}}
(1\cdot1+0\cdot1+1\cdot0)
=
\frac{1}{\sqrt{2}},
$$

$$
P_{\mathbf{u}_1}(\mathbf{v}_2)
=
\frac{1}{\sqrt{2}}
\left(
\frac{1}{\sqrt{2}},
\frac{1}{\sqrt{2}},
0
\right)^T
=
\left(
\frac12,
\frac12,
0
\right)^T.
$$

Der orthogonale Anteil ist:

$$
\mathbf{w}_2
=
\mathbf{v}_2
-
P_{\mathbf{u}_1}(\mathbf{v}_2)
=
(1,0,1)^T
-
\left(
\frac12,
\frac12,
0
\right)^T
=
\left(
\frac12,
-\frac12,
1
\right)^T.
$$

Normalisierung von $\mathbf{w}_2$:

$$
\|\mathbf{w}_2\|
=
\sqrt{
\left(\frac12\right)^2+
\left(-\frac12\right)^2+
1^2
}
=
\sqrt{\frac14+\frac14+1}
=
\sqrt{\frac32},
$$

$$
\mathbf{u}_2
=
\frac{\mathbf{w}_2}{\|\mathbf{w}_2\|}
=
\frac{
\left(
\frac12,-\frac12,1
\right)^T
}{
\sqrt{\frac32}
}
=
\left(
\frac{1}{\sqrt6},
-\frac{1}{\sqrt6},
\frac{2}{\sqrt6}
\right)^T.
$$

### Beispiel: Gram-Schmidt in $\mathbb{R}^3$ – Schritt 3

Berechnung von $\mathbf{u}_3$:

$$
P_{\mathbf{u}_1}(\mathbf{v}_3)
=
\langle\mathbf{v}_3,\mathbf{u}_1\rangle\mathbf{u}_1,
$$

$$
\langle\mathbf{v}_3,\mathbf{u}_1\rangle
=
\frac{1}{\sqrt2}
(0\cdot1+1\cdot1+1\cdot0)
=
\frac{1}{\sqrt2},
$$

$$
P_{\mathbf{u}_1}(\mathbf{v}_3)
=
\frac{1}{\sqrt2}
\left(
\frac{1}{\sqrt2},
\frac{1}{\sqrt2},
0
\right)^T
=
\left(
\frac12,
\frac12,
0
\right)^T.
$$

Weiter gilt:

$$
P_{\mathbf{u}_2}(\mathbf{v}_3)
=
\langle\mathbf{v}_3,\mathbf{u}_2\rangle\mathbf{u}_2,
$$

$$
\langle\mathbf{v}_3,\mathbf{u}_2\rangle
=
\left(
0\cdot\frac{1}{\sqrt6}
+
1\cdot\left(-\frac{1}{\sqrt6}\right)
+
1\cdot\frac{2}{\sqrt6}
\right)
=
\frac{1}{\sqrt6},
$$

$$
P_{\mathbf{u}_2}(\mathbf{v}_3)
=
\frac{1}{\sqrt6}
\left(
\frac{1}{\sqrt6},
-\frac{1}{\sqrt6},
\frac{2}{\sqrt6}
\right)^T
=
\left(
\frac16,
-\frac16,
\frac26
\right)^T.
$$

Der orthogonale Anteil ist:

$$
\begin{aligned}
\mathbf{w}_3
&=
\mathbf{v}_3
-
P_{\mathbf{u}_1}(\mathbf{v}_3)
-
P_{\mathbf{u}_2}(\mathbf{v}_3)
\\
&=
(0,1,1)^T
-
\left(\frac12,\frac12,0\right)^T
-
\left(\frac16,-\frac16,\frac13\right)^T,
\end{aligned}
$$

also

$$
\mathbf{w}_3
=
\left(
-\frac23,
\frac23,
\frac23
\right)^T.
$$

Normalisierung:

$$
\mathbf{u}_3
=
\frac{\mathbf{w}_3}{\|\mathbf{w}_3\|}
=
\left(
-\frac{1}{\sqrt3},
\frac{1}{\sqrt3},
\frac{1}{\sqrt3}
\right)^T.
$$

Wir haben also ausgehend von einer nicht-orthogonalen Basis eine orthonormale Basis des Vektorraums bestimmt.

```python
# Ergänzung: Gram-Schmidt-Algorithmus mit exakter Rechnung

import sympy as sp

v1 = sp.Matrix([1, 1, 0])
v2 = sp.Matrix([1, 0, 1])
v3 = sp.Matrix([0, 1, 1])

vektoren = [v1, v2, v3]
orthonormal = []

for v in vektoren:
    w = v.copy()

    for u in orthonormal:
        w = sp.simplify(w - (v.dot(u)) * u)

    u = sp.simplify(w / sp.sqrt(w.dot(w)))
    orthonormal.append(u)

for i, u in enumerate(orthonormal, start=1):
    print(f"u_{i} =")
    sp.pprint(u)
    print()
```

## Orthogonales Komplement eines Vektorraums

### Das orthogonale Komplement eines Vektorraums

**Definition:**

Das orthogonale Komplement eines Unterraums
$U\subseteq\mathbb{R}^n$ ist die Menge aller Vektoren in
$\mathbb{R}^n$, die zu allen Vektoren in $U$ orthogonal sind.

**Notation:**

$$
U^\perp
=
\left\{
\mathbf{v}\in\mathbb{R}^n
\mid
\langle\mathbf{v},\mathbf{u}\rangle=0
\;\forall\,\mathbf{u}\in U
\right\}.
$$

**Wichtig:**

- $U^\perp$ ist ebenfalls ein Unterraum von $\mathbb{R}^n$.
- Die Dimensionen von $U$ und $U^\perp$ erfüllen:

$$
\dim(U)+\dim(U^\perp)=n.
$$

### Geometrische Intuition

**Anschaulich:**

- Das orthogonale Komplement $U^\perp$ enthält alle Vektoren, die senkrecht zu jedem Vektor im Unterraum $U$ stehen.
- Beispiel in $\mathbb{R}^3$:
  - $U$: Eine Ebene durch den Ursprung.
  - $U^\perp$: Die Gerade senkrecht zur Ebene durch den Ursprung.

**Eigenschaften:**

- Jeder Vektor $\mathbf{v}\in\mathbb{R}^n$ kann eindeutig geschrieben werden als:

$$
\mathbf{v}
=
\mathbf{u}+\mathbf{w},
\qquad
\mathbf{u}\in U,
\quad
\mathbf{w}\in U^\perp.
$$

- Dies ist die Projektion von $\mathbf{v}$ auf $U$ und $U^\perp$.

### Beispiel: Orthogonales Komplement in $\mathbb{R}^3$

**Gegeben:**

Der Unterraum $U$ wird durch den Vektor

$$
\mathbf{a}
=
(1,2,3)^T
$$

aufgespannt:

$$
U=\operatorname{span}(\mathbf{a}).
$$

**Aufgabe:** Bestimme $U^\perp$.

Ein Vektor

$$
\mathbf{v}=(x,y,z)
$$

liegt in $U^\perp$, wenn

$$
\langle\mathbf{v},\mathbf{a}\rangle=0.
$$

Damit gilt:

$$
x\cdot1+y\cdot2+z\cdot3=0.
$$

Die Gleichung

$$
x+2y+3z=0
$$

beschreibt $U^\perp$, das eine Ebene in $\mathbb{R}^3$ ist.

```python
# Ergänzung: Basis des orthogonalen Komplements bestimmen

import sympy as sp

A = sp.Matrix([[1, 2, 3]])

basis_U_perp = A.nullspace()

print("Eine Basis von U^⊥ ist:")
for v in basis_U_perp:
    sp.pprint(v)
```

## Direkte Summe von Unterräumen

### Summe und direkte Summe von Unterräumen

**Definition (Summe von Unterräumen):**

Seien $U$ und $W$ Unterräume eines Vektorraums $V$.

Die *Summe* der Unterräume ist definiert als

$$
U+W
=
\{
\mathbf{u}+\mathbf{w}
\mid
\mathbf{u}\in U,\;
\mathbf{w}\in W
\}.
$$

**Definition (direkte Summe):**

Die Summe $U+W$ heißt *direkte Summe* und wird mit
$U\oplus W$ bezeichnet, wenn

$$
U\cap W=\{\mathbf{0}\}.
$$

In diesem Fall schreiben wir

$$
V=U\oplus W.
$$

**Eigenschaft:**

Ist

$$
V=U\oplus W,
$$

so lässt sich jeder Vektor $\mathbf{v}\in V$ eindeutig als

$$
\mathbf{v}=\mathbf{u}+\mathbf{w}
$$

mit $\mathbf{u}\in U$ und $\mathbf{w}\in W$ schreiben.

### Geometrische Intuition

**Anschaulich:**

- Die direkte Summe $U\oplus W$ ist die Kombination zweier Unterräume, die nur den Nullvektor gemeinsam haben.
- Beispiel in $\mathbb{R}^3$:
  - $U$: Eine Gerade durch den Ursprung.
  - $W$: Eine Ebene durch den Ursprung, die orthogonal zu $U$ ist.
  - $U\oplus W=\mathbb{R}^3$.

**Dimension der direkten Summe:**

$$
\dim(U\oplus W)
=
\dim(U)+\dim(W).
$$

### Dimensionssatz für Summen von Unterräumen

**Satz (Dimensionsformel):**

Seien $U$ und $W$ Unterräume eines endlichdimensionalen Vektorraums $V$. Dann gilt

$$
\dim(U+W)
=
\dim(U)+\dim(W)-\dim(U\cap W).
$$

**Folgerung (direkte Summe):**

Ist

$$
V=U\oplus W,
$$

d.h.

$$
V=U+W
\qquad\text{und}\qquad
U\cap W=\{0\},
$$

so folgt

$$
\dim(V)
=
\dim(U\oplus W)
=
\dim(U)+\dim(W).
$$

**Interpretation:**

- Die Dimension des Schnitts $U\cap W$ wird doppelt gezählt und muss daher abgezogen werden.
- Bei einer direkten Summe ist der Schnitt trivial, daher muss nichts abgezogen werden.

### Beispiel: Direkte Summe in $\mathbb{R}^3$

**Gegeben:**

- Unterraum

$$
U=\operatorname{span}(\mathbf{u}),
\qquad
\mathbf{u}=(1,0,0)^T.
$$

- Unterraum

$$
W=\operatorname{span}(\mathbf{v},\mathbf{w}),
$$

mit

$$
\mathbf{v}=(0,1,0)^T,
\qquad
\mathbf{w}=(0,0,1)^T.
$$

**Direkte Summe:**

$$
U\oplus W=\mathbb{R}^3,
\qquad
\dim(U\oplus W)
=
\dim(U)+\dim(W)
=
1+2
=
3.
$$

**Interpretation:**

Jeder Vektor in $\mathbb{R}^3$ lässt sich eindeutig schreiben als

$$
\mathbf{v}
=
c_1\mathbf{u}
+
c_2\mathbf{v}
+
c_3\mathbf{w},
\qquad
c_1,c_2,c_3\in\mathbb{R}.
$$

### Weiteres Beispiel: Direkte Summe in $\mathbb{R}^3$

**Gegeben:**

- Unterraum

$$
U=\operatorname{span}(\mathbf{u}_1,\mathbf{u}_2),
$$

mit

$$
\mathbf{u}_1=(1,1,0)^T,
\qquad
\mathbf{u}_2=(0,1,1)^T.
$$

- Unterraum

$$
W=\operatorname{span}(\mathbf{w}),
$$

mit

$$
\mathbf{w}=(1,-1,1)^T.
$$

**Direkte Summe:**

- Überprüfung:

$$
U\cap W=\{\mathbf{0}\},
$$

da kein nichttrivialer Vektor aus $U$ auch in $W$ liegt.

- Dimensionen:

$$
\dim(U)=2,
\qquad
\dim(W)=1,
\qquad
\dim(U\oplus W)
=
\dim(U)+\dim(W)
=
3.
$$

**Interpretation:**

Jeder Vektor in $\mathbb{R}^3$ lässt sich eindeutig schreiben als

$$
\mathbf{v}
=
c_1\mathbf{u}_1
+
c_2\mathbf{u}_2
+
c_3\mathbf{w},
\qquad
c_1,c_2,c_3\in\mathbb{R}.
$$

**Anschaulich:**

Der Raum $\mathbb{R}^3$ ist die direkte Summe von $U$ (einer Ebene) und $W$ (einer Geraden, die nicht in der Ebene $U$ liegt).

### Überprüfung: $U\cap W=\{\mathbf{0}\}$

**Zu prüfen:**

$$
U\cap W=\{\mathbf{0}\}.
$$

Ein Vektor $\mathbf{v}\in U\cap W$ ist sowohl Linearkombination von
$\mathbf{u}_1,\mathbf{u}_2$ als auch von $\mathbf{w}$:

$$
\mathbf{v}
=
\lambda_1\mathbf{u}_1
+
\lambda_2\mathbf{u}_2
=
\lambda_3\mathbf{w}.
$$

Komponentenweise:

$$
\lambda_1
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
0\\
1\\
1
\end{pmatrix}
-
\lambda_3
\begin{pmatrix}
1\\
-1\\
1
\end{pmatrix}
=
\begin{pmatrix}
0\\
0\\
0
\end{pmatrix}.
$$

Daraus folgt das Gleichungssystem:

$$
\begin{aligned}
\lambda_1-\lambda_3 &= 0,\\
\lambda_1+\lambda_2+\lambda_3 &= 0,\\
\lambda_2-\lambda_3 &= 0.
\end{aligned}
$$

### Lösung des Systems mit Gauß-Elimination

Matrixdarstellung des Systems:

$$
\left[
\begin{array}{c|ccc|c}
\text{I}   & 1 & 0 & -1 & 0\\
\text{II}  & 1 & 1 &  1 & 0\\
\text{III} & 0 & 1 & -1 & 0
\end{array}
\right].
$$

Erster Eliminationsschritt:

$$
\begin{array}{r|ccc|c|l}
\text{I}    & 1 & 0 & -1 & 0 &\\
\text{II'}  & 0 & 1 & 2  & 0 & \text{II'}=\text{II}-\text{I}\\
\text{III}  & 0 & 1 & -1 & 0 &
\end{array}
$$

Danach:

$$
\begin{array}{r|ccc|c|l}
\text{I}    & 1 & 0 & -1 & 0 &\\
\text{II'}  & 0 & 1 & 2  & 0 &\\
\text{III'} & 0 & 0 & -3 & 0 & \text{III'}=\text{III}-\text{II'}
\end{array}.
$$

### Schlussfolgerung

- Aus Zeile $\text{III'}$:

$$
-3\lambda_3=0
\quad\Rightarrow\quad
\lambda_3=0.
$$

- Einsetzen in Zeile $\text{II'}$:

$$
\lambda_2+2\cdot0=0
\quad\Rightarrow\quad
\lambda_2=0.
$$

- Einsetzen in Zeile $\text{I}$:

$$
\lambda_1-0=0
\quad\Rightarrow\quad
\lambda_1=0.
$$

Der einzige Vektor, der sowohl in $U$ als auch in $W$ liegt, ist der Nullvektor.

Es folgt:

$$
U\cap W=\{\mathbf{0}\}.
$$

$U$ und $W$ haben keine nichttrivialen gemeinsamen Vektoren. Die direkte Summe $U\oplus W$ ist wohldefiniert.

```python
# Ergänzung: Überprüfung der direkten Summe

import sympy as sp

u1 = sp.Matrix([1, 1, 0])
u2 = sp.Matrix([0, 1, 1])
w  = sp.Matrix([1, -1, 1])

A = sp.Matrix.hstack(u1, u2, w)

print("Matrix aus den Basisvektoren:")
sp.pprint(A)

print("\nRang:", A.rank())
print("Determinante:", A.det())

if A.rank() == 3:
    print("\nDie drei Vektoren sind linear unabhängig.")
    print("Damit gilt U ⊕ W = R^3.")
```

## Dimensionstheorem für Unterräume

### Das Dimensionstheorem und seine Implikationen

**Dimensionstheorem:**

Für einen Unterraum $U\subseteq\mathbb{R}^n$ und dessen orthogonales Komplement $U^\perp$ gilt:

$$
\dim(U)+\dim(U^\perp)=n.
$$

**Wichtige Fragen:**

- Welche Konsequenzen hat dies für die Struktur von $\mathbb{R}^n$?
- Wie hilft das Dimensionstheorem bei der Analyse von Unterräumen?

### Zerlegung von $\mathbb{R}^n$

**Hauptaussage:**

Jeder Vektor $\mathbf{v}\in\mathbb{R}^n$ kann eindeutig in einen Teil in $U$ und einen Teil in $U^\perp$ zerlegt werden:

$$
\mathbf{v}
=
\mathbf{u}
+
\mathbf{w},
\qquad
\mathbf{u}\in U,
\quad
\mathbf{w}\in U^\perp.
$$

**Konsequenzen:**

- Der Raum $\mathbb{R}^n$ ist die direkte Summe der beiden Unterräume:

$$
\mathbb{R}^n
=
U\oplus U^\perp.
$$

- Projektionen: Die Projektion eines Vektors auf $U$ und $U^\perp$ ist wohldefiniert.

### Erweiterung der Basis

**Hauptaussage:**

Eine Basis für $U\subset V$ kann erweitert werden, um eine Basis für $V$ zu bilden.

- Gegeben eine Basis

$$
\{\mathbf{u}_1,\ldots,\mathbf{u}_k\}
$$

von $U$, können wir

$$
\{\mathbf{v}_1,\ldots,\mathbf{v}_{n-k}\}
$$

für $U^\perp$ finden.

- Die Vereinigung

$$
\{
\mathbf{u}_1,\ldots,\mathbf{u}_k,
\mathbf{v}_1,\ldots,\mathbf{v}_{n-k}
\}
$$

ist dann eine Basis für $V$.

## Dimensionstheorems für lineare Gleichungssysteme

Für eine Matrix

$$
A\in\mathbb{R}^{m\times n}
$$

gilt:

$$
\dim(\operatorname{Kern}(A))
+
\dim(\operatorname{Bild}(A))
=
n.
$$

Dabei ist

- $\operatorname{Kern}(A)$: der Lösungsraum des homogenen Systems

$$
A\mathbf{x}=\mathbf{0},
$$

- $\operatorname{Bild}(A)$: der Raum der durch $A$ erreichbaren Vektoren (Spaltenraum).

### Implikationen für lineare Systeme

1. **Homogenes System**

   Das homogene System

   $$
   A\mathbf{x}=\mathbf{0}
   $$

   hat Lösungen, wenn

   $$
   \dim(\operatorname{Kern}(A))>0.
   $$

   Die Anzahl der Freiheitsgrade der Lösung entspricht

   $$
   \dim(\operatorname{Kern}(A)).
   $$

2. **Inhomogenes System**

   Das inhomogene System

   $$
   A\mathbf{x}=\mathbf{b}
   $$

   ist lösbar, wenn

   $$
   \mathbf{b}\in\operatorname{Bild}(A).
   $$

   Die Lösungsmenge hat dieselbe Dimension wie $\operatorname{Kern}(A)$.

   Wenn

   $$
   \operatorname{Rang}(A)=n,
   $$

   gibt es genau eine Lösung; andernfalls ist die Lösungsmenge eine affine Untermenge mit der Dimension

   $$
   \dim(\operatorname{Kern}(A)).
   $$

3. **Volle Rang-Bedingung**

   Wenn

   $$
   \operatorname{Rang}(A)=n,
   $$

   ist

   $$
   \dim(\operatorname{Kern}(A))=0,
   $$

   d.h. das homogene System hat nur die triviale Lösung.

### Geometrische Interpretation

**Homogenes System:**

- Die Lösungen von

$$
A\mathbf{x}=\mathbf{0}
$$

bilden einen Unterraum, $\operatorname{Kern}(A)$, im $\mathbb{R}^n$.

- Dieser Unterraum ist orthogonal zum Raum $\operatorname{Bild}(A^\top)$, der von den Zeilen der Matrix $\mathbf{A}$ aufgespannt wird:

$$
\mathbb{R}^n
=
\operatorname{Kern}(A)
\oplus
\operatorname{Bild}(A^\top).
$$

**Inhomogenes System:**

- Für

$$
A\mathbf{x}=\mathbf{b}
$$

ist

$$
\mathbf{b}\in\operatorname{Bild}(A)
$$

eine notwendige Bedingung für die Lösbarkeit.

- Jede Lösung des inhomogenen Systems hat die Form

$$
\mathbf{x}
=
\mathbf{x}_p
+
\mathbf{x}_h,
$$

wobei $\mathbf{x}_p$ eine spezielle Lösung ist und
$\mathbf{x}_h\in\operatorname{Kern}(A)$ eine Lösung des homogenen Systems.

### Beispiel: Lösung durch Gauß-Elimination

**Gegeben:**

$$
A=
\begin{bmatrix}
1 & 2 & -1\\
2 & 4 & -2\\
1 & 0 & 1
\end{bmatrix},
\qquad
\mathbf{x}
=
\begin{bmatrix}
x_1\\
x_2\\
x_3
\end{bmatrix},
\qquad
\mathbf{b}
=
\begin{bmatrix}
1\\
2\\
1
\end{bmatrix}.
$$

Wir lösen das inhomogene System

$$
A\mathbf{x}=\mathbf{b}
$$

mit Gauß-Elimination:

$$
\begin{array}{ccc|c}
1 & 2 & -1 & 1\\
2 & 4 & -2 & 2\\
1 & 0 & 1 & 1
\end{array}
$$

$$
\begin{array}{ccc|c|l}
1 & 2 & -1 & 1 &\\
0 & 0 & 0 & 0 & \text{II'}=\text{II}-2\text{I}\\
0 & -2 & 2 & 0 & \text{III'}=\text{III}-\text{I}
\end{array}.
$$

### Lösung des Gleichungssystems

Aus der Zeilenstufenform lesen wir:

$$
\begin{aligned}
x_1+2x_2-x_3 &= 1,\\
-2x_2+2x_3 &= 0.
\end{aligned}
$$

**Parameterisierung:**

Setze

$$
x_3=t,
\qquad
t\in\mathbb{R}.
$$

Aus der zweiten Gleichung folgt

$$
x_2=x_3=t.
$$

Einsetzen in die erste Gleichung:

$$
x_1+2t-t=1
\quad\Rightarrow\quad
x_1=1-t.
$$

**Lösungsmenge:**

$$
\mathbf{x}
=
\begin{bmatrix}
1-t\\
t\\
t
\end{bmatrix}
=
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix}
+
t
\begin{bmatrix}
-1\\
1\\
1
\end{bmatrix},
\qquad
t\in\mathbb{R}.
$$

### Interpretation der Lösung

**Partikuläre Lösung:**

$$
\mathbf{x}_p
=
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix},
$$

so dass

$$
A\mathbf{x}_p=\mathbf{b}
$$

gilt.

**Kern von $A$:**

Die allgemeine Lösung des homogenen Systems

$$
A\mathbf{x}=\mathbf{0}
$$

ist

$$
\mathbf{x}_h
=
t
\begin{bmatrix}
-1\\
1\\
1
\end{bmatrix},
\qquad
t\in\mathbb{R},
$$

also

$$
\ker(A)
=
\operatorname{span}
\left\{
\begin{bmatrix}
-1\\
1\\
1
\end{bmatrix}
\right\}.
$$

**Gesamte Lösungsmenge:**

$$
\mathbf{x}
=
\mathbf{x}_p+\mathbf{x}_h
=
\mathbf{x}_p
+
t
\begin{bmatrix}
-1\\
1\\
1
\end{bmatrix},
\qquad
t\in\mathbb{R}.
$$

**Geometrische Interpretation:**

- $A$ ist eine quadratische Matrix mit nichttrivialem Kern (eine Gerade in $\mathbb{R}^3$).
- Die Lösungsmenge von $A\mathbf{x}=\mathbf{b}$ ist eine zu $\ker(A)$ parallele Gerade, die durch die partikuläre Lösung $\mathbf{x}_p$ verläuft.

```python
# Ergänzung: Lösung des linearen Gleichungssystems mit SymPy

import sympy as sp

A = sp.Matrix([
    [1, 2, -1],
    [2, 4, -2],
    [1, 0,  1]
])

b = sp.Matrix([1, 2, 1])

x1, x2, x3 = sp.symbols("x1 x2 x3")

loesung = sp.linsolve((A, b), (x1, x2, x3))

print("Lösungsmenge:")
sp.pprint(loesung)

print("\nKern von A:")
for v in A.nullspace():
    sp.pprint(v)

print("\nRang von A:", A.rank())
print("Dimension des Kerns:", len(A.nullspace()))
```
