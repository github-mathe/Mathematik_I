## Lösen eines linearen Gleichungssystems mit dem Gauß-Eliminationsverfahren

Gegeben ist das lineare Gleichungssystem

$$
\begin{aligned}
x_1 + 2x_2 - x_3 &= 3,\\
2x_1 + 5x_2 + x_3 &= 8,\\
-\,x_1 - x_2 + 2x_3 &= -1.
\end{aligned}
$$

Stellen Sie zunächst die zugehörige erweiterte Koeffizientenmatrix auf.

Führen Sie anschließend das **Gauß-Eliminationsverfahren** durch und bringen Sie die Matrix durch elementare Zeilenumformungen in eine **reduzierte Dreiecksform**.

Verwenden Sie dabei nur die folgenden elementaren Zeilenoperationen:

- Vertauschen zweier Zeilen,
- Multiplikation einer Zeile mit einer von null verschiedenen Zahl,
- Addition eines Vielfachen einer Zeile zu einer anderen Zeile.

Dokumentieren Sie jeden Eliminationsschritt.

Bestimmen Sie anschließend die Lösung des linearen Gleichungssystems durch Rückwärtseinsetzen.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Schritt 1 – Erster Eintrag unterhalb des Pivotelements

Ausgangsmatrix:

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
2 & 5 & 1 & 8\\
-1 & -1 & 2 & -1
\end{array}
\right)
$$

Ziel ist es zunächst, den Eintrag $2$ in der ersten Spalte zu eliminieren.

Dazu verwenden wir die Zeilenoperation

$$
Z_2 \leftarrow Z_2-2Z_1.
$$

Es ergibt sich

$$
Z_2:
\qquad
(2,\;5,\;1\mid8)
-
2(1,\;2,\;-1\mid3)
$$

und damit

$$
Z_2=(0,\;1,\;3\mid2).
$$

Die Matrix lautet nun:

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & 3 & 2\\
-1 & -1 & 2 & -1
\end{array}
\right)
}
$$
:::

:::{dropdown} Schritt 2 – Zweiter Eintrag in der ersten Spalte eliminieren

Ausgehend von

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & 3 & 2\\
-1 & -1 & 2 & -1
\end{array}
\right)
$$

soll nun der Eintrag $-1$ in der ersten Spalte eliminiert werden.

Dazu verwenden wir

$$
Z_3 \leftarrow Z_3+Z_1.
$$

Es ergibt sich

$$
Z_3:
\qquad
(-1,\;-1,\;2\mid-1)
+
(1,\;2,\;-1\mid3)
$$

und damit

$$
Z_3=(0,\;1,\;1\mid2).
$$

Die Matrix lautet jetzt:

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & 3 & 2\\
0 & 1 & 1 & 2
\end{array}
\right)
}
$$
:::

:::{dropdown} Schritt 3 – Eintrag unterhalb des zweiten Pivotelements eliminieren

Nun soll der Eintrag $1$ in der zweiten Spalte der dritten Zeile eliminiert werden.

Dazu verwenden wir

$$
Z_3 \leftarrow Z_3-Z_2.
$$

Es ergibt sich

$$
Z_3:
\qquad
(0,\;1,\;1\mid2)
-
(0,\;1,\;3\mid2)
$$

und damit

$$
Z_3=(0,\;0,\;-2\mid0).
$$

Damit erhalten wir die obere Dreiecksmatrix

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & 3 & 2\\
0 & 0 & -2 & 0
\end{array}
\right)
}
$$
:::
