## Kern und Bild einer linearen Abbildung

Gegeben sei die lineare Abbildung

$$
T:\mathbb{R}^3\to\mathbb{R}^3
$$

mit

$$
T
\begin{pmatrix}
x\\
y\\
z
\end{pmatrix}
=
\begin{pmatrix}
x+2y+3z\\
y+z\\
x+y+2z
\end{pmatrix}.
$$

1. Bestimmen Sie die Matrix $\mathbf A$, die die lineare Abbildung $T$ beschreibt.
2. Bestimmen Sie den Kern von $T$ sowie eine Basis und die Dimension des Kerns.
3. Bestimmen Sie das Bild von $T$ sowie eine Basis und die Dimension des Bildes.

## Lösung

Die lineare Abbildung wird zunächst durch ihre Matrix $\mathbf A$ dargestellt.

Anschließend bestimmen wir den Kern über das homogene Gleichungssystem

$$
\mathbf A\vec{x}=\vec{0}
$$

und das Bild über die linear unabhängigen Spalten von $\mathbf A$.

:::{dropdown} Matrix der linearen Abbildung

Aus

$$
T
\begin{pmatrix}
x\\
y\\
z
\end{pmatrix}
=
\begin{pmatrix}
x+2y+3z\\
y+z\\
x+y+2z
\end{pmatrix}
$$

erhalten wir die Matrix

$$
\boxed{
\mathbf A
=
\begin{pmatrix}
1 & 2 & 3\\
0 & 1 & 1\\
1 & 1 & 2
\end{pmatrix}.
}
$$

Damit gilt

$$
T(\vec{x})=\mathbf A\vec{x}.
$$
:::

:::{dropdown} Kern bestimmen

Für den Kern suchen wir alle Vektoren $\vec{x}\in\mathbb{R}^3$, für die

$$
T(\vec{x})=\vec{0}
$$

gilt.

Damit lösen wir das homogene Gleichungssystem

$$
\mathbf A\vec{x}=\vec{0}.
$$

Die erweiterte Koeffizientenmatrix lautet

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & 1 & 1 & 0\\
1 & 1 & 2 & 0
\end{array}
\right).
$$

Der Eintrag in der ersten Spalte der dritten Zeile wird eliminiert:

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & 1 & 1 & 0\\
1 & 1 & 2 & 0
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}'=\mathrm{III}-\mathrm{I}
\end{array}
$$

Damit ergibt sich

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & 1 & 1 & 0\\
0 & -1 & -1 & 0
\end{array}
\right).
$$

Nun wird der Eintrag in der zweiten Spalte der dritten Zeile eliminiert:

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & 1 & 1 & 0\\
0 & -1 & -1 & 0
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}''=\mathrm{III}'+\mathrm{II}
\end{array}
$$

Damit erhalten wir

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & 1 & 1 & 0\\
0 & 0 & 0 & 0
\end{array}
\right).
}
$$

Das zugehörige Gleichungssystem lautet

$$
\begin{aligned}
x+2y+3z&=0,\\
y+z&=0.
\end{aligned}
$$

Die Variable $z$ ist frei. Wir setzen

$$
z=t.
$$

Aus

$$
y+z=0
$$

folgt

$$
y=-t.
$$

Aus

$$
x+2y+3z=0
$$

folgt

$$
x+2(-t)+3t=0
$$

und damit

$$
x=-t.
$$

Somit ist die Lösungsmenge

$$
\left\{
\begin{pmatrix}
-t\\
-t\\
t
\end{pmatrix}
\;\middle|\;
t\in\mathbb{R}
\right\}.
$$

Damit gilt

$$
\boxed{
\operatorname{Kern}(T)
=
\operatorname{span}
\left\{
\begin{pmatrix}
-1\\
-1\\
1
\end{pmatrix}
\right\}.
}
$$

Eine Basis des Kerns ist

$$
\boxed{
\left\{
\begin{pmatrix}
-1\\
-1\\
1
\end{pmatrix}
\right\}.
}
$$

Damit ist

$$
\boxed{
\dim(\operatorname{Kern}(T))=1.
}
$$
:::

:::{dropdown} Bild bestimmen

Das Bild der linearen Abbildung entspricht dem Spaltenraum der Matrix

$$
\mathbf A
=
\begin{pmatrix}
1 & 2 & 3\\
0 & 1 & 1\\
1 & 1 & 2
\end{pmatrix}.
$$

Zur Bestimmung der linear unabhängigen Spalten führen wir die Gauß-Elimination durch.

$$
\left(
\begin{array}{ccc}
1 & 2 & 3\\
0 & 1 & 1\\
1 & 1 & 2
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}'=\mathrm{III}-\mathrm{I}
\end{array}
$$

Damit ergibt sich

$$
\left(
\begin{array}{ccc}
1 & 2 & 3\\
0 & 1 & 1\\
0 & -1 & -1
\end{array}
\right).
$$

Nun gilt

$$
\left(
\begin{array}{ccc}
1 & 2 & 3\\
0 & 1 & 1\\
0 & -1 & -1
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}''=\mathrm{III}'+\mathrm{II}
\end{array}
$$

und damit

$$
\boxed{
\left(
\begin{array}{ccc}
1 & 2 & 3\\
0 & 1 & 1\\
0 & 0 & 0
\end{array}
\right).
}
$$

Die Pivot-Spalten sind die erste und die zweite Spalte.

Damit ist

$$
\operatorname{Rang}(\mathbf A)=2.
$$

Für eine Basis des Bildes verwenden wir die entsprechenden Spalten der **ursprünglichen Matrix** $\mathbf A$.

Somit ist eine Basis von $\operatorname{Bild}(T)$

$$
\boxed{
\left\{
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\begin{pmatrix}
2\\
1\\
1
\end{pmatrix}
\right\}.
}
$$

Damit gilt

$$
\boxed{
\operatorname{Bild}(T)
=
\operatorname{span}
\left\{
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\begin{pmatrix}
2\\
1\\
1
\end{pmatrix}
\right\}.
}
$$

Die Dimension des Bildes ist

$$
\boxed{
\dim(\operatorname{Bild}(T))=2.
}
$$
:::

Für den Kern ergibt sich

$$
\boxed{

\operatorname{span}
\left\{
\begin{pmatrix}
-1\\
-1\\
1
\end{pmatrix}
\right\}

$$

mit

$$
\boxed{
\dim(\operatorname{Kern}(T))=1.
}

$$
\boxed{

1
\end{pmatrix},
\begin{pmatrix}

1

\right\}
}
$$

$$
\boxed{
\dim(\operatorname{Bild}(T))=2.
}
$$
