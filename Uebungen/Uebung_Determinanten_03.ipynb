## Determinante einer Dreiecksmatrix

Gegeben sei die Matrix

$$
\mathbf{A}
=
\begin{pmatrix}
2 & -1 & 3 & 4\\
0 & -3 & 2 & 1\\
0 & 0 & 4 & -2\\
0 & 0 & 0 & 5
\end{pmatrix}.
$$

1. Begründen Sie, warum es sich bei $\mathbf{A}$ um eine Dreiecksmatrix handelt.
2. Berechnen Sie die Determinante

$$
\det(\mathbf{A}).
$$

## Lösung

Bei einer oberen Dreiecksmatrix sind alle Einträge unterhalb der Hauptdiagonalen gleich null.

Für eine Dreiecksmatrix ist die Determinante das Produkt der Diagonalelemente.

:::{dropdown} Berechnung der Determinante

Die Matrix

$$
\mathbf{A}
=
\begin{pmatrix}
2 & -1 & 3 & 4\\
0 & -3 & 2 & 1\\
0 & 0 & 4 & -2\\
0 & 0 & 0 & 5
\end{pmatrix}
$$

ist eine **obere Dreiecksmatrix**, da alle Einträge unterhalb der Hauptdiagonalen gleich null sind.

Die Diagonalelemente sind

$$
2,\qquad -3,\qquad 4,\qquad 5.
$$

Damit ergibt sich die Determinante als Produkt der Diagonalelemente:

$$
\begin{aligned}
\det(\mathbf{A})
&=
2\cdot(-3)\cdot4\cdot5\\
&=
-120.
\end{aligned}
$$

Somit gilt

$$
\boxed{
\det(\mathbf{A})=-120.
}
$$
:::
