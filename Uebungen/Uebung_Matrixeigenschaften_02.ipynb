## Eigenschaften von Matrizen

Gegeben seien die Matrizen

$$
\mathbf A
=
\begin{pmatrix}
2 & -1\\
-1 & 3
\end{pmatrix},
\qquad
\mathbf B
=
\begin{pmatrix}
0 & -1\\
1 & 0
\end{pmatrix},
$$

$$
\mathbf C
=
\begin{pmatrix}
2 & 1+i\\
1-i & 3
\end{pmatrix},
\qquad
\mathbf D
=
\frac{1}{\sqrt{2}}
\begin{pmatrix}
1 & i\\
i & 1
\end{pmatrix}.
$$

Untersuchen Sie die Matrizen auf folgende Eigenschaften:

- symmetrisch,
- orthogonal,
- hermitesch,
- unitär.

Begründen Sie Ihre Entscheidung jeweils mithilfe der entsprechenden Definition.

Bestimmen Sie zusätzlich, soweit sinnvoll,

- die transponierte bzw. adjungierte Matrix,
- die inverse Matrix.

## Lösung

Die Matrizen werden anhand der jeweiligen Definition überprüft.

:::{dropdown} Matrix $\mathbf A$

Gegeben ist

$$
\mathbf A
=
\begin{pmatrix}
2 & -1\\
-1 & 3
\end{pmatrix}.
$$

Die transponierte Matrix lautet

$$
\mathbf A^T
=
\begin{pmatrix}
2 & -1\\
-1 & 3
\end{pmatrix}
=
\mathbf A.
$$

Damit ist $\mathbf A$ symmetrisch.

Zur Überprüfung der Orthogonalität berechnen wir

$$
\mathbf A^T\mathbf A
=
\begin{pmatrix}
5 & -5\\
-5 & 10
\end{pmatrix}
\neq
\mathbf I_2.
$$

Damit ist $\mathbf A$ nicht orthogonal.

Die inverse Matrix ist

$$
\boxed{
\mathbf A^{-1}
=
\frac{1}{5}
\begin{pmatrix}
3 & 1\\
1 & 2
\end{pmatrix}.
}
$$
:::

:::{dropdown} Matrix $\mathbf B$

Gegeben ist

$$
\mathbf B
=
\begin{pmatrix}
0 & -1\\
1 & 0
\end{pmatrix}.
$$

Die transponierte Matrix lautet

$$
\mathbf B^T
=
\begin{pmatrix}
0 & 1\\
-1 & 0
\end{pmatrix}.
$$

Damit gilt

$$
\mathbf B^T\neq\mathbf B,
$$

also ist $\mathbf B$ nicht symmetrisch.

Weiter gilt

$$
\mathbf B^T\mathbf B
=
\begin{pmatrix}
1 & 0\\
0 & 1
\end{pmatrix}
=
\mathbf I_2
$$

und

$$
\mathbf B\mathbf B^T
=
\mathbf I_2.
$$

Damit ist $\mathbf B$ orthogonal.

Für eine orthogonale Matrix gilt

$$
\boxed{
\mathbf B^{-1}
=
\mathbf B^T
=
\begin{pmatrix}
0 & 1\\
-1 & 0
\end{pmatrix}.
}
$$
:::

:::{dropdown} Matrix $\mathbf C$

Gegeben ist

$$
\mathbf C
=
\begin{pmatrix}
2 & 1+i\\
1-i & 3
\end{pmatrix}.
$$

Die konjugierte Matrix lautet

$$
\overline{\mathbf C}
=
\begin{pmatrix}
2 & 1-i\\
1+i & 3
\end{pmatrix}.
$$

Damit ergibt sich die adjungierte Matrix

$$
\mathbf C^*
=
\overline{\mathbf C}^{\,T}
=
\begin{pmatrix}
2 & 1+i\\
1-i & 3
\end{pmatrix}.
$$

Es gilt also

$$
\mathbf C^*=\mathbf C.
$$

Damit ist $\mathbf C$ hermitesch.

Zur Überprüfung der Unitarität berechnen wir

$$
\mathbf C^*\mathbf C
=
\begin{pmatrix}
6 & 5+5i\\
5-5i & 11
\end{pmatrix}
\neq
\mathbf I_2.
$$

Damit ist $\mathbf C$ nicht unitär.

Die inverse Matrix ist

$$
\boxed{
\mathbf C^{-1}
=
\frac{1}{4}
\begin{pmatrix}
3 & -1-i\\
-1+i & 2
\end{pmatrix}.
}
$$
:::

:::{dropdown} Matrix $\mathbf D$

Gegeben ist

$$
\mathbf D
=
\frac{1}{\sqrt{2}}
\begin{pmatrix}
1 & i\\
i & 1
\end{pmatrix}.
$$

Die adjungierte Matrix lautet

$$
\mathbf D^*
=
\frac{1}{\sqrt{2}}
\begin{pmatrix}
1 & -i\\
-i & 1
\end{pmatrix}.
$$

Damit gilt

$$
\begin{aligned}
\mathbf D^*\mathbf D
&=
\frac{1}{2}
\begin{pmatrix}
1 & -i\\
-i & 1
\end{pmatrix}
\begin{pmatrix}
1 & i\\
i & 1
\end{pmatrix}\\
&=
\begin{pmatrix}
1 & 0\\
0 & 1
\end{pmatrix}\\
&=
\mathbf I_2.
\end{aligned}
$$

Ebenso gilt

$$
\mathbf D\mathbf D^*
=
\mathbf I_2.
$$

Damit ist $\mathbf D$ unitär.

Außerdem gilt

$$
\mathbf D^*\neq\mathbf D,
$$

also ist $\mathbf D$ nicht hermitesch.

Für die inverse Matrix gilt

$$
\boxed{
\mathbf D^{-1}
=
\mathbf D^*
=
\frac{1}{\sqrt{2}}
\begin{pmatrix}
1 & -i\\
-i & 1
\end{pmatrix}.
}
$$
:::
