## Komplexes Skalarprodukt

Gegeben seien die Vektoren

$$
\vec{x}
=
\begin{pmatrix}
1+i\\
2
\end{pmatrix},
\qquad
\vec{y}
=
\begin{pmatrix}
1\\
1-i
\end{pmatrix}
\in\mathbb{C}^2.
$$

Bearbeiten Sie die folgenden Aufgaben:

1. Berechnen Sie das komplexe Skalarprodukt

   $$
   \langle\vec{x},\vec{y}\rangle
   =
   \vec{x}^{\,*}\vec{y}.
   $$

2. Bestimmen Sie die Normen der beiden Vektoren mithilfe des Skalarprodukts:

   $$
   \|\vec{x}\|
   =
   \sqrt{\langle\vec{x},\vec{x}\rangle},
   \qquad
   \|\vec{y}\|
   =
   \sqrt{\langle\vec{y},\vec{y}\rangle}.
   $$

3. Überprüfen Sie, ob die beiden Vektoren orthogonal zueinander sind.

Verwenden Sie bei der Multiplikation komplexer Zahlen die Rechenregel

$$
(a+bi)(c+di)
=
(ac-bd)+(ad+bc)i.
$$

## Lösung

Für das komplexe Skalarprodukt gilt

$$
\langle\vec{x},\vec{y}\rangle
=
\vec{x}^{\,*}\vec{y}.
$$

Dazu wird zunächst der erste Vektor konjugiert-transponiert.

:::{dropdown} Bestimmung von $\vec{x}^{\,*}$

Gegeben ist

$$
\vec{x}
=
\begin{pmatrix}
1+i\\
2
\end{pmatrix}.
$$

Die komplex konjugierten Einträge sind

$$
\overline{1+i}=1-i
$$

und

$$
\overline{2}=2.
$$

Damit ergibt sich

$$
\boxed{
\vec{x}^{\,*}
=
\begin{pmatrix}
1-i & 2
\end{pmatrix}.
}
$$
:::

:::{dropdown} Berechnung des Skalarprodukts

Nach der Definition gilt

$$
\langle\vec{x},\vec{y}\rangle
=
\vec{x}^{\,*}\vec{y}.
$$

Damit erhalten wir

$$
\langle\vec{x},\vec{y}\rangle
=
\begin{pmatrix}
1-i & 2
\end{pmatrix}
\begin{pmatrix}
1\\
1-i
\end{pmatrix}.
$$

Also

$$
\langle\vec{x},\vec{y}\rangle
=
(1-i)\cdot1
+
2\cdot(1-i).
$$

Für den ersten Term gilt

$$
(1-i)\cdot1
=
1-i.
$$

Für den zweiten Term gilt

$$
2\cdot(1-i)
=
2-2i.
$$

Damit folgt

$$
\begin{aligned}
\langle\vec{x},\vec{y}\rangle
&=
(1-i)+(2-2i)\\
&=
3-3i.
\end{aligned}
$$

Somit gilt

$$
\boxed{
\langle\vec{x},\vec{y}\rangle
=
3-3i.
}
$$
:::

:::{dropdown} Berechnung der Norm von $\vec{x}$

Für die Norm gilt

$$
\|\vec{x}\|
=
\sqrt{\langle\vec{x},\vec{x}\rangle}.
$$

Zunächst berechnen wir

$$
\langle\vec{x},\vec{x}\rangle
=
\vec{x}^{\,*}\vec{x}.
$$

Damit ergibt sich

$$
\langle\vec{x},\vec{x}\rangle
=
\begin{pmatrix}
1-i & 2
\end{pmatrix}
\begin{pmatrix}
1+i\\
2
\end{pmatrix}.
$$

Also

$$
\langle\vec{x},\vec{x}\rangle
=
(1-i)(1+i)+2\cdot2.
$$

Für das Produkt

$$
(1-i)(1+i)
$$

verwenden wir die Rechenregel

$$
(a+bi)(c+di)
=
(ac-bd)+(ad+bc)i.
$$

Damit gilt

$$
\begin{aligned}
(1-i)(1+i)
&=
(1\cdot1-(-1)\cdot1)
+
(1\cdot1+(-1)\cdot1)i\\
&=
2.
\end{aligned}
$$

Somit erhalten wir

$$
\langle\vec{x},\vec{x}\rangle
=
2+4
=
6.
$$

Daher ist

$$
\boxed{
\|\vec{x}\|
=
\sqrt{6}.
}
$$
:::

:::{dropdown} Berechnung der Norm von $\vec{y}$

Gegeben ist

$$
\vec{y}
=
\begin{pmatrix}
1\\
1-i
\end{pmatrix}.
$$

Der konjugiert-transponierte Vektor lautet

$$
\vec{y}^{\,*}
=
\begin{pmatrix}
1 & 1+i
\end{pmatrix}.
$$

Damit gilt

$$
\langle\vec{y},\vec{y}\rangle
=
\vec{y}^{\,*}\vec{y}.
$$

Also

$$
\langle\vec{y},\vec{y}\rangle
=
1\cdot1
+
(1+i)(1-i).
$$

Für das komplexe Produkt gilt

$$
\begin{aligned}
(1+i)(1-i)
&=
(1\cdot1-1\cdot(-1))
+
(1\cdot(-1)+1\cdot1)i\\
&=
2.
\end{aligned}
$$

Damit folgt

$$
\langle\vec{y},\vec{y}\rangle
=
1+2
=
3.
$$

Daher ist

$$
\boxed{
\|\vec{y}\|
=
\sqrt{3}.
}
$$
:::

:::{dropdown} Überprüfung der Orthogonalität

Zwei komplexe Vektoren sind orthogonal zueinander, wenn

$$
\langle\vec{x},\vec{y}\rangle=0.
$$

Wir haben erhalten:

$$
\langle\vec{x},\vec{y}\rangle
=
3-3i.
$$

Da

$$
3-3i\neq0,
$$

sind die beiden Vektoren nicht orthogonal zueinander.

Damit gilt

$$
\boxed{
\vec{x}\not\perp\vec{y}.
}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Für

$$
\vec{x}
=
\begin{pmatrix}
1+i\\
2
\end{pmatrix},
\qquad
\vec{y}
=
\begin{pmatrix}
1\\
1-i
\end{pmatrix}
$$

gilt nach

$$
\langle\vec{x},\vec{y}\rangle
=
\vec{x}^{\,*}\vec{y}
$$

das Skalarprodukt

$$
\boxed{
\langle\vec{x},\vec{y}\rangle
=
3-3i.
}
$$

Die Normen sind

$$
\boxed{
\|\vec{x}\|=\sqrt{6},
\qquad
\|\vec{y}\|=\sqrt{3}.
}
$$

Da

$$
\langle\vec{x},\vec{y}\rangle\neq0,
$$

sind die Vektoren nicht orthogonal:

$$
\boxed{
\vec{x}\not\perp\vec{y}.
}
$$
:::
