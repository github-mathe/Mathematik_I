## Reelles Skalarprodukt

Gegeben seien die Vektoren

$$
\vec{x}
=
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\qquad
\vec{y}
=
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
\in\mathbb{R}^3.
$$

Bearbeiten Sie die folgenden Aufgaben:

1. Berechnen Sie das Skalarprodukt

   $$
   \langle \vec{x},\vec{y}\rangle.
   $$

2. Bestimmen Sie die Normen der beiden Vektoren mithilfe des Skalarprodukts:

   $$
   \|\vec{x}\|
   =
   \sqrt{\langle\vec{x},\vec{x}\rangle},
   \qquad
   \|\vec{y}\|
   =
   \sqrt{\langle\vec{y},\vec{y}\rangle}.
   $$

3. Überprüfen Sie, ob die beiden Vektoren orthogonal zueinander sind.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Berechnung des Skalarprodukts

Für zwei reelle Vektoren

$$
\vec{x}
=
\begin{pmatrix}
x_1\\
x_2\\
x_3
\end{pmatrix},
\qquad
\vec{y}
=
\begin{pmatrix}
y_1\\
y_2\\
y_3
\end{pmatrix}
$$

wird das Skalarprodukt komponentenweise berechnet:

$$
\langle\vec{x},\vec{y}\rangle
=
x_1y_1+x_2y_2+x_3y_3.
$$

Für die gegebenen Vektoren ergibt sich

$$
\begin{aligned}
\langle\vec{x},\vec{y}\rangle
&=
1\cdot1
+
0\cdot1
+
1\cdot0\\
&=
1.
\end{aligned}
$$

Damit gilt

$$
\boxed{
\langle\vec{x},\vec{y}\rangle=1
}
$$
:::

:::{dropdown} Berechnung der Normen mithilfe des Skalarprodukts

Die Norm eines Vektors kann mithilfe des Skalarprodukts berechnet werden:

$$
\|\vec{x}\|
=
\sqrt{
\langle\vec{x},\vec{x}\rangle
}.
$$

Für $\vec{x}$ gilt

$$
\begin{aligned}
\langle\vec{x},\vec{x}\rangle
&=
1^2+0^2+1^2\\
&=
2.
\end{aligned}
$$

Damit ist

$$
\boxed{
\|\vec{x}\|=\sqrt{2}.
}
$$

Für $\vec{y}$ gilt entsprechend

$$
\begin{aligned}
\langle\vec{y},\vec{y}\rangle
&=
1^2+1^2+0^2\\
&=
2.
\end{aligned}
$$

Damit ist

$$
\boxed{
\|\vec{y}\|=\sqrt{2}.
}
$$
:::

:::{dropdown} Überprüfung der Orthogonalität

Zwei Vektoren sind orthogonal zueinander, wenn ihr Skalarprodukt gleich null ist:

$$
\langle\vec{x},\vec{y}\rangle=0.
$$

Wir haben jedoch erhalten:

$$
\langle\vec{x},\vec{y}\rangle=1.
$$

Da

$$
1\neq0,
$$

sind die beiden Vektoren nicht orthogonal zueinander.

Damit gilt

$$
\boxed{
\vec{x}\not\perp\vec{y}.
}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Für

$$
\vec{x}
=
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\qquad
\vec{y}
=
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
$$

ergibt sich das Skalarprodukt zu

$$
\boxed{
\langle\vec{x},\vec{y}\rangle=1.
}
$$

Die Normen sind

$$
\boxed{
\|\vec{x}\|=\sqrt{2},
\qquad
\|\vec{y}\|=\sqrt{2}.
}
$$

Da

$$
\langle\vec{x},\vec{y}\rangle\neq0,
$$

sind die Vektoren nicht orthogonal zueinander.
:::
