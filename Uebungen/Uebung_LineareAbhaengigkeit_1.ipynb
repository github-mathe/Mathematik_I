## Lineare Abhängigkeit von Vektoren 

Gegeben seien die Vektoren

$$
\vec{v}_1=
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix},
\qquad
\vec{v}_2=
\begin{pmatrix}
2\\
1\\
1
\end{pmatrix},
\qquad
\vec{v}_3=
\begin{pmatrix}
3\\
3\\
0
\end{pmatrix}.
$$

Untersuchen Sie mithilfe des **Gauß-Eliminationsverfahrens**, ob die Vektoren

$$
\vec{v}_1,\vec{v}_2,\vec{v}_3
$$

linear abhängig oder linear unabhängig sind.

Stellen Sie dazu die Gleichung

$$
\lambda_1\vec{v}_1
+
\lambda_2\vec{v}_2
+
\lambda_3\vec{v}_3
=
\vec{0}
$$

auf und lösen Sie das entstehende homogene lineare Gleichungssystem.

Begründen Sie das Ergebnis zusätzlich mithilfe des Rangs der Matrix.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Aufstellen des homogenen linearen Gleichungssystems

Zur Untersuchung der linearen Abhängigkeit betrachten wir

$$
\lambda_1\vec{v}_1
+
\lambda_2\vec{v}_2
+
\lambda_3\vec{v}_3
=
\vec{0}.
$$

Einsetzen der Vektoren liefert

$$
\lambda_1
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
2\\
1\\
1
\end{pmatrix}
+
\lambda_3
\begin{pmatrix}
3\\
3\\
0
\end{pmatrix}
=
\begin{pmatrix}
0\\
0\\
0
\end{pmatrix}.
$$

Daraus ergibt sich das homogene lineare Gleichungssystem

$$
\begin{aligned}
\lambda_1+2\lambda_2+3\lambda_3 &=0,\\
2\lambda_1+\lambda_2+3\lambda_3 &=0,\\
-\lambda_1+\lambda_2 &=0.
\end{aligned}
$$

Die zugehörige erweiterte Koeffizientenmatrix lautet

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
2 & 1 & 3 & 0\\
-1 & 1 & 0 & 0
\end{array}
\right)
}
$$

:::

::::{dropdown} Durchführung der Elimination

Ausgangspunkt ist

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
2 & 1 & 3 & 0\\
-1 & 1 & 0 & 0
\end{array}
\right).
$$


:::{dropdown} Schritt 1 – Ersten Eintrag eliminieren

Der Eintrag $2$ in der ersten Spalte wird eliminiert.

Wir verwenden

$$
Z_2\leftarrow Z_2-2Z_1.
$$

Es ergibt sich

$$
Z_2:
\qquad
(2,\;1,\;3\mid0)
-
2(1,\;2,\;3\mid0)
$$

und damit

$$
Z_2=(0,\;-3,\;-3\mid0).
$$

Die Matrix lautet nun

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & -3 & -3 & 0\\
-1 & 1 & 0 & 0
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 2 – Zweiten Eintrag in der ersten Spalte eliminieren

Nun wird der Eintrag $-1$ in der ersten Spalte eliminiert.

Wir verwenden

$$
Z_3\leftarrow Z_3+Z_1.
$$

Es ergibt sich

$$
Z_3:
\qquad
(-1,\;1,\;0\mid0)
+
(1,\;2,\;3\mid0)
$$

und damit

$$
Z_3=(0,\;3,\;3\mid0).
$$

Die Matrix lautet nun

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & -3 & -3 & 0\\
0 & 3 & 3 & 0
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 3 – Eintrag unterhalb des zweiten Pivotelements eliminieren

Wir verwenden

$$
Z_3\leftarrow Z_3+Z_2.
$$

Damit ergibt sich

$$
Z_3=(0,\;0,\;0\mid0).
$$

Die Matrix lautet

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & -3 & -3 & 0\\
0 & 0 & 0 & 0
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 4 – Zweites Pivotelement normieren

Wir verwenden

$$
Z_2\leftarrow-\frac{1}{3}Z_2.
$$

Damit erhalten wir

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & 1 & 1 & 0\\
0 & 0 & 0 & 0
\end{array}
\right)
}
$$

:::

::::

:::{dropdown} Betrachtung der linearen Abhängigkeit

Nach der Gauß-Elimination erhalten wir

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & 1 & 1 & 0\\
0 & 0 & 0 & 0
\end{array}
\right).
$$

Die Matrix besitzt zwei Pivotelemente. Daher gilt

$$
\operatorname{rang}(A)=2.
$$

Es werden jedoch drei Vektoren untersucht. Somit gilt

$$
\operatorname{rang}(A)=2<3.
$$

Das homogene Gleichungssystem besitzt daher neben der trivialen Lösung auch **nichttriviale Lösungen**.

Es gibt einen Freiheitsgrad:

$$
3-\operatorname{rang}(A)
=
3-2
=
1.
$$

Damit sind die drei Vektoren **linear abhängig**.
:::

:::{dropdown} Bestimmung einer Abhängigkeitsbeziehung

Aus der zweiten Zeile folgt

$$
\lambda_2+\lambda_3=0.
$$

Wir wählen

$$
\lambda_3=t,
\qquad
t\in\mathbb{R}.
$$

Damit ist

$$
\lambda_2=-t.
$$

Aus der ersten Zeile folgt

$$
\lambda_1+2\lambda_2+3\lambda_3=0.
$$

Einsetzen ergibt

$$
\lambda_1-2t+3t=0
$$

und somit

$$
\lambda_1=-t.
$$

Damit lautet die Lösungsmenge

$$
\boxed{
L=
\left\{
\begin{pmatrix}
-t\\
-t\\
t
\end{pmatrix}
\;\middle|\;
t\in\mathbb{R}
\right\}.
}
$$

Wählen wir beispielsweise

$$
t=1,
$$

so erhalten wir die nichttriviale Lösung

$$
\lambda_1=-1,
\qquad
\lambda_2=-1,
\qquad
\lambda_3=1.
$$

Damit gilt

$$
-\vec{v}_1-\vec{v}_2+\vec{v}_3=\vec{0}.
$$

Also

$$
\boxed{
\vec{v}_3=\vec{v}_1+\vec{v}_2
}
$$

und die Vektoren sind somit **linear abhängig**.
:::
