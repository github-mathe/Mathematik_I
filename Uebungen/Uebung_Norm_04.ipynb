## Vektornormen mit komplexen Zahlen

Gegeben sei der komplexe Vektor

$$
\mathbf{x}
=
\begin{pmatrix}
3+4i\\
1-i\\
-2i
\end{pmatrix}
\in\mathbb{C}^3.
$$

Berechnen Sie für den Vektor $\mathbf{x}$

1. die $L_1$-Norm

   $$
   \|\mathbf{x}\|_1
   =
   \sum_{i=1}^{n}|x_i|,
   $$

2. die $L_\infty$-Norm

   $$
   \|\mathbf{x}\|_\infty
   =
   \max_{1\leq i\leq n}|x_i|,
   $$

3. die $L_2$-Norm

   $$
   \|\mathbf{x}\|_2
   =
   \sqrt{
   \sum_{i=1}^{n}|x_i|^2
   }.
   $$

Bestimmen Sie zunächst die Beträge der einzelnen komplexen Komponenten und vergleichen Sie anschließend die drei Normen.

## Lösung

Bei komplexen Vektoren müssen zunächst die Beträge der einzelnen komplexen Komponenten bestimmt werden.

Für eine komplexe Zahl

$$
z=a+bi
$$

gilt

$$
|z|
=
\sqrt{a^2+b^2}.
$$

Anschließend können die verschiedenen Vektornormen berechnet werden.

:::{dropdown} Beträge der komplexen Komponenten

Für die erste Komponente gilt

$$
x_1=3+4i.
$$

Damit ist

$$
\begin{aligned}
|x_1|
&=
|3+4i|\\
&=
\sqrt{3^2+4^2}\\
&=
\sqrt{9+16}\\
&=
5.
\end{aligned}
$$

Für die zweite Komponente gilt

$$
x_2=1-i.
$$

Damit ist

$$
\begin{aligned}
|x_2|
&=
|1-i|\\
&=
\sqrt{1^2+(-1)^2}\\
&=
\sqrt{2}.
\end{aligned}
$$

Für die dritte Komponente gilt

$$
x_3=-2i.
$$

Damit ist

$$
\begin{aligned}
|x_3|
&=
|-2i|\\
&=
\sqrt{0^2+(-2)^2}\\
&=
2.
\end{aligned}
$$

Somit erhalten wir

$$
\boxed{
|x_1|=5,
\qquad
|x_2|=\sqrt{2},
\qquad
|x_3|=2.
}
$$
:::

:::{dropdown} Berechnung der $L_1$-Norm

Die $L_1$-Norm ist definiert durch

$$
\|\mathbf{x}\|_1
=
\sum_{i=1}^{n}|x_i|.
$$

Damit ergibt sich

$$
\begin{aligned}
\|\mathbf{x}\|_1
&=
|3+4i|
+
|1-i|
+
|-2i|\\
&=
5+\sqrt{2}+2\\
&=
7+\sqrt{2}.
\end{aligned}
$$

Somit gilt

$$
\boxed{
\|\mathbf{x}\|_1=7+\sqrt{2}
}
$$
:::

:::{dropdown} Berechnung der $L_\infty$-Norm

Die $L_\infty$-Norm ist definiert durch

$$
\|\mathbf{x}\|_\infty
=
\max_{1\leq i\leq n}|x_i|.
$$

Die Beträge der Komponenten sind

$$
5,
\qquad
\sqrt{2},
\qquad
2.
$$

Damit gilt

$$
\|\mathbf{x}\|_\infty
=
\max
\left\{
5,\sqrt{2},2
\right\}.
$$

Der größte Betrag ist $5$.

Somit ergibt sich

$$
\boxed{
\|\mathbf{x}\|_\infty=5
}
$$
:::

:::{dropdown} Berechnung der $L_2$-Norm

Für einen komplexen Vektor wird die $L_2$-Norm mit den Beträgen der komplexen Komponenten berechnet:

$$
\|\mathbf{x}\|_2
=
\sqrt{
\sum_{i=1}^{n}|x_i|^2
}.
$$

Damit ergibt sich

$$
\begin{aligned}
\|\mathbf{x}\|_2
&=
\sqrt{
|3+4i|^2
+
|1-i|^2
+
|-2i|^2
}\\
&=
\sqrt{
5^2
+
(\sqrt{2})^2
+
2^2
}\\
&=
\sqrt{
25+2+4
}\\
&=
\sqrt{31}.
\end{aligned}
$$

Somit gilt

$$
\boxed{
\|\mathbf{x}\|_2=\sqrt{31}
}
$$
:::

:::{dropdown} Vergleich der Vektornormen

Für den komplexen Vektor

$$
\mathbf{x}
=
\begin{pmatrix}
3+4i\\
1-i\\
-2i
\end{pmatrix}
$$

haben wir erhalten:

$$
\boxed{
\|\mathbf{x}\|_1=7+\sqrt{2}
}
$$

$$
\boxed{
\|\mathbf{x}\|_\infty=5
}
$$

und

$$
\boxed{
\|\mathbf{x}\|_2=\sqrt{31}.
}
$$

Näherungsweise gilt

$$
7+\sqrt{2}\approx8{,}41
$$

und

$$
\sqrt{31}\approx5{,}57.
$$

Damit ergibt sich

$$
\boxed{
\|\mathbf{x}\|_\infty
<
\|\mathbf{x}\|_2
<
\|\mathbf{x}\|_1.
}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Die Beträge der Komponenten sind

$$
|3+4i|=5,
\qquad
|1-i|=\sqrt{2},
\qquad
|-2i|=2.
$$

Damit ergeben sich die Vektornormen zu

$$
\boxed{
\|\mathbf{x}\|_1=7+\sqrt{2},
\qquad
\|\mathbf{x}\|_\infty=5,
\qquad
\|\mathbf{x}\|_2=\sqrt{31}.
}
$$
:::
