## Vektor-Vektor-Multiplikation

Gegeben seien der Zeilenvektor

$$
\vec{a}^{\,T}
=
\begin{pmatrix}
2 & -1 & 3
\end{pmatrix}
$$

und der Spaltenvektor

$$
\vec{b}
=
\begin{pmatrix}
1\\
4\\
-2
\end{pmatrix}.
$$

Berechnen Sie die beiden Produkte

$$
\vec{a}^{\,T}\vec{b}
$$

und

$$
\vec{b}\vec{a}^{\,T}.
$$

Gehen Sie dabei wie folgt vor:

1. Bestimmen Sie zunächst die Dimensionen der beiden Vektoren.
2. Überprüfen Sie für beide Produkte, ob die Multiplikation möglich ist.
3. Bestimmen Sie jeweils die Dimension des Ergebnisses.
4. Berechnen Sie beide Produkte.
5. Vergleichen Sie die Ergebnisse und erklären Sie, warum

$$
\vec{a}^{\,T}\vec{b}
\neq
\vec{b}\vec{a}^{\,T}.
$$

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Bestimmung der Dimensionen

Der Zeilenvektor

$$
\vec{a}^{\,T}
=
\begin{pmatrix}
2 & -1 & 3
\end{pmatrix}
$$

besitzt die Dimension

$$
1\times 3.
$$

Der Spaltenvektor

$$
\vec{b}
=
\begin{pmatrix}
1\\
4\\
-2
\end{pmatrix}
$$

besitzt die Dimension

$$
3\times 1.
$$

Für das Produkt

$$
\vec{a}^{\,T}\vec{b}
$$

ergibt sich daher

$$
(1\times3)\cdot(3\times1).
$$

Die inneren Dimensionen stimmen überein. Das Produkt ist somit definiert und besitzt die Dimension

$$
\boxed{1\times1}.
$$

Für das Produkt

$$
\vec{b}\vec{a}^{\,T}
$$

ergibt sich

$$
(3\times1)\cdot(1\times3).
$$

Auch hier stimmen die inneren Dimensionen überein. Das Ergebnis besitzt jedoch die Dimension

$$
\boxed{3\times3}.
$$
:::

:::{dropdown} Berechnung von $\vec{a}^{\,T}\vec{b}$

Wir berechnen

$$
\vec{a}^{\,T}\vec{b}
=
\begin{pmatrix}
2 & -1 & 3
\end{pmatrix}
\begin{pmatrix}
1\\
4\\
-2
\end{pmatrix}.
$$

Die Komponenten werden paarweise multipliziert und anschließend addiert:

$$
\vec{a}^{\,T}\vec{b}
=
2\cdot1
+
(-1)\cdot4
+
3\cdot(-2).
$$

Damit ergibt sich

$$
2-4-6=-8.
$$

Also

$$
\boxed{
\vec{a}^{\,T}\vec{b}=-8
}
$$

Das Ergebnis ist ein Skalar.
:::

:::{dropdown} Berechnung von $\vec{b}\vec{a}^{\,T}$

Nun betrachten wir die umgekehrte Reihenfolge:

$$
\vec{b}\vec{a}^{\,T}
=
\begin{pmatrix}
1\\
4\\
-2
\end{pmatrix}
\begin{pmatrix}
2 & -1 & 3
\end{pmatrix}.
$$

Jede Komponente des Spaltenvektors wird mit dem gesamten Zeilenvektor multipliziert:

$$
\vec{b}\vec{a}^{\,T}
=
\begin{pmatrix}
1\cdot2 & 1\cdot(-1) & 1\cdot3\\
4\cdot2 & 4\cdot(-1) & 4\cdot3\\
(-2)\cdot2 & (-2)\cdot(-1) & (-2)\cdot3
\end{pmatrix}.
$$

Damit erhalten wir

$$
\boxed{
\vec{b}\vec{a}^{\,T}
=
\begin{pmatrix}
2 & -1 & 3\\
8 & -4 & 12\\
-4 & 2 & -6
\end{pmatrix}
}
$$

Das Ergebnis ist eine $3\times3$-Matrix.
:::

:::{dropdown} Vergleich der beiden Produkte

Die beiden Produkte liefern unterschiedliche Arten von Ergebnissen.

Für

$$
\vec{a}^{\,T}\vec{b}
$$

gilt

$$
(1\times3)\cdot(3\times1)
\longrightarrow
1\times1,
$$

also

$$
\boxed{
\vec{a}^{\,T}\vec{b}=-8.
}
$$

Für die umgekehrte Reihenfolge gilt dagegen

$$
(3\times1)\cdot(1\times3)
\longrightarrow
3\times3,
$$

also

$$
\boxed{
\vec{b}\vec{a}^{\,T}
=
\begin{pmatrix}
2 & -1 & 3\\
8 & -4 & 12\\
-4 & 2 & -6
\end{pmatrix}.
}
$$

Somit gilt

$$
\boxed{
\vec{a}^{\,T}\vec{b}
\neq
\vec{b}\vec{a}^{\,T}.
}
$$

Die Reihenfolge bei der Multiplikation ist also entscheidend.
:::

:::{dropdown} Gesamtlösung anzeigen

Es gilt

$$
\boxed{
\vec{a}^{\,T}\vec{b}=-8
}
$$

mit einem skalaren Ergebnis und

$$
\boxed{
\vec{b}\vec{a}^{\,T}
=
\begin{pmatrix}
2 & -1 & 3\\
8 & -4 & 12\\
-4 & 2 & -6
\end{pmatrix}
}
$$

mit einem $3\times3$-Matrixergebnis.

Die beiden Multiplikationen unterscheiden sich somit sowohl in der Dimension als auch in der Art des Ergebnisses.
:::
