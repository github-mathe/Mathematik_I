## Matrix-Matrix-Multiplikation

Gegeben seien die Matrizen

$$
A=
\begin{pmatrix}
1 & 2 & -1\\
0 & 3 & 2
\end{pmatrix}
$$

und

$$
B=
\begin{pmatrix}
2 & 1\\
-1 & 0\\
3 & 4
\end{pmatrix}.
$$

Berechnen Sie das Matrixprodukt

$$
AB.
$$

Gehen Sie dabei wie folgt vor:

1. Bestimmen Sie die Dimensionen der Matrizen $A$ und $B$.
2. Überprüfen Sie, ob das Produkt $AB$ definiert ist.
3. Bestimmen Sie die Dimension der Ergebnismatrix.
4. Berechnen Sie die Einträge der Ergebnismatrix schrittweise.
5. Geben Sie anschließend das vollständige Matrixprodukt an.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Dimensionen der Matrizen

Die Matrix

$$
A=
\begin{pmatrix}
1 & 2 & -1\\
0 & 3 & 2
\end{pmatrix}
$$

besitzt die Dimension

$$
2\times3.
$$

Die Matrix

$$
B=
\begin{pmatrix}
2 & 1\\
-1 & 0\\
3 & 4
\end{pmatrix}
$$

besitzt die Dimension

$$
3\times2.
$$

Für das Produkt

$$
AB
$$

ergibt sich somit

$$
(2\times3)\cdot(3\times2).
$$

Die inneren Dimensionen stimmen überein.

Das Produkt ist daher definiert.

Die Ergebnismatrix besitzt die Dimension

$$
\boxed{
2\times2
}
$$
:::

Wir bezeichnen die Ergebnismatrix mit

$$
C=AB.
$$

Da

$$
A\in\mathbb{R}^{2\times3}
\qquad\text{und}\qquad
B\in\mathbb{R}^{3\times2},
$$

hat $C$ die Dimension

$$
C\in\mathbb{R}^{2\times2}.
$$

Wir schreiben

$$
C=
\begin{pmatrix}
c_{11} & c_{12}\\
c_{21} & c_{22}
\end{pmatrix}.
$$

Die Einträge von $C$ werden jeweils durch die Multiplikation einer Zeile von $A$ mit einer Spalte von $B$ bestimmt.

:::{dropdown} Berechnung von $c_{11}$

Der Eintrag $c_{11}$ entsteht aus der ersten Zeile von $A$ und der ersten Spalte von $B$:

$$
c_{11}
=
\begin{pmatrix}
1 & 2 & -1
\end{pmatrix}
\begin{pmatrix}
2\\
-1\\
3
\end{pmatrix}.
$$

Damit gilt

$$
c_{11}
=
1\cdot2
+
2\cdot(-1)
+
(-1)\cdot3.
$$

Also

$$
c_{11}
=
2-2-3
=
-3.
$$
:::

:::{dropdown} Berechnung von $c_{12}$

Der Eintrag $c_{12}$ entsteht aus der ersten Zeile von $A$ und der zweiten Spalte von $B$:

$$
c_{12}
=
\begin{pmatrix}
1 & 2 & -1
\end{pmatrix}
\begin{pmatrix}
1\\
0\\
4
\end{pmatrix}.
$$

Damit gilt

$$
c_{12}
=
1\cdot1
+
2\cdot0
+
(-1)\cdot4.
$$

Also

$$
c_{12}
=
1-4
=
-3.
$$
:::

:::{dropdown} Berechnung von $c_{21}$

Der Eintrag $c_{21}$ entsteht aus der zweiten Zeile von $A$ und der ersten Spalte von $B$:

$$
c_{21}
=
\begin{pmatrix}
0 & 3 & 2
\end{pmatrix}
\begin{pmatrix}
2\\
-1\\
3
\end{pmatrix}.
$$

Damit gilt

$$
c_{21}
=
0\cdot2
+
3\cdot(-1)
+
2\cdot3.
$$

Also

$$
c_{21}
=
-3+6
=
3.
$$
:::

:::{dropdown} Berechnung von $c_{22}$

Der Eintrag $c_{22}$ entsteht aus der zweiten Zeile von $A$ und der zweiten Spalte von $B$:

$$
c_{22}
=
\begin{pmatrix}
0 & 3 & 2
\end{pmatrix}
\begin{pmatrix}
1\\
0\\
4
\end{pmatrix}.
$$

Damit gilt

$$
c_{22}
=
0\cdot1
+
3\cdot0
+
2\cdot4.
$$

Also

$$
c_{22}
=
8.
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Die Ergebnismatrix hat die Dimension

$$
2\times2.
$$

Mit

$$
c_{11}=-3,\qquad
c_{12}=-3,
$$

$$
c_{21}=3,\qquad
c_{22}=8
$$

ergibt sich

$$
\boxed{
AB=
\begin{pmatrix}
-3 & -3\\
3 & 8
\end{pmatrix}
}
$$
:::
