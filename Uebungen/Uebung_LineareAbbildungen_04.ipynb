## Projektion auf eine Linie

Gegeben seien

$$
\vec{x}
=
\begin{pmatrix}
4\\
1
\end{pmatrix}
$$

und die durch den Richtungsvektor

$$
\vec{a}
=
\begin{pmatrix}
2\\
1
\end{pmatrix}
$$

aufgespannte Linie

$$
L=\operatorname{span}\{\vec{a}\}.
$$

1. Bestimmen Sie die Projektionsmatrix

$$
P_{\vec a}
=
\frac{\vec a\otimes\vec a}{\|\vec a\|^2}.
$$

2. Berechnen Sie die orthogonale Projektion $P_{\vec a}(\vec{x})$ von $\vec{x}$ auf die Linie $L$.

3. Bestimmen Sie den Differenzvektor

$$
\vec r=\vec{x}-P_{\vec a}(\vec{x})
$$

und überprüfen Sie, dass $\vec r$ orthogonal zu $\vec a$ ist.

## Lösung

Zunächst wird die Projektionsmatrix bestimmt. Anschließend wird diese auf den Vektor $\vec{x}$ angewendet.

:::{dropdown} Projektionsmatrix bestimmen

Gegeben ist

$$
\vec a
=
\begin{pmatrix}
2\\
1
\end{pmatrix}.
$$

Zunächst berechnen wir das Tensorprodukt

$$
\begin{aligned}
\vec a\otimes\vec a
&=
\vec a\,\vec a^{\,T}\\[1ex]
&=
\begin{pmatrix}
2\\
1
\end{pmatrix}
\begin{pmatrix}
2 & 1
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
4 & 2\\
2 & 1
\end{pmatrix}.
\end{aligned}
$$

Weiter gilt

$$
\|\vec a\|^2
=
2^2+1^2
=
5.
$$

Damit ergibt sich die Projektionsmatrix

$$
\boxed{
P_{\vec a}
=
\frac{1}{5}
\begin{pmatrix}
4 & 2\\
2 & 1
\end{pmatrix}.
}
$$
:::

:::{dropdown} Projektion berechnen

Die Projektion des Vektors

$$
\vec{x}
=
\begin{pmatrix}
4\\
1
\end{pmatrix}
$$

auf die Linie wird durch

$$
P_{\vec a}(\vec{x})
=
P_{\vec a}\vec{x}
$$

bestimmt.

Damit gilt

$$
\begin{aligned}
P_{\vec a}(\vec{x})
&=
\frac{1}{5}
\begin{pmatrix}
4 & 2\\
2 & 1
\end{pmatrix}
\begin{pmatrix}
4\\
1
\end{pmatrix}\\[1ex]
&=
\frac{1}{5}
\begin{pmatrix}
18\\
9
\end{pmatrix}.
\end{aligned}
$$

Somit ist

$$
\boxed{
P_{\vec a}(\vec{x})
=
\begin{pmatrix}
\frac{18}{5}\\[1ex]
\frac{9}{5}
\end{pmatrix}.
}
$$
:::

:::{dropdown} Differenzvektor und Orthogonalität

Der Differenzvektor ist

$$
\vec r
=
\vec{x}-P_{\vec a}(\vec{x}).
$$

Damit ergibt sich

$$
\begin{aligned}
\vec r
&=
\begin{pmatrix}
4\\
1
\end{pmatrix}
-
\begin{pmatrix}
\frac{18}{5}\\[1ex]
\frac{9}{5}
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
\frac{2}{5}\\[1ex]
-\frac{4}{5}
\end{pmatrix}.
\end{aligned}
$$

Zur Überprüfung der Orthogonalität berechnen wir das Skalarprodukt:

$$
\begin{aligned}
\langle\vec r,\vec a\rangle
&=
\frac{2}{5}\cdot2
+
\left(-\frac{4}{5}\right)\cdot1\\[1ex]
&=
\frac{4}{5}-\frac{4}{5}\\
&=
0.
\end{aligned}
$$

Damit gilt

$$
\boxed{
\vec r\perp\vec a.
}
$$

Der Differenzvektor steht somit senkrecht auf der Linie $L$.
:::

:::{dropdown} Gesamtlösung anzeigen

Für

$$
\vec{x}
=
\begin{pmatrix}
4\\
1
\end{pmatrix},
\qquad
\vec a
=
\begin{pmatrix}
2\\
1
\end{pmatrix}
$$

ergibt sich die Projektionsmatrix

$$
\boxed{
P_{\vec a}
=
\frac{1}{5}
\begin{pmatrix}
4 & 2\\
2 & 1
\end{pmatrix}.
}
$$

Die Projektion von $\vec{x}$ auf die Linie

$$
L=\operatorname{span}\{\vec a\}
$$

lautet

$$
\boxed{
P_{\vec a}(\vec{x})
=
\begin{pmatrix}
\frac{18}{5}\\[1ex]
\frac{9}{5}
\end{pmatrix}.
}
$$

Der Differenzvektor ist

$$
\boxed{
\vec r
=
\begin{pmatrix}
\frac{2}{5}\\[1ex]
-\frac{4}{5}
\end{pmatrix}.
}
$$

Da

$$
\langle\vec r,\vec a\rangle=0,
$$

gilt

$$
\boxed{
\vec r\perp\vec a.
}
$$
:::
