## Abbildungsmatrizen aufstellen

Bestimmen Sie zu den folgenden linearen Abbildungen jeweils die zugehörige Standardmatrix.

### a)

$$
T_1:\mathbb{R}^2\to\mathbb{R}^2,
\qquad
T_1
\begin{pmatrix}
x\\
y
\end{pmatrix}
=
\begin{pmatrix}
2x-y\\
x+3y
\end{pmatrix}.
$$

### b)

$$
T_2:\mathbb{R}^3\to\mathbb{R}^2,
\qquad
T_2
\begin{pmatrix}
x\\
y\\
z
\end{pmatrix}
=
\begin{pmatrix}
x+2y-z\\
3x+z
\end{pmatrix}.
$$

### c)

$$
T_3:\mathbb{R}^2\to\mathbb{R}^3,
\qquad
T_3
\begin{pmatrix}
x\\
y
\end{pmatrix}
=
\begin{pmatrix}
x\\
-2x+y\\
3y
\end{pmatrix}.
$$

### d)

$$
T_4:\mathbb{R}^3\to\mathbb{R}^3,
\qquad
T_4
\begin{pmatrix}
x\\
y\\
z
\end{pmatrix}
=
\begin{pmatrix}
2x+y\\
-x+z\\
y-2z
\end{pmatrix}.
$$

## Lösung

Die Spalten der Standardmatrix bestehen aus den Bildern der Einheitsvektoren.

:::{dropdown} Lösung a)

Es gilt

$$
T_1(\mathbf e_1)
=
\begin{pmatrix}
2\\
1
\end{pmatrix},
\qquad
T_1(\mathbf e_2)
=
\begin{pmatrix}
-1\\
3
\end{pmatrix}.
$$

Damit lautet die Standardmatrix

$$
\boxed{
\mathbf A_1
=
\begin{pmatrix}
2 & -1\\
1 & 3
\end{pmatrix}.
}
$$
:::

:::{dropdown} Lösung b)

Die Bilder der Einheitsvektoren sind

$$
T_2(\mathbf e_1)
=
\begin{pmatrix}
1\\
3
\end{pmatrix},
\qquad
T_2(\mathbf e_2)
=
\begin{pmatrix}
2\\
0
\end{pmatrix},
\qquad
T_2(\mathbf e_3)
=
\begin{pmatrix}
-1\\
1
\end{pmatrix}.
$$

Damit lautet die Standardmatrix

$$
\boxed{
\mathbf A_2
=
\begin{pmatrix}
1 & 2 & -1\\
3 & 0 & 1
\end{pmatrix}.
}
$$
:::

:::{dropdown} Lösung c)

Es gilt

$$
T_3(\mathbf e_1)
=
\begin{pmatrix}
1\\
-2\\
0
\end{pmatrix},
\qquad
T_3(\mathbf e_2)
=
\begin{pmatrix}
0\\
1\\
3
\end{pmatrix}.
$$

Damit lautet die Standardmatrix

$$
\boxed{
\mathbf A_3
=
\begin{pmatrix}
1 & 0\\
-2 & 1\\
0 & 3
\end{pmatrix}.
}
$$
:::

:::{dropdown} Lösung d)

Die Bilder der Einheitsvektoren sind

$$
T_4(\mathbf e_1)
=
\begin{pmatrix}
2\\
-1\\
0
\end{pmatrix},
\qquad
T_4(\mathbf e_2)
=
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\qquad
T_4(\mathbf e_3)
=
\begin{pmatrix}
0\\
1\\
-2
\end{pmatrix}.
$$

Damit lautet die Standardmatrix

$$
\boxed{
\mathbf A_4
=
\begin{pmatrix}
2 & 1 & 0\\
-1 & 0 & 1\\
0 & 1 & -2
\end{pmatrix}.
}
$$
:::
