# Übungen

In diesem Abschnitt finden Sie Übungsaufgaben zu den Inhalten der Vorlesung **Mathematik I**.

Die Aufgaben dienen dazu, die behandelten Begriffe und Rechenverfahren zu vertiefen und selbstständig anzuwenden. Sie sind thematisch nach den einzelnen Kapiteln der Vorlesung gegliedert.

Es empfiehlt sich, die Aufgaben zunächst eigenständig zu bearbeiten und dabei die Herleitungen und Rechenschritte vollständig nachzuvollziehen.
