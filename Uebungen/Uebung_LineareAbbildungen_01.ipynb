## Überprüfung einer Abbildung auf Linearität

Gegeben sei die Abbildung

$$
T:\mathbb{R}^2\to\mathbb{R}^2
$$

mit

$$
T
\begin{pmatrix}
x\\
y
\end{pmatrix}
=
\begin{pmatrix}
2x-y\\
3x+y
\end{pmatrix}.
$$

Überprüfen Sie mithilfe der Definition, ob $T$ eine lineare Abbildung ist.

Prüfen Sie dazu:

$$
T(\vec v+\vec w)=T(\vec v)+T(\vec w)
$$

und

$$
T(c\vec v)=cT(\vec v).
$$

## Lösung

Wir überprüfen die beiden Bedingungen **Additivität** und **Homogenität**.

:::{dropdown} Additivität überprüfen

Seien

$$
\vec v=
\begin{pmatrix}
v_1\\
v_2
\end{pmatrix},
\qquad
\vec w=
\begin{pmatrix}
w_1\\
w_2
\end{pmatrix}.
$$

Dann gilt

$$
\vec v+\vec w
=
\begin{pmatrix}
v_1+w_1\\
v_2+w_2
\end{pmatrix}.
$$

Damit ist

$$
\begin{aligned}
T(\vec v+\vec w)
&=
T
\begin{pmatrix}
v_1+w_1\\
v_2+w_2
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
2(v_1+w_1)-(v_2+w_2)\\
3(v_1+w_1)+(v_2+w_2)
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
2v_1-v_2+2w_1-w_2\\
3v_1+v_2+3w_1+w_2
\end{pmatrix}.
\end{aligned}
$$

Andererseits gilt

$$
\begin{aligned}
T(\vec v)+T(\vec w)
&=
\begin{pmatrix}
2v_1-v_2\\
3v_1+v_2
\end{pmatrix}
+
\begin{pmatrix}
2w_1-w_2\\
3w_1+w_2
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
2v_1-v_2+2w_1-w_2\\
3v_1+v_2+3w_1+w_2
\end{pmatrix}.
\end{aligned}
$$

Somit gilt

$$
\boxed{
T(\vec v+\vec w)=T(\vec v)+T(\vec w).
}
$$

Die Additivität ist erfüllt.
:::

:::{dropdown} Homogenität überprüfen

Sei

$$
\vec v=
\begin{pmatrix}
v_1\\
v_2
\end{pmatrix}
$$

und $c\in\mathbb{R}$.

Dann gilt

$$
c\vec v
=
\begin{pmatrix}
cv_1\\
cv_2
\end{pmatrix}.
$$

Damit ist

$$
\begin{aligned}
T(c\vec v)
&=
T
\begin{pmatrix}
cv_1\\
cv_2
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
2cv_1-cv_2\\
3cv_1+cv_2
\end{pmatrix}\\[1ex]
&=
c
\begin{pmatrix}
2v_1-v_2\\
3v_1+v_2
\end{pmatrix}\\[1ex]
&=
cT(\vec v).
\end{aligned}
$$

Somit gilt

$$
\boxed{
T(c\vec v)=cT(\vec v).
}
$$

Die Homogenität ist erfüllt.
:::

:::{dropdown} Gesamtlösung anzeigen

Für die Abbildung

$$
T
\begin{pmatrix}
x\\
y
\end{pmatrix}
=
\begin{pmatrix}
2x-y\\
3x+y
\end{pmatrix}
$$

gilt für alle $\vec v,\vec w\in\mathbb{R}^2$

$$
T(\vec v+\vec w)
=
T(\vec v)+T(\vec w)
$$

und für alle $c\in\mathbb{R}$

$$
T(c\vec v)
=
cT(\vec v).
$$

Damit sind sowohl **Additivität** als auch **Homogenität** erfüllt.

Also ist

$$
\boxed{
T:\mathbb{R}^2\to\mathbb{R}^2
\text{ eine lineare Abbildung.}
}
$$
:::
