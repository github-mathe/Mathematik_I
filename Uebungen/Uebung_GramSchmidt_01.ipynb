## Gram-Schmidt-Verfahren

Gegeben sei die Basis

$$
\left\{
\mathbf v_1=
\begin{pmatrix}
1\\
1\\
1
\end{pmatrix},
\quad
\mathbf v_2=
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\quad
\mathbf v_3=
\begin{pmatrix}
0\\
1\\
1
\end{pmatrix}
\right\}.
$$

Bestimmen Sie mit dem Gram-Schmidt-Verfahren eine orthonormale Basis.

## Lösung

Wir führen das Gram-Schmidt-Verfahren schrittweise durch.

:::{dropdown} Schritt 1 – Bestimmung von $\mathbf u_1$

Nach dem Gram-Schmidt-Verfahren wird zunächst $\mathbf v_1$ normiert:

$$
\mathbf u_1
=
\frac{\mathbf v_1}{\|\mathbf v_1\|}.
$$

Es gilt

$$
\|\mathbf v_1\|
=
\sqrt{1^2+1^2+1^2}
=
\sqrt{3}.
$$

Damit erhalten wir

$$
\boxed{
\mathbf u_1
=
\frac{1}{\sqrt{3}}
\begin{pmatrix}
1\\
1\\
1
\end{pmatrix}.
}
$$
:::

:::{dropdown} Schritt 2 – Bestimmung von $\mathbf u_2$

Zunächst bestimmen wir die Projektion von $\mathbf v_2$ auf $\mathbf u_1$:

$$
P_{\mathbf u_1}(\mathbf v_2)
=
\langle \mathbf v_2,\mathbf u_1\rangle\,\mathbf u_1.
$$

Das Skalarprodukt ist

$$
\begin{aligned}
\langle \mathbf v_2,\mathbf u_1\rangle
&=
\begin{pmatrix}
1 & 0 & 1
\end{pmatrix}
\frac{1}{\sqrt{3}}
\begin{pmatrix}
1\\
1\\
1
\end{pmatrix}\\
&=
\frac{2}{\sqrt{3}}.
\end{aligned}
$$

Damit ist

$$
P_{\mathbf u_1}(\mathbf v_2)
=
\frac{2}{\sqrt{3}}
\frac{1}{\sqrt{3}}
\begin{pmatrix}
1\\
1\\
1
\end{pmatrix}
=
\begin{pmatrix}
\frac23\\
\frac23\\
\frac23
\end{pmatrix}.
$$

Der orthogonale Anteil lautet

$$
\begin{aligned}
\mathbf w_2
&=
\mathbf v_2-P_{\mathbf u_1}(\mathbf v_2)\\[1ex]
&=
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}
-
\begin{pmatrix}
\frac23\\
\frac23\\
\frac23
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
\frac13\\
-\frac23\\
\frac13
\end{pmatrix}.
\end{aligned}
$$

Nun wird $\mathbf w_2$ normiert:

$$
\|\mathbf w_2\|
=
\sqrt{
\frac19+\frac49+\frac19
}
=
\frac{\sqrt6}{3}.
$$

Somit ergibt sich

$$
\boxed{
\mathbf u_2
=
\frac{\mathbf w_2}{\|\mathbf w_2\|}
=
\frac{1}{\sqrt6}
\begin{pmatrix}
1\\
-2\\
1
\end{pmatrix}.
}
$$
:::

:::{dropdown} Schritt 3 – Bestimmung von $\mathbf u_3$

Für den dritten Vektor müssen die Projektionen auf $\mathbf u_1$ und $\mathbf u_2$ abgezogen werden:

$$
\mathbf w_3
=
\mathbf v_3
-
P_{\mathbf u_1}(\mathbf v_3)
-
P_{\mathbf u_2}(\mathbf v_3).
$$

Zunächst gilt

$$
\begin{aligned}
\langle\mathbf v_3,\mathbf u_1\rangle
&=
\begin{pmatrix}
0 & 1 & 1
\end{pmatrix}
\frac{1}{\sqrt3}
\begin{pmatrix}
1\\
1\\
1
\end{pmatrix}\\
&=
\frac{2}{\sqrt3}.
\end{aligned}
$$

Damit ist

$$
P_{\mathbf u_1}(\mathbf v_3)
=
\begin{pmatrix}
\frac23\\
\frac23\\
\frac23
\end{pmatrix}.
$$

Für die zweite Projektion gilt

$$
\begin{aligned}
\langle\mathbf v_3,\mathbf u_2\rangle
&=
\begin{pmatrix}
0 & 1 & 1
\end{pmatrix}
\frac{1}{\sqrt6}
\begin{pmatrix}
1\\
-2\\
1
\end{pmatrix}\\
&=
-\frac{1}{\sqrt6}.
\end{aligned}
$$

Damit ergibt sich

$$
P_{\mathbf u_2}(\mathbf v_3)
=
-\frac{1}{\sqrt6}
\frac{1}{\sqrt6}
\begin{pmatrix}
1\\
-2\\
1
\end{pmatrix}
=
\begin{pmatrix}
-\frac16\\
\frac13\\
-\frac16
\end{pmatrix}.
$$

Der orthogonale Anteil ist somit

$$
\begin{aligned}
\mathbf w_3
&=
\begin{pmatrix}
0\\
1\\
1
\end{pmatrix}
-
\begin{pmatrix}
\frac23\\
\frac23\\
\frac23
\end{pmatrix}
-
\begin{pmatrix}
-\frac16\\
\frac13\\
-\frac16
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
-\frac12\\
0\\
\frac12
\end{pmatrix}.
\end{aligned}
$$

Die Norm ist

$$
\|\mathbf w_3\|
=
\sqrt{\frac14+\frac14}
=
\frac{1}{\sqrt2}.
$$

Damit erhalten wir

$$
\boxed{
\mathbf u_3
=
\frac{\mathbf w_3}{\|\mathbf w_3\|}
=
\frac{1}{\sqrt2}
\begin{pmatrix}
-1\\
0\\
1
\end{pmatrix}.
}
$$
:::

:::{dropdown} Orthonormalität überprüfen

Die berechneten Vektoren sind

$$
\mathbf u_1
=
\frac{1}{\sqrt3}
\begin{pmatrix}
1\\
1\\
1
\end{pmatrix},
\qquad
\mathbf u_2
=
\frac{1}{\sqrt6}
\begin{pmatrix}
1\\
-2\\
1
\end{pmatrix},
\qquad
\mathbf u_3
=
\frac{1}{\sqrt2}
\begin{pmatrix}
-1\\
0\\
1
\end{pmatrix}.
$$

Für die Normen gilt

$$
\|\mathbf u_1\|
=
\|\mathbf u_2\|
=
\|\mathbf u_3\|
=
1.
$$

Außerdem gilt

$$
\langle\mathbf u_1,\mathbf u_2\rangle=0,
$$

$$
\langle\mathbf u_1,\mathbf u_3\rangle=0
$$

und

$$
\langle\mathbf u_2,\mathbf u_3\rangle=0.
$$

Damit bilden die drei Vektoren eine orthonormale Basis.
:::

\end{pmatrix},

\frac{1}{\sqrt6}
\begin{pmatrix}

}
$$
:::
