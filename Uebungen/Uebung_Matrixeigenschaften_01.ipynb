## Konjugierte und adjungierte Matrix

Gegeben sei die komplexe Matrix

$$
\mathbf{A}
=
\begin{pmatrix}
1+2i & -i & 3\\
2-i & 1+i & -2i\\
i & 4 & -1+3i
\end{pmatrix}.
$$

1. Bestimmen Sie die **konjugierte Matrix**

$$
\overline{\mathbf{A}}.
$$

2. Bestimmen Sie anschließend die **adjungierte Matrix**

$$
\mathbf{A}^*
=
\overline{\mathbf{A}}^{\,T}.
$$

## Lösung

Zur Berechnung der konjugierten Matrix wird jeder Eintrag von $\mathbf{A}$ durch seinen komplexen Konjugierten ersetzt.

Anschließend wird die konjugierte Matrix transponiert, um die adjungierte Matrix zu erhalten:

$$
\mathbf{A}^*
=
\overline{\mathbf{A}}^{\,T}.
$$

:::{dropdown} Konjugierte Matrix bestimmen

Gegeben ist

$$
\mathbf{A}
=
\begin{pmatrix}
1+2i & -i & 3\\
2-i & 1+i & -2i\\
i & 4 & -1+3i
\end{pmatrix}.
$$

Wir ersetzen jeden Eintrag durch seinen komplexen Konjugierten:

$$
\overline{1+2i}=1-2i,
\qquad
\overline{-i}=i,
\qquad
\overline{3}=3,
$$

$$
\overline{2-i}=2+i,
\qquad
\overline{1+i}=1-i,
\qquad
\overline{-2i}=2i,
$$

$$
\overline{i}=-i,
\qquad
\overline{4}=4,
\qquad
\overline{-1+3i}=-1-3i.
$$

Damit ergibt sich

$$
\boxed{
\overline{\mathbf{A}}
=
\begin{pmatrix}
1-2i & i & 3\\
2+i & 1-i & 2i\\
-i & 4 & -1-3i
\end{pmatrix}
}
$$

:::

:::{dropdown} Adjungierte Matrix bestimmen

Die adjungierte Matrix ist die konjugierte Transponierte von $\mathbf{A}$:

$$
\mathbf{A}^*
=
\overline{\mathbf{A}}^{\,T}.
$$

Aus

$$
\overline{\mathbf{A}}
=
\begin{pmatrix}
1-2i & i & 3\\
2+i & 1-i & 2i\\
-i & 4 & -1-3i
\end{pmatrix}
$$

wird durch Vertauschen von Zeilen und Spalten

$$
\boxed{
\mathbf{A}^*
=
\begin{pmatrix}
1-2i & 2+i & -i\\
i & 1-i & 4\\
3 & 2i & -1-3i
\end{pmatrix}.
}
$$

:::

:::{dropdown} Gesamtlösung anzeigen

Für

$$
\mathbf{A}
=
\begin{pmatrix}
1+2i & -i & 3\\
2-i & 1+i & -2i\\
i & 4 & -1+3i
\end{pmatrix}
$$

lautet die konjugierte Matrix

$$
\boxed{
\overline{\mathbf{A}}
=
\begin{pmatrix}
1-2i & i & 3\\
2+i & 1-i & 2i\\
-i & 4 & -1-3i
\end{pmatrix}.
}
$$

Die adjungierte Matrix erhält man durch Transponieren der konjugierten Matrix:

$$
\mathbf{A}^*
=
\overline{\mathbf{A}}^{\,T}.
$$

Damit folgt

$$
\boxed{
\mathbf{A}^*
=
\begin{pmatrix}
1-2i & 2+i & -i\\
i & 1-i & 4\\
3 & 2i & -1-3i
\end{pmatrix}.
}
$$

:::
