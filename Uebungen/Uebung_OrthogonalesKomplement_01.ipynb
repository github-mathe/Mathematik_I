## Orthogonales Komplement

Gegeben sei der Unterraum

$$
U
=
\operatorname{span}
\left\{
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix},
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}
\right\}
\subset\mathbb R^3.
$$

1. Bestimmen Sie das orthogonale Komplement $U^\perp$.
2. Geben Sie eine Basis von $U^\perp$ an.
3. Bestimmen Sie $\dim(U^\perp)$ und überprüfen Sie das Dimensionstheorem

$$
\dim(U)+\dim(U^\perp)=3.
$$

## Lösung

Ein Vektor gehört zum orthogonalen Komplement $U^\perp$, wenn er zu allen Basisvektoren von $U$ orthogonal ist.

:::{dropdown} Orthogonalitätsbedingungen aufstellen

Sei

$$
\mathbf v
=
\begin{pmatrix}
x\\
y\\
z
\end{pmatrix}
\in U^\perp.
$$

Dann muss $\mathbf v$ zu beiden Basisvektoren von $U$ orthogonal sein.

Für

$$
\mathbf u_1
=
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
$$

gilt daher

$$
\langle\mathbf v,\mathbf u_1\rangle=0.
$$

Damit erhalten wir

$$
x+y=0.
$$

Für

$$
\mathbf u_2
=
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}
$$

gilt entsprechend

$$
\langle\mathbf v,\mathbf u_2\rangle=0,
$$

also

$$
x+z=0.
$$

Damit erhalten wir das Gleichungssystem

$$
\boxed{
\begin{aligned}
x+y&=0,\\
x+z&=0.
\end{aligned}
}
$$
:::

:::{dropdown} Gleichungssystem lösen

Aus

$$
x+y=0
$$

folgt

$$
y=-x.
$$

Aus

$$
x+z=0
$$

folgt

$$
z=-x.
$$

Wir setzen

$$
x=t,
\qquad t\in\mathbb R.
$$

Damit ist

$$
\mathbf v
=
\begin{pmatrix}
t\\
-t\\
-t
\end{pmatrix}
=
t
\begin{pmatrix}
1\\
-1\\
-1
\end{pmatrix}.
$$

Somit lautet das orthogonale Komplement

$$
\boxed{
U^\perp
=
\operatorname{span}
\left\{
\begin{pmatrix}
1\\
-1\\
-1
\end{pmatrix}
\right\}.
}
$$
:::

:::{dropdown} Basis und Dimension bestimmen

Eine Basis des orthogonalen Komplements ist

$$
\boxed{
B_{U^\perp}
=
\left\{
\begin{pmatrix}
1\\
-1\\
-1
\end{pmatrix}
\right\}.
}
$$

Daher gilt

$$
\boxed{
\dim(U^\perp)=1.
}
$$

Die beiden gegebenen Vektoren von $U$ sind linear unabhängig. Somit ist

$$
\dim(U)=2.
$$

Damit folgt

$$
\dim(U)+\dim(U^\perp)
=
2+1
=
3.
$$

Das Dimensionstheorem ist somit erfüllt.
:::

:::{dropdown} Orthogonalität überprüfen

Für den Basisvektor

$$
\mathbf w
=
\begin{pmatrix}
1\\
-1\\
-1
\end{pmatrix}
$$

von $U^\perp$ gilt

$$
\begin{aligned}
\langle\mathbf w,\mathbf u_1\rangle
&=
1\cdot1+(-1)\cdot1+(-1)\cdot0\\
&=0
\end{aligned}
$$

und

$$
\begin{aligned}
\langle\mathbf w,\mathbf u_2\rangle
&=
1\cdot1+(-1)\cdot0+(-1)\cdot1\\
&=0.
\end{aligned}
$$

Damit ist $\mathbf w$ zu beiden Basisvektoren von $U$ orthogonal und somit

$$
\boxed{
\mathbf w\in U^\perp.
}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Aus den Orthogonalitätsbedingungen

$$
x+y=0
$$

und

$$
x+z=0
$$

folgt

$$
y=-x,
\qquad
z=-x.
$$

Damit gilt

$$
\boxed{
U^\perp
=
\operatorname{span}
\left\{
\begin{pmatrix}
1\\
-1\\
-1
\end{pmatrix}
\right\}.
}
$$

Eine Basis ist somit

$$
\boxed{
B_{U^\perp}
=
\left\{
\begin{pmatrix}
1\\
-1\\
-1
\end{pmatrix}
\right\}.
}
$$

Weiter gilt

$$
\boxed{
\dim(U^\perp)=1.
}
$$

Da

$$
\dim(U)=2,
$$

ist

$$
\boxed{
\dim(U)+\dim(U^\perp)=2+1=3.
}
$$
:::
