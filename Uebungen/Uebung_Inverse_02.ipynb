## Berechnung der Inversen mittels LU-Zerlegung

Gegeben sei die Matrix

$$
\mathbf A
=
\begin{pmatrix}
1 & 1 & 0\\
2 & 3 & 1\\
-1 & 0 & 2
\end{pmatrix}.
$$

1. Bestimmen Sie die LU-Zerlegung

$$
\mathbf A=\mathbf L\mathbf U.
$$

2. Berechnen Sie anschließend die inverse Matrix $\mathbf A^{-1}$ mithilfe von Vorwärts- und Rückwärtssubstitution.

3. Überprüfen Sie das Ergebnis durch Berechnung von

$$
\mathbf A\mathbf A^{-1}.
$$

## Lösung

Zur Bestimmung der Inversen wird zunächst die LU-Zerlegung der Matrix berechnet.

Anschließend lösen wir

$$
\mathbf L\mathbf Y=\mathbf I_3
$$

mittels Vorwärtssubstitution und danach

$$
\mathbf U\mathbf X=\mathbf Y
$$

mittels Rückwärtssubstitution.

Die Matrix $\mathbf X$ ist schließlich die gesuchte inverse Matrix $\mathbf A^{-1}$.

::::{dropdown} Durchführung der LU-Zerlegung

Ausgangspunkt ist

$$
\mathbf A
=
\begin{pmatrix}
1 & 1 & 0\\
2 & 3 & 1\\
-1 & 0 & 2
\end{pmatrix}.
$$

:::{dropdown} Schritt 1

Die Einträge unterhalb des ersten Pivotelements werden eliminiert.

$$
\left(
\begin{array}{ccc}
1 & 1 & 0\\
2 & 3 & 1\\
-1 & 0 & 2
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'=\mathrm{II}-2\cdot\mathrm{I}\\
\mathrm{III}'=\mathrm{III}+\mathrm{I}
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\begin{pmatrix}
1 & 1 & 0\\
0 & 1 & 1\\
0 & 1 & 2
\end{pmatrix}
}
$$

Die verwendeten Multiplikatoren sind

$$
m_{21}=2,
\qquad
m_{31}=-1.
$$

:::

:::{dropdown} Schritt 2

Nun wird der Eintrag unterhalb des zweiten Pivotelements eliminiert.

$$
\left(
\begin{array}{ccc}
1 & 1 & 0\\
0 & 1 & 1\\
0 & 1 & 2
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}''=\mathrm{III}'-\mathrm{II}'
\end{array}
$$

Damit ergibt sich die obere Dreiecksmatrix

$$
\boxed{
\mathbf U
=
\begin{pmatrix}
1 & 1 & 0\\
0 & 1 & 1\\
0 & 0 & 1
\end{pmatrix}.
}
$$

Der weitere Multiplikator ist

$$
m_{32}=1.
$$

Die Matrix $\mathbf L$ enthält die während der Elimination verwendeten Multiplikatoren:

$$
\boxed{
\mathbf L
=
\begin{pmatrix}
1 & 0 & 0\\
2 & 1 & 0\\
-1 & 1 & 1
\end{pmatrix}.
}
$$

Damit gilt

$$
\boxed{
\mathbf A=\mathbf L\mathbf U.
}
$$

:::

::::

:::{dropdown} Vorwärtssubstitution

Wir lösen

$$
\mathbf L\mathbf Y=\mathbf I_3.
$$

Mit

$$
\mathbf L
=
\begin{pmatrix}
1 & 0 & 0\\
2 & 1 & 0\\
-1 & 1 & 1
\end{pmatrix}
$$

ergibt sich

$$
\begin{pmatrix}
1 & 0 & 0\\
2 & 1 & 0\\
-1 & 1 & 1
\end{pmatrix}
\begin{pmatrix}
y_{11} & y_{12} & y_{13}\\
y_{21} & y_{22} & y_{23}\\
y_{31} & y_{32} & y_{33}
\end{pmatrix}
=
\begin{pmatrix}
1 & 0 & 0\\
0 & 1 & 0\\
0 & 0 & 1
\end{pmatrix}.
$$

Aus der ersten Zeile folgt

$$
y_{11}=1,
\qquad
y_{12}=0,
\qquad
y_{13}=0.
$$

Aus der zweiten Zeile folgt

$$
2y_{11}+y_{21}=0
\quad\Rightarrow\quad
2\cdot1+y_{21}=0
\quad\Rightarrow\quad
y_{21}=-2,
$$

$$
2y_{12}+y_{22}=1
\quad\Rightarrow\quad
2\cdot0+y_{22}=1
\quad\Rightarrow\quad
y_{22}=1,
$$

$$
2y_{13}+y_{23}=0
\quad\Rightarrow\quad
2\cdot0+y_{23}=0
\quad\Rightarrow\quad
y_{23}=0.
$$

Aus der dritten Zeile folgt

$$
-y_{11}+y_{21}+y_{31}=0,
$$

also

$$
-1-2+y_{31}=0
\quad\Rightarrow\quad
y_{31}=3.
$$

Weiter gilt

$$
-y_{12}+y_{22}+y_{32}=0,
$$

also

$$
-0+1+y_{32}=0
\quad\Rightarrow\quad
y_{32}=-1.
$$

Schließlich gilt

$$
-y_{13}+y_{23}+y_{33}=1,
$$

also

$$
-0+0+y_{33}=1
\quad\Rightarrow\quad
y_{33}=1.
$$

Damit erhalten wir

$$
\boxed{
\mathbf Y
=
\begin{pmatrix}
1 & 0 & 0\\
-2 & 1 & 0\\
3 & -1 & 1
\end{pmatrix}.
}
$$
:::

:::{dropdown} Rückwärtssubstitution

Nun lösen wir

$$
\mathbf U\mathbf X=\mathbf Y.
$$

Mit

$$
\mathbf U
=
\begin{pmatrix}
1 & 1 & 0\\
0 & 1 & 1\\
0 & 0 & 1
\end{pmatrix}
$$

und

$$
\mathbf Y
=
\begin{pmatrix}
1 & 0 & 0\\
-2 & 1 & 0\\
3 & -1 & 1
\end{pmatrix}
$$

ergibt sich

$$
\begin{pmatrix}
1 & 1 & 0\\
0 & 1 & 1\\
0 & 0 & 1
\end{pmatrix}
\begin{pmatrix}
x_{11} & x_{12} & x_{13}\\
x_{21} & x_{22} & x_{23}\\
x_{31} & x_{32} & x_{33}
\end{pmatrix}
=
\begin{pmatrix}
1 & 0 & 0\\
-2 & 1 & 0\\
3 & -1 & 1
\end{pmatrix}.
$$

Wir beginnen mit der dritten Zeile:

$$
x_{31}=3,
\qquad
x_{32}=-1,
\qquad
x_{33}=1.
$$

Aus der zweiten Zeile folgt

$$
x_{21}+x_{31}=-2
\quad\Rightarrow\quad
x_{21}+3=-2
\quad\Rightarrow\quad
x_{21}=-5,
$$

$$
x_{22}+x_{32}=1
\quad\Rightarrow\quad
x_{22}-1=1
\quad\Rightarrow\quad
x_{22}=2,
$$

$$
x_{23}+x_{33}=0
\quad\Rightarrow\quad
x_{23}+1=0
\quad\Rightarrow\quad
x_{23}=-1.
$$

Aus der ersten Zeile folgt

$$
x_{11}+x_{21}=1
\quad\Rightarrow\quad
x_{11}-5=1
\quad\Rightarrow\quad
x_{11}=6,
$$

$$
x_{12}+x_{22}=0
\quad\Rightarrow\quad
x_{12}+2=0
\quad\Rightarrow\quad
x_{12}=-2,
$$

$$
x_{13}+x_{23}=0
\quad\Rightarrow\quad
x_{13}-1=0
\quad\Rightarrow\quad
x_{13}=1.
$$

Damit ergibt sich

$$
\boxed{
\mathbf X
=
\begin{pmatrix}
6 & -2 & 1\\
-5 & 2 & -1\\
3 & -1 & 1
\end{pmatrix}.
}
$$

Da $\mathbf X=\mathbf A^{-1}$ gilt,

$$
\boxed{
\mathbf A^{-1}
=
\begin{pmatrix}
6 & -2 & 1\\
-5 & 2 & -1\\
3 & -1 & 1
\end{pmatrix}.
}
$$
:::

:::{dropdown} Inverse überprüfen

Zur Kontrolle berechnen wir

$$
\mathbf A\mathbf A^{-1}.
$$

Es gilt

$$
\begin{aligned}
\mathbf A\mathbf A^{-1}
&=
\begin{pmatrix}
1 & 1 & 0\\
2 & 3 & 1\\
-1 & 0 & 2
\end{pmatrix}
\begin{pmatrix}
6 & -2 & 1\\
-5 & 2 & -1\\
3 & -1 & 1
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
1 & 0 & 0\\
0 & 1 & 0\\
0 & 0 & 1
\end{pmatrix}.
\end{aligned}
$$

Damit gilt

$$
\boxed{
\mathbf A\mathbf A^{-1}
=
\mathbf I_3.
}
$$

Die berechnete inverse Matrix ist damit bestätigt.
:::

:::{dropdown} Gesamtlösung anzeigen

Für

$$
\mathbf A
=
\begin{pmatrix}
1 & 1 & 0\\
2 & 3 & 1\\
-1 & 0 & 2
\end{pmatrix}
$$

ergibt sich die LU-Zerlegung

$$
\boxed{
\mathbf L
=
\begin{pmatrix}
1 & 0 & 0\\
2 & 1 & 0\\
-1 & 1 & 1
\end{pmatrix},
\qquad
\mathbf U
=
\begin{pmatrix}
1 & 1 & 0\\
0 & 1 & 1\\
0 & 0 & 1
\end{pmatrix}.
}
$$

Aus

$$
\mathbf L\mathbf Y=\mathbf I_3
$$

folgt

$$
\mathbf Y
=
\begin{pmatrix}
1 & 0 & 0\\
-2 & 1 & 0\\
3 & -1 & 1
\end{pmatrix}.
$$

Durch Lösung von

$$
\mathbf U\mathbf X=\mathbf Y
$$

erhält man

$$
\boxed{
\mathbf A^{-1}
=
\mathbf X
=
\begin{pmatrix}
6 & -2 & 1\\
-5 & 2 & -1\\
3 & -1 & 1
\end{pmatrix}.
}
$$

Zur Überprüfung gilt

$$
\boxed{
\mathbf A\mathbf A^{-1}
=
\mathbf I_3.
}
$$
:::
