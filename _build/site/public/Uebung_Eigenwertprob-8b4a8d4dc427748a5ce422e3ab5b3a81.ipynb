## Eigenwerte und Eigenvektoren

Gegeben sei die Matrix

$$
\mathbf A
=
\begin{pmatrix}
2 & 1 & 0\\
1 & 2 & 0\\
0 & 0 & 4
\end{pmatrix}.
$$

1. Bestimmen Sie das charakteristische Polynom $p(\lambda)$.
2. Bestimmen Sie die Eigenwerte von $\mathbf A$.
3. Bestimmen Sie zu jedem Eigenwert die zugehörigen Eigenvektoren. Verwenden Sie hierzu die Gauß-Elimination.
4. Überprüfen Sie die gefundenen Eigenvektoren mit

$$
\mathbf A\mathbf v=\lambda\mathbf v.
$$

## Lösung

Das Eigenwertproblem wird in zwei Schritten gelöst:

1. Bestimmung der Eigenwerte aus dem charakteristischen Polynom.
2. Bestimmung der zugehörigen Eigenvektoren aus dem Kern von $\mathbf A-\lambda\mathbf I$.

:::{dropdown} Charakteristisches Polynom bestimmen

Das charakteristische Polynom ist definiert durch

$$
p(\lambda)
=
\det(\mathbf A-\lambda\mathbf I).
$$

Es gilt

$$
\mathbf A-\lambda\mathbf I
=
\begin{pmatrix}
2-\lambda & 1 & 0\\
1 & 2-\lambda & 0\\
0 & 0 & 4-\lambda
\end{pmatrix}.
$$

Damit ergibt sich

$$
\begin{aligned}
p(\lambda)
&=
\det
\begin{pmatrix}
2-\lambda & 1 & 0\\
1 & 2-\lambda & 0\\
0 & 0 & 4-\lambda
\end{pmatrix}\\[1ex]
&=
(4-\lambda)
\left(
(2-\lambda)^2-1
\right)\\[1ex]
&=
(4-\lambda)
\left(
\lambda^2-4\lambda+3
\right)\\[1ex]
&=
\boxed{
(4-\lambda)(\lambda-1)(\lambda-3)
}.
\end{aligned}
$$
:::

:::{dropdown} Eigenwerte bestimmen

Die Eigenwerte sind die Nullstellen des charakteristischen Polynoms:

$$
(4-\lambda)(\lambda-1)(\lambda-3)=0.
$$

Daraus folgen die drei Eigenwerte

$$
\boxed{
\lambda_1=1,
\qquad
\lambda_2=3,
\qquad
\lambda_3=4.
}
$$
:::

:::{dropdown} Eigenvektoren zu $\lambda_1=1$

Zur Bestimmung der Eigenvektoren lösen wir

$$
(\mathbf A-\lambda_1\mathbf I)\mathbf v=\mathbf 0.
$$

Für $\lambda_1=1$ gilt

$$
\mathbf A-\mathbf I
=
\begin{pmatrix}
1 & 1 & 0\\
1 & 1 & 0\\
0 & 0 & 3
\end{pmatrix}.
$$

Wir wenden die Gauß-Elimination an:

$$
\left(
\begin{array}{ccc}
1 & 1 & 0\\
1 & 1 & 0\\
0 & 0 & 3
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'=\mathrm{II}-\mathrm{I}\\
\\
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\begin{pmatrix}
1 & 1 & 0\\
0 & 0 & 0\\
0 & 0 & 3
\end{pmatrix}.
}
$$

Aus dem Gleichungssystem folgt

$$
x_1+x_2=0
$$

und

$$
x_3=0.
$$

Die Variable $x_2$ ist frei. Wir setzen

$$
x_2=t,
\qquad t\in\mathbb R.
$$

Dann folgt

$$
x_1=-t
$$

und

$$
x_3=0.
$$

Damit gilt

$$
\mathbf v
=
t
\begin{pmatrix}
-1\\
1\\
0
\end{pmatrix}.
$$

Eine Basis des zugehörigen Kerns ist somit

$$
\boxed{
\left\{
\begin{pmatrix}
-1\\
1\\
0
\end{pmatrix}
\right\}.
}
$$
:::

:::{dropdown} Eigenvektoren zu $\lambda_2=3$

Für $\lambda_2=3$ lösen wir

$$
(\mathbf A-3\mathbf I)\mathbf v=\mathbf 0.
$$

Es gilt

$$
\mathbf A-3\mathbf I
=
\begin{pmatrix}
-1 & 1 & 0\\
1 & -1 & 0\\
0 & 0 & 1
\end{pmatrix}.
$$

Wir wenden die Gauß-Elimination an:

$$
\left(
\begin{array}{ccc}
-1 & 1 & 0\\
1 & -1 & 0\\
0 & 0 & 1
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'=\mathrm{II}+\mathrm{I}\\
\\
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\begin{pmatrix}
-1 & 1 & 0\\
0 & 0 & 0\\
0 & 0 & 1
\end{pmatrix}.
}
$$

Daraus folgt

$$
-x_1+x_2=0
$$

und

$$
x_3=0.
$$

Also gilt

$$
x_2=x_1.
$$

Mit

$$
x_1=t
$$

erhalten wir

$$
\mathbf v
=
t
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}.
$$

Eine Basis des zugehörigen Kerns ist

$$
\boxed{
\left\{
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
\right\}.
}
$$
:::

:::{dropdown} Eigenvektoren zu $\lambda_3=4$

Für $\lambda_3=4$ lösen wir

$$
(\mathbf A-4\mathbf I)\mathbf v=\mathbf 0.
$$

Es gilt

$$
\mathbf A-4\mathbf I
=
\begin{pmatrix}
-2 & 1 & 0\\
1 & -2 & 0\\
0 & 0 & 0
\end{pmatrix}.
$$

Wir wenden die Gauß-Elimination an:

$$
\left(
\begin{array}{ccc}
-2 & 1 & 0\\
1 & -2 & 0\\
0 & 0 & 0
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'=2\cdot\mathrm{II}+\mathrm{I}\\
\\
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\begin{pmatrix}
-2 & 1 & 0\\
0 & -3 & 0\\
0 & 0 & 0
\end{pmatrix}.
}
$$

Aus der zweiten Zeile folgt

$$
x_2=0.
$$

Damit ergibt die erste Zeile

$$
x_1=0.
$$

Die Variable $x_3$ ist frei. Wir setzen

$$
x_3=t.
$$

Somit ist

$$
\mathbf v
=
t
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}.
$$

Eine Basis des zugehörigen Kerns ist

$$
\boxed{
\left\{
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}
\right\}.
}
$$
:::

:::{dropdown} Eigenvektoren überprüfen

Für

$$
\lambda_1=1,
\qquad
\mathbf v_1=
\begin{pmatrix}
1\\
-1\\
0
\end{pmatrix}
$$

gilt

$$
\mathbf A\mathbf v_1
=
\begin{pmatrix}
1\\
-1\\
0
\end{pmatrix}
=
1\cdot\mathbf v_1.
$$

Für

$$
\lambda_2=3,
\qquad
\mathbf v_2=
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
$$

gilt

$$
\mathbf A\mathbf v_2
=
\begin{pmatrix}
3\\
3\\
0
\end{pmatrix}
=
3\cdot\mathbf v_2.
$$

Für

$$
\lambda_3=4,
\qquad
\mathbf v_3=
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}
$$

gilt

$$
\mathbf A\mathbf v_3
=
\begin{pmatrix}
0\\
0\\
4
\end{pmatrix}
=
4\cdot\mathbf v_3.
$$

Damit sind die gefundenen Eigenwerte und Eigenvektoren bestätigt.
:::

:::{dropdown} Gesamtlösung anzeigen

Das charakteristische Polynom lautet

$$
\boxed{
p(\lambda)
=
(4-\lambda)(\lambda-1)(\lambda-3).
}
$$

Die Eigenwerte sind

$$
\boxed{
\lambda_1=1,
\qquad
\lambda_2=3,
\qquad
\lambda_3=4.
}
$$

Dazu können beispielsweise die folgenden Eigenvektoren gewählt werden:

$$
\boxed{
\mathbf v_1=
\begin{pmatrix}
1\\
-1\\
0
\end{pmatrix},
\qquad
\mathbf v_2=
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix},
\qquad
\mathbf v_3=
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}.
}
$$
:::
