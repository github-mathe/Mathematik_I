## Von Mengen zu Gruppen

Mengen bilden die Grundlage vieler algebraischer Strukturen. Eine der einfachsten, aber wichtigsten Strukturen ist die Gruppe. Gruppen fügen zusätzliche Eigenschaften zu Mengen und Verknüpfungen hinzu, wie wir im Folgenden sehen werden.

<a id="gruppe"></a>

## Gruppe

### Definition einer Gruppe

Eine Gruppe $(G,\cdot)$ besteht aus einer Menge $G$ und einer Verknüpfung $\cdot$, die die folgenden Eigenschaften erfüllt:

1. **Abgeschlossenheit**:

   $$
   a\cdot b\in G
   $$

   für alle $a,b\in G$.

2. **Assoziativität**:

   $$
   (a\cdot b)\cdot c
   =
   a\cdot(b\cdot c)
   $$

   für alle $a,b,c\in G$.

3. **Neutrales Element**:

   Es gibt ein Element $e\in G$, sodass

   $$
   e\cdot a
   =
   a\cdot e
   =
   a
   $$

   für alle $a\in G$.

4. **Inverses Element**:

   Für jedes $a\in G$ existiert ein $a^{-1}\in G$, sodass

   $$
   a\cdot a^{-1}
   =
   a^{-1}\cdot a
   =
   e.
   $$

Eine **Verknüpfung** oder auch **Operation** ist in der Mathematik eine Regel, die zwei Elemente einer Menge nimmt und ihnen ein drittes Element derselben Menge zuordnet.

### Kommutativität (Abelsche Gruppe)

- Eine Gruppe ist *abelsch* oder *kommutativ*, wenn zusätzlich gilt:

$$
a\cdot b
=
b\cdot a
\qquad
\text{für alle }
a,b\in G.
$$

- Diese Bedingung ist optional. Nicht alle Gruppen sind abelsch.

### Beispiel 1: Ganze Zahlen unter Addition

Die Menge der ganzen Zahlen $\mathbb Z$ mit der Addition $+$ bildet eine Gruppe:

- Die Menge:

$$
\mathbb Z
=
\{\ldots,-3,-2,-1,0,1,2,3,\ldots\}.
$$

- Die Operation: Addition $+$.
- Das neutrale Element: $0$, da

$$
a+0=a.
$$

- Inverses Element: $-a$, da

$$
a+(-a)=0.
$$

- Diese Gruppe ist **abelsch**, weil

$$
a+b=b+a
$$

für alle $a,b\in\mathbb Z$.

### Beispiel 2: Reelle Zahlen ohne Null unter Multiplikation

Die Menge der reellen Zahlen ohne $0$,

$$
\mathbb R\setminus\{0\},
$$

mit der Multiplikation bildet eine Gruppe:

- Die Menge:

$$
\mathbb R\setminus\{0\}.
$$

- Die Operation: Multiplikation $\cdot$.
- Das neutrale Element: $1$, da

$$
a\cdot1=a.
$$

- Inverses Element:

$$
\frac1a,
$$

da

$$
a\cdot\frac1a=1.
$$

- Diese Gruppe ist **abelsch**, weil

$$
a\cdot b=b\cdot a.
$$

## Von Gruppen zu Ringen und Körpern

Während Gruppen eine einzelne Verknüpfung betrachten, beinhalten **Ringe** und **Körper** zwei Verknüpfungen: Addition und Multiplikation. Diese Strukturen sind in der Algebra ebenso wichtig und erweitern die Konzepte der Gruppen.
