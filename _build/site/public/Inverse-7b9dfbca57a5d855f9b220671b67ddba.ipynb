# Berechnung der Inverse einer Matrix

Die Berechnung der Inversen einer Matrix ist ein zentrales Verfahren in der linearen Algebra und wird in zahlreichen Anwendungen verwendet.


### Bedingungen für die Invertierbarkeit

- Die Matrix muss quadratisch sein.
- Die Matrix muss **regulär** sein, d. h.

  $$
  \det(\mathbf{A}) \neq 0.
  $$

- Für eine **singuläre Matrix**

  $$
  \det(\mathbf{A}) = 0
  $$

  existiert keine Inverse.

### Anwendungen der Inversen einer Matrix

- **Lösen linearer Gleichungssysteme:** Für ein System

  $$
  \mathbf{A}\mathbf{x}=\mathbf{b}
  $$

  ist die Lösung

  $$
  \mathbf{x}=\mathbf{A}^{-1}\mathbf{b},
  $$

  falls $\mathbf{A}^{-1}$ existiert.

- **Transformationen in der Geometrie und Physik:** Die Inverse wird verwendet, um Koordinaten zurückzutransformieren oder Transformationen rückgängig zu machen.

- **Statistik:** Bei Berechnungen von Varianzen und Kovarianzen in der Statistik, insbesondere in der multivariaten Analyse.

- **Optimierung:** Für iterative Methoden zur Lösung nichtlinearer Optimierungsprobleme.

## Methoden zur Berechnung der Inversen

### Gauß-Jordan-Algorithmus

- Erweiterung der Matrix $\vec{A}$ mit der Identitätsmatrix $\vec{I}_n$ zu

  $$
  [\vec{A}\mid\vec{I}_n].
  $$

- Durchführung von Zeilenoperationen, um $\vec{A}$ in die Identitätsmatrix zu überführen.
- Die transformierte rechte Hälfte ergibt dann $\vec{A}^{-1}$.

### Adjungierte Matrix

- Berechnung der Kofaktorenmatrix von $\vec{A}$.
- Transponieren der Kofaktorenmatrix ergibt die Adjunkte $\operatorname{adj}(\vec{A})$.
- Die Inverse ist dann gegeben durch:

$$
\vec{A}^{-1}
=
\frac{1}{\det(\vec{A})}
\cdot
\operatorname{adj}(\vec{A}).
$$

### LU-Zerlegung

- Zerlegung der Matrix $\vec{A}$ in das Produkt einer unteren Dreiecksmatrix $\vec{L}$ und einer oberen Dreiecksmatrix $\vec{U}$:

  $$
  \vec{A}=\vec{L}\vec{U}.
  $$

- Lösung der Gleichung

  $$
  \vec{A}\vec{X}=\vec{I}_n
  $$

  durch Vorwärtssubstitution und Rückwärtssubstitution.

- Ergebnis ist die Inverse $\vec{A}^{-1}$.

## Gauß-Jordan-Algorithmus

### Schritte

1. Erweiterung der Matrix $\vec{A}$ mit der Identitätsmatrix $\vec{I}_n$ zu

   $$
   [\vec{A}\mid\vec{I}_n].
   $$

2. Durchführung von Zeilenoperationen, um $\vec{A}$ in die Identitätsmatrix zu überführen.

3. Die transformierte rechte Hälfte ergibt dann $\vec{A}^{-1}$.

### Beispiel

$$
\left(
\begin{array}{cc|cc}
1&2&1&0\\
3&4&0&1
\end{array}
\right)
\quad\longrightarrow\quad
\left(
\begin{array}{cc|cc}
1&0&-2&1\\
0&1&1.5&-0.5
\end{array}
\right)
$$

Damit ergibt sich:

$$
\vec{A}^{-1}
=
\begin{pmatrix}
-2&1\\
1.5&-0.5
\end{pmatrix}.
$$

### Beispiel: Gauß-Jordan-Algorithmus

Ausgangspunkt:

$$
\begin{array}{c|cc|cc}
\text{I}&1&2&1&0\\
\text{II}&3&4&0&1
\end{array}
$$

### Schritt 1

$$
\begin{array}{c|cc|cc|l}
\text{I}&1&2&1&0&\\
\text{II'}&0&-2&-3&1&
\text{II'}=\text{II}-3\text{I}
\end{array}
$$

### Schritt 2

$$
\begin{array}{c|cc|cc|l}
\text{I}&1&2&1&0&\\
\text{II''}&0&1&3/2&-1/2&
\text{II''}=-1/2\,\text{II'}
\end{array}
$$

### Schritt 3

$$
\begin{array}{c|cc|cc|l}
\text{I'}&1&0&-2&1&
\text{I'}=\text{I}-2\text{II''}\\
\text{II''}&0&1&3/2&-1/2&
\end{array}
$$

### Endergebnis

$$
\boxed{
\vec{A}^{-1}
=
\begin{pmatrix}
-2&1\\
3/2&-1/2
\end{pmatrix}
}
$$

```python
import sympy as sp

A = sp.Matrix([
    [1, 2],
    [3, 4]
])

A_inv = A.inv()

print("A^(-1) =")
sp.pprint(A_inv)

print("\nÜberprüfung A * A^(-1) =")
sp.pprint(A * A_inv)
```

### Gauß-Jordan-Algorithmus: Beispiel $3\times3$

Gegeben sei:

$$
\begin{array}{c|ccc|ccc}
\text{I}&2&3&-1&1&0&0\\
\text{II}&4&7&-3&0&1&0\\
\text{III}&6&5&2&0&0&1
\end{array}
$$

### Schritt 1

$$
\begin{array}{c|ccc|ccc|l}
\text{I}&2&3&-1&1&0&0&\\
\text{II'}&0&1&-1&-2&1&0&
\text{II'}=\text{II}-2\text{I}\\
\text{III'}&0&-4&5&-3&0&1&
\text{III'}=\text{III}-3\text{I}
\end{array}
$$

### Schritt 2

$$
\begin{array}{c|ccc|ccc|l}
\text{I}&2&3&-1&1&0&0&\\
\text{II'}&0&1&-1&-2&1&0&\\
\text{III''}&0&0&1&-11&4&1&
\text{III''}=\text{III'}+4\text{II'}
\end{array}
$$

### Schritt 3

$$
\begin{array}{c|ccc|ccc|l}
\text{I'}&2&3&0&-10&4&1&
\text{I'}=\text{I}+\text{III'''}\\
\text{II''}&0&1&0&-13&5&1&
\text{II''}=\text{II'}+\text{III'''}\\
\text{III'''}&0&0&1&-11&4&1&
\end{array}
$$

### Schritt 4

$$
\begin{array}{c|ccc|ccc|l}
\text{I''}&2&0&0&29&-11&-2&
\text{I''}=\text{I'}-3\text{II'}\\
\text{II''}&0&1&0&-13&5&1&\\
\text{III'''}&0&0&1&-11&4&1&
\end{array}
$$

### Schritt 5

$$
\begin{array}{c|ccc|ccc|l}
\text{I'''}&1&0&0&29/2&-11/2&-1&
\text{I'''}=1/2\,\text{I''}\\
\text{I''}&0&1&0&-13&5&1&\\
\text{III'''}&0&0&1&-11&4&1&
\end{array}
$$

### Endergebnis

$$
\boxed{
\vec{A}^{-1}
=
\begin{pmatrix}
29/2&-11/2&-1\\
-13&5&1\\
-11&4&1
\end{pmatrix}
}
$$

```python
import sympy as sp

A = sp.Matrix([
    [2, 3, -1],
    [4, 7, -3],
    [6, 5, 2]
])

A_inv = A.inv()

print("A^(-1) =")
sp.pprint(A_inv)

print("\nÜberprüfung:")
sp.pprint(A * A_inv)
```

## LU-Zerlegung

### Definition

Zerlegung der Matrix $\vec{A}$ in das Produkt einer unteren Dreiecksmatrix $\vec{L}$ und einer oberen Dreiecksmatrix $\vec{U}$:

$$
\boxed{
\vec{A}=\vec{L}\vec{U}
}
$$

### Schritte

1. Durchführung von Gauß-Elimination, um $\vec{A}$ in $\vec{U}$ zu überführen.
2. Die Matrix $\vec{L}$ enthält die Multiplikatoren, die während der Elimination verwendet wurden.
3. Lösung der Gleichung

   $$
   \vec{A}\vec{X}=\vec{I}_n
   $$

   durch Vorwärtssubstitution und Rückwärtssubstitution.

4. Ergebnis ist die Inverse $\vec{A}^{-1}$.

### Beispiel

$$
\vec{A}
=
\begin{pmatrix}
2&3\\
4&7
\end{pmatrix}
$$

mit

$$
\vec{L}
=
\begin{pmatrix}
1&0\\
2&1
\end{pmatrix},
\qquad
\vec{U}
=
\begin{pmatrix}
2&3\\
0&1
\end{pmatrix}.
$$

Im Skript wird anschließend angegeben:

$$
\vec{A}^{-1}
=
\vec{U}^{-1}\vec{L}^{-1}
=
\begin{pmatrix}
0.5&-1.5\\
0&1
\end{pmatrix}
\begin{pmatrix}
1&0\\
-2&1
\end{pmatrix}
=
\begin{pmatrix}
0.5&-1.5\\
-2&1
\end{pmatrix}.
$$

### Beispiel zur LU-Zerlegung

Gegeben:

$$
\vec{A}
=
\begin{pmatrix}
2&3\\
4&7
\end{pmatrix}.
$$

Ziel: Zerlegung von $\vec{A}$ in $\vec{L}$ und $\vec{U}$ durch Gaussian Elimination.

$$
\begin{array}{c|cc}
\text{I}&2&3\\
\text{II}&4&7
\end{array}
\quad
\longrightarrow
\quad
\begin{array}{c|cc|l}
\text{I}&2&3&\\
\text{II'}&0&1&
\text{II'}=\text{II}-2\text{I}
\end{array}
$$

Die obere Dreiecksmatrix $\vec{U}$ hat nun die Form:

$$
\vec{U}
=
\begin{pmatrix}
2&3\\
0&1
\end{pmatrix}.
$$

### Matrix $\vec{L}$

Die Matrix $\vec{L}$ enthält die Multiplikatoren, die während der Elimination verwendet wurden.

$\vec{L}$ ist eine untere Dreiecksmatrix mit Einsen auf der Diagonale.

Der Multiplikator zur Elimination von $a_{21}$ ist:

$$
m_{21}=2.
$$

Damit ergibt sich:

$$
\vec{L}
=
\begin{pmatrix}
1&0\\
2&1
\end{pmatrix}.
$$

Die Zerlegung lautet:

$$
\boxed{
\vec{A}
=
\vec{L}\vec{U}
=
\begin{pmatrix}
1&0\\
2&1
\end{pmatrix}
\begin{pmatrix}
2&3\\
0&1
\end{pmatrix}
}
$$

### Gestalt der Matrix $\vec{L}$

Die Matrix $\vec{L}$ enthält die Multiplikatoren $m_{ij}$, die während der Elimination verwendet werden.

Der Multiplikator $m_{ij}$ wird verwendet, um das Element $a_{ij}$ mit $i>j$ zu eliminieren.

Berechnung des Multiplikators:

$$
m_{ij}
=
\frac{a_{ij}}{a_{jj}}.
$$

### Allgemeine Form der Matrix $\vec{L}$

$$
\vec{L}
=
\begin{pmatrix}
1&0&0&\cdots&0\\
m_{21}&1&0&\cdots&0\\
m_{31}&m_{32}&1&\cdots&0\\
\vdots&\vdots&\vdots&\ddots&0\\
m_{n1}&m_{n2}&m_{n3}&\cdots&1
\end{pmatrix}.
$$

## LU-Zerlegung zur Invertierung einer Matrix

Die LU-Zerlegung teilt eine Matrix $\vec{A}$ in das Produkt einer unteren Dreiecksmatrix $\vec{L}$ und einer oberen Dreiecksmatrix $\vec{U}$:

$$
\vec{A}
=
\vec{L}\vec{U}.
$$

Ziel ist die Lösung des Gleichungssystems

$$
\vec{A}\vec{X}
=
\vec{I}_n
$$

zur Bestimmung der **Inversen** $\vec{A}^{-1}$.

### Vorwärtssubstitution

Gegeben:

$$
\vec{A}
=
\vec{L}\vec{U}
$$

und

$$
\vec{A}\vec{X}
=
\vec{I}_n.
$$

Damit gilt:

$$
\vec{L}\vec{U}\vec{X}
=
\vec{I}_n.
$$

Setze

$$
\vec{U}\vec{X}
=
\vec{Y}.
$$

Dann folgt:

$$
\boxed{
\vec{L}\vec{Y}
=
\vec{I}_n
}
$$

**Ziel:** Bestimmung von $\vec{Y}$ mittels Vorwärtssubstitution.

### Vorwärtssubstitution: Vorgehensweise

1. Löse das Gleichungssystem

   $$
   \vec{L}\vec{Y}
   =
   \vec{I}_n.
   $$

2. Da $\vec{L}$ eine untere Dreiecksmatrix ist, kann $\vec{Y}$ Schritt für Schritt berechnet werden.

### Beispiel

$$
\vec{L}
=
\begin{pmatrix}
1&0\\
2&1
\end{pmatrix},
\qquad
\vec{I}_2
=
\begin{pmatrix}
1&0\\
0&1
\end{pmatrix}.
$$

Gegeben ist:

$$
\begin{pmatrix}
1&0\\
2&1
\end{pmatrix}
\begin{pmatrix}
y_{11}&y_{12}\\
y_{21}&y_{22}
\end{pmatrix}
=
\begin{pmatrix}
1&0\\
0&1
\end{pmatrix}.
$$

### Schritt 1: Bestimme $y_{11}$ und $y_{12}$

$$
1\cdot y_{11}+0\cdot y_{21}=1
\quad\Rightarrow\quad
y_{11}=1
$$

$$
1\cdot y_{12}+0\cdot y_{22}=0
\quad\Rightarrow\quad
y_{12}=0
$$

### Schritt 2: Bestimme $y_{21}$ und $y_{22}$

$$
2\cdot y_{11}+1\cdot y_{21}=0
$$

$$
2\cdot1+y_{21}=0
\quad\Rightarrow\quad
y_{21}=-2
$$

und

$$
2\cdot y_{12}+1\cdot y_{22}=1
$$

$$
2\cdot0+y_{22}=1
\quad\Rightarrow\quad
y_{22}=1.
$$

### Ergebnis

$$
\boxed{
\vec{Y}
=
\begin{pmatrix}
1&0\\
-2&1
\end{pmatrix}
}
$$

### Rückwärtssubstitution

Gegeben:

$$
\vec{U}\vec{X}
=
\vec{Y}.
$$

**Ziel:** Bestimmung von $\vec{X}$ mittels Rückwärtssubstitution.

### Vorgehensweise

1. Löse das Gleichungssystem

   $$
   \vec{U}\vec{X}
   =
   \vec{Y}.
   $$

2. Da $\vec{U}$ eine obere Dreiecksmatrix ist, kann $\vec{X}$ Schritt für Schritt von unten nach oben berechnet werden.

### Beispiel

$$
\vec{U}
=
\begin{pmatrix}
2&3\\
0&1
\end{pmatrix},
\qquad
\vec{Y}
=
\begin{pmatrix}
1&0\\
-2&1
\end{pmatrix}.
$$

Damit:

$$
\begin{pmatrix}
2&3\\
0&1
\end{pmatrix}
\begin{pmatrix}
x_{11}&x_{12}\\
x_{21}&x_{22}
\end{pmatrix}
=
\begin{pmatrix}
1&0\\
-2&1
\end{pmatrix}.
$$

### Schritt 1: Bestimme $x_{22}$

$$
1\cdot x_{22}=1
\quad\Rightarrow\quad
x_{22}=1.
$$

### Schritt 2: Bestimme $x_{12}$

$$
2x_{12}+3x_{22}=0
$$

$$
2x_{12}+3\cdot1=0
\quad\Rightarrow\quad
x_{12}=-\frac{3}{2}.
$$

### Schritt 3: Bestimme $x_{21}$

$$
0\cdot x_{11}+1\cdot x_{21}=-2
\quad\Rightarrow\quad
x_{21}=-2.
$$

### Schritt 4: Bestimme $x_{11}$

$$
2x_{11}+3x_{21}=1
$$

$$
\Rightarrow
x_{11}
=
\frac{7}{2}.
$$

### Ergebnis

$$
\vec{X}
=
\begin{pmatrix}
7/2&-3/2\\
-2&1
\end{pmatrix}.
$$

Die Matrix $\vec{X}$ ist die gesuchte Inverse, also:

$$
\boxed{
\vec{A}^{-1}
=
\begin{pmatrix}
7/2&-3/2\\
-2&1
\end{pmatrix}
}
$$

### Überprüfung der Inversen

Ziel ist die Bestätigung, dass

$$
\vec{A}\vec{A}^{-1}
=
\vec{I}.
$$

Es gilt:

$$
\begin{pmatrix}
2&3\\
4&7
\end{pmatrix}
\begin{pmatrix}
7/2&-3/2\\
-2&1
\end{pmatrix}
=
\begin{pmatrix}
1&0\\
0&1
\end{pmatrix}.
$$

Damit:

$$
\boxed{
\vec{A}\vec{A}^{-1}
=
\vec{I}
}
$$

Da das Produkt $\vec{A}\vec{A}^{-1}$ die Identitätsmatrix $\vec{I}$ ergibt, ist die berechnete Inverse korrekt.

```python
import sympy as sp

A = sp.Matrix([
    [2, 3],
    [4, 7]
])

L = sp.Matrix([
    [1, 0],
    [2, 1]
])

U = sp.Matrix([
    [2, 3],
    [0, 1]
])

print("L * U =")
sp.pprint(L * U)

print("\nA^(-1) =")
sp.pprint(A.inv())

print("\nA * A^(-1) =")
sp.pprint(A * A.inv())
```

## Zusammenfassung

- **LU-Zerlegung:** Eine Matrix $\vec{A}$ wird in das Produkt einer unteren Dreiecksmatrix $\vec{L}$ und einer oberen Dreiecksmatrix $\vec{U}$ zerlegt.

- **Vorwärtssubstitution:** Das Gleichungssystem

  $$
  \vec{L}\vec{Y}
  =
  \vec{I}_n
  $$

  wird gelöst, um $\vec{Y}$ zu bestimmen.

- **Rückwärtssubstitution:** Das Gleichungssystem

  $$
  \vec{U}\vec{X}
  =
  \vec{Y}
  $$

  wird gelöst, um die Inverse

  $$
  \vec{A}^{-1}
  =
  \vec{X}
  $$

  zu erhalten.

### Vorteile

- Effizient für kleine bis mittelgroße Matrizen.
- Bietet numerische Stabilität bei geeigneten Matrizen.
- Erlaubt die Wiederverwendung der Zerlegung für verschiedene rechte Seiten.

**Hinweis:** Die Schritte und Berechnungen sollten stets überprüft werden, um Fehler zu vermeiden.

## Zusammenhang zwischen der Inversen und der Lösung linearer Gleichungssysteme

Die Anwendung der Inversen einer Matrix auf einen Vektor liefert die Lösung des entsprechenden linearen Gleichungssystems.

Mathematisch ist dies äquivalent, da:

$$
\boxed{
\vec{x}
=
\vec{A}^{-1}\vec{b}
\quad
\text{genau dann gilt, wenn}
\quad
\vec{A}\vec{x}
=
\vec{b}
}
$$

### Fazit

Die Inverse einer Matrix ermöglicht die direkte Berechnung der Lösung eines linearen Gleichungssystems.
