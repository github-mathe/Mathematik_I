# Komplexe Zahlen

### Warum brauchen wir die Erweiterung der reelen Zahlen?

- Eine Erweiterung der reelen Zahlen wird eingeführt, um Lösungen für Gleichungen wie

$$
x^2 + 1 = 0
$$

zu ermöglichen.

- In den reellen Zahlen existiert keine Zahl $x$, für die

$$
x^2 = -1
$$

gilt.

- Die Erweiterung ist nicht nur notwendig, um das Quadrat einer negativen Zahl zu definieren, sondern auch für Wurzeln aus negativen Zahlen.

- Beispiel:

$$
\sqrt[4]{-16}
$$

erfordert neue Zahlen, da es keine reelle Zahl gibt, deren vierte Potenz $-16$ ergibt.

- Für rationale Exponenten, wie

$$
x^{1/2}
\qquad \text{oder} \qquad
x^{3/4},
$$

müssen negative Basen ebenfalls durch die Erweiterung der reelen Zahlen definiert werden.

### Gleichung $x^2 + 1 = 0$

Wir betrachten die Gleichung

$$
x^2 + 1 = 0.
$$

- Diese Gleichung entspricht der Suche nach den Nullstellen des Polynoms $x^2 + 1$.
- Wie im Graphen auf der nächsten Abbildung zu sehen ist, gibt es im Bereich der reellen Zahlen keine Lösungen für diese Gleichung.
- Die Lösung muss daher in einer Erweiterung der reellen Zahlen gefunden werden, die als komplexe Zahlen bekannt ist. Diese wird im Folgenden definiert.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 400)

f = x**2 + 1
g = x**2 - 1

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(x, f, color="blue", label=r"$f(x)=x^2+1$")
ax.plot(x, g, color="red", label=r"$g(x)=x^2-1$")

ax.axhline(0, color="black", linewidth=1)
ax.axvline(0, color="black", linewidth=1)

ax.set_xlim(-3, 3)
ax.set_ylim(-2, 5)

ax.set_xlabel("$x$")
ax.set_ylabel("$y$")

ax.grid(alpha=0.3)
ax.legend()

plt.show()
```

### Graph von $f(x)=x^2+1$ und $g(x)=x^2-1$

- Der Graph zeigt die Funktionen $f(x)=x^2+1$ (blau) und $g(x)=x^2-1$ (rot).
- Die Parabel $f(x)$ schneidet die x-Achse nicht, während $g(x)$ bei

$$
x=\pm 1
$$

Nullstellen hat.

### Die imaginäre Einheit $i$

- Um Lösungen für Gleichungen wie

$$
x^2=-1
$$

zu finden, wird die imaginäre Einheit $i$ eingeführt.

- Die **imaginäre Einheit** ist definiert als

$$
i=\sqrt{-1}.
$$

- Es gilt

$$
i^2=-1.
$$

- Mit der Einführung von $i$ können wir die Lösungen der Gleichung

$$
x^2+1=0
$$

wie folgt ausdrücken:

$$
\begin{aligned}
x &= \pm\sqrt{-1}\\
  &= \pm i.
\end{aligned}
$$

### Definition

Eine **komplexe Zahl** ist eine Zahl der Form

$$
z=a+bi,
$$

wobei

- $a$ der **reelle Teil** $(\operatorname{Re}(z))$ ist,
- $b$ der **imaginäre Teil** $(\operatorname{Im}(z))$ ist,
- $i$ die **imaginäre Einheit** ist, definiert durch

$$
i^2=-1.
$$
