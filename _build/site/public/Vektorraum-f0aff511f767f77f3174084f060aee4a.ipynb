## Vektorräume

### Was ist ein Vektorraum?

Ein **Vektorraum** $V$ über einem Körper $K$ ist eine Menge von Objekten, genannt *Vektoren*, zusammen mit zwei Operationen:

- **Vektoraddition**:

$$
+ : V\times V\rightarrow V
$$

- **Skalarmultiplikation**:

$$
\cdot : K\times V\rightarrow V
$$

sodass die folgenden Axiome gelten:

1. **Assoziativität der Addition**:

$$
(\vec u+\vec v)+\vec w
=
\vec u+(\vec v+\vec w)
$$

2. **Kommutativität der Addition**:

$$
\vec u+\vec v
=
\vec v+\vec u
$$

3. **Existenz des Nullvektors**:

Es gibt ein $\vec 0_V\in V$ mit

$$
\vec u+\vec 0_V
=
\vec u.
$$

4. **Existenz des inversen Elements**:

Zu jedem $\vec u\in V$ existiert ein $-\vec u\in V$ mit

$$
\vec u+(-\vec u)
=
\vec 0_V.
$$

5. **Assoziativität der Skalarmultiplikation**:

$$
a(b\vec u)
=
(ab)\vec u
\qquad
\text{mit } a,b\in K.
$$

6. **Einheitselement der Skalarmultiplikation**:

$$
1_K\vec u
=
\vec u,
$$

wobei $1_K$ das Einselement von $K$ ist.

7. **Distributivität der Skalarmultiplikation bezüglich der Vektoraddition**:

$$
a(\vec u+\vec v)
=
a\vec u+a\vec v
$$

und

$$
(a+b)\vec u
=
a\vec u+b\vec u.
$$

### Beispiele für Vektorräume

- **Reeller Koordinatenraum** $\mathbb R^n$:
  - Vektoren sind $n$-Tupel reeller Zahlen.
  - Addition und Skalarmultiplikation komponentenweise definiert.

- **Polynomraum** $P_n$:
  - Menge aller Polynome vom Grad höchstens $n$.
  - Addition und Skalarmultiplikation wie üblich für Polynome.

- **Funktionenraum** $C([a,b])$:
  - Menge der stetigen Funktionen auf dem Intervall $[a,b]$.
  - Addition und Skalarmultiplikation punktweise definiert.

### Vektoren als Spaltenvektoren

In diesen Vorlesungsunterlagen werden alle Vektoren, sofern nicht anders angegeben, als **vertikale Vektoren** (Spaltenvektoren) betrachtet.

Das bedeutet, dass ein Vektor

$$
\mathbf v\in\mathbb R^n
$$

in der Form

$$
\mathbf v
=
\begin{pmatrix}
v_1\\
v_2\\
\vdots\\
v_n
\end{pmatrix}
$$

dargestellt wird.

Dies ist wichtig für die Anwendung in Matrixoperationen und erleichtert die Notation bei Berechnungen.

Bei der Addition und Skalarmultiplikation von Vektoren in $\mathbb R^n$ wird die komponentenweise Darstellung wie gewohnt verwendet, allerdings als Spaltenstruktur, um die Form und Ausrichtung beizubehalten.
