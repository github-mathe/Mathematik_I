## Linearkombination von Vektoren

Gegeben seien die Vektoren

$$
\vec{v}_1=
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix},
\qquad
\vec{v}_2=
\begin{pmatrix}
2\\
-1\\
3
\end{pmatrix}
$$

und der Vektor

$$
\vec{w}=
\begin{pmatrix}
4\\
3\\
1
\end{pmatrix}.
$$

Untersuchen Sie, ob sich der Vektor $\vec{w}$ als **Linearkombination** der Vektoren $\vec{v}_1$ und $\vec{v}_2$ darstellen lässt.

Bestimmen Sie dazu die Skalare $\lambda_1,\lambda_2\in\mathbb{R}$, sodass

$$
\vec{w}
=
\lambda_1\vec{v}_1
+
\lambda_2\vec{v}_2.
$$

Geben Sie anschließend die Linearkombination explizit an.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Aufstellen des linearen Gleichungssystems

Gesucht sind $\lambda_1,\lambda_2\in\mathbb{R}$ mit

$$
\lambda_1\vec{v}_1
+
\lambda_2\vec{v}_2
=
\vec{w}.
$$

Einsetzen der Vektoren ergibt

$$
\lambda_1
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
2\\
-1\\
3
\end{pmatrix}
=
\begin{pmatrix}
4\\
3\\
1
\end{pmatrix}.
$$

Durch komponentenweisen Vergleich erhalten wir

$$
\begin{aligned}
\lambda_1+2\lambda_2 &=4,\\
2\lambda_1-\lambda_2 &=3,\\
-\lambda_1+3\lambda_2 &=1.
\end{aligned}
$$

Die erweiterte Koeffizientenmatrix lautet

$$
\boxed{
\left(
\begin{array}{cc|c}
1 & 2 & 4\\
2 & -1 & 3\\
-1 & 3 & 1
\end{array}
\right)
}
$$
:::

::::{dropdown} Durchführung der Elimination

Ausgangspunkt ist die erweiterte Koeffizientenmatrix

$$
\left(
\begin{array}{cc|c}
1 & 2 & 4\\
2 & -1 & 3\\
-1 & 3 & 1
\end{array}
\right).
$$


:::{dropdown} Schritt 1 – Eintrag in der ersten Spalte eliminieren

Der Eintrag $2$ in der ersten Spalte der zweiten Zeile wird eliminiert.

$$
\left(
\begin{array}{cc|c}
1 & 2 & 4\\
2 & -1 & 3\\
-1 & 3 & 1
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'=\mathrm{II}-2\cdot\mathrm{I}\\
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\left(
\begin{array}{cc|c}
1 & 2 & 4\\
0 & -5 & -5\\
-1 & 3 & 1
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 2 – Weiteren Eintrag in der ersten Spalte eliminieren

Nun wird der Eintrag $-1$ in der ersten Spalte der dritten Zeile eliminiert.

$$
\left(
\begin{array}{cc|c}
1 & 2 & 4\\
0 & -5 & -5\\
-1 & 3 & 1
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}'=\mathrm{III}+\mathrm{I}
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\left(
\begin{array}{cc|c}
1 & 2 & 4\\
0 & -5 & -5\\
0 & 5 & 5
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 3 – Eintrag unterhalb des zweiten Pivotelements eliminieren

Nun wird der Eintrag $5$ in der zweiten Spalte der dritten Zeile eliminiert.

Dabei werden bereits veränderte Zeilen mit einem Strich gekennzeichnet.

$$
\left(
\begin{array}{cc|c}
1 & 2 & 4\\
0 & -5 & -5\\
0 & 5 & 5
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}''=\mathrm{III}'+\mathrm{II}'
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\left(
\begin{array}{cc|c}
1 & 2 & 4\\
0 & -5 & -5\\
0 & 0 & 0
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 4 – Zweites Pivotelement normieren

Das zweite Pivotelement wird auf $1$ normiert.

$$
\left(
\begin{array}{cc|c}
1 & 2 & 4\\
0 & -5 & -5\\
0 & 0 & 0
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}''=-\frac{1}{5}\mathrm{II}'\\
\end{array}
$$

Damit erhalten wir

$$
\boxed{
\left(
\begin{array}{cc|c}
1 & 2 & 4\\
0 & 1 & 1\\
0 & 0 & 0
\end{array}
\right)
}
$$

:::

::::

:::{dropdown} Rückwärtseinsetzen

Aus der zweiten Zeile folgt

$$
\lambda_2=1.
$$

Aus der ersten Zeile ergibt sich

$$
\lambda_1+2\lambda_2=4.
$$

Mit $\lambda_2=1$ folgt

$$
\lambda_1+2=4
$$

und damit

$$
\lambda_1=2.
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Es gilt

$$
\lambda_1=2,
\qquad
\lambda_2=1.
$$

Damit lässt sich $\vec{w}$ als Linearkombination von $\vec{v}_1$ und $\vec{v}_2$ darstellen:

$$
\boxed{
\vec{w}
=
2\vec{v}_1+\vec{v}_2
}
$$

Zur Kontrolle:

$$
2
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix}
+
\begin{pmatrix}
2\\
-1\\
3
\end{pmatrix}
=
\begin{pmatrix}
4\\
3\\
1
\end{pmatrix}.
$$

Somit gilt

$$
\boxed{
\vec{w}\in
\operatorname{span}\{\vec{v}_1,\vec{v}_2\}.
}
$$
:::
