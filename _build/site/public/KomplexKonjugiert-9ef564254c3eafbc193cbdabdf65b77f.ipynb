## Konjugiert komplexe Zahl

Die **konjugiert komplexe Zahl** zu einer komplexen Zahl $z = a + bi$ ist definiert als:

$$
\overline{z} = a - bi
$$

- Der konjugiert komplexe Teil vertauscht das Vorzeichen des imaginären Teils.

**Beispiele:**

- Für $z = 3 + 4i$ ist die konjugierte Zahl $\overline{z} = 3 - 4i$.
- Für $z = -2 + 5i$ ist die konjugierte Zahl $\overline{z} = -2 - 5i$.
- Für $z = 7 - i$ ist die konjugierte Zahl $\overline{z} = 7 + i$.

### Komplexe Zahl und ihre konjugierte Zahl

- Die komplexe Zahl $z = a + bi$ und ihre konjugierte Zahl $\bar{z} = a - bi$ liegen symmetrisch zur reellen Achse.
- Die folgenden Segmente zeigen die Positionen von $z = 2 + i$ und $\bar{z} = 2 - i$ in der Gauss-Ebene:

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 5))

# Achsen
ax.annotate("", xy=(4, 0), xytext=(-1.5, 0),
            arrowprops=dict(arrowstyle="->", linewidth=1.8))
ax.annotate("", xy=(0, 2), xytext=(0, -2),
            arrowprops=dict(arrowstyle="->", linewidth=1.8))

# Punkte z und konjugiert komplexe Zahl
Z = (2, 1)
ConjZ = (2, -1)

ax.plot(Z[0], Z[1], "ko")
ax.plot(ConjZ[0], ConjZ[1], "ko")

# gestrichelte Linien zur reellen Achse
ax.plot([2, 2], [0, 1], linestyle="--")
ax.plot([2, 2], [0, -1], linestyle="--")

# Strecken vom Ursprung
ax.plot([0, Z[0]], [0, Z[1]], linewidth=2)
ax.plot([0, ConjZ[0]], [0, ConjZ[1]], linewidth=2)

# Beschriftungen
ax.text(4.1, -0.1, "Re", fontsize=11)
ax.text(0.1, 2.05, "Im", fontsize=11)

ax.text(2.1, 1.1, r"$z = 2 + i$", fontsize=12)
ax.text(2.1, -1.25, r"$\bar{z} = 2 - i$", fontsize=12)

ax.text(2, -0.2, "2", ha="center", fontsize=11)
ax.text(0.12, 1, "1", va="center", fontsize=11)
ax.text(0.12, -1, "-1", va="center", fontsize=11)

ax.set_xlim(-1.5, 4.3)
ax.set_ylim(-2.2, 2.2)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```
