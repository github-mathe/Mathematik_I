## Berechnung verschiedener Vektornormen

Gegeben sei der Vektor

$$
\mathbf{x}
=
\begin{pmatrix}
1\\
-2\\
2
\end{pmatrix}
\in\mathbb{R}^3.
$$

Berechnen Sie für den Vektor $\mathbf{x}$ die folgenden Vektornormen:

1. die $L_1$-Norm

   $$
   \|\mathbf{x}\|_1
   =
   \sum_{i=1}^{n}|x_i|,
   $$

2. die $L_\infty$-Norm

   $$
   \|\mathbf{x}\|_\infty
   =
   \max_{1\leq i\leq n}|x_i|,
   $$

3. die $L_2$-Norm

   $$
   \|\mathbf{x}\|_2
   =
   \sqrt{\sum_{i=1}^{n}x_i^2}.
   $$

Vergleichen Sie anschließend die erhaltenen Werte.

## Lösung

Die verschiedenen Normen messen die Größe eines Vektors auf unterschiedliche Weise.

Berechnen Sie zunächst jede Norm selbstständig und öffnen Sie anschließend die jeweilige Lösung zur Kontrolle.

:::{dropdown} Berechnung der $L_1$-Norm

Die $L_1$-Norm ist definiert durch

$$
\|\mathbf{x}\|_1
=
\sum_{i=1}^{n}|x_i|.
$$

Für

$$
\mathbf{x}
=
\begin{pmatrix}
1\\
-2\\
2
\end{pmatrix}
$$

ergibt sich

$$
\begin{aligned}
\|\mathbf{x}\|_1
&=
|1|+|-2|+|2|\\
&=
1+2+2\\
&=
5.
\end{aligned}
$$

Damit gilt

$$
\boxed{
\|\mathbf{x}\|_1=5
}
$$
:::

:::{dropdown} Berechnung der $L_\infty$-Norm

Die $L_\infty$-Norm ist definiert durch

$$
\|\mathbf{x}\|_\infty
=
\max_{1\leq i\leq n}|x_i|.
$$

Wir betrachten zunächst die Beträge der Komponenten:

$$
|1|=1,
\qquad
|-2|=2,
\qquad
|2|=2.
$$

Der größte Betrag ist

$$
\max\{1,2,2\}=2.
$$

Damit gilt

$$
\boxed{
\|\mathbf{x}\|_\infty=2
}
$$
:::

:::{dropdown} Berechnung der $L_2$-Norm

Die $L_2$-Norm ist definiert durch

$$
\|\mathbf{x}\|_2
=
\sqrt{\sum_{i=1}^{n}x_i^2}.
$$

Für den gegebenen Vektor ergibt sich

$$
\begin{aligned}
\|\mathbf{x}\|_2
&=
\sqrt{1^2+(-2)^2+2^2}\\
&=
\sqrt{1+4+4}\\
&=
\sqrt{9}\\
&=
3.
\end{aligned}
$$

Damit gilt

$$
\boxed{
\|\mathbf{x}\|_2=3
}
$$
:::

:::{dropdown} Vergleich der Vektornormen

Für den Vektor

$$
\mathbf{x}
=
\begin{pmatrix}
1\\
-2\\
2
\end{pmatrix}
$$

haben wir erhalten:

$$
\boxed{
\|\mathbf{x}\|_1=5
}
$$

$$
\boxed{
\|\mathbf{x}\|_\infty=2
}
$$

und

$$
\boxed{
\|\mathbf{x}\|_2=3.
}
$$

Die verschiedenen Normen liefern unterschiedliche Werte, da sie die Größe eines Vektors unterschiedlich bestimmen:

- Die $L_1$-Norm addiert die Beträge aller Komponenten.
- Die $L_\infty$-Norm betrachtet nur den größten Betrag einer Komponente.
- Die $L_2$-Norm bestimmt die euklidische Länge des Vektors.

Für diesen Vektor gilt somit

$$
\boxed{
\|\mathbf{x}\|_\infty
<
\|\mathbf{x}\|_2
<
\|\mathbf{x}\|_1.
}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Für

$$
\mathbf{x}
=
\begin{pmatrix}
1\\
-2\\
2
\end{pmatrix}
$$

ergeben sich die drei Vektornormen zu

$$
\boxed{
\|\mathbf{x}\|_1=5,
\qquad
\|\mathbf{x}\|_\infty=2,
\qquad
\|\mathbf{x}\|_2=3.
}
$$
:::
