## Linearkombination

Eine Linearkombination von Vektoren

$$
\mathbf{v}_1,\mathbf{v}_2,\dots,\mathbf{v}_k
$$

ist ein Ausdruck der Form:

$$
c_1\mathbf{v}_1
+
c_2\mathbf{v}_2
+
\dots
+
c_k\mathbf{v}_k,
$$

wobei

$$
c_1,c_2,\dots,c_k
$$

Skalare sind.

**Beispiel für Vektoren:**

$$
2\mathbf{v}_1-\mathbf{v}_2+3\mathbf{v}_3
$$

Eine Linearkombination kann verwendet werden, um einen Vektor als Summe anderer Vektoren zu schreiben.

### Linearkombinationen anderer mathematischer Objekte

Linearkombinationen können jedoch nicht nur für Vektoren, sondern auch für Matrizen, Funktionen oder andere mathematische Objekte definiert werden.

**Beispiel für Matrizen:**

$$
c_1\vec A_1+c_2\vec A_2
$$

wobei $\vec A_1$ und $\vec A_2$ Matrizen sind und $c_1,c_2$ Skalare.

**Beispiel für Funktionen:**

$$
c_1f_1(x)+c_2f_2(x)
$$

wobei $f_1(x)$ und $f_2(x)$ Funktionen sind, und $c_1,c_2$ Skalare.

**Beispiel:**

$$
3\sin(x)-2\cos(x)
$$

### Linearkombinationen in der Geometrie

- Eine Linearkombination kann geometrisch als Skalierung und Addition von Vektoren betrachtet werden.

- Beispiel im $\mathbb{R}^2$: Die Linearkombination

$$
c_1\mathbf{v}_1+c_2\mathbf{v}_2
$$

beschreibt die Position eines Punkts in der Ebene, der durch die Gewichtung der Vektoren $\mathbf{v}_1$ und $\mathbf{v}_2$ bestimmt wird.

- Wenn $\mathbf{v}_1$ und $\mathbf{v}_2$ nicht kollinear sind, spannen sie den gesamten $\mathbb{R}^2$-Raum auf, und jede Kombination

$$
c_1\mathbf{v}_1+c_2\mathbf{v}_2
$$

beschreibt eine eindeutige Position in der Ebene.
