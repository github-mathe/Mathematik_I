<a id="definition-lineare-abbildung"></a>

## Definition einer linearen Abbildung

Eine Abbildung

$$
T:\mathbb{R}^n\rightarrow\mathbb{R}^m
$$

ist linear, wenn:

$$
T(\mathbf{v}+\mathbf{w})
=
T(\mathbf{v})+T(\mathbf{w})
$$

für alle

$$
\mathbf{v},\mathbf{w}\in\mathbb{R}^n
$$

und

$$
T(c\mathbf{v})
=
cT(\mathbf{v})
$$

für alle

$$
c\in\mathbb{R}
\qquad\text{und}\qquad
\mathbf{v}\in\mathbb{R}^n.
$$

Eine Matrix

$$
A\in\mathbb{R}^{m\times n}
$$

definiert eine lineare Abbildung

$$
T:\mathbb{R}^n\rightarrow\mathbb{R}^m
$$

durch:

$$
\boxed{
T(\mathbf{x})=A\mathbf{x}
}
$$

$A\mathbf{x}$ ist das Matrix-Vektor-Produkt, das einen Vektor

$$
\mathbf{x}\in\mathbb{R}^n
$$

in einen neuen Vektor

$$
\mathbf{y}=A\mathbf{x}\in\mathbb{R}^m
$$

abbildet.

### Eigenschaften von linearen Abbildungen

Eine Abbildung

$$
T:\mathbb{R}^n\rightarrow\mathbb{R}^m
$$

ist **genau dann linear**, wenn sie die folgenden zwei Eigenschaften erfüllt:

#### Homogenität – skalare Multiplikation

Eine Abbildung $T$ ist homogen, das heißt, für jeden Vektor

$$
\vec v\in\mathbb{R}^n
$$

und jeden Skalar

$$
c\in\mathbb{R}
$$

gilt:

$$
\boxed{
T(c\vec v)=cT(\vec v)
}
$$

**Beispiel:** Bei einer Skalierungsabbildung wird der Vektor $\vec v$ durch $T(c\vec v)$ um den Faktor $c$ skaliert.

#### Additivität – Vektoraddition

Eine Abbildung $T$ ist additiv, das heißt, für alle Vektoren

$$
\vec v,\vec w\in\mathbb{R}^n
$$

gilt:

$$
\boxed{
T(\vec v+\vec w)
=
T(\vec v)+T(\vec w)
}
$$

**Beispiel:** Wenn $T$ eine Rotationsabbildung ist, dann dreht $T(\vec v+\vec w)$ die Summe der Vektoren genauso, als ob man jeden Vektor einzeln dreht und dann addiert.

<a id="geometrische-interpretation"></a>

## Geometrische Interpretation

Eine Matrix $A$ kann Vektoren durch verschiedene Transformationen verändern, wie z. B. Dehnen, Komprimieren, Drehen, Spiegeln oder andere geometrische Manipulationen.

### Beispiele für Transformationen

- **Rotationsmatrix:** Dreht Vektoren im 2D- oder 3D-Raum um einen bestimmten Winkel.
- **Diagonalmatrix:** Skaliert Vektoren entlang der Koordinatenachsen (Dehnen oder Komprimieren).
- **Projektionsmatrix:** Projiziert Vektoren auf Unterräume, z. B. auf eine Linie oder Ebene.
- **Reflexionsmatrix:** Spiegelt Vektoren an einer bestimmten Achse oder Ebene.
- **Shear-Matrix (Schermatrix):** Verschiebt die Komponenten eines Vektors in Abhängigkeit voneinander, ohne die Länge zu verändern.
- **Skalierungs-Matrix:** Verändert die Länge eines Vektors, ohne seine Richtung zu ändern (homogene Skalierung).
- **Householder-Matrix:** Spiegelt einen Vektor an einer Spiegelungsebene, die durch einen anderen Vektor definiert wird.

## Beispiel: Rotation in der Ebene

Die Rotationsmatrix um den Winkel $\theta$ im 2D-Raum ist:

$$
\boxed{
\vec R_\theta
=
\begin{pmatrix}
\cos\theta&-\sin\theta\\
\sin\theta&\cos\theta
\end{pmatrix}
}
$$

Die Anwendung von $\vec R(\theta)$ auf einen Vektor

$$
\mathbf v=
\begin{pmatrix}
x\\
y
\end{pmatrix}
$$

dreht den Vektor um den Winkel $\theta$.

### Beispiel: Rotation um $90^\circ$

Für

$$
\theta=90^\circ
$$

ergibt sich:

$$
\vec R_{90^\circ}
=
\begin{pmatrix}
0&-1\\
1&0
\end{pmatrix}.
$$

Der Vektor

$$
\mathbf v=
\begin{pmatrix}
1\\
0
\end{pmatrix}
$$

wird um $90^\circ$ gedreht:

$$
\vec R_{90^\circ}\mathbf v
=
\begin{pmatrix}
0&-1\\
1&0
\end{pmatrix}
\begin{pmatrix}
1\\
0
\end{pmatrix}
=
\begin{pmatrix}
0\\
1
\end{pmatrix}.
$$

```python
import numpy as np
import matplotlib.pyplot as plt

v = np.array([1, 0])
v_neu = np.array([0, 1])

plt.figure(figsize=(6, 6))

plt.axhline(0)
plt.axvline(0)

plt.quiver(0, 0, v[0], v[1],
           angles="xy", scale_units="xy", scale=1)

plt.quiver(0, 0, v_neu[0], v_neu[1],
           angles="xy", scale_units="xy", scale=1)

winkel = np.linspace(0, np.pi / 2, 100)
r = 0.35

plt.plot(r * np.cos(winkel), r * np.sin(winkel))
plt.text(0.35, 0.35, r"$90^\circ$")

plt.text(1.05, 0, r"$\mathbf{v}=(1,0)$")
plt.text(0.05, 1.05, r"$\mathbf{v}'=(0,1)$")

plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.gca().set_aspect("equal")
plt.grid()

plt.show()
```

### Interaktive Darstellung einer Rotation

Mit dem Schieberegler kann der Rotationswinkel $\theta$ verändert werden.

Der Ausgangsvektor

$$
\mathbf v=
\begin{pmatrix}
1\\
0
\end{pmatrix}
$$

wird mit der Rotationsmatrix

$$
\vec R_\theta=
\begin{pmatrix}
\cos\theta&-\sin\theta\\
\sin\theta&\cos\theta
\end{pmatrix}
$$

transformiert:

$$
\mathbf v'
=
\vec R_\theta\mathbf v.
$$

Durch Veränderung des Winkels $\theta$ ändert sich die Lage des transformierten Vektors $\mathbf v'$.

```python
%pip install ipywidgets
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Ausgangsvektor
v = np.array([1.0, 0.0])

# Schieberegler
winkel_slider = widgets.IntSlider(
    value=45,
    min=0,
    max=360,
    step=1,
    description="θ:",
    continuous_update=True
)

def rotation_zeichnen(theta):
    # Winkel in Radiant
    theta_rad = np.radians(theta)

    # Rotationsmatrix
    R = np.array([
        [np.cos(theta_rad), -np.sin(theta_rad)],
        [np.sin(theta_rad),  np.cos(theta_rad)]
    ])

    # Transformierter Vektor
    v_neu = R @ v

    # Abbildung
    fig, ax = plt.subplots(figsize=(6, 6))

    ax.axhline(0)
    ax.axvline(0)

    # Ausgangsvektor
    ax.quiver(
        0, 0,
        v[0], v[1],
        angles="xy",
        scale_units="xy",
        scale=1,
        label=r"$\mathbf{v}$"
    )

    # Rotierter Vektor
    ax.quiver(
        0, 0,
        v_neu[0], v_neu[1],
        angles="xy",
        scale_units="xy",
        scale=1,
        label=r"$\mathbf{v}'$"
    )

    # Winkelbogen
    winkel = np.linspace(0, theta_rad, 200)
    radius = 0.35

    ax.plot(
        radius * np.cos(winkel),
        radius * np.sin(winkel)
    )

    # Winkelbeschriftung
    theta_mitte = theta_rad / 2

    ax.text(
        0.48 * np.cos(theta_mitte),
        0.48 * np.sin(theta_mitte),
        rf"$\theta={theta}^\circ$",
        ha="center",
        va="center"
    )

    # Beschriftung der Vektoren
    ax.text(
        1.08 * v[0],
        1.08 * v[1],
        r"$\mathbf{v}$",
        fontsize=13
    )

    ax.text(
        1.08 * v_neu[0],
        1.08 * v_neu[1],
        r"$\mathbf{v}'$",
        fontsize=13
    )

    # Koordinatensystem
    ax.set_xlim(-1.4, 1.4)
    ax.set_ylim(-1.4, 1.4)
    ax.set_aspect("equal")
    ax.grid()

    ax.set_title(
        rf"Rotation des Vektors um $\theta={theta}^\circ$"
    )

    plt.show()

    # Rotationsmatrix und Ergebnis ausgeben
    print("Rotationsmatrix:")
    print(np.round(R, 3))

    print("\nTransformierter Vektor:")
    print(np.round(v_neu, 3))


widgets.interactive(rotation_zeichnen, theta=winkel_slider)
```

## Beispiel: Anwendung linearer Transformationen auf Positionsvektoren

In den nächsten Abschnitten werden verschiedene geometrische Transformationen auf die „L“-Form angewendet.

Diese Transformationen werden durch Matrizen dargestellt und verändern die Position, Größe und Orientierung der „L“-Form.

Die betrachteten Transformationen sind:

- Skalierung
- Drehung
- Scherung
- Spiegelung

### Das ursprüngliche L

Die Form des „L“ entsteht durch das Verbinden von Positionsvektoren:

- $O(0,0)$
- $A(0,2)$
- $B(0.4,2)$
- $C(0.4,0.4)$
- $D(1.4,0.4)$
- $E(1.4,0)$

```python
import numpy as np
import matplotlib.pyplot as plt

L = np.array([
    [0.0, 0.0],
    [0.0, 2.0],
    [0.4, 2.0],
    [0.4, 0.4],
    [1.4, 0.4],
    [1.4, 0.0],
    [0.0, 0.0]
])

plt.figure(figsize=(6, 6))

plt.axhline(0)
plt.axvline(0)

plt.fill(L[:, 0], L[:, 1], alpha=0.2)
plt.plot(L[:, 0], L[:, 1])

punkte = ["O", "A", "B", "C", "D", "E"]

for punkt, name in zip(L[:-1], punkte):
    plt.text(punkt[0] + 0.05, punkt[1] + 0.05, name)

plt.xlim(-0.5, 2)
plt.ylim(-0.5, 2.5)
plt.gca().set_aspect("equal")
plt.grid()

plt.show()
```

### Transformation von Positionsvektoren

Eine lineare Transformation wirkt auf alle Positionsvektoren, die die „L“-Form definieren.

Jeder Vektor $\mathbf v$ wird durch die Matrix-Vektor-Multiplikation transformiert:

$$
\boxed{
\mathbf v'=T\mathbf v
}
$$

wobei $T$ die Transformationsmatrix ist.

Die transformierten Vektoren $\mathbf v'$ zeigen auf neue Positionen, wodurch sich die gesamte Form verändert.

Dies führt zu einer neuen Darstellung der „L“-Form, die durch die Transformationsmatrix bestimmt wird.

### Skalierung des L

Die Skalierung wird durch die Matrix definiert:

$$
\vec S
=
\begin{pmatrix}
1.5&0\\
0&1.5
\end{pmatrix}.
$$

Diese Matrix vergrößert das „L“ um das 1.5-fache.

```python
S = np.array([
    [1.5, 0],
    [0, 1.5]
])

L_skaliert = (S @ L.T).T

plt.figure(figsize=(6, 6))

plt.axhline(0)
plt.axvline(0)

plt.plot(L[:, 0], L[:, 1], linestyle=":")
plt.fill(L_skaliert[:, 0], L_skaliert[:, 1], alpha=0.2)
plt.plot(L_skaliert[:, 0], L_skaliert[:, 1])

plt.xlim(-0.5, 2.5)
plt.ylim(-0.5, 3.5)
plt.gca().set_aspect("equal")
plt.grid()

plt.show()
```

### Interaktive Darstellung einer Skalierung

Mit dem Schieberegler kann der Skalierungsfaktor $s$ verändert werden.

Die Skalierung wird durch die Matrix

$$
\vec S=
\begin{pmatrix}
s & 0\\
0 & s
\end{pmatrix}
$$

beschrieben.

Jeder Positionsvektor $\mathbf v$ der „L“-Form wird transformiert durch

$$
\mathbf v'=\vec S \mathbf v.
$$

Verändert man den Faktor $s$, so wird die gesamte Form entsprechend vergrößert oder verkleinert.

```python
%pip install ipywidgets
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

# Punkte der L-Form
L = np.array([
    [0.0, 0.0],
    [0.0, 2.0],
    [0.4, 2.0],
    [0.4, 0.4],
    [1.4, 0.4],
    [1.4, 0.0],
    [0.0, 0.0]
])

# Schieberegler für den Skalierungsfaktor
s_slider = widgets.FloatSlider(
    value=1.5,
    min=0.2,
    max=3.0,
    step=0.1,
    description="s:",
    continuous_update=True
)

def skaliere_L(s):
    # Skalierungsmatrix
    S = np.array([
        [s, 0],
        [0, s]
    ])

    # transformierte Punkte
    L_skaliert = (S @ L.T).T

    # Plot
    fig, ax = plt.subplots(figsize=(6, 6))

    ax.axhline(0)
    ax.axvline(0)

    # Originalform
    ax.plot(L[:, 0], L[:, 1], linestyle=":", label="Original")

    # Skalierte Form
    ax.fill(L_skaliert[:, 0], L_skaliert[:, 1], alpha=0.2)
    ax.plot(L_skaliert[:, 0], L_skaliert[:, 1], label="Skaliert")

    ax.set_xlim(-0.5, 5)
    ax.set_ylim(-0.5, 5)
    ax.set_aspect("equal")
    ax.grid()

    ax.set_title(f"Skalierung der L-Form mit s = {s}")
    ax.legend()

    plt.show()

    print("Skalierungsmatrix:")
    print(np.round(S, 2))

widgets.interactive(skaliere_L, s=s_slider)
```

### Drehung des L

Die Drehung um $45^\circ$ erfolgt durch die Matrix:

$$
\vec R
=
\begin{pmatrix}
\cos45^\circ&-\sin45^\circ\\
\sin45^\circ&\cos45^\circ
\end{pmatrix}
$$

und damit:

$$
\boxed{
\vec R
=
\begin{pmatrix}
\frac{\sqrt2}{2}&-\frac{\sqrt2}{2}\\
\frac{\sqrt2}{2}&\frac{\sqrt2}{2}
\end{pmatrix}
}
$$

```python
theta = np.radians(45)

R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

L_rotiert = (R @ L.T).T

plt.figure(figsize=(6, 6))

plt.axhline(0)
plt.axvline(0)

plt.fill(L_rotiert[:, 0], L_rotiert[:, 1], alpha=0.2)
plt.plot(L_rotiert[:, 0], L_rotiert[:, 1])

plt.xlim(-2, 2)
plt.ylim(-0.5, 3)
plt.gca().set_aspect("equal")
plt.grid()

plt.show()
```

### Interaktive Darstellung einer Drehung

Mit dem Schieberegler kann der Drehwinkel $\theta$ verändert werden.

Die Drehung wird durch die Matrix

$$
\vec R=
\begin{pmatrix}
\cos(\theta) & -\sin(\theta)\\
\sin(\theta) & \cos(\theta)
\end{pmatrix}
$$

beschrieben.

Jeder Positionsvektor $\mathbf v$ der „L“-Form wird transformiert durch

$$
\mathbf v'=\vec R \mathbf v.
$$

Verändert man den Winkel $\theta$, so dreht sich die gesamte Form um den Ursprung.

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

# Punkte der L-Form
L = np.array([
    [0.0, 0.0],
    [0.0, 2.0],
    [0.4, 2.0],
    [0.4, 0.4],
    [1.4, 0.4],
    [1.4, 0.0],
    [0.0, 0.0]
])

# Schieberegler für den Winkel
theta_slider = widgets.FloatSlider(
    value=45,
    min=0,
    max=360,
    step=5,
    description="θ:",
    continuous_update=True
)

def drehe_L(theta):
    # Winkel in Radiant
    theta_rad = np.radians(theta)

    # Rotationsmatrix
    R = np.array([
        [np.cos(theta_rad), -np.sin(theta_rad)],
        [np.sin(theta_rad),  np.cos(theta_rad)]
    ])

    # transformierte Punkte
    L_rotiert = (R @ L.T).T

    # Plot
    fig, ax = plt.subplots(figsize=(6, 6))

    ax.axhline(0)
    ax.axvline(0)

    # Originalform
    ax.plot(L[:, 0], L[:, 1], linestyle=":", label="Original")

    # Rotierte Form
    ax.fill(L_rotiert[:, 0], L_rotiert[:, 1], alpha=0.2)
    ax.plot(L_rotiert[:, 0], L_rotiert[:, 1], label="Rotiert")

    ax.set_xlim(-3, 3)
    ax.set_ylim(-3, 3)
    ax.set_aspect("equal")
    ax.grid()

    ax.set_title(f"Drehung der L-Form um θ = {theta}°")
    ax.legend()

    plt.show()

    print("Rotationsmatrix:")
    print(np.round(R, 3))

widgets.interactive(drehe_L, theta=theta_slider)
```

### Scheren des L

Die Scherung erfolgt durch die Matrix:

$$
\boxed{
\vec H
=
\begin{pmatrix}
1&0.5\\
0&1
\end{pmatrix}
}
$$

```python
H = np.array([
    [1, 0.5],
    [0, 1]
])

L_geschert = (H @ L.T).T

plt.figure(figsize=(6, 6))

plt.axhline(0)
plt.axvline(0)

plt.fill(L_geschert[:, 0], L_geschert[:, 1], alpha=0.2)
plt.plot(L_geschert[:, 0], L_geschert[:, 1])

plt.xlim(-0.5, 2.5)
plt.ylim(-0.5, 2.5)
plt.gca().set_aspect("equal")
plt.grid()

plt.show()
```

### Interaktive Darstellung einer Scherung

Mit dem Schieberegler kann der Scherfaktor $h$ verändert werden.

Die Scherung wird durch die Matrix

$$
\vec H=
\begin{pmatrix}
1 & h\\
0 & 1
\end{pmatrix}
$$

beschrieben.

Jeder Positionsvektor $\mathbf v$ der „L“-Form wird transformiert durch

$$
\mathbf v'=\vec H \mathbf v.
$$

Verändert man den Scherfaktor $h$, so wird die Form entlang der $x$-Richtung geschert.

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

# Punkte der L-Form
L = np.array([
    [0.0, 0.0],
    [0.0, 2.0],
    [0.4, 2.0],
    [0.4, 0.4],
    [1.4, 0.4],
    [1.4, 0.0],
    [0.0, 0.0]
])

# Schieberegler für den Scherfaktor
h_slider = widgets.FloatSlider(
    value=0.5,
    min=-2.0,
    max=2.0,
    step=0.1,
    description="h:",
    continuous_update=True
)

def schere_L(h):
    # Scherungsmatrix
    H = np.array([
        [1, h],
        [0, 1]
    ])

    # transformierte Punkte
    L_geschert = (H @ L.T).T

    # Plot
    fig, ax = plt.subplots(figsize=(6, 6))

    ax.axhline(0)
    ax.axvline(0)

    # Originalform
    ax.plot(L[:, 0], L[:, 1], linestyle=":", label="Original")

    # Gescherte Form
    ax.fill(L_geschert[:, 0], L_geschert[:, 1], alpha=0.2)
    ax.plot(L_geschert[:, 0], L_geschert[:, 1], label="Geschert")

    ax.set_xlim(-5, 5)
    ax.set_ylim(-1, 3)
    ax.set_aspect("equal")
    ax.grid()

    ax.set_title(f"Scherung der L-Form mit h = {h}")
    ax.legend()

    plt.show()

    print("Scherungsmatrix:")
    print(np.round(H, 2))

widgets.interactive(schere_L, h=h_slider)
```

### Spiegelung des L

Die Spiegelung an der $y$-Achse wird durch die Matrix definiert:

$$
\boxed{
\vec M
=
\begin{pmatrix}
-1&0\\
0&1
\end{pmatrix}
}
$$

Diese Matrix spiegelt das „L“ entlang der $y$-Achse, indem sie die $x$-Koordinaten invertiert.

```python
M = np.array([
    [-1, 0],
    [0, 1]
])

L_gespiegelt = (M @ L.T).T

plt.figure(figsize=(6, 6))

plt.axhline(0)
plt.axvline(0)

plt.plot(L[:, 0], L[:, 1], linestyle=":")
plt.fill(L_gespiegelt[:, 0], L_gespiegelt[:, 1], alpha=0.2)
plt.plot(L_gespiegelt[:, 0], L_gespiegelt[:, 1])

plt.text(0.5, -0.3, "Original")
plt.text(-1.3, -0.3, "Spiegelbild")

plt.xlim(-2, 2)
plt.ylim(-0.5, 2.5)
plt.gca().set_aspect("equal")
plt.grid()

plt.show()
```

## Standardmatrix einer linearen Abbildung

Jede lineare Abbildung

$$
T:\mathbb{R}^n\rightarrow\mathbb{R}^m
$$

kann durch eine Matrix dargestellt werden.

Diese Matrix, die die Wirkung der linearen Abbildung beschreibt, wird als die **Standardmatrix** von $T$ bezeichnet.

Die Spalten der Standardmatrix bestehen aus den Bildern der Einheitsvektoren

$$
\mathbf e_1,\mathbf e_2,\dots,\mathbf e_n
$$

unter der Abbildung $T$.

Formal gilt:

$$
\boxed{
T(\mathbf e_i)
=
\text{die $i$-te Spalte der Standardmatrix}
}
$$

### Beispiel: Lineare Abbildung und Standardbasis

Betrachte die lineare Abbildung

$$
T:\mathbb{R}^3\rightarrow\mathbb{R}^3,
$$

definiert durch:

$$
T(\vec x)
=
T
\begin{pmatrix}
x\\
y\\
z
\end{pmatrix}
=
\begin{pmatrix}
2x+y\\
x-z\\
y+3z
\end{pmatrix}.
$$

Die **Standardbasis** von $\mathbb{R}^3$ ist:

$$
\vec e_1
=
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix},
\qquad
\vec e_2
=
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix},
\qquad
\vec e_3
=
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}.
$$

### Bilder der Standardbasisvektoren

Das Bild von $\vec e_1$ unter $T$ ist:

$$
T(\vec e_1)
=
T
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}
=
\begin{pmatrix}
2\\
1\\
0
\end{pmatrix}.
$$

Das Bild von $\vec e_2$ unter $T$ ist:

$$
T(\vec e_2)
=
T
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix}
=
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}.
$$

Das Bild von $\vec e_3$ unter $T$ ist:

$$
T(\vec e_3)
=
T
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}
=
\begin{pmatrix}
0\\
-1\\
3
\end{pmatrix}.
$$

### Standardmatrix der linearen Abbildung

Die **Standardmatrix** von $T$ wird durch die Bilder der Standardbasisvektoren gebildet:

$$
\boxed{
\vec A
=
\begin{pmatrix}
2&1&0\\
1&0&-1\\
0&1&3
\end{pmatrix}
}
$$

Diese Matrix repräsentiert die lineare Abbildung $T$, sodass

$$
\boxed{
T(\vec v)=\vec A\vec v
}
$$

für jeden Vektor

$$
\vec v\in\mathbb{R}^3.
$$

## Lineare Abbildung: Rotation eines Vektors

Gegeben ist eine lineare Abbildung

$$
T:\mathbb{R}^2\rightarrow\mathbb{R}^2,
$$

die jeden Vektor

$$
\vec v\in\mathbb{R}^2
$$

um den Winkel $\theta$ gegen den Uhrzeigersinn dreht.

Die Abbildung $T$ wirkt auf jeden Vektor

$$
\vec v=
\begin{pmatrix}
x\\
y
\end{pmatrix},
$$

sodass der Vektor in den neuen Punkt rotiert wird.

Um die Standardmatrix dieser linearen Abbildung zu bestimmen, wenden wir $T$ auf die **Standardbasisvektoren**

$$
\vec e_1=
\begin{pmatrix}
1\\
0
\end{pmatrix},
\qquad
\vec e_2=
\begin{pmatrix}
0\\
1
\end{pmatrix}
$$

an.

### Berechnung der Rotationsmatrix

Wenden wir $T$ auf $\vec e_1$ an:

$$
T(\vec e_1)
=
\text{Rotation um }\theta
\text{ von }
\begin{pmatrix}
1\\
0
\end{pmatrix}.
$$

Nach der Drehung um den Winkel $\theta$ erhalten wir:

$$
\boxed{
T(\vec e_1)
=
\begin{pmatrix}
\cos\theta\\
\sin\theta
\end{pmatrix}
}
$$

Wenden wir nun $T$ auf $\vec e_2$ an:

$$
T(\vec e_2)
=
\text{Rotation um }\theta
\text{ von }
\begin{pmatrix}
0\\
1
\end{pmatrix}.
$$

Nach der Drehung um den Winkel $\theta$ erhalten wir:

$$
\boxed{
T(\vec e_2)
=
\begin{pmatrix}
-\sin\theta\\
\cos\theta
\end{pmatrix}
}
$$

### Herleitung der Rotationsmatrix

Die Standardmatrix von $T$, deren Spalten die Bilder der Einheitsvektoren sind, lautet:

$$
\vec A
=
\begin{pmatrix}
T(\vec e_1)&T(\vec e_2)
\end{pmatrix}
=
\boxed{
\begin{pmatrix}
\cos\theta&-\sin\theta\\
\sin\theta&\cos\theta
\end{pmatrix}
}
$$

Diese Matrix beschreibt eine Drehung um den Winkel $\theta$ gegen den Uhrzeigersinn im $\mathbb{R}^2$.

Für einen Vektor

$$
\vec v=
\begin{pmatrix}
x\\
y
\end{pmatrix}
$$

ergibt sich das Bild der Rotation durch:

$$
T(\vec v)
=
\vec A\vec v
=
\begin{pmatrix}
x\cos\theta-y\sin\theta\\
x\sin\theta+y\cos\theta
\end{pmatrix}.
$$

## Überblick über Transformationen

Eine **Transformation** ist eine Zuordnung, bei der ein Vektor in einen anderen Vektor überführt wird.

**Lineare Transformationen** sind spezielle Transformationen, die durch Matrizen beschrieben werden können und die Eigenschaften der Vektoraddition und der skalaren Multiplikation bewahren.

Häufige Transformationen im 2D- und 3D-Raum umfassen:

- **Rotation:** Dreht Vektoren um einen bestimmten Winkel.
- **Skalierung:** Vergrößert oder verkleinert Vektoren entlang bestimmter Achsen.
- **Sherung:** Ändert die Komponenten eines Vektors in Abhängigkeit von seinen anderen Komponenten.
- **Reflexion:** Spiegelt Vektoren an einer Achse oder Ebene.
- **Projektion:** Bildet Vektoren auf einen Unterraum ab.

Diese Transformationen können alle als Matrix-Vektor-Produkte beschrieben werden.

## Homogenität und Additivität: Grundlagen linearer Abbildungen

Die beiden Eigenschaften der **Homogenität** und **Additivität** definieren eine lineare Abbildung und garantieren, dass lineare Transformationen durch Matrizen beschrieben werden können.

Diese beiden Eigenschaften sind **notwendige und hinreichende Bedingungen** für die Linearität einer Abbildung.

Eine Matrix $\vec A$ erfüllt sowohl die Homogenität als auch die Additivität, was bedeutet:

$$
T(\vec v)
=
\vec A\vec v.
$$

Damit sind lineare Abbildungen vollständig durch Matrizen repräsentiert.
