## Basisbestimmung und Gram-Schmidt-Verfahren

Gegeben seien die Vektoren

$$
\mathbf v_1=
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\qquad
\mathbf v_2=
\begin{pmatrix}
2\\
1\\
1
\end{pmatrix},
\qquad
\mathbf v_3=
\begin{pmatrix}
3\\
1\\
2
\end{pmatrix},
\qquad
\mathbf v_4=
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix}.
$$

Sie spannen den Unterraum

$$
U=\operatorname{span}\{\mathbf v_1,\mathbf v_2,\mathbf v_3,\mathbf v_4\}
$$

auf.

1. Bestimmen Sie mit dem Gauß-Eliminationsverfahren eine vereinfachte Basis von $U$.  
   Schreiben Sie dazu die gegebenen Vektoren als Zeilen einer Matrix.

2. Wenden Sie auf die erhaltenen Basisvektoren das Gram-Schmidt-Verfahren an und bestimmen Sie eine orthonormale Basis von $U$.

## Lösung

Zunächst wird mit dem Gauß-Eliminationsverfahren eine Basis des aufgespannten Unterraums bestimmt.

Anschließend wird diese vereinfachte Basis mit dem Gram-Schmidt-Verfahren orthonormalisiert.

::::{dropdown} Durchführung der Gauß-Elimination

Die Vektoren werden als Zeilen in eine Matrix geschrieben:

$$
\begin{pmatrix}
1 & 0 & 1\\
2 & 1 & 1\\
3 & 1 & 2\\
1 & 2 & -1
\end{pmatrix}.
$$

:::{dropdown} Schritt 1

Wir eliminieren die Einträge unterhalb des ersten Pivotelements:

$$
\left(
\begin{array}{ccc}
1 & 0 & 1\\
2 & 1 & 1\\
3 & 1 & 2\\
1 & 2 & -1
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'=\mathrm{II}-2\cdot\mathrm{I}\\
\mathrm{III}'=\mathrm{III}-3\cdot\mathrm{I}\\
\mathrm{IV}'=\mathrm{IV}-\mathrm{I}
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\begin{pmatrix}
1 & 0 & 1\\
0 & 1 & -1\\
0 & 1 & -1\\
0 & 2 & -2
\end{pmatrix}.
}
$$

:::

:::{dropdown} Schritt 2

Nun werden die Einträge unterhalb des zweiten Pivotelements eliminiert:

$$
\left(
\begin{array}{ccc}
1 & 0 & 1\\
0 & 1 & -1\\
0 & 1 & -1\\
0 & 2 & -2
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}''=\mathrm{III}'-\mathrm{II}'\\
\mathrm{IV}''=\mathrm{IV}'-2\cdot\mathrm{II}'
\end{array}
$$

Damit erhalten wir

$$
\boxed{
\begin{pmatrix}
1 & 0 & 1\\
0 & 1 & -1\\
0 & 0 & 0\\
0 & 0 & 0
\end{pmatrix}.
}
$$

:::

::::

:::{dropdown} Vereinfachte Basis bestimmen

Nach der Gauß-Elimination bleiben zwei von Null verschiedene Zeilen übrig.

Da die Gauß-Elimination den von den Zeilen aufgespannten Raum nicht verändert, können diese als vereinfachte Basis von $U$ verwendet werden:

$$
\mathbf b_1=
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\qquad
\mathbf b_2=
\begin{pmatrix}
0\\
1\\
-1
\end{pmatrix}.
$$

Damit ist

$$
\boxed{
B=
\left\{
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\begin{pmatrix}
0\\
1\\
-1
\end{pmatrix}
\right\}.
}
$$

Der Rang beträgt

$$
\operatorname{Rang}=2,
$$

also gilt

$$
\boxed{\dim(U)=2.}
$$
:::

:::{dropdown} Gram-Schmidt – Bestimmung von $\mathbf u_1$

Der erste Basisvektor wird normiert:

$$
\mathbf u_1
=
\frac{\mathbf b_1}{\|\mathbf b_1\|}.
$$

Für die Norm gilt

$$
\|\mathbf b_1\|
=
\sqrt{1^2+0^2+1^2}
=
\sqrt{2}.
$$

Damit ist

$$
\boxed{
\mathbf u_1
=
\frac{1}{\sqrt{2}}
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}.
}
$$
:::

:::{dropdown} Gram-Schmidt – Bestimmung von $\mathbf u_2$

Zunächst wird die Projektion von $\mathbf b_2$ auf $\mathbf u_1$ bestimmt:

$$
P_{\mathbf u_1}(\mathbf b_2)
=
\langle\mathbf b_2,\mathbf u_1\rangle\,\mathbf u_1.
$$

Das Skalarprodukt ist

$$
\begin{aligned}
\langle\mathbf b_2,\mathbf u_1\rangle
&=
\begin{pmatrix}
0 & 1 & -1
\end{pmatrix}
\frac{1}{\sqrt{2}}
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}\\[1ex]
&=
-\frac{1}{\sqrt{2}}.
\end{aligned}
$$

Damit ergibt sich

$$
\begin{aligned}
P_{\mathbf u_1}(\mathbf b_2)
&=
-\frac{1}{\sqrt{2}}
\frac{1}{\sqrt{2}}
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
-\frac12\\
0\\
-\frac12
\end{pmatrix}.
\end{aligned}
$$

Der orthogonale Anteil ist

$$
\begin{aligned}
\mathbf w_2
&=
\mathbf b_2-P_{\mathbf u_1}(\mathbf b_2)\\[1ex]
&=
\begin{pmatrix}
0\\
1\\
-1
\end{pmatrix}
-
\begin{pmatrix}
-\frac12\\
0\\
-\frac12
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
\frac12\\
1\\
-\frac12
\end{pmatrix}.
\end{aligned}
$$

Die Norm von $\mathbf w_2$ ist

$$
\begin{aligned}
\|\mathbf w_2\|
&=
\sqrt{
\left(\frac12\right)^2
+
1^2
+
\left(-\frac12\right)^2
}\\
&=
\sqrt{\frac32}
=
\frac{\sqrt6}{2}.
\end{aligned}
$$

Nun wird $\mathbf w_2$ normiert:

$$
\boxed{
\mathbf u_2
=
\frac{\mathbf w_2}{\|\mathbf w_2\|}
=
\frac{1}{\sqrt6}
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix}.
}
$$
:::

:::{dropdown} Orthonormalität überprüfen

Die beiden erhaltenen Vektoren lauten

$$
\mathbf u_1
=
\frac{1}{\sqrt2}
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\qquad
\mathbf u_2
=
\frac{1}{\sqrt6}
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix}.
$$

Beide Vektoren sind normiert:

$$
\|\mathbf u_1\|=1,
\qquad
\|\mathbf u_2\|=1.
$$

Außerdem gilt

$$
\begin{aligned}
\langle\mathbf u_1,\mathbf u_2\rangle
&=
\frac{1}{\sqrt2}
\frac{1}{\sqrt6}
\left(
1\cdot1+0\cdot2+1\cdot(-1)
\right)\\
&=0.
\end{aligned}
$$

Damit sind die beiden Vektoren orthonormal.
:::

:::{dropdown} Gesamtlösung anzeigen

Durch die Gauß-Elimination ergibt sich zunächst die vereinfachte Basis

$$
\boxed{
B=
\left\{
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\begin{pmatrix}
0\\
1\\
-1
\end{pmatrix}
\right\}.
}
$$

Damit gilt

$$
\dim(U)=2.
$$

Auf diese Basis wird anschließend das Gram-Schmidt-Verfahren angewendet.

Die resultierende orthonormale Basis lautet

$$
\boxed{
B_{\mathrm{ON}}
=
\left\{
\frac{1}{\sqrt2}
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\quad
\frac{1}{\sqrt6}
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix}
\right\}.
}
$$
:::
