## Unterräume

### Definition eines Unterraums

Ein **Unterraum** eines Vektorraums $V$ ist eine Teilmenge

$$
U\subseteq V,
$$

die selbst ein Vektorraum ist.

Das bedeutet, dass $U$ die folgenden Eigenschaften erfüllt:

- **Abgeschlossenheit unter Addition:** Für alle

$$
\vec{u}_1,\vec{u}_2\in U
$$

gilt:

$$
\vec{u}_1+\vec{u}_2\in U.
$$

- **Abgeschlossenheit unter Skalarmultiplikation:** Für alle

$$
\vec{u}\in U
$$

und

$$
\lambda\in\mathbb{R}
$$

gilt:

$$
\lambda\vec{u}\in U.
$$

- **Nullvektor:** Der Nullvektor

$$
\vec{0}\in V
$$

gehört zu $U$.

### Beispiele von Unterräumen

- **Nullraum:** Der Vektorraum, der nur den Nullvektor $\vec{0}$ enthält, ist ein Unterraum jedes Vektorraums.

- **Die $xy$-Ebene in $\mathbb{R}^3$:** Der Raum aller Vektoren der Form

$$
\begin{pmatrix}
x\\
y\\
0
\end{pmatrix},
$$

wobei $x,y\in\mathbb{R}$, ist ein Unterraum von $\mathbb{R}^3$, da er die Bedingungen für Addition und Skalarmultiplikation erfüllt.

- **Geraden durch den Ursprung:** Jede Gerade, die durch den Ursprung verläuft, ist ein Unterraum von $\mathbb{R}^2$ oder $\mathbb{R}^3$.

Zum Beispiel ist der Raum der Vektoren der Form

$$
\lambda
\begin{pmatrix}
1\\
2
\end{pmatrix}
$$

mit $\lambda\in\mathbb{R}$ ein Unterraum von $\mathbb{R}^2$.

### Weitere Beispiele von Unterräumen

- **Ebenen durch den Ursprung in $\mathbb{R}^3$:** Eine Ebene, die den Ursprung enthält, ist ein Unterraum von $\mathbb{R}^3$.

Zum Beispiel der Raum, der alle Vektoren der Form

$$
\lambda_1
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix}
$$

enthält, ist eine Ebene in $\mathbb{R}^3$.

- **Lösungsraum eines homogenen Gleichungssystems:** Der Raum aller Lösungen eines homogenen linearen Gleichungssystems

$$
\vec A\vec{x}=\vec 0
$$

bildet einen Unterraum.

Dieser Raum wird auch als **Nullraum** der Matrix $A$ bezeichnet.

- **Polynomräume:** Der Raum der Polynome vom Grad $\leq n$ ist ein Unterraum des Vektorraums aller Polynome.

Zum Beispiel ist der Raum der Polynome vom Grad $\leq 2$ der Unterraum

$$
P_2
=
\left\{
a_0+a_1x+a_2x^2
\mid
a_0,a_1,a_2\in\mathbb{R}
\right\}.
$$
