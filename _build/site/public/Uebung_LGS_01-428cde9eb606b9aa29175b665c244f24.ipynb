## Lösen eines linearen Gleichungssystems mit dem Gauß-Eliminationsverfahren

Gegeben ist das lineare Gleichungssystem

$$
\begin{aligned}
x_1 + 2x_2 - x_3 &= 3,\\
2x_1 + 5x_2 + x_3 &= 8,\\
-\,x_1 - x_2 + 2x_3 &= -1.
\end{aligned}
$$

Stellen Sie zunächst die zugehörige erweiterte Koeffizientenmatrix auf.

Führen Sie anschließend das **Gauß-Eliminationsverfahren** durch und bringen Sie die Matrix durch elementare Zeilenumformungen in eine **reduzierte Dreiecksform**.

Verwenden Sie dabei nur die folgenden elementaren Zeilenoperationen:

- Vertauschen zweier Zeilen,
- Multiplikation einer Zeile mit einer von null verschiedenen Zahl,
- Addition eines Vielfachen einer Zeile zu einer anderen Zeile.

Dokumentieren Sie jeden Eliminationsschritt.

Bestimmen Sie anschließend die Lösung des linearen Gleichungssystems durch Rückwärtseinsetzen.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Umwandlung des LGS in Matrixschreibweise

Gegeben ist das lineare Gleichungssystem

$$
\begin{aligned}
x_1 + 2x_2 - x_3 &= 3,\\
2x_1 + 5x_2 + x_3 &= 8,\\
-\,x_1 - x_2 + 2x_3 &= -1.
\end{aligned}
$$

Die Koeffizienten der Unbekannten werden in der Matrix

$$
A=
\begin{pmatrix}
1 & 2 & -1\\
2 & 5 & 1\\
-1 & -1 & 2
\end{pmatrix}
$$

zusammengefasst.

Der Vektor der Unbekannten lautet

$$
\vec{x}
=
\begin{pmatrix}
x_1\\
x_2\\
x_3
\end{pmatrix},
$$

und die rechte Seite wird durch

$$
\vec{b}
=
\begin{pmatrix}
3\\
8\\
-1
\end{pmatrix}
$$

beschrieben.

Damit erhält man die Matrixschreibweise

$$
\boxed{
A\vec{x}=\vec{b}
}
$$

also

$$
\begin{pmatrix}
1 & 2 & -1\\
2 & 5 & 1\\
-1 & -1 & 2
\end{pmatrix}
\begin{pmatrix}
x_1\\
x_2\\
x_3
\end{pmatrix}
=
\begin{pmatrix}
3\\
8\\
-1
\end{pmatrix}.
$$

Für das Gauß-Eliminationsverfahren verwenden wir die erweiterte Koeffizientenmatrix

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
2 & 5 & 1 & 8\\
-1 & -1 & 2 & -1
\end{array}
\right)
}
$$
:::

::::{dropdown} Durchführung der Elimination

Ausgangspunkt ist die erweiterte Koeffizientenmatrix

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
2 & 5 & 1 & 8\\
-1 & -1 & 2 & -1
\end{array}
\right).
$$


:::{dropdown} Schritt 1 – Ersten Eintrag eliminieren

Der Eintrag $2$ in der ersten Spalte wird eliminiert.

Wir verwenden

$$
Z_2 \leftarrow Z_2-2Z_1.
$$

Es ergibt sich

$$
Z_2:
\qquad
(2,\;5,\;1\mid8)
-
2(1,\;2,\;-1\mid3)
$$

und damit

$$
Z_2=(0,\;1,\;3\mid2).
$$

Die Matrix lautet nun

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & 3 & 2\\
-1 & -1 & 2 & -1
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 2 – Zweiten Eintrag eliminieren

Nun wird der Eintrag $-1$ in der ersten Spalte eliminiert.

Wir verwenden

$$
Z_3 \leftarrow Z_3+Z_1.
$$

Es ergibt sich

$$
Z_3:
\qquad
(-1,\;-1,\;2\mid-1)
+
(1,\;2,\;-1\mid3)
$$

und damit

$$
Z_3=(0,\;1,\;1\mid2).
$$

Die Matrix lautet nun

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & 3 & 2\\
0 & 1 & 1 & 2
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 3 – Eintrag unterhalb des zweiten Pivotelements eliminieren

Nun wird der Eintrag $1$ in der zweiten Spalte der dritten Zeile eliminiert.

Wir verwenden

$$
Z_3 \leftarrow Z_3-Z_2.
$$

Es ergibt sich

$$
Z_3:
\qquad
(0,\;1,\;1\mid2)
-
(0,\;1,\;3\mid2)
$$

und damit

$$
Z_3=(0,\;0,\;-2\mid0).
$$

Die Matrix lautet nun

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & 3 & 2\\
0 & 0 & -2 & 0
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 4 – Letztes Pivotelement normieren

Das letzte Pivotelement wird auf $1$ normiert.

Wir verwenden

$$
Z_3 \leftarrow -\frac{1}{2}Z_3.
$$

Damit ergibt sich

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & 3 & 2\\
0 & 0 & 1 & 0
\end{array}
\right)
}
$$

:::

::::

:::{dropdown} Betrachtung der Lösungsmenge

Nach der Gauß-Elimination erhalten wir die Matrix

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & 3 & 2\\
0 & 0 & 1 & 0
\end{array}
\right).
$$

In jeder der drei Zeilen befindet sich ein Pivotelement. Daher gilt

$$
\operatorname{rang}(A)=3.
$$

Da das Gleichungssystem drei Unbekannte besitzt, ist

$$
n=3.
$$

Somit gilt

$$
\operatorname{rang}(A)=n.
$$

Auch für die erweiterte Koeffizientenmatrix gilt

$$
\operatorname{rang}(A\mid \vec b)=3.
$$

Damit ist

$$
\operatorname{rang}(A)
=
\operatorname{rang}(A\mid \vec b)
=
n.
$$

Das lineare Gleichungssystem besitzt daher **genau eine Lösung**.

Es gibt keine freien Variablen und somit keine Freiheitsgrade:

$$
n-\operatorname{rang}(A)
=
3-3
=
0.
$$

Die eindeutige Lösung wird anschließend durch Rückwärtseinsetzen bestimmt.
:::

:::{dropdown} Rückwärtseinsetzen

Aus der oberen Dreiecksmatrix

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & 3 & 2\\
0 & 0 & 1 & 0
\end{array}
\right)
$$

bestimmen wir die Unbekannten von unten nach oben.

Aus der dritten Zeile folgt direkt

$$
x_3=0.
$$

Aus der zweiten Zeile ergibt sich

$$
x_2+3x_3=2.
$$

Mit $x_3=0$ folgt

$$
x_2=2.
$$

Aus der ersten Zeile ergibt sich

$$
x_1+2x_2-x_3=3.
$$

Einsetzen von $x_2=2$ und $x_3=0$ liefert

$$
x_1+4=3,
$$

also

$$
x_1=-1.
$$

:::

:::{dropdown} Gesamtlösung anzeigen

Die Lösung des linearen Gleichungssystems lautet

$$
\boxed{
x_1=-1,\qquad
x_2=2,\qquad
x_3=0
}
$$

bzw. in Vektorschreibweise

$$
\boxed{
\vec{x}
=
\begin{pmatrix}
-1\\
2\\
0
\end{pmatrix}
}
$$
:::
