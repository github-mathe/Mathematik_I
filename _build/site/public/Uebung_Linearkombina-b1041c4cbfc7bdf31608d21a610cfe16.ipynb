## Lineare Abhängigkeit von Polynomen

Gegeben seien die Polynome

$$
p_1(x)=1+x,
$$

$$
p_2(x)=2-x+x^2,
$$

und

$$
p_3(x)=3+x^2.
$$

Untersuchen Sie, ob die Polynome

$$
p_1(x),\qquad p_2(x),\qquad p_3(x)
$$

**linear abhängig oder linear unabhängig** sind.

Stellen Sie dazu die Gleichung

$$
\lambda_1p_1(x)
+
\lambda_2p_2(x)
+
\lambda_3p_3(x)
=
0
$$

auf.

Vergleichen Sie anschließend die Koeffizienten der Potenzen

$$
1,\qquad x,\qquad x^2
$$

und lösen Sie das entstehende homogene lineare Gleichungssystem mit dem **Gauß-Eliminationsverfahren**.

Begründen Sie das Ergebnis zusätzlich mithilfe des Rangs der Koeffizientenmatrix.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Aufstellen der Linearkombination

Zur Untersuchung der linearen Abhängigkeit betrachten wir

$$
\lambda_1p_1(x)
+
\lambda_2p_2(x)
+
\lambda_3p_3(x)
=
0.
$$

Einsetzen der Polynome ergibt

$$
\lambda_1(1+x)
+
\lambda_2(2-x+x^2)
+
\lambda_3(3+x^2)
=
0.
$$

Wir sortieren nach den Potenzen von $x$:

$$
(\lambda_1+2\lambda_2+3\lambda_3)
+
(\lambda_1-\lambda_2)x
+
(\lambda_2+\lambda_3)x^2
=
0.
$$

Damit das Polynom für alle $x$ gleich null ist, müssen alle Koeffizienten gleich null sein.

Somit erhalten wir das homogene lineare Gleichungssystem

$$
\begin{aligned}
\lambda_1+2\lambda_2+3\lambda_3 &=0,\\
\lambda_1-\lambda_2 &=0,\\
\lambda_2+\lambda_3 &=0.
\end{aligned}
$$

Die zugehörige erweiterte Koeffizientenmatrix lautet

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
1 & -1 & 0 & 0\\
0 & 1 & 1 & 0
\end{array}
\right)
}
$$
:::

::::{dropdown} Durchführung der Elimination

Ausgangspunkt ist die erweiterte Koeffizientenmatrix

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
1 & -1 & 0 & 0\\
0 & 1 & 1 & 0
\end{array}
\right).
$$


:::{dropdown} Schritt 1 – Eintrag in der ersten Spalte eliminieren

Der Eintrag $1$ in der ersten Spalte der zweiten Zeile wird eliminiert.

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
1 & -1 & 0 & 0\\
0 & 1 & 1 & 0
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'=\mathrm{II}-\mathrm{I}\\
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & -3 & -3 & 0\\
0 & 1 & 1 & 0
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 2 – Eintrag unterhalb des zweiten Pivotelements eliminieren

Um ohne Brüche zu rechnen, wird der Eintrag $1$ in der zweiten Spalte der dritten Zeile eliminiert.

Dabei wird die bereits veränderte zweite Zeile mit einem Strich gekennzeichnet.

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & -3 & -3 & 0\\
0 & 1 & 1 & 0
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}'=3\cdot\mathrm{III}+\mathrm{II}'
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & -3 & -3 & 0\\
0 & 0 & 0 & 0
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 3 – Zweites Pivotelement normieren

Das zweite Pivotelement wird auf $1$ normiert.

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & -3 & -3 & 0\\
0 & 0 & 0 & 0
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}''=-\frac{1}{3}\mathrm{II}'\\
\end{array}
$$

Damit erhalten wir

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & 1 & 1 & 0\\
0 & 0 & 0 & 0
\end{array}
\right)
}
$$

:::

::::

:::{dropdown} Betrachtung der linearen Abhängigkeit

Nach der Gauß-Elimination erhalten wir

$$
\left(
\begin{array}{ccc|c}
1 & 2 & 3 & 0\\
0 & 1 & 1 & 0\\
0 & 0 & 0 & 0
\end{array}
\right).
$$

Die Koeffizientenmatrix besitzt zwei Pivotelemente.

Daher gilt

$$
\boxed{
\operatorname{rang}(A)=2
}
$$

Es werden jedoch drei Polynome untersucht.

Somit gilt

$$
\operatorname{rang}(A)=2<3.
$$

Das homogene lineare Gleichungssystem besitzt daher neben der trivialen Lösung auch **nichttriviale Lösungen**.

Damit sind die Polynome

$$
\boxed{
p_1(x),\;p_2(x),\;p_3(x)
\text{ linear abhängig.}
}
$$
:::

:::{dropdown} Bestimmung einer Abhängigkeitsbeziehung

Aus der zweiten Zeile folgt

$$
\lambda_2+\lambda_3=0.
$$

Wir setzen

$$
\lambda_3=t,
\qquad
t\in\mathbb{R}.
$$

Damit gilt

$$
\lambda_2=-t.
$$

Aus der ersten Zeile folgt

$$
\lambda_1+2\lambda_2+3\lambda_3=0.
$$

Einsetzen ergibt

$$
\lambda_1-2t+3t=0
$$

und damit

$$
\lambda_1=-t.
$$

Wir wählen beispielsweise

$$
t=-1.
$$

Dann erhalten wir

$$
\lambda_1=1,
\qquad
\lambda_2=1,
\qquad
\lambda_3=-1.
$$

Somit gilt die nichttriviale Linearkombination

$$
\boxed{
p_1(x)+p_2(x)-p_3(x)=0.
}
$$

Tatsächlich ist

$$
(1+x)+(2-x+x^2)-(3+x^2)=0.
$$

Daraus folgt

$$
\boxed{
p_3(x)=p_1(x)+p_2(x).
}
$$

Die drei Polynome sind daher **linear abhängig**.
:::
