## Überprüfung einer Norm

Auf dem Vektorraum $\mathbb{R}^2$ sei für

$$
\mathbf{x}
=
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}
$$

die folgende Abbildung definiert:

$$
\|\mathbf{x}\|_*
=
|x_1|+|x_2|^2.
$$

Untersuchen Sie, ob $\|\cdot\|_*$ tatsächlich eine Norm auf $\mathbb{R}^2$ ist.

Überprüfen Sie dazu die drei Eigenschaften einer Norm:

1. **Positive Definitheit**

   $$
   \|\mathbf{x}\|_*\geq 0
   \qquad\text{und}\qquad
   \|\mathbf{x}\|_*=0
   \iff
   \mathbf{x}=\mathbf{0}.
   $$

2. **Homogenität**

   $$
   \|\alpha\mathbf{x}\|_*
   =
   |\alpha|\cdot\|\mathbf{x}\|_*
   $$

   für alle $\alpha\in\mathbb{R}$ und alle $\mathbf{x}\in\mathbb{R}^2$.

3. **Dreiecksungleichung**

   $$
   \|\mathbf{u}+\mathbf{v}\|_*
   \leq
   \|\mathbf{u}\|_*+\|\mathbf{v}\|_*
   $$

   für alle $\mathbf{u},\mathbf{v}\in\mathbb{R}^2$.

Entscheiden Sie anschließend, ob $\|\cdot\|_*$ eine Norm auf $\mathbb{R}^2$ ist.

## Lösung

Damit $\|\cdot\|_*$ eine Norm ist, müssen **alle drei Eigenschaften** erfüllt sein.

Wir überprüfen die Eigenschaften nacheinander.

:::{dropdown} 1. Positive Definitheit

Gegeben ist

$$
\|\mathbf{x}\|_*
=
|x_1|+|x_2|^2.
$$

Da

$$
|x_1|\geq0
\qquad\text{und}\qquad
|x_2|^2\geq0,
$$

gilt für jeden Vektor $\mathbf{x}\in\mathbb{R}^2$

$$
\boxed{
\|\mathbf{x}\|_*\geq0.
}
$$

Nun betrachten wir den Fall

$$
\|\mathbf{x}\|_*=0.
$$

Dann gilt

$$
|x_1|+|x_2|^2=0.
$$

Da beide Summanden nicht-negativ sind, kann die Summe nur dann null sein, wenn

$$
|x_1|=0
\qquad\text{und}\qquad
|x_2|^2=0.
$$

Daraus folgt

$$
x_1=0
\qquad\text{und}\qquad
x_2=0.
$$

Somit gilt

$$
\|\mathbf{x}\|_*=0
\iff
\mathbf{x}=\mathbf{0}.
$$

Die **Positive Definitheit ist erfüllt**.

$$
\boxed{\text{Positive Definitheit: erfüllt}}
$$
:::

:::{dropdown} 2. Homogenität

Für die Homogenität müsste für alle $\alpha\in\mathbb{R}$ und alle $\mathbf{x}\in\mathbb{R}^2$ gelten:

$$
\|\alpha\mathbf{x}\|_*
=
|\alpha|\cdot\|\mathbf{x}\|_*.
$$

Es ist

$$
\alpha\mathbf{x}
=
\begin{pmatrix}
\alpha x_1\\
\alpha x_2
\end{pmatrix}.
$$

Damit gilt

$$
\begin{aligned}
\|\alpha\mathbf{x}\|_*
&=
|\alpha x_1|+|\alpha x_2|^2\\
&=
|\alpha|\,|x_1|+|\alpha|^2|x_2|^2.
\end{aligned}
$$

Auf der anderen Seite erhalten wir

$$
\begin{aligned}
|\alpha|\cdot\|\mathbf{x}\|_*
&=
|\alpha|
\left(
|x_1|+|x_2|^2
\right)\\
&=
|\alpha|\,|x_1|
+
|\alpha|\,|x_2|^2.
\end{aligned}
$$

Die beiden Ausdrücke stimmen im Allgemeinen nicht überein, da

$$
|\alpha|^2|x_2|^2
\neq
|\alpha||x_2|^2.
$$

Ein konkretes Gegenbeispiel ist

$$
\alpha=2
\qquad\text{und}\qquad
\mathbf{x}
=
\begin{pmatrix}
0\\
1
\end{pmatrix}.
$$

Dann gilt

$$
\|\mathbf{x}\|_*
=
0+1^2
=
1.
$$

Für die linke Seite erhalten wir

$$
\left\|
2
\begin{pmatrix}
0\\
1
\end{pmatrix}
\right\|_*
=
\left\|
\begin{pmatrix}
0\\
2
\end{pmatrix}
\right\|_*
=
4.
$$

Für die rechte Seite gilt dagegen

$$
|2|\cdot\|\mathbf{x}\|_*
=
2\cdot1
=
2.
$$

Somit ist

$$
4\neq2.
$$

Die **Homogenität ist nicht erfüllt**.

$$
\boxed{\text{Homogenität: nicht erfüllt}}
$$
:::

:::{dropdown} 3. Dreiecksungleichung

Für eine Norm müsste für alle Vektoren $\mathbf{u}$ und $\mathbf{v}$ gelten:

$$
\|\mathbf{u}+\mathbf{v}\|_*
\leq
\|\mathbf{u}\|_*+\|\mathbf{v}\|_*.
$$

Wir wählen

$$
\mathbf{u}
=
\begin{pmatrix}
0\\
1
\end{pmatrix},
\qquad
\mathbf{v}
=
\begin{pmatrix}
0\\
1
\end{pmatrix}.
$$

Dann gilt

$$
\mathbf{u}+\mathbf{v}
=
\begin{pmatrix}
0\\
2
\end{pmatrix}.
$$

Für die linke Seite der Dreiecksungleichung erhalten wir

$$
\begin{aligned}
\|\mathbf{u}+\mathbf{v}\|_*
&=
|0|+|2|^2\\
&=4.
\end{aligned}
$$

Für die rechte Seite gilt

$$
\|\mathbf{u}\|_*
=
1
$$

und

$$
\|\mathbf{v}\|_*
=
1.
$$

Damit ist

$$
\|\mathbf{u}\|_*+\|\mathbf{v}\|_*
=
2.
$$

Die Dreiecksungleichung würde also verlangen

$$
4\leq2.
$$

Diese Aussage ist falsch.

Somit ist die **Dreiecksungleichung nicht erfüllt**.

$$
\boxed{\text{Dreiecksungleichung: nicht erfüllt}}
$$
:::

:::{dropdown} Ist $\|\cdot\|_*$ eine Norm?

Wir haben die drei Eigenschaften überprüft:

| Eigenschaft | erfüllt? |
|---|---|
| Positive Definitheit | ja |
| Homogenität | nein |
| Dreiecksungleichung | nein |

Damit eine Abbildung eine Norm ist, müssen **alle drei Eigenschaften** erfüllt sein.

Da die Homogenität und die Dreiecksungleichung verletzt werden, ist

$$
\boxed{
\|\mathbf{x}\|_*
=
|x_1|+|x_2|^2
}
$$

**keine Norm auf $\mathbb{R}^2$.**

$$
\boxed{\|\cdot\|_*\text{ ist keine Norm.}}
$$
:::
