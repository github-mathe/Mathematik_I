## Komplexe Gleichung mit Substitution

Wir betrachten die komplexe Gleichung:

$$
z^6-7z^3+6=0.
$$

**Ziel:** Bestimmen Sie alle komplexen Lösungen dieser Gleichung.

**Hinweis:** Die Gleichung ist vom Grad $6$ und weist die Potenzen $z^6$ und $z^3$ auf. Durch geeignete Substitution können wir die Gleichung vereinfachen und lösen.

### Schritt 1: Substitution

Da die Gleichung Terme mit $z^6$ und $z^3$ enthält, führen wir die Substitution ein:

$$
u=z^3.
$$

Dadurch wird die ursprüngliche Gleichung zu einer quadratischen Gleichung in $u$:

$$
u^2-7u+6=0.
$$

### Schritt 2: Lösen der quadratischen Gleichung

Die quadratische Gleichung lautet:

$$
u^2-7u+6=0.
$$

Anwendung der Mitternachtsformel:

$$
u_{1,2}
=
\frac{7\pm\sqrt{49-24}}{2}
=
\frac{7\pm5}{2}.
$$

Damit:

$$
u_1=6,
\qquad
u_2=1.
$$

### Schritt 3: Rücksubstitution

Wir erhalten zwei Gleichungen:

- $z^3=6$
- $z^3=1$

### Schritt 4: Lösen von $z^3=6$

**Berechnung des Betrags und Arguments von $w=6$:**

$$
\begin{aligned}
|w| &= 6,\\
\varphi &= 0.
\end{aligned}
$$

**Berechnung des Betrags der Lösungen:**

$$
r
=
|w|^{1/3}
=
\sqrt[3]{6}.
$$

**Berechnung der Argumente:**

$$
\theta_k
=
\frac{0+2\pi k}{3}
=
\frac{2\pi k}{3},
\qquad
k=0,1,2.
$$

**Lösungen:**

$$
z_k
=
\sqrt[3]{6}\,
e^{i\frac{2\pi k}{3}}.
$$

### Schritt 5: Lösen von $z^3=1$

**Berechnung des Betrags und Arguments von $w=1$:**

$$
\begin{aligned}
|w| &= 1,\\
\varphi &= 0.
\end{aligned}
$$

**Berechnung des Betrags der Lösungen:**

$$
r
=
|w|^{1/3}
=
1.
$$

**Berechnung der Argumente:**

$$
\theta_k
=
\frac{0+2\pi k}{3}
=
\frac{2\pi k}{3},
\qquad
k=0,1,2.
$$

**Lösungen:**

$$
z_k
=
e^{i\frac{2\pi k}{3}}.
$$

### Zusammenfassung der Lösungen

Insgesamt haben wir sechs Lösungen.

Aus

$$
z^3=6
$$

erhalten wir:

$$
z_0
=
\sqrt[3]{6}\,e^{i0},
\qquad
z_1
=
\sqrt[3]{6}\,e^{i\frac{2\pi}{3}},
\qquad
z_2
=
\sqrt[3]{6}\,e^{i\frac{4\pi}{3}}.
$$

Aus

$$
z^3=1
$$

erhalten wir:

$$
z_3=1,
\qquad
z_4=e^{i\frac{2\pi}{3}},
\qquad
z_5=e^{i\frac{4\pi}{3}}.
$$

### Graphische Darstellung

```python
import numpy as np
import matplotlib.pyplot as plt

# Winkel der drei kubischen Wurzeln
k = np.arange(3)
theta = 2 * np.pi * k / 3

# Radien
r1 = 1
r2 = np.cbrt(6)

# Lösungen von z^3 = 1
x1 = r1 * np.cos(theta)
y1 = r1 * np.sin(theta)

# Lösungen von z^3 = 6
x2 = r2 * np.cos(theta)
y2 = r2 * np.sin(theta)

fig, ax = plt.subplots(figsize=(7, 7))

t = np.linspace(0, 2*np.pi, 500)

# Kreise
ax.plot(r1*np.cos(t), r1*np.sin(t), linestyle="--")
ax.plot(r2*np.cos(t), r2*np.sin(t), linestyle="--")

# Achsen
ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

# Punkte
ax.scatter(x1, y1)
ax.scatter(x2, y2)

# Beschriftungen
for j in range(3):
    ax.text(
        1.12*x2[j],
        1.12*y2[j],
        rf"$z_{j}$",
        ha="center",
        va="center"
    )

    ax.text(
        1.18*x1[j],
        1.18*y1[j],
        rf"$z_{j+3}$",
        ha="center",
        va="center"
    )

ax.set_xlabel("Re")
ax.set_ylabel("Im")

lim = 2.5
ax.set_xlim(-lim, lim)
ax.set_ylim(-lim, lim)

ax.set_aspect("equal")
ax.grid(alpha=0.3)

plt.show()
```

### Zusammenfassung: Lösung mit Substitution

Die Methode der Substitution eignet sich zur Lösung von Polynomgleichungen höheren Grades:

- **Identifizieren von Mustern:** Wiederkehrende Potenzen oder Strukturen suchen, die eine Substitution ermöglichen (z. B. $z^{2k},z^k$).
- **Einführen der Substitution:** Eine geeignete Variable $u=z^k$ oder Funktion von $z$ setzen, um die Gleichung zu vereinfachen.
- **Reduzieren der Gleichung:** Umformen der ursprünglichen Gleichung in eine Gleichung niedrigeren Grades in $u$ (oft eine quadratische Gleichung).
- **Lösen der reduzierten Gleichung:** Bestimmen der Lösungen für $u$.
- **Rücksubstitution:** Einsetzen der gefundenen Werte von $u$ in die Substitutionsgleichung, um Gleichungen in $z$ zu erhalten.
- **Lösen der Gleichungen in $z$:** Ermitteln der Lösungen für $z$ aus den Gleichungen geringeren Grades.
- **Anzahl der Lösungen:** Für eine Polynomgleichung $n$-ten Grades existieren insgesamt $n$ Lösungen in $\mathbb C$.
