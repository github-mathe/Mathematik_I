## Kern einer linearen Abbildung (Engl. Kernel)

Der Kern einer linearen Abbildung

$$
T:V\to W,
$$

bezeichnet als $\operatorname{Kern}(T)$, ist die Menge aller Vektoren in $V$, die auf den Nullvektor abgebildet werden:

$$
\boxed{
\operatorname{Kern}(T)
=
\{v\in V:T(v)=0\}
}
$$

Der Kern beschreibt also alle Vektoren, die durch die Abbildung $T$ „verschwinden“.

Der Kern ist immer ein Unterraum von $V$ und wird auch Nullraum genannt.

<a id="bild-lineare-abbildung"></a>

## Bild einer linearen Abbildung (Engl. Range)

Das Bild einer linearen Abbildung

$$
T:V\to W,
$$

bezeichnet als $\operatorname{Bild}(T)$, ist die Menge aller Vektoren in $W$, die durch $T$ aus Vektoren von $V$ erzeugt werden:

$$
\boxed{
\operatorname{Bild}(T)
=
\{w\in W:w=T(v)\text{ für ein }v\in V\}
}
$$

Das Bild beschreibt also, wohin die Vektoren durch $T$ abgebildet werden.

Es ist ein Unterraum von $W$.

## Beispiel für Kern und Bild einer linearen Abbildung

Betrachte die lineare Abbildung

$$
T:\mathbb{R}^3\to\mathbb{R}^3
$$

definiert durch:

$$
T(x,y,z)
=
(2x+y,\;x+2z,\;y-z)^T.
$$

### Kern

Wir suchen die Vektoren

$$
(x,y,z)\in\mathbb{R}^3,
$$

für die

$$
T(x,y,z)=\vec 0.
$$

Das bedeutet:

$$
\begin{aligned}
2x+y&=0\\
x+2z&=0\\
y-z&=0
\end{aligned}
$$

Die Lösung ist

$$
\boxed{
\operatorname{Kern}(T)
=
\{(0,0,0)^T\}.
}
$$

Man sagt auch dass der Kern leer ist.

## Kern für lineare Abbildungen

### Kern mit Dimension 1

Betrachten wir eine lineare Abbildung

$$
T:\mathbb{R}^3\to\mathbb{R}^2
$$

definiert durch:

$$
T(x,y,z)
=
(x+y,\;y-z).
$$

Die Bedingung

$$
T(x,y,z)=\vec 0
$$

führt auf:

$$
\begin{aligned}
x+y&=0\\
y-z&=0
\end{aligned}
$$

Die Lösungsmenge des Systems ist

$$
\mathcal L
=
\left\{
\begin{pmatrix}
-t\\
t\\
t
\end{pmatrix}
:
t\in\mathbb R
\right\}.
$$

Der Kern ist:

$$
\boxed{
\operatorname{Kern}(T)
=
\operatorname{span}
\left\{
(-1,1,1)^T
\right\}
}
$$

also eindimensional.

<a id="parametrischer-kern"></a>

## Parametrische Abbildung und Dimension des Kerns

Sei

$$
T_\alpha:\mathbb{R}^3\to\mathbb{R}^3
$$

eine Familie linearer Abbildungen, definiert durch:

$$
T_\alpha(x,y,z)
=
(\alpha x+y,\;x-z,\;\alpha z)^T.
$$

### Fall $\alpha=0$

Das Gleichungssystem

$$
T_\alpha(x,y,z)=\vec 0
$$

ergibt:

$$
\begin{aligned}
y&=0\\
x-z&=0\\
0&=0
\end{aligned}
$$

Die Lösungsmenge des homogenen Systems ist

$$
\mathcal L
=
\left\{
(t,0,t)^T,\;
t\in\mathbb R
\right\}.
$$

Der eindimensionale Kern ist dann:

$$
\boxed{
\operatorname{Kern}(T_0)
=
\operatorname{span}
\left\{
(1,0,1)^T
\right\}.
}
$$

### Fall $\alpha\neq0$

Das Gleichungssystem ist:

$$
\begin{aligned}
\alpha x+y&=0\\
x-z&=0\\
\alpha z&=0
\end{aligned}
$$

Für $\alpha\neq0$ folgt

$$
x=z=y=0,
$$

daher:

$$
\operatorname{Kern}(T_\alpha)
=
\operatorname{span}\{\emptyset\}.
$$

Der Kern ist trivial mit Dimension $0$.

## Parametrische Abbildung und Dimension des Kerns

Sei

$$
T_\alpha:\mathbb{R}^3\to\mathbb{R}^3
$$

eine Familie linearer Abbildungen, definiert durch:

$$
T_\alpha(x,y,z)
=
(x-y+z,\;2x+\alpha z,\;3x-y+2z)^T.
$$

### Fall $\alpha\neq0$

Das Gleichungssystem

$$
T_\alpha(x,y,z)=\vec 0
$$

ist:

$$
\begin{aligned}
x-y+z&=0\\
2x+\alpha z&=0\\
3x-y+2z&=0
\end{aligned}
$$

### Schritt 1: Matrixdarstellung des Systems

$$
\begin{pmatrix}
1&-1&1\\
2&0&\alpha\\
3&-1&2
\end{pmatrix}
\begin{pmatrix}
x\\
y\\
z
\end{pmatrix}
=
\begin{pmatrix}
0\\
0\\
0
\end{pmatrix}.
$$

### Schritt 2: Gauß-Elimination

$$
\begin{bmatrix}
1&-1&1\\
2&0&\alpha\\
3&-1&2
\end{bmatrix}
\rightarrow
\begin{bmatrix}
1&-1&1\\
0&2&-2+\alpha\\
0&0&1-\alpha
\end{bmatrix}.
$$

### Schritt 3: Analyse

- Für $\alpha\neq1$:

  Die Matrix ist invertierbar, und

  $$
  x=y=z=0.
  $$

  Der Kern ist trivial:

  $$
  \operatorname{Kern}(T_{\alpha\neq1})
  =
  \operatorname{span}\{\emptyset\}.
  $$

  Dimension:

  $$
  \dim(\operatorname{Kern}(T_\alpha))
  =
  0.
  $$

- Für $\alpha=1$:

  Die letzte Zeile wird

  $$
  0=0,
  $$

  was einen freien Parameter erlaubt.

  Die Lösungsmenge ist

  $$
  \mathcal L
  =
  \left\{
  (-t/2,t/2,t)^T,\;
  t\in\mathbb R
  \right\}.
  $$

  Der Kern ist:

  $$
  \boxed{
  \operatorname{Kern}(T_1)
  =
  \operatorname{span}
  \left\{
  (-1/2,1/2,1)^T
  \right\}.
  }
  $$

  Dimension:

  $$
  \dim(\operatorname{Kern}(T_1))
  =
  1.
  $$

<a id="bestimmung-bild"></a>

## Bestimmung des Bildes einer linearen Abbildung

Eine lineare Abbildung $T$ kann durch eine Matrix

$$
\mathbf A\in\mathbb R^{m\times n}
$$

dargestellt werden.

Das **Bild** der Abbildung entspricht dem Spaltenraum der Matrix $\mathbf A$.

Der Spaltenraum ist der Vektorraum, der von den Spalten der Matrix $\mathbf A$ aufgespannt wird.

Eine Basis des Bildes wird durch die linear unabhängigen Spalten von $\mathbf A$ gebildet.

Die Dimension des Bildes, auch **Rang der Matrix** genannt, gibt die Anzahl der linear unabhängigen Spalten an.

Um das Bild von $T$ zu bestimmen, prüfen wir die lineare Unabhängigkeit der Spalten der Matrix $\mathbf A$.

Zum Beispiel:

$$
\mathbf A
=
\begin{pmatrix}
2&1&0\\
1&0&2\\
0&1&-1
\end{pmatrix}.
$$

In diesem Fall hat die Matrix den Rang $3$, da alle Spalten linear unabhängig sind.

Somit ist:

$$
\boxed{
\dim(\operatorname{Bild}(T))
=
3.
}
$$

## Bild der Abbildung durch Gauß-Elimination

**Gegeben:** Die Matrix, die die lineare Abbildung $T$ repräsentiert:

$$
\vec A
=
\begin{pmatrix}
2&1&0\\
1&0&2\\
0&1&-1
\end{pmatrix}.
$$

### Schritt 1: Gauß-Elimination zur Bestimmung der Pivot-Spalten

$$
\begin{array}{l|ccc|l}
\text{I}&2&1&0&\\
\text{II'}&0&-1/2&2&
\text{II'}=\text{II}-1/2\,\text{I}\\
\text{III'}&0&0&3&
\text{III'}=\text{III}+2\,\text{I}
\end{array}
$$

### Schritt 2: Pivot-Spalten bestimmen

- Die Pivot-Spalten sind die 1., 2. und 3. Spalte.
- Der Rang der Matrix ist:

$$
\operatorname{Rang}(\vec A)=3.
$$

**Feststellung:** Das Bild von $T$ ist der gesamte $\mathbb R^3$-Raum, und:

$$
\boxed{
\dim(\operatorname{Bild}(T))
=
3.
}
$$

## Beispiel: Lineare Abbildung $T:\mathbb R^4\to\mathbb R^3$

**Gegeben:** Die Matrix, die die lineare Abbildung $T$ repräsentiert:

$$
\vec A
=
\begin{pmatrix}
1&2&3&4\\
0&1&1&2\\
1&1&2&2
\end{pmatrix}.
$$

**Ziel:** Finde eine Basis für das Bild der Abbildung.

### Schritt 1: Gauß-Elimination zur Bestimmung der Pivot-Spalten

$$
\begin{array}{l|cccc|l}
\text{I}&1&2&3&4&\\
\text{II}&0&1&1&2&\\
\text{III'}&0&-1&-1&-2&
\text{III'}=\text{III}-\text{I}
\end{array}
\qquad
\begin{array}{l|cccc|l}
\text{I}&1&2&3&4&\\
\text{II}&0&1&1&2&\\
\text{III''}&0&0&0&0&
\text{III''}=\text{III'}+\text{II}
\end{array}
$$

### Schritt 2: Pivot-Spalten identifizieren

- Die Pivot-Spalten sind die 1. und 2. Spalte.
- Der Rang der Matrix ist:

$$
\operatorname{Rang}(\vec A)
=
2.
$$

### Schritt 3: Basis des Bildes bestimmen

Die Basis des Bildes besteht aus den Spalten-Vektoren von $\vec A$, die den Pivot-Spalten entsprechen:

$$
\boxed{
\text{Basis}
=
\left\{
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\begin{pmatrix}
2\\
1\\
1
\end{pmatrix}
\right\}.
}
$$

Die **Dimension** des Bildes ist:

$$
\dim(\operatorname{Bild}(T))
=
2.
$$

> **Hinweis:** Die Pivot-Spalten haben wir in der Stufenform nach den Eliminationsschritten identifiziert. Für das Bild von $\vec A$ verwenden wir jedoch die entsprechenden Spalten der ursprünglichen Matrix.

## Basis des Bildes

- **Wichtig:** Die Gauß-Elimination verändert den aufgespannten Raum der Zeilen nicht, aber sie verändert im Allgemeinen den aufgespannten Raum der Spalten. Daher können die Spalten der Stufenform nicht direkt als Basis des Bildes von $A$ verwendet werden.

- Stattdessen identifizieren wir die Positionen der Pivot-Spalten in der Zeilenstufenform und nehmen die entsprechenden Spalten der ursprünglichen Matrix, um eine Basis des Bildes zu bestimmen.

- Die Gauß-Elimination lässt sich als Multiplikation mit einer invertierbaren Matrix $E$ schreiben:

  $$
  R=EA.
  $$

  Dadurch werden zwar die Spalten von $A$ linear transformiert, aber eine Menge von Spalten ist genau dann linear unabhängig, wenn die entsprechenden Spalten in $R$ linear unabhängig sind.

- **Schlussfolgerung:** Die Positionen der Pivot-Spalten bestimmen die linear unabhängigen Spalten der ursprünglichen Matrix $A$, und diese Spalten bilden eine Basis des Bildes von $A$.

<a id="bild-transponierte"></a>

## Alternative Methode: Transponierte Matrix

**Alternative Methode zur Bestimmung des Bildes:**

- Statt die Pivot-Spalten in der ursprünglichen Matrix zu identifizieren, können wir die **Transponierte** der Matrix verwenden:

$$
\vec A^\top.
$$

- Wir führen die Gauß-Elimination auf $\vec A^\top$ durch.
- Die **Pivot-Zeilen** der transponierten Matrix entsprechen den **linear unabhängigen Spalten** der ursprünglichen Matrix.

## Gauß-Elimination auf $\vec A^\top$

**Gegeben:** Die transponierte Matrix:

$$
\vec A^\top
=
\begin{pmatrix}
1&0&1\\
2&1&1\\
3&1&2\\
4&2&2
\end{pmatrix}.
$$

### Schritt 1: Eliminieren der ersten Spalte

$$
\begin{array}{l|ccc|l}
\text{I}&1&0&1&\\
\text{II'}&0&1&-1&
\text{II'}=\text{II}-2\,\text{I}\\
\text{III'}&0&1&-1&
\text{III'}=\text{III}-3\,\text{I}\\
\text{IV'}&0&2&-2&
\text{IV'}=\text{IV}-4\,\text{I}
\end{array}
$$

### Schritt 2: Eliminieren der zweiten Spalte

$$
\begin{array}{l|ccc|l}
\text{I}&1&0&1&\\
\text{II'}&0&1&-1&\\
\text{III''}&0&0&0&
\text{III''}=\text{III'}-\text{II'}\\
\text{IV''}&0&0&0&
\text{IV''}=\text{IV'}-2\,\text{II'}
\end{array}
$$

### Schritt 3: Ergebnis der Elimination

- Die Matrix ist jetzt in reduzierter Zeilenstufenform.
- Die Pivot-Zeilen sind die erste und zweite Zeile.
- Die Positionen der Pivot-Zeilen bestimmen die unabhängigen Spalten der ursprünglichen Matrix.

## Basis des Bildes

- Die Pivot-Zeilen entsprechen den linear unabhängigen Spalten der ursprünglichen Matrix $\vec A$. Sie ergeben sich aus Linearkombinationen der ursprünglichen Spalten, die durch die Zeilenoperationen auf der transponierten Matrix während der Gauß-Elimination erzeugt werden.

- Wir nehmen die erste und zweite Zeile als Basisvektoren für das Bild:

$$
\boxed{
\text{Basis von }\operatorname{Bild}(\vec A)
=
\left\{
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
\begin{pmatrix}
0\\
1\\
-1
\end{pmatrix}
\right\}.
}
$$

- Die Dimension des Bildes ist:

$$
\dim(\operatorname{Bild}(\vec A))
=
2.
$$

## Begründung der Methode

**Warum funktioniert diese Methode?**

- Die Gauß-Elimination verändert den aufgespannten Raum der Zeilen nicht.
- Daher bleiben die Pivot-Zeilen der transponierten Matrix linear unabhängig und können als Basis des Bildes verwendet werden.
- Die Positionen der Pivot-Zeilen in $\vec A^\top$ entsprechen den Positionen der unabhängigen Spalten in der ursprünglichen Matrix $\vec A$.
- Diese Methode bietet eine alternative Perspektive, indem wir die Zeilen und Spalten der Matrix austauschen.

<a id="fazit-kern-bild"></a>

## Fazit: Bild und Kern einer linearen Abbildung

Gegeben sei eine lineare Abbildung

$$
T:V\to W
$$

zwischen zwei Vektorräumen.

Zwei fundamentale Begriffe von $T$ sind:

1. Das **Bild** von $T$, also die Menge der Vektoren, die durch $T$ erreicht werden können.
2. Der **Kern** von $T$, also die Menge der Vektoren, die durch $T$ auf den Nullvektor abgebildet werden.

Das **Bild** von $T$ ist definiert als:

$$
\boxed{
\operatorname{Bild}(T)
=
\{T(\vec v)\mid\vec v\in V\}
\subseteq W.
}
$$

Der **Kern** von $T$ ist definiert als:

$$
\boxed{
\operatorname{Kern}(T)
=
\{\vec v\in V\mid T(\vec v)=\vec0\}
\subseteq V.
}
$$

<a id="bild-kern-matrix"></a>

## Bild und Kern einer Matrix

Jede lineare Abbildung

$$
T:\mathbb R^n\to\mathbb R^m
$$

kann durch eine Matrix

$$
\mathbf A\in\mathbb R^{m\times n}
$$

dargestellt werden.

Die Begriffe Bild und Kern gelten auch für Matrizen:

### Bild

Das **Bild** von $\mathbf A$ entspricht der Spaltenraum der Matrix:

$$
\boxed{
\operatorname{Bild}(\mathbf A)
=
\operatorname{span}
\{\text{Spalten von }\mathbf A\}.
}
$$

### Kern

Der **Kern** von $\mathbf A$ entspricht der Lösungsmenge des homogenen Systems:

$$
\boxed{
\operatorname{Kern}(\mathbf A)
=
\operatorname{span}
\{
\vec x\in\mathbb R^n
\mid
\mathbf A\vec x=\vec0
\}.
}
$$

## Beispiel: Bild und Kern einer Matrix

**Gegeben:** Die Matrix

$$
\mathbf A
=
\begin{pmatrix}
1&2&3\\
0&1&4\\
0&0&0
\end{pmatrix}.
$$

### Bild

Die Spalten der Matrix spannen den Bildraum auf:

$$
\boxed{
\operatorname{Bild}(\mathbf A)
=
\operatorname{span}
\left\{
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix},
\begin{pmatrix}
2\\
1\\
0
\end{pmatrix}
\right\}.
}
$$

Die Dimension des Bildes ist:

$$
\dim(\operatorname{Bild}(\mathbf A))
=
2.
$$

### Basis des Kerns

Lösungsmenge des homogenen Systems

$$
\mathbf A\vec x=\vec0.
$$

Es gilt:

$$
\begin{pmatrix}
1&2&3\\
0&1&4\\
0&0&0
\end{pmatrix}
\begin{pmatrix}
x_1\\
x_2\\
x_3
\end{pmatrix}
=
\vec0.
$$

Dies führt auf die Gleichungen:

$$
x_1+2x_2+3x_3=0,
\qquad
x_2+4x_3=0.
$$

Daraus ergibt sich:

$$
x_1=-2x_2-3x_3,
\qquad
x_2=-4x_3,
\qquad
x_3=t,
\quad
t\in\mathbb R.
$$

Damit:

$$
\vec x
=
t(5,-4,1)^T,
\qquad
t\in\mathbb R.
$$

### Kern

$$
\boxed{
\operatorname{Kern}(\mathbf A)
=
\operatorname{span}
\left\{
\begin{pmatrix}
5\\
-4\\
1
\end{pmatrix}
\right\}.
}
$$
