## Einige Symbole der Mengenlehre

- **Elemente der Mengen**
  - $a \in A$ bedeutet „$a$ ist ein Element der Menge $A$.“
  - $b \notin A$ bedeutet „$b$ ist kein Element der Menge $A$.“

- **Teilmengen**
  - $A \subseteq B$ bedeutet „$A$ ist eine Teilmenge von $B$.“
  - $A \subset B$ bedeutet „$A$ ist eine echte Teilmenge von $B$.“
  - $A \nsubseteq B$ bedeutet „$A$ ist keine Teilmenge von $B$.“

- **Mengenoperationen**
  - $A \cup B$ bedeutet „Vereinigung von $A$ und $B$.“
  - $A \cap B$ bedeutet „Schnittmenge von $A$ und $B$.“
  - $A \setminus B$ bedeutet „Differenzmenge von $A$ und $B$ (Elemente in $A$, aber nicht in $B$).“

- **Kardinalität und Leere Menge**
  - $|A|$ bedeutet „Kardinalität der Menge $A$ (Anzahl der Elemente in $A$).“
  - $\emptyset$ ist die leere Menge (eine Menge ohne Elemente).
