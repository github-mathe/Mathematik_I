## Polarform und Eulersche Formel

### Polarform

Komplexe Zahlen können auch in Polarform ausgedrückt werden:

$$
z = r(\cos\theta + i\sin\theta)
$$

wobei:

- $r = |z| = \sqrt{a^2+b^2}$ der **Betrag** oder **Modul** ist,
- $\theta = \arg(z) = \tan^{-1}\left(\frac{b}{a}\right)$ das **Argument** oder der **Winkel** ist.

### Eulersche Formel

Die Eulersche Formel verbindet trigonometrische Funktionen mit der Exponentialfunktion:

$$
e^{i\theta}
=
\cos\theta+i\sin\theta.
$$

Damit kann die Polarform geschrieben werden als:

$$
z = re^{i\theta}.
$$

### Graph der Tangensfunktion

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 5))

# Die Tangensfunktion wird abschnittsweise gezeichnet,
# damit nicht über die Polstellen hinweg verbunden wird.
intervalle = [
    (-2*np.pi, -3*np.pi/2 - 0.01),
    (-3*np.pi/2 + 0.01, -np.pi/2 - 0.01),
    (-np.pi/2 + 0.01, np.pi/2 - 0.01),
    (np.pi/2 + 0.01, 3*np.pi/2 - 0.01),
    (3*np.pi/2 + 0.01, 2*np.pi)
]

for a, b in intervalle:
    x = np.linspace(a, b, 500)
    y = np.tan(x)
    ax.plot(x, y)

# Vertikalasymptoten
for x0 in [-3*np.pi/2, -np.pi/2, np.pi/2, 3*np.pi/2]:
    ax.axvline(x0, linestyle="--")

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

ax.set_xlim(-2*np.pi, 2*np.pi)
ax.set_ylim(-5, 5)

ax.set_xticks([
    -2*np.pi,
    -3*np.pi/2,
    -np.pi,
    -np.pi/2,
    0,
    np.pi/2,
    np.pi,
    3*np.pi/2,
    2*np.pi
])

ax.set_xticklabels([
    r"$-2\pi$",
    r"$-\frac{3\pi}{2}$",
    r"$-\pi$",
    r"$-\frac{\pi}{2}$",
    r"$0$",
    r"$\frac{\pi}{2}$",
    r"$\pi$",
    r"$\frac{3\pi}{2}$",
    r"$2\pi$"
])

ax.set_yticks([-4, -2, 0, 2, 4])

ax.set_xlabel("$x$")
ax.set_ylabel(r"$y=\tan x$")

ax.grid()

plt.show()
```

### Die Umkehrfunktion der Tangensfunktion

- Die **Tangensfunktion** $y=\tan x$ ist periodisch und hat Vertikalasymptoten bei

$$
x=\pm\frac{\pi}{2},
\pm\frac{3\pi}{2},
\ldots,
$$

wo die Funktion unendlich wird.

- Um die **Umkehrfunktion** (Arcustangens oder $\arctan$) zu definieren, betrachten wir nur den Bereich der Tangensfunktion, in dem sie **streng monoton** und somit umkehrbar ist.

- Wir beschränken die Tangensfunktion auf den Bereich

$$
\left(-\frac{\pi}{2},\frac{\pi}{2}\right).
$$

In diesem Bereich ist die Tangensfunktion **injektiv** und durchläuft alle möglichen Werte von $-\infty$ bis $\infty$.

- Die **Umkehrfunktion** der Tangensfunktion, die als $\arctan(y)$ oder $\tan^{-1}(y)$ geschrieben wird, gibt uns für jeden Wert $y$ die entsprechende Winkelgröße $x$ im Intervall

$$
\left(-\frac{\pi}{2},\frac{\pi}{2}\right).
$$

### Graph der Arcustangensfunktion ($\arctan$)

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 500)
y = np.arctan(x)

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(x, y)

# Horizontale Asymptoten
ax.axhline(np.pi/2, linestyle="--")
ax.axhline(-np.pi/2, linestyle="--")

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

ax.set_xlim(-5, 5)
ax.set_ylim(-np.pi/2, np.pi/2)

ax.set_xticks([-4, -3, -2, -1, 0, 1, 2, 3, 4])

ax.set_yticks([
    -np.pi/2,
    -np.pi/4,
    0,
    np.pi/4,
    np.pi/2
])

ax.set_yticklabels([
    r"$-\frac{\pi}{2}$",
    r"$-\frac{\pi}{4}$",
    r"$0$",
    r"$\frac{\pi}{4}$",
    r"$\frac{\pi}{2}$"
])

ax.set_xlabel("$x$")
ax.set_ylabel(r"$y=\arctan x$")

ax.grid()

plt.show()
```
