## Determinante einer $4\times4$-Matrix mit der Laplace-Entwicklung

Gegeben sei die Matrix

$$
\mathbf{A}
=
\begin{pmatrix}
1 & 2 & 0 & -1\\
0 & 3 & 0 & 2\\
2 & 1 & 1 & 0\\
1 & 0 & 2 & 1
\end{pmatrix}.
$$

Berechnen Sie die Determinante

$$
\det(\mathbf{A})
$$

mit der **Laplace-Entwicklung**.

Entwickeln Sie die Determinante entlang der zweiten Zeile.

Verwenden Sie dabei die Schreibweise

$$
\det(A)
=
\sum_{j=1}^{n}a_{ij}s_{ij}
$$

mit

$$
s_{ij}
=
(-1)^{i+j}\det(M_{ij}),
$$

wobei $M_{ij}$ die Streichungsmatrix des Elements $a_{ij}$ bezeichnet.

## Lösung

Wir entwickeln die Determinante entlang der zweiten Zeile.

Die zweite Zeile lautet

$$
\begin{pmatrix}
0 & 3 & 0 & 2
\end{pmatrix}.
$$

Nach der Laplace-Entwicklung gilt

$$
\det(A)
=
a_{21}s_{21}
+
a_{22}s_{22}
+
a_{23}s_{23}
+
a_{24}s_{24}.
$$

Da

$$
a_{21}=0
\qquad\text{und}\qquad
a_{23}=0,
$$

müssen nur die Faktoren $s_{22}$ und $s_{24}$ bestimmt werden.

:::{dropdown} Berechnung des Faktors $s_{22}$

Für den Faktor $s_{22}$ gilt

$$
s_{22}
=
(-1)^{2+2}\det(M_{22}).
$$

Die Streichungsmatrix $M_{22}$ entsteht, indem die zweite Zeile und die zweite Spalte von $\mathbf{A}$ entfernt werden:

$$
M_{22}
=
\begin{pmatrix}
1 & 0 & -1\\
2 & 1 & 0\\
1 & 2 & 1
\end{pmatrix}.
$$

Da

$$
(-1)^{2+2}=1,
$$

gilt

$$
s_{22}
=
\det
\begin{pmatrix}
1 & 0 & -1\\
2 & 1 & 0\\
1 & 2 & 1
\end{pmatrix}.
$$

Wir entwickeln diese Determinante entlang der ersten Zeile:

$$
\begin{aligned}
s_{22}
={}&
1\cdot(-1)^{1+1}
\det
\begin{pmatrix}
1 & 0\\
2 & 1
\end{pmatrix}\\
&+
0\cdot(-1)^{1+2}
\det
\begin{pmatrix}
2 & 0\\
1 & 1
\end{pmatrix}\\
&+
(-1)\cdot(-1)^{1+3}
\det
\begin{pmatrix}
2 & 1\\
1 & 2
\end{pmatrix}.
\end{aligned}
$$

Damit ergibt sich

$$
\begin{aligned}
s_{22}
&=
1\cdot(1\cdot1-0\cdot2)
-
(2\cdot2-1\cdot1)\\
&=
1-3\\
&=
-2.
\end{aligned}
$$

Somit gilt

$$
\boxed{
s_{22}=-2.
}
$$
:::

:::{dropdown} Berechnung des Faktors $s_{24}$

Für den Faktor $s_{24}$ gilt

$$
s_{24}
=
(-1)^{2+4}\det(M_{24}).
$$

Die Streichungsmatrix $M_{24}$ entsteht, indem die zweite Zeile und die vierte Spalte von $\mathbf{A}$ entfernt werden:

$$
M_{24}
=
\begin{pmatrix}
1 & 2 & 0\\
2 & 1 & 1\\
1 & 0 & 2
\end{pmatrix}.
$$

Da

$$
(-1)^{2+4}=1,
$$

gilt

$$
s_{24}
=
\det
\begin{pmatrix}
1 & 2 & 0\\
2 & 1 & 1\\
1 & 0 & 2
\end{pmatrix}.
$$

Wir entwickeln erneut entlang der ersten Zeile:

$$
\begin{aligned}
s_{24}
={}&
1\cdot(-1)^{1+1}
\det
\begin{pmatrix}
1 & 1\\
0 & 2
\end{pmatrix}\\
&+
2\cdot(-1)^{1+2}
\det
\begin{pmatrix}
2 & 1\\
1 & 2
\end{pmatrix}\\
&+
0\cdot(-1)^{1+3}
\det
\begin{pmatrix}
2 & 1\\
1 & 0
\end{pmatrix}.
\end{aligned}
$$

Damit ergibt sich

$$
\begin{aligned}
s_{24}
&=
1\cdot(1\cdot2-1\cdot0)
-
2\cdot(2\cdot2-1\cdot1)\\
&=
2-2\cdot3\\
&=
2-6\\
&=
-4.
\end{aligned}
$$

Somit gilt

$$
\boxed{
s_{24}=-4.
}
$$
:::

:::{dropdown} Einsetzen in die Laplace-Entwicklung

Die Entwicklung entlang der zweiten Zeile lautet

$$
\det(A)
=
a_{21}s_{21}
+
a_{22}s_{22}
+
a_{23}s_{23}
+
a_{24}s_{24}.
$$

Mit

$$
a_{21}=0,
\qquad
a_{22}=3,
\qquad
a_{23}=0,
\qquad
a_{24}=2
$$

und

$$
s_{22}=-2,
\qquad
s_{24}=-4
$$

ergibt sich

$$
\begin{aligned}
\det(A)
&=
0\cdot s_{21}
+
3\cdot(-2)
+
0\cdot s_{23}
+
2\cdot(-4)\\
&=
-6-8\\
&=
-14.
\end{aligned}
$$

Damit gilt

$$
\boxed{
\det(\mathbf{A})=-14.
}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Für

$$
\mathbf{A}
=
\begin{pmatrix}
1 & 2 & 0 & -1\\
0 & 3 & 0 & 2\\
2 & 1 & 1 & 0\\
1 & 0 & 2 & 1
\end{pmatrix}
$$

wird entlang der zweiten Zeile entwickelt:

$$
\det(A)
=
a_{21}s_{21}
+
a_{22}s_{22}
+
a_{23}s_{23}
+
a_{24}s_{24}.
$$

Da

$$
a_{21}=a_{23}=0,
$$

bleibt

$$
\det(A)
=
3s_{22}+2s_{24}.
$$

Die Faktoren sind

$$
\boxed{
s_{22}=-2
}
$$

und

$$
\boxed{
s_{24}=-4.
}
$$

Damit ergibt sich

$$
\begin{aligned}
\det(A)
&=
3\cdot(-2)
+
2\cdot(-4)\\
&=
-6-8\\
&=
-14.
\end{aligned}
$$

Somit gilt

$$
\boxed{
\det(\mathbf{A})=-14.
}
$$
:::
