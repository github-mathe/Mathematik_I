# Vektor- und Matrixnormen


Normen sind ein wichtiges Werkzeug der linearen Algebra, um die **Größe** beziehungsweise **Länge** von Vektoren und Matrizen zu beschreiben.

Bei Vektoren ermöglichen Normen beispielsweise die Bestimmung von Abständen und die quantitative Beschreibung ihrer Länge. Zu den wichtigsten Vektornormen gehören die $L_1$-Norm, die $L_2$-Norm und die $L_\infty$-Norm. Sie spielen unter anderem bei Fehlerabschätzungen in numerischen Verfahren sowie bei Regularisierungsmethoden im maschinellen Lernen eine wichtige Rolle.

Auch Matrizen können mithilfe von Normen charakterisiert werden. Matrixnormen geben Auskunft über die Größe einer Matrix und darüber, wie stark eine lineare Abbildung Vektoren verändern kann. Sie werden beispielsweise bei der Untersuchung der Kondition von Gleichungssystemen, bei Approximationen und Optimierungsproblemen sowie in der Bildverarbeitung eingesetzt.

Im Folgenden werden zunächst wichtige Vektor- und Matrixnormen vorgestellt und anschließend ihre Eigenschaften, Zusammenhänge und Anwendungen näher betrachtet.
Normen dienen dazu, die **Größe** beziehungsweise **Länge** von Vektoren und Matrizen zu beschreiben.

## Definition einer Norm

Eine **Norm** ist eine Abbildung, die einem Vektor oder einer Matrix eine nicht-negative reelle Zahl zuordnet, die dessen „Größe“ oder „Länge“ misst.

Für Vektoren:

$$
\|\cdot\|:\mathbb{R}^n\rightarrow\mathbb{R},
\qquad
\|\vec{v}\|\geq 0
\quad
\text{für alle }
\vec{v}\in\mathbb{R}^n.
$$

Für Matrizen:

$$
\|\cdot\|:\mathbb{R}^{m\times n}\rightarrow\mathbb{R},
\qquad
\|A\|\geq 0
\quad
\text{für alle }
A\in\mathbb{R}^{m\times n}.
$$

### Eigenschaften einer Norm

Eine Norm erfüllt insbesondere folgende Eigenschaften:

- **Positivität**

  $$
  \|\vec{v}\|\geq 0
  $$

  und

  $$
  \|\vec{v}\|=0
  \quad\Longleftrightarrow\quad
  \vec{v}=\vec{0}.
  $$

- **Absolute Homogenität**

  Für alle $\lambda\in\mathbb{R}$ und $\vec{v}\in\mathbb{R}^n$ gilt:

  $$
  \|\lambda\vec{v}\|
  =
  |\lambda|\,\|\vec{v}\|.
  $$

- **Dreiecksungleichung**

  Für alle $\vec{v},\vec{w}\in\mathbb{R}^n$ gilt:

  $$
  \|\vec{v}+\vec{w}\|
  \leq
  \|\vec{v}\|+\|\vec{w}\|.
  $$

<a id="vektornormen"></a>

## Vektornormen

Für einen Vektor

$$
\vec{v}
=
\begin{pmatrix}
v_1\\
v_2\\
\vdots\\
v_n
\end{pmatrix}
\in\mathbb{R}^n
$$

werden häufig die folgenden Normen verwendet.

### $L_1$-Norm

Die $L_1$-Norm, auch **Manhattan-Norm** genannt, ist definiert durch

$$
\|\vec{v}\|_1
=
\sum_{i=1}^{n}|v_i|.
$$

Sie entspricht der Summe der Beträge aller Komponenten des Vektors.

### $L_2$-Norm

Die $L_2$-Norm, auch **euklidische Norm** genannt, ist definiert durch

$$
\|\vec{v}\|_2
=
\sqrt{\sum_{i=1}^{n}v_i^2}.
$$

Sie entspricht der gewöhnlichen geometrischen Länge eines Vektors.

### $L_\infty$-Norm

Die $L_\infty$-Norm, auch **Maximum-Norm** genannt, ist definiert durch

$$
\|\vec{v}\|_\infty
=
\max_{1\leq i\leq n}|v_i|.
$$

Sie entspricht dem größten Betrag einer Komponente des Vektors.

<a id="matrixnormen"></a>

## Matrixnormen

Für eine Matrix

$$
A=(a_{ij})\in\mathbb{R}^{m\times n}
$$

können ebenfalls unterschiedliche Normen definiert werden.

### $L_1$-Norm – Spaltensummennorm

Die $L_1$-Norm einer Matrix ist die maximale Spaltensumme:

$$
\|A\|_1
=
\max_{1\leq j\leq n}
\sum_{i=1}^{m}|a_{ij}|.
$$

Dazu werden für jede Spalte die Beträge der Elemente addiert. Die größte dieser Summen bestimmt die Norm.

### $L_\infty$-Norm – Zeilensummennorm

Die $L_\infty$-Norm einer Matrix ist die maximale Zeilensumme:

$$
\|A\|_\infty
=
\max_{1\leq i\leq m}
\sum_{j=1}^{n}|a_{ij}|.
$$

Dazu werden für jede Zeile die Beträge der Elemente addiert. Die größte dieser Summen bestimmt die Norm.

### Frobenius-Norm

Die Frobenius-Norm ist definiert durch

$$
\|A\|_F
=
\sqrt{
\sum_{i=1}^{m}
\sum_{j=1}^{n}
|a_{ij}|^2
}.
$$

Die Frobenius-Norm misst die Größe einer Matrix, indem die Quadrate aller Matrixelemente addiert und anschließend die Quadratwurzel gebildet wird.

Sie wird beispielsweise verwendet:

- in der Regularisierung, beispielsweise bei Ridge-Regression oder Tikhonov-Regularisierung,
- bei der Untersuchung der Kondition einer Matrix,
- bei der Analyse der Fehlerfortpflanzung in numerischen Methoden.

<a id="beispiel-vektornormen"></a>

## Numerisches Beispiel – Vektornormen

Gegeben sei der Vektor

$$
\vec{v}
=
\begin{pmatrix}
1\\
-2\\
3
\end{pmatrix}
\in\mathbb{R}^3.
$$

### $L_1$-Norm

$$
\|\vec{v}\|_1
=
|1|+|-2|+|3|
=
1+2+3
=
6.
$$

### $L_2$-Norm

$$
\|\vec{v}\|_2
=
\sqrt{1^2+(-2)^2+3^2}
$$

$$
=
\sqrt{1+4+9}
=
\sqrt{14}
\approx 3{,}74.
$$

### $L_\infty$-Norm

$$
\|\vec{v}\|_\infty
=
\max(|1|,|-2|,|3|)
=
3.
$$

```python
import sympy as sp

v = sp.Matrix([1, -2, 3])

norm_1 = sum(abs(x) for x in v)
norm_2 = sp.sqrt(sum(x**2 for x in v))
norm_inf = max(abs(x) for x in v)

print("L1-Norm =", norm_1)
print("L2-Norm =", norm_2)
print("L∞-Norm =", norm_inf)
```

<a id="beispiel-matrixnormen"></a>

## Numerisches Beispiel – Matrixnormen

Gegeben sei die Matrix

$$
A
=
\begin{pmatrix}
1 & -2 & 3\\
4 & 0 & -1\\
-1 & 2 & 5
\end{pmatrix}
\in\mathbb{R}^{3\times3}.
$$

### $L_1$-Norm – Spaltensummennorm

Wir bestimmen zunächst die Summen der Beträge in den einzelnen Spalten:

$$
\|A\|_1
=
\max
\left(
|1|+|4|+|-1|,
|-2|+|0|+|2|,
|3|+|-1|+|5|
\right).
$$

Damit ergibt sich

$$
\|A\|_1
=
\max(6,4,9)
=
9.
$$

### $L_\infty$-Norm – Zeilensummennorm

Für die Zeilensummennorm betrachten wir die Summen der Beträge in den einzelnen Zeilen:

$$
\|A\|_\infty
=
\max
\left(
|1|+|-2|+|3|,
|4|+|0|+|-1|,
|-1|+|2|+|5|
\right).
$$

Damit folgt

$$
\|A\|_\infty
=
\max(6,5,8)
=
8.
$$

### Frobenius-Norm

Für die Frobenius-Norm werden alle Matrixelemente berücksichtigt:

$$
\|A\|_F
=
\sqrt{
1^2+(-2)^2+3^2+
4^2+0^2+(-1)^2+
(-1)^2+2^2+5^2
}.
$$

Somit gilt

$$
\|A\|_F
=
\sqrt{61}
\approx
7{,}81.
$$

```python
import sympy as sp

A = sp.Matrix([
    [1, -2,  3],
    [4,  0, -1],
    [-1, 2,  5]
])

# L1-Norm: maximale Spaltensumme
spaltensummen = [
    sum(abs(A[i, j]) for i in range(A.rows))
    for j in range(A.cols)
]

norm_1 = max(spaltensummen)

# L∞-Norm: maximale Zeilensumme
zeilensummen = [
    sum(abs(A[i, j]) for j in range(A.cols))
    for i in range(A.rows)
]

norm_inf = max(zeilensummen)

# Frobenius-Norm
norm_f = sp.sqrt(
    sum(A[i, j]**2 for i in range(A.rows) for j in range(A.cols))
)

print("Spaltensummen:", spaltensummen)
print("L1-Norm =", norm_1)

print("\nZeilensummen:", zeilensummen)
print("L∞-Norm =", norm_inf)

print("\nFrobenius-Norm =", norm_f)
```

## Beweis der Eigenschaften der $L_1$-Norm

Eine **Norm** ist eine Funktion, die jedem Vektor in einem Vektorraum eine nicht-negative reelle Zahl zuordnet.

Sie muss bestimmte Eigenschaften erfüllen:

1. Positive Definitheit
2. Homogenität
3. Dreiecksungleichung

Die $L_1$-Norm, auch bekannt als Manhattan- oder Taxicab-Norm, ist definiert als:

$$
\|\mathbf{x}\|_1
=
|x_1|+|x_2|+\dots+|x_n|.
$$

### Positive Definitheit

Eine Norm $\|\cdot\|$ erfüllt die **Positive Definitheit**, wenn für alle Vektoren $\mathbf{x}$ gilt:

$$
\|\mathbf{x}\|\geq 0
\quad\text{und}\quad
\|\mathbf{x}\|=0
\iff
\mathbf{x}=\mathbf{0}.
$$

**Satz:** Die $L_1$-Norm erfüllt die Positive Definitheit.

**Beweis:**

1. **Nicht-Negativität:**

$$
\|\mathbf{x}\|_1
=
|x_1|+|x_2|+\dots+|x_n|
\geq 0.
$$

Da Beträge stets nicht-negativ sind.

2. **Nur der Nullvektor hat Norm Null:**

$$
\|\mathbf{x}\|_1=0
\implies
|x_1|=|x_2|=\dots=|x_n|=0
$$

$$
\implies
x_1=x_2=\dots=x_n=0
\implies
\mathbf{x}=\mathbf{0}.
$$

### Homogenität

Eine Norm $\|\cdot\|$ erfüllt die **Homogenität**, wenn für alle Skalare $\alpha$ und alle Vektoren $\mathbf{x}$ gilt:

$$
\|\alpha\mathbf{x}\|
=
|\alpha|\cdot\|\mathbf{x}\|.
$$

**Satz:** Die $L_1$-Norm erfüllt die Homogenität.

**Beweis:**

$$
\begin{aligned}
\|\alpha\mathbf{x}\|_1
&=
|\alpha x_1|
+
|\alpha x_2|
+
\dots
+
|\alpha x_n|
\\
&=
|\alpha|\cdot|x_1|
+
|\alpha|\cdot|x_2|
+
\dots
+
|\alpha|\cdot|x_n|
\\
&=
|\alpha|\cdot\|\mathbf{x}\|_1.
\end{aligned}
$$

### Dreiecksungleichung

Eine Norm $\|\cdot\|$ erfüllt die **Dreiecksungleichung**, wenn für alle Vektoren $\mathbf{u}$ und $\mathbf{v}$ gilt:

$$
\|\mathbf{u}+\mathbf{v}\|
\leq
\|\mathbf{u}\|+\|\mathbf{v}\|.
$$

**Satz:** Die $L_1$-Norm erfüllt die Dreiecksungleichung.

**Beweis:**

Für

$$
\mathbf{u}=(u_1,u_2,\dots,u_n)^T
$$

und

$$
\mathbf{v}=(v_1,v_2,\dots,v_n)^T
$$

gilt:

$$
\begin{aligned}
\|\mathbf{u}+\mathbf{v}\|_1
&=
\sum_{i=1}^{n}|u_i+v_i|
\\
&\leq
\sum_{i=1}^{n}(|u_i|+|v_i|)
\\
&=
\sum_{i=1}^{n}|u_i|
+
\sum_{i=1}^{n}|v_i|
\\
&=
\|\mathbf{u}\|_1+\|\mathbf{v}\|_1.
\end{aligned}
$$

Die Ungleichung

$$
|u_i+v_i|
\leq
|u_i|+|v_i|
$$

folgt aus der Dreiecksungleichung der Betragsfunktion.

## Überblick über die Vektor-Normen

Die $L_1$-Norm erfüllt alle drei notwendigen Eigenschaften einer Norm:

1. Positive Definitheit
2. Homogenität
3. Dreiecksungleichung

Somit ist die $L_1$-Norm eine gültige Norm im Vektorraum $\mathbb{R}^n$.

Der Einheitskreis besteht aus allen Punkten, bei denen

$$
|x|+|y|=1.
$$

Der Einheitskreis der $L_1$-Norm bildet in zwei Dimensionen einen Diamanten (Rhombus) mit den Scheitelpunkten

$$
(1,0),\quad
(0,1),\quad
(-1,0),\quad
(0,-1).
$$

### Warum ergibt $|x|+|y|=1$ eine diamantförmige Gestalt?

Die Gleichung

$$
|x|+|y|=1
$$

beschreibt in $\mathbb{R}^2$ den Einheitskreis der $L_1$-Norm.

Diese Gleichung führt zu einer diamantförmigen Kurve, auch Rhombus genannt.

Um dies zu verstehen, analysieren wir die Gleichung in allen vier Quadranten.

### Fall 1: Erster Quadrant ($x\geq0$ und $y\geq0$)

In diesem Quadranten sind $x\geq0$ und $y\geq0$, daher können die Beträge entfernt werden:

$$
|x|+|y|=1
\implies
x+y=1.
$$

Dies ist die Gleichung einer Geraden mit Steigung $-1$ und Achsenabschnitten $(1,0)$ und $(0,1)$.

Der Teil der Linie im ersten Quadranten verläuft von $(1,0)$ nach $(0,1)$.

### Fall 2: Zweiter Quadrant ($x\leq0$ und $y\geq0$)

In diesem Quadranten ist $x\leq0$ und $y\geq0$, daher:

$$
|x|+|y|=1
\implies
-x+y=1.
$$

Dies ist ebenfalls die Gleichung einer Geraden, aber mit Steigung $1$, die durch die Punkte $(-1,0)$ und $(0,1)$ verläuft.

Der Teil der Linie im zweiten Quadranten verläuft von $(-1,0)$ nach $(0,1)$.

### Fall 3: Dritter Quadrant ($x\leq0$ und $y\leq0$)

In diesem Quadranten sind $x\leq0$ und $y\leq0$, daher:

$$
|x|+|y|=1
\implies
-x-y=1.
$$

Dies ist die Gleichung einer Geraden mit Steigung $-1$, die durch die Punkte $(-1,0)$ und $(0,-1)$ verläuft.

Der Teil der Linie im dritten Quadranten verläuft von $(-1,0)$ nach $(0,-1)$.

### Fall 4: Vierter Quadrant ($x\geq0$ und $y\leq0$)

In diesem Quadranten ist $x\geq0$ und $y\leq0$, daher:

$$
|x|+|y|=1
\implies
x-y=1.
$$

Dies ist die Gleichung einer Geraden mit Steigung $1$, die durch die Punkte $(1,0)$ und $(0,-1)$ verläuft.

Der Teil der Linie im vierten Quadranten verläuft von $(1,0)$ nach $(0,-1)$.

```python
import matplotlib.pyplot as plt

x = [1, 0, -1, 0, 1]
y = [0, 1, 0, -1, 0]

plt.figure(figsize=(5, 5))
plt.plot(x, y)
plt.fill(x, y, alpha=0.2)

plt.axhline(0)
plt.axvline(0)

plt.scatter([1, 0, -1, 0], [0, 1, 0, -1])

plt.text(1, 0, "(1,0)")
plt.text(0, 1, "(0,1)")
plt.text(-1, 0, "(-1,0)")
plt.text(0, -1, "(0,-1)")

plt.xlabel("x")
plt.ylabel("y")
plt.axis("equal")
plt.grid()
plt.show()
```

## Kompatibilität von Vektor- und Matrixnormen

Die **Kompatibilität von Vektor- und Matrixnormen** bedeutet, dass die Norm einer Matrix im Zusammenhang mit der Norm eines Vektors konsistent berechnet wird.

**Verträgliche Normen** bezeichnen die Beziehung zwischen Vektor- und Matrixnormen, sodass die Ungleichung

$$
\|\vec{A}\mathbf{x}\|
\leq
\|\vec{A}\|\cdot\|\mathbf{x}\|
$$

für eine Matrix

$$
\vec{A}\in\mathbb{R}^{m\times n}
$$

und einen Vektor

$$
\mathbf{x}\in\mathbb{R}^n
$$

erfüllt ist.

### Bedeutung der Kompatibilität

- **Stabilität:** Diese Eigenschaft hilft dabei, sicherzustellen, dass lineare Abbildungen durch Matrizen die Größe eines Vektors nicht zu stark verzerren.
- **Effiziente Berechnung:** Kompatible Normen ermöglichen eine einfache Abschätzung der Wirkung einer Matrix auf einen Vektor.

### Verträgliche Normen für die $L_1$-Norm

Vektor-Norm: $L_1$-Norm, definiert als

$$
\|\mathbf{x}\|_1
=
\sum_{i=1}^{n}|x_i|.
$$

**Verträgliche Matrix-Norm:** Maximale Spaltensummennorm:

$$
\|A\|_1
=
\max_{1\leq j\leq n}
\sum_{i=1}^{m}|a_{ij}|.
$$

Diese Norm summiert die Beträge der Elemente jeder Spalte der Matrix und nimmt das Maximum dieser Summen.

### Verträgliche Normen für die $L_\infty$-Norm

Vektor-Norm: $L_\infty$-Norm, definiert als

$$
\|\mathbf{x}\|_\infty
=
\max_{1\leq i\leq n}|x_i|.
$$

**Verträgliche Matrix-Norm:** Maximale Zeilensummennorm:

$$
\|A\|_\infty
=
\max_{1\leq i\leq m}
\sum_{j=1}^{n}|a_{ij}|.
$$

Diese Norm summiert die Beträge der Elemente jeder Zeile und nimmt das Maximum dieser Summen.

### Verträgliche Normen für die $L_2$-Norm

Vektor-Norm: $L_2$-Norm, definiert als

$$
\|\mathbf{x}\|_2
=
\sqrt{\sum_{i=1}^{n}x_i^2}.
$$

**Verträgliche Matrix-Norm:** Spektralnorm, definiert als

$$
\|A\|_2
=
\sup_{\mathbf{x}\neq0}
\frac{\|A\mathbf{x}\|_2}{\|\mathbf{x}\|_2}.
$$

Die Spektralnorm ist die größte Singularzahl der Matrix und misst, wie stark eine Matrix einen Vektor dehnen kann.

### Die Frobenius-Norm und ihre Verträglichkeit

Die **Frobenius-Norm** ist eine Matrix-Norm, definiert als

$$
\|\vec{A}\|_F
=
\sqrt{
\sum_{i=1}^{m}
\sum_{j=1}^{n}
|a_{ij}|^2
}.
$$

### Verträglichkeit

Es gilt:

$$
\|\vec{A}\vec{x}\|_2^2
=
\sum_{i=1}^{m}
\left(
\vec{a}_i^{\top}\vec{x}
\right)^2,
$$

wobei $\vec{a}_i^{\top}$ die $i$-te Zeile von $\vec{A}$ bezeichnet.

Wendet man nun die **Cauchy-Schwarz-Ungleichung** an, so erhält man:

$$
\left(
\vec{a}_i^{\top}\vec{x}
\right)^2
\leq
\|\vec{a}_i\|_2^2
\,
\|\vec{x}\|_2^2.
$$

Durch Summation über alle Zeilen folgt:

$$
\begin{aligned}
\|\vec{A}\vec{x}\|_2^2
&\leq
\left(
\sum_{i=1}^{m}
\|\vec{a}_i\|_2^2
\right)
\|\vec{x}\|_2^2
\\
&=
\|\vec{A}\|_F^2
\,
\|\vec{x}\|_2^2.
\end{aligned}
$$

Damit folgt:

$$
\boxed{
\|\vec{A}\vec{x}\|_2
\leq
\|\vec{A}\|_F
\,
\|\vec{x}\|_2
}
$$

## Wahl der Norm in der Praxis

Die Wahl der richtigen Norm hängt stark von der spezifischen Anwendung ab. Unterschiedliche Normen erfassen verschiedene Aspekte der Messung:

- Die $L_1$-Norm eignet sich gut für sparse Probleme, z. B. Lasso-Regression, oder Situationen, in denen man robust gegenüber Ausreißern sein möchte.

- Die $L_\infty$-Norm wird häufig verwendet, wenn der maximale Fehler interessiert, z. B. in der Worst-Case-Analyse.

- Die $L_2$-Norm wird oft bei Problemen verwendet, bei denen es um den euklidischen Abstand geht, z. B. bei der linearen Regression oder bei Fehlerabschätzungen in numerischen Algorithmen.

- Die Auswahl der passenden Norm ist entscheidend für die Analyse und Effizienz eines Algorithmus. Verträgliche Normen gewährleisten, dass mathematische Operationen wie Matrix-Vektor-Multiplikationen stabil durchgeführt werden können.

## Fehlerberechnung mit der $L_2$-Norm

Die $L_2$-Norm kann verwendet werden, um den **Fehler** zwischen Messdaten und einem Modell zu messen.

Für eine Datenmenge

$$
\mathbf{y}_{\text{meas}}
$$

und ein Modell

$$
\mathbf{y}_{\text{model}}
$$

misst die $L_2$-Norm den Unterschied (Fehler) zwischen den beiden.

Der Fehlervektor ist definiert als:

$$
\mathbf{e}
=
\mathbf{y}_{\text{meas}}
-
\mathbf{y}_{\text{model}}.
$$

Der Fehler in der $L_2$-Norm wird berechnet als:

$$
\|\mathbf{e}\|_2
=
\sqrt{
\sum_{i=1}^{n}
\left(
y_{\text{meas},i}
-
y_{\text{model},i}
\right)^2
}.
$$

Diese Formel gibt den **euklidischen Abstand** zwischen den gemessenen Werten und den modellierten Werten an.

### Beispiel: Fehler in einem linearen Modell

Nehmen wir an, wir haben eine Menge von Messdaten

$$
\mathbf{y}_{\text{meas}}
=
[y_1,y_2,\dots,y_n]
$$

und ein **lineares Modell** der Form:

$$
\hat{y}_i
=
ax_i+b,
$$

wobei $a$ die Steigung und $b$ der Achsenabschnitt ist.

Der Fehlervektor $\mathbf{e}$ ist:

$$
e_i
=
y_{\text{meas},i}
-
\hat{y}_i
=
y_{\text{meas},i}
-
(ax_i+b).
$$

Der Fehler in der $L_2$-Norm wird dann berechnet als:

$$
\|\mathbf{e}\|_2
=
\sqrt{
\sum_{i=1}^{n}
\left(
y_{\text{meas},i}
-
(ax_i+b)
\right)^2
}.
$$

Dieser Ausdruck misst den quadratischen Fehler zwischen den Messpunkten und dem Modell.

### Minimierung des $L_2$-Fehlers: Lineare Regression

Die Minimierung des $L_2$-Fehlers führt zur **linearen Regression**, einer Methode, um die besten Werte für $a$ und $b$ zu finden.

Ziel ist es, $a$ und $b$ so zu wählen, dass der Fehler in der $L_2$-Norm minimiert wird:

$$
\min_{a,b}
\sum_{i=1}^{n}
\left(
y_{\text{meas},i}
-
(ax_i+b)
\right)^2.
$$

Die Lösung dieses Minimierungsproblems liefert die optimalen Parameter $a$ (Steigung) und $b$ (Achsenabschnitt), sodass das Modell die Daten am besten beschreibt.

### Grafische Darstellung des $L_2$-Fehlers

Der Fehler zwischen einem Messwert und dem zugehörigen Modellwert kann grafisch als vertikaler Abstand dargestellt werden.

Für jeden Messpunkt wird dabei die Differenz

$$
e_i
=
y_{\text{meas},i}
-
y_{\text{model},i}
$$

betrachtet.

Die roten gestrichelten Linien zeigen den Abstand (Fehler) zwischen den Messdaten $y_{\text{meas}}$ und den Modellwerten $y_{\text{model}}$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Messdaten
x = np.array([0, 1, 2, 3, 4], dtype=float)
y_meas = np.array([1, 2, 1.5, 2.8, 4], dtype=float)

# Lineare Regression
a, b = np.polyfit(x, y_meas, 1)

# Modellwerte
y_model = a * x + b

# Fehlervektor
e = y_meas - y_model

# L2-Norm des Fehlers
fehler_l2 = np.sqrt(np.sum(e**2))

print("Steigung a =", a)
print("Achsenabschnitt b =", b)
print("L2-Fehler =", fehler_l2)

# Darstellung
plt.figure(figsize=(6, 4))

plt.scatter(x, y_meas, label="Messdaten")
plt.plot(x, y_model, label="Modell")

for xi, yi, ymi in zip(x, y_meas, y_model):
    plt.plot([xi, xi], [yi, ymi], linestyle="--")

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```
