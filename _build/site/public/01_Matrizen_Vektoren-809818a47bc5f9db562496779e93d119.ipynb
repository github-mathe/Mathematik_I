# Vektoren und Matrizen

## Motivation

In den Ingenieurwissenschaften stoßen wir häufig auf komplexe Probleme, die mithilfe mathematischer Modelle beschrieben werden.

Ein zentrales Werkzeug der **Linearen Algebra** sind **Matrizen** und **lineare Gleichungssysteme**. Sie ermöglichen es, viele technische Problemstellungen kompakt darzustellen und effizient zu lösen.

## Was ist eine Matrix?

Eine **Matrix** ist eine rechteckige Anordnung von Zahlen, Symbolen oder Ausdrücken, die in **Zeilen** $Z_i$ (horizontal) und **Spalten** $S_i$ (vertikal) organisiert sind. Matrizen dienen als kompakte Methode zur Darstellung von Daten und Beziehungen zwischen Variablen.

Beispiel:

$$
\begin{array}{c | c c c}
     & \textbf{S1} & \textbf{S2} & \textbf{S3} \\
    \hline
    \textbf{Z1} & a_{11} & a_{12} & a_{13} \\
    \textbf{Z2} & a_{21} & a_{22} & a_{23} \\
    \textbf{Z3} & a_{31} & a_{32} & a_{33} \\
\end{array}
$$

## Zeilen einer Matrix

**Zeile $Z_2$**:

$$
\begin{array}{c | c c c}
     & \textbf{S1} & \textbf{S2} & \textbf{S3} \\
    \hline
    \textbf{Z1} & a_{11} & a_{12} & a_{13} \\
    \color{red}{\textbf{Z2}} &
    \color{red}{a_{21}} &
    \color{red}{a_{22}} &
    \color{red}{a_{23}} \\
    \textbf{Z3} & a_{31} & a_{32} & a_{33} \\
\end{array}
$$

## Spalten einer Matrix

**Spalte $S_1$**

$$
\begin{array}{c | c c c}
     & \color{red}{\textbf{S1}} & \textbf{S2} & \textbf{S3} \\
    \hline
    \textbf{Z1} & \color{red}{a_{11}} & a_{12} & a_{13} \\
    \textbf{Z2} & \color{red}{a_{21}} & a_{22} & a_{23} \\
    \textbf{Z3} & \color{red}{a_{31}} & a_{32} & a_{33} \\
\end{array}
$$

## Elemente und Dimensionen einer Matrix

- **Elemente**: Jedes Element einer Matrix wird durch seine Position angegeben, z. B. $a_{ij}$ für das Element in der $i$-ten Zeile und $j$-ten Spalte.
- **Dimensionen**: Eine Matrix mit $m$ Zeilen und $n$ Spalten wird als $m \times n$-Matrix bezeichnet.
