## Körper

### Definition eines Körpers

Ein **Körper** ist ein Ring, bei dem zusätzlich jedes Element außer $0$ ein Inverses hat und die Multiplikation kommutativ ist.

Das bedeutet, dass für alle

$$
a \in K,
\qquad
a \neq 0,
$$

ein $a^{-1} \in K$ existiert, sodass

$$
a \cdot a^{-1} = 1.
$$

### Beispiele für Körper

- Die rationalen Zahlen

$$
\mathbb{Q}.
$$

- Die reellen Zahlen

$$
\mathbb{R}.
$$

- Die komplexen Zahlen

$$
\mathbb{C}.
$$
