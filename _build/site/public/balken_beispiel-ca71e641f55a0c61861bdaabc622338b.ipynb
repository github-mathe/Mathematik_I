## Anwendung von Gauß-Eliminationsverfahren: Balken-Beispiel

Wir betrachten einen einfach unterstützten Balken mit einer Länge von $L = 10\,\text{m}$. Der Balken hat zwei Auflager:

- Ein festes Auflager an Punkt $A$ (linkes Ende)
- Ein Rollenlager an Punkt $B$ (rechtes Ende)

Auf den Balken wirkt die folgende Kraft:

- $F = 15\,\text{kN}$ wirkt nach unten bei $x_1 = 5\,\text{m}$ von Punkt $A$

**Ziel:**

Zu bestimmen sind die Reaktionen an den Auflagern $A$ (Reaktionskraft $R_A$ und Moment $M_A$) sowie $B$ (Reaktionskraft $R_B$) unter Verwendung der Gleichgewichtsbedingungen.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Arc

fig, ax = plt.subplots(figsize=(12, 5))

# Balken
ax.plot([0, 10], [0, 0], linewidth=4)

# x- und y-Achsen
ax.annotate(
    "",
    xy=(11, 0),
    xytext=(-1, 0),
    arrowprops=dict(arrowstyle="->")
)
ax.text(11.1, 0, r"$x$", fontsize=14)

ax.annotate(
    "",
    xy=(0, 2),
    xytext=(0, -2),
    arrowprops=dict(arrowstyle="->")
)
ax.text(0.1, 2, r"$y$", fontsize=14)

# Festes Auflager A
ax.plot([0, 0], [-0.2, 0.2], linewidth=2)
ax.text(0.2, -0.45, "A", fontsize=13)

for y in [-0.2, -0.1, 0, 0.1, 0.2]:
    ax.plot([0, -0.3], [y, y - 0.2], linewidth=1.5)

# Rollenlager B
rolle = Circle((10, -0.2), 0.15, fill=False, linewidth=2)
ax.add_patch(rolle)
ax.text(10, -0.7, "B", ha="center", fontsize=13)

# Reaktionskraft R_A
ax.annotate(
    "",
    xy=(0, 1.0),
    xytext=(0, 0.2),
    arrowprops=dict(arrowstyle="->", linewidth=2)
)
ax.text(0.15, 0.7, r"$R_A$", fontsize=14)

# Reaktionskraft R_B
ax.annotate(
    "",
    xy=(10, 0.8),
    xytext=(10, 0),
    arrowprops=dict(arrowstyle="->", linewidth=2)
)
ax.text(10.15, 0.55, r"$R_B$", fontsize=14)

# Kraft F
ax.annotate(
    "",
    xy=(5, -1.5),
    xytext=(5, 0),
    arrowprops=dict(arrowstyle="->", linewidth=2)
)
ax.text(4.7, -1.0, r"$F$", fontsize=14)

# Abstand
ax.text(5.7, -0.5, r"$x_1=\frac{L}{2}$", fontsize=14)

# Moment M_A
arc = Arc((0, 0), 1.2, 1.2, theta1=0, theta2=180, linewidth=2)
ax.add_patch(arc)

ax.annotate(
    "",
    xy=(-0.6, 0),
    xytext=(-0.58, 0.15),
    arrowprops=dict(arrowstyle="->", linewidth=2)
)
ax.text(-0.9, 0.65, r"$M_A$", fontsize=14)

# Balkenlänge
ax.text(5, 0.5, "Länge = 10 m", ha="center", fontsize=13)

ax.set_xlim(-1.5, 11.5)
ax.set_ylim(-2.2, 2.2)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

### Moment in der Statik

Ein **Moment** ist eine Größe, die beschreibt, wie eine Kraft einen Körper um einen Punkt oder eine Achse dreht.

**Beispiel:**

Wenn eine Kraft an einem Balken an einem bestimmten Punkt wirkt, erzeugt sie ein Moment, das den Balken um den Punkt dreht. Das Moment ist größer, wenn die Kraft größer ist oder wenn sie weiter vom Drehpunkt entfernt wirkt.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Arc

fig, ax = plt.subplots(figsize=(9, 6))

# Balken
ax.plot([0, 6], [0, 0], linewidth=2)
ax.text(3, 0.25, "Balken", ha="center", fontsize=13)

# Auflager
ax.plot([0, 0], [-0.1, 0.1], linewidth=2)
ax.text(0, -0.35, "A", ha="center", fontsize=13)

ax.plot([6, 6], [-0.1, 0.1], linewidth=2)
ax.text(6, -0.35, "B", ha="center", fontsize=13)

# Kraft F
ax.annotate(
    "",
    xy=(7, 2),
    xytext=(6, 0),
    arrowprops=dict(arrowstyle="->", linewidth=2)
)
ax.text(6.65, 1.1, r"$F$", fontsize=14)

# Verlängerung der Wirkungslinie
ax.plot([6, 4], [0, -4], linestyle="--")
ax.text(4.8, -2.5, "Linie der Kraftwirkung", fontsize=11)

# Orthogonaler Abstand r
ax.plot([0, 4.8], [0, -2.4], linestyle=":")
ax.text(2.3, -1.5, r"$r$", fontsize=14)
ax.text(4.25, -2.55, r"$\perp$", fontsize=14)

# Moment M
arc = Arc((0, 0), 1.2, 1.2, theta1=0, theta2=180, linewidth=2)
ax.add_patch(arc)

ax.annotate(
    "",
    xy=(-0.6, 0),
    xytext=(-0.58, 0.15),
    arrowprops=dict(arrowstyle="->", linewidth=2)
)
ax.text(-0.9, 0.65, r"$M$", fontsize=14)

ax.set_xlim(-1.5, 7.5)
ax.set_ylim(-4.5, 2.5)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

### Mathematische Definition des Moments

Das Moment $M$ einer Kraft $F$ bezüglich eines Punktes (oder einer Achse) wird berechnet als das Produkt der Kraft und des senkrechten Abstands $r$ von der Linie der Kraftwirkung zum Drehpunkt:

$$
M = F \cdot r
$$

- $M$: Moment,
- $F$: Kraft,
- $r$: Hebelarm, Abstand zwischen der Linie der Kraft und dem Drehpunkt.

**Einheit:**

Das Moment wird in der Einheit **Newtonmeter (Nm)** gemessen.

### Gleichgewicht der Kräfte und Momente

Ein System befindet sich im Gleichgewicht, wenn die folgenden Bedingungen erfüllt sind:

- **Summe der vertikalen und horizontalen Kräfte:**

$$
\sum F_y = 0
\quad \text{und} \quad
\sum F_x = 0
$$

- **Summe der Momente:**

$$
\sum M = 0
$$

Die Summe aller Momente um jeden Punkt (z. B. um einen Auflagerpunkt) muss null sein.

Diese Bedingungen gewährleisten, dass das System stabil bleibt und sich nicht bewegt. Im statischen Gleichgewicht sind translationale und rotatorische Bewegungen ausgeschlossen.

### Gleichgewichtsbedingungen

- **Summe der vertikalen Kräfte:**

$$
R_A + R_B = F
$$

- **Summe der Momente um Punkt A:**

$$
M_A + R_B \cdot L - F \cdot \dfrac{L}{2} = 0
$$

### Kompatibilitätsbedingung

- Die Durchbiegung am Punkt $B$ ($x = L$) muss Null sein:

$$
y(L) = 0
$$

- Berechnung der Durchbiegung bei $x = L$ aufgrund von:
  - Punktlast $F$
  - Reaktion $R_B$

### Berechnung der Durchbiegung

- **Durchbiegung aufgrund von $F$:**

$$
\delta_F = \dfrac{5 F L^3}{48 EI}
$$

- **Durchbiegung aufgrund von $R_B$:**

$$
\delta_{R_B} = \dfrac{R_B L^3}{3 EI}
$$

- **Kompatibilitätsbedingung:**

$$
\delta_F - \delta_{R_B} = 0
$$

- **Ergibt:**

$$
\dfrac{5 F L^3}{48 EI} - \dfrac{R_B L^3}{3 EI} = 0
$$

$$
\dfrac{5 F}{48} = \dfrac{R_B}{3}
$$

$$
16 R_B - 5 F = 0
$$

### Beschreibung der Symbole

- $\delta$ — **Durchbiegung**: Die Auslenkung oder Verformung eines Balkens aufgrund einer aufgebrachten Kraft oder eines Moments.

- $E$ — **Elastizitätsmodul (Young's Modulus)**: Ein Materialparameter, der die Steifigkeit eines Materials beschreibt. Je höher der Elastizitätsmodul, desto steifer ist das Material.

- $I$ — **Flächenträgheitsmoment**: Ein geometrischer Parameter, der die Widerstandsfähigkeit eines Balkens gegen Biegung angibt. Abhängig von der Querschnittsform des Balkens.

- $F$ — **Kraft**: Die auf den Balken wirkende äußere Last, die eine Durchbiegung verursacht.

- $R_B$ — **Auflagerreaktion**: Die vertikale Kraft, die durch das Auflager des Balkens bereitgestellt wird, um das Gleichgewicht zu gewährleisten.

- $L$ — **Länge des Balkens**: Die Gesamtlänge des Balkens, die in den Berechnungen der Durchbiegung und Auflagerkräfte berücksichtigt wird.

### System von linearen Gleichungen

Die drei Gleichungen lauten:

$$
R_A + R_B = F
$$

$$
M_A + R_B L - F \cdot \dfrac{L}{2} = 0
$$

$$
16 R_B - 5 F = 0
$$

### Matrixdarstellung

Schreiben wir das System in Matrixform:

$$
\begin{bmatrix}
1 & 1 & 0 \\
0 & L & 1 \\
0 & 16 & 0
\end{bmatrix}
\begin{bmatrix}
R_A \\
R_B \\
M_A
\end{bmatrix}
=
\begin{bmatrix}
F \\
\dfrac{F L}{2} \\
5 F
\end{bmatrix}
$$

Ersetzen der Konstanten ergibt:

$$
\begin{bmatrix}
1 & 1 & 0 \\
0 & 10 & 1 \\
0 & 16 & 0
\end{bmatrix}
\begin{bmatrix}
R_A \\
R_B \\
M_A
\end{bmatrix}
=
\begin{bmatrix}
15 \\
75 \\
75
\end{bmatrix}
$$

*Obwohl das System leicht durch Substitution ab der letzten Zeile gelöst werden kann, verwenden wir exemplarisch das Gaußeliminationsverfahren.*

### Das gegebene Gleichungssystem

Wir haben das lineare Gleichungssystem:

$$
\begin{bmatrix}
1 & 1 & 0 \\
0 & 10 & 1 \\
0 & 16 & 0
\end{bmatrix}
\begin{bmatrix}
R_A \\
R_B \\
M_A
\end{bmatrix}
=
\begin{bmatrix}
15 \\
75 \\
75
\end{bmatrix}
$$

und die erweiterte Koeffizientenmatrix:

$$
\left[
\begin{array}{ccc|c}
1 & 1 & 0 & 15 \\
0 & 10 & 1 & 75 \\
0 & 16 & 0 & 75
\end{array}
\right]
$$

### Erste Schritte des Eliminationsverfahrens

***Schritt 1: Identifikation des ersten Pivot-Elements***

Das erste Pivot-Element ist $a_{11} = 1$.

Da die Elemente unterhalb bereits Null sind, müssen wir keine Eliminierung in der ersten Spalte durchführen.

***Schritt 2: Eliminierung in der zweiten Spalte***

**Pivot-Element in Zeile 2:** $a_{22} = 10$

Wir eliminieren das Element unterhalb des Pivot-Elements in Spalte 2 ($a_{32} = 16$).

**Berechnung des Multiplikators:**

$$
m = \dfrac{a_{32}}{a_{22}}
= \dfrac{16}{10}
$$

### Erste Schritte des Eliminationsverfahrens

***Schritt 2: Zeilenoperation zur Eliminierung***

$$
\text{Zeile 3}
=
\text{Zeile 3}
-
m \cdot \text{Zeile 2}
$$

**Berechnung:**

$$
a_{31}'
=
a_{31}
-
m a_{21}
=
0
-
\frac{16}{10}\cdot0
=
0
$$

$$
a_{32}'
=
a_{32}
-
m a_{22}
=
16
-
\frac{16}{10}\cdot10
=
16-16
=
0
$$

$$
a_{33}'
=
a_{33}
-
m a_{23}
=
0
-
\frac{16}{10}\cdot1
=
-\dfrac{16}{10}
$$

$$
b_3'
=
b_3
-
m b_2
=
75
-
\frac{16}{10}\cdot75
=
75-120
=
-45
$$

Die neue dritte Zeile ist:

$$
\left[
0
\quad
0
\quad
-\frac{16}{10}
\quad | \quad
-45
\right]
$$

### Stufenform und Schritt 3: Lösen für $M_A$

Die Matrix nach der Eliminierung hat eine **Stufenform**. Das System kann nun von unten gelöst werden:

$$
\left[
\begin{array}{ccc|c}
1 & 1 & 0 & 15 \\
0 & 10 & 1 & 75 \\
0 & 0 & -\frac{16}{10} & -45
\end{array}
\right]
$$

Aus der dritten Gleichung:

$$
-\frac{16}{10}M_A=-45
\implies
M_A=\frac{450}{16}
=
28{,}125\,\text{kN}\cdot\text{m}
$$

### Schritt 4: Rückwärtseinsetzen zur Bestimmung von $R_B$

Aus der zweiten Gleichung:

$$
10R_B+M_A=75
$$

Einsetzen von $M_A=\frac{450}{16}$:

$$
10R_B+\frac{450}{16}=75
\implies
10R_B
=
75-\frac{450}{16}
=
\frac{375}{8}
$$

$$
R_B
=
\frac{375}{80}
=
4{,}6875\,\text{kN}
$$

### Schritt 5: Bestimmung von $R_A$

Aus der ersten Gleichung:

$$
R_A+R_B=15
$$

Einsetzen von $R_B=\frac{375}{80}$:

$$
R_A
=
15-\frac{375}{80}
=
\frac{825}{80}
=
10{,}3125\,\text{kN}
$$

```python
import sympy as sp
from IPython.display import display, Math

A = sp.Matrix([
    [1, 1, 0],
    [0, 10, 1],
    [0, 16, 0]
])

b = sp.Matrix([
    15,
    75,
    75
])

RA, RB, MA = A.LUsolve(b)

display(Math(rf"R_A = {sp.latex(RA)}\,\mathrm{{kN}}"))
display(Math(rf"R_B = {sp.latex(RB)}\,\mathrm{{kN}}"))
display(Math(rf"M_A = {sp.latex(MA)}\,\mathrm{{kN}}\cdot\mathrm{{m}}"))
```

### Lösung

- Die Lösung ist:
  - $R_A = 10{,}3125\,\text{kN}$
  - $R_B = 4{,}6875\,\text{kN}$
  - $M_A = 28{,}125\,\text{kN}\cdot\text{m}$

- Wir haben aus dem physikalischen Modell der Statik ein lineares Gleichungssystem formuliert.
- Das System ist lösbar und hat eine eindeutige Lösung.
- Es kann in Matrixform geschrieben werden und mit dem Gauß-Eliminationsverfahren gelöst werden.
