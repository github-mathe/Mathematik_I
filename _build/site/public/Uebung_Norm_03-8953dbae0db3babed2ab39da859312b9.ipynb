## Berechnung verschiedener Matrixnormen

Gegeben sei die Matrix

$$
\vec A=
\begin{pmatrix}
1 & 1\\
2 & 2
\end{pmatrix}.
$$

Berechnen Sie für die Matrix $\vec A$ die folgenden Matrixnormen:

1. die maximale Spaltensummennorm

   $$
   \|\vec A\|_1
   =
   \max_{1\leq j\leq n}
   \sum_{i=1}^{m}|a_{ij}|,
   $$

2. die maximale Zeilensummennorm

   $$
   \|\vec A\|_\infty
   =
   \max_{1\leq i\leq m}
   \sum_{j=1}^{n}|a_{ij}|,
   $$

3. die Spektralnorm

   $$
   \|\vec A\|_2,
   $$

   wobei die Spektralnorm der größten Singularzahl der Matrix entspricht,

4. die Frobenius-Norm

   $$
   \|\vec A\|_F
   =
   \sqrt{
   \sum_{i=1}^{m}
   \sum_{j=1}^{n}
   |a_{ij}|^2
   }.
   $$

Vergleichen Sie anschließend die erhaltenen Werte.

## Lösung

Die verschiedenen Matrixnormen messen die Größe einer Matrix auf unterschiedliche Weise.

Berechnen Sie zunächst jede Norm selbstständig und öffnen Sie anschließend die jeweilige Lösung zur Kontrolle.

:::{dropdown} Berechnung der maximalen Spaltensummennorm

Die maximale Spaltensummennorm ist definiert durch

$$
\|\vec A\|_1
=
\max_{1\leq j\leq n}
\sum_{i=1}^{m}|a_{ij}|.
$$

Für

$$
\vec A=
\begin{pmatrix}
1 & 1\\
2 & 2
\end{pmatrix}
$$

werden zunächst die Beträge der Einträge jeder Spalte addiert.

Für die erste Spalte gilt

$$
|1|+|2|=3.
$$

Für die zweite Spalte gilt

$$
|1|+|2|=3.
$$

Damit ergibt sich

$$
\|\vec A\|_1
=
\max\{3,3\}.
$$

Also

$$
\boxed{
\|\vec A\|_1=3
}
$$
:::

:::{dropdown} Berechnung der maximalen Zeilensummennorm

Die maximale Zeilensummennorm ist definiert durch

$$
\|\vec A\|_\infty
=
\max_{1\leq i\leq m}
\sum_{j=1}^{n}|a_{ij}|.
$$

Nun werden die Beträge der Einträge jeder Zeile addiert.

Für die erste Zeile gilt

$$
|1|+|1|=2.
$$

Für die zweite Zeile gilt

$$
|2|+|2|=4.
$$

Damit ergibt sich

$$
\|\vec A\|_\infty
=
\max\{2,4\}.
$$

Also

$$
\boxed{
\|\vec A\|_\infty=4
}
$$
:::

:::{dropdown} Berechnung der Spektralnorm

Die Spektralnorm

$$
\|\vec A\|_2
$$

entspricht der größten Singularzahl der Matrix.

Die Singularwerte erhält man aus den Eigenwerten der Matrix

$$
\vec A^T\vec A.
$$

Zunächst gilt

$$
\vec A^T
=
\begin{pmatrix}
1 & 2\\
1 & 2
\end{pmatrix}.
$$

Damit ergibt sich

$$
\vec A^T\vec A
=
\begin{pmatrix}
1 & 2\\
1 & 2
\end{pmatrix}
\begin{pmatrix}
1 & 1\\
2 & 2
\end{pmatrix}
=
\begin{pmatrix}
5 & 5\\
5 & 5
\end{pmatrix}.
$$

Die Eigenwerte dieser Matrix sind

$$
\lambda_1=10
\qquad\text{und}\qquad
\lambda_2=0.
$$

Die Singularwerte von $\vec A$ sind die Quadratwurzeln dieser Eigenwerte:

$$
\sigma_1=\sqrt{10},
\qquad
\sigma_2=0.
$$

Die größte Singularzahl ist daher

$$
\sqrt{10}.
$$

Somit gilt

$$
\boxed{
\|\vec A\|_2=\sqrt{10}
}
$$
:::

:::{dropdown} Berechnung der Frobenius-Norm

Die Frobenius-Norm ist definiert durch

$$
\|\vec A\|_F
=
\sqrt{
\sum_{i=1}^{m}
\sum_{j=1}^{n}
|a_{ij}|^2
}.
$$

Für

$$
\vec A=
\begin{pmatrix}
1 & 1\\
2 & 2
\end{pmatrix}
$$

ergibt sich

$$
\begin{aligned}
\|\vec A\|_F
&=
\sqrt{
1^2+1^2+2^2+2^2
}\\
&=
\sqrt{
1+1+4+4
}\\
&=
\sqrt{10}.
\end{aligned}
$$

Damit gilt

$$
\boxed{
\|\vec A\|_F=\sqrt{10}
}
$$
:::

:::{dropdown} Vergleich der Matrixnormen

Für die Matrix

$$
\vec A=
\begin{pmatrix}
1 & 1\\
2 & 2
\end{pmatrix}
$$

haben wir erhalten:

$$
\boxed{
\|\vec A\|_1=3
}
$$

$$
\boxed{
\|\vec A\|_\infty=4
}
$$

$$
\boxed{
\|\vec A\|_2=\sqrt{10}
}
$$

und

$$
\boxed{
\|\vec A\|_F=\sqrt{10}.
}
$$

Mit

$$
\sqrt{10}\approx3{,}16
$$

ergibt sich für diese Matrix

$$
\boxed{
\|\vec A\|_1
<
\|\vec A\|_2
=
\|\vec A\|_F
<
\|\vec A\|_\infty.
}
$$

Die verschiedenen Matrixnormen liefern im Allgemeinen unterschiedliche Werte, da sie unterschiedliche Eigenschaften der Matrix zur Bestimmung ihrer Größe verwenden.
:::

:::{dropdown} Gesamtlösung anzeigen

Für

$$
\vec A=
\begin{pmatrix}
1 & 1\\
2 & 2
\end{pmatrix}
$$

ergeben sich die Matrixnormen zu

$$
\boxed{
\|\vec A\|_1=3,
\qquad
\|\vec A\|_\infty=4,
\qquad
\|\vec A\|_2=\sqrt{10},
\qquad
\|\vec A\|_F=\sqrt{10}.
}
$$
:::
