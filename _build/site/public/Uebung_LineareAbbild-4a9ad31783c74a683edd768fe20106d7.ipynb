## Projektion auf eine Hyperebene

Gegeben sei die Hyperebene

$$
H
=
\left\{
\vec{x}\in\mathbb{R}^3
\;\middle|\;
\left\langle
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix},
\vec{x}
\right\rangle
=
0
\right\}
$$

mit dem Normalenvektor

$$
\vec n
=
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix}.
$$

Weiter sei

$$
\vec{x}
=
\begin{pmatrix}
3\\
0\\
0
\end{pmatrix}.
$$

1. Bestimmen Sie den Einheitsnormalenvektor $\vec n_0$.
2. Bestimmen Sie die Projektionsmatrix

$$
P_H
=
\mathbf I-\vec n_0\otimes\vec n_0.
$$

3. Berechnen Sie die Projektion $P_H\vec{x}$.
4. Überprüfen Sie, dass der projizierte Vektor in $H$ liegt.

## Lösung

Zur Projektion auf die Hyperebene wird zunächst der Normalenvektor normiert.

Anschließend wird die Projektionsmatrix

$$
P_H
=
\mathbf I-\vec n_0\otimes\vec n_0
$$

bestimmt.

:::{dropdown} Einheitsnormalenvektor bestimmen

Gegeben ist

$$
\vec n
=
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix}.
$$

Die Norm des Normalenvektors ist

$$
\|\vec n\|
=
\sqrt{1^2+2^2+2^2}
=
\sqrt{9}
=
3.
$$

Damit ergibt sich der Einheitsnormalenvektor zu

$$
\boxed{
\vec n_0
=
\frac{\vec n}{\|\vec n\|}
=
\frac13
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix}.
}
$$
:::

:::{dropdown} Projektionsmatrix bestimmen

Zunächst berechnen wir das Tensorprodukt

$$
\vec n_0\otimes\vec n_0.
$$

Mit

$$
\vec n_0
=
\frac13
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix}
$$

gilt

$$
\begin{aligned}
\vec n_0\otimes\vec n_0
&=
\frac13
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix}
\frac13
\begin{pmatrix}
1 & 2 & 2
\end{pmatrix}\\[1ex]
&=
\frac19
\begin{pmatrix}
1 & 2 & 2\\
2 & 4 & 4\\
2 & 4 & 4
\end{pmatrix}.
\end{aligned}
$$

Die Projektionsmatrix ist

$$
P_H
=
\mathbf I-\vec n_0\otimes\vec n_0.
$$

Damit ergibt sich

$$
\begin{aligned}
P_H
&=
\begin{pmatrix}
1 & 0 & 0\\
0 & 1 & 0\\
0 & 0 & 1
\end{pmatrix}
-
\frac19
\begin{pmatrix}
1 & 2 & 2\\
2 & 4 & 4\\
2 & 4 & 4
\end{pmatrix}\\[1ex]
&=
\frac19
\begin{pmatrix}
8 & -2 & -2\\
-2 & 5 & -4\\
-2 & -4 & 5
\end{pmatrix}.
\end{aligned}
$$

Somit ist

$$
\boxed{
P_H
=
\frac19
\begin{pmatrix}
8 & -2 & -2\\
-2 & 5 & -4\\
-2 & -4 & 5
\end{pmatrix}.
}
$$
:::

:::{dropdown} Projektion berechnen

Die Projektion von

$$
\vec{x}
=
\begin{pmatrix}
3\\
0\\
0
\end{pmatrix}
$$

auf die Hyperebene ist

$$
P_H\vec{x}.
$$

Damit gilt

$$
\begin{aligned}
P_H\vec{x}
&=
\frac19
\begin{pmatrix}
8 & -2 & -2\\
-2 & 5 & -4\\
-2 & -4 & 5
\end{pmatrix}
\begin{pmatrix}
3\\
0\\
0
\end{pmatrix}\\[1ex]
&=
\frac19
\begin{pmatrix}
24\\
-6\\
-6
\end{pmatrix}.
\end{aligned}
$$

Somit ergibt sich

$$
\boxed{
P_H\vec{x}
=
\begin{pmatrix}
\frac83\\[1ex]
-\frac23\\[1ex]
-\frac23
\end{pmatrix}.
}
$$
:::

:::{dropdown} Zugehörigkeit zur Hyperebene überprüfen

Ein Vektor liegt in der Hyperebene $H$, wenn

$$
\langle \vec n,\vec{x}\rangle=0
$$

gilt.

Für den projizierten Vektor erhalten wir

$$
\left\langle
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix},
\begin{pmatrix}
\frac83\\[1ex]
-\frac23\\[1ex]
-\frac23
\end{pmatrix}
\right\rangle
=
1\cdot\frac83
+
2\cdot\left(-\frac23\right)
+
2\cdot\left(-\frac23\right).
$$

Damit gilt

$$
\frac83-\frac43-\frac43=0.
$$

Also liegt der projizierte Vektor in der Hyperebene:

$$
\boxed{
P_H\vec{x}\in H.
}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Der Normalenvektor lautet

$$
\vec n
=
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix}
$$

mit

$$
\|\vec n\|=3.
$$

Damit ist

$$
\boxed{
\vec n_0
=
\frac13
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix}.
}
$$

Die Projektionsmatrix lautet

$$
\boxed{
P_H
=
\frac19
\begin{pmatrix}
8 & -2 & -2\\
-2 & 5 & -4\\
-2 & -4 & 5
\end{pmatrix}.
}
$$

Für

$$
\vec{x}
=
\begin{pmatrix}
3\\
0\\
0
\end{pmatrix}
$$

ergibt sich

$$
\boxed{
P_H\vec{x}
=
\begin{pmatrix}
\frac83\\[1ex]
-\frac23\\[1ex]
-\frac23
\end{pmatrix}.
}
$$

Da

$$
\left\langle
\vec n,
P_H\vec{x}
\right\rangle
=
0,
$$

liegt der projizierte Vektor in der Hyperebene $H$.
:::
