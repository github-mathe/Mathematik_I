## Projektion auf einen Unterraum

Gegeben sei der Unterraum

$$
U=\operatorname{span}\{\vec u_1,\vec u_2\}\subset\mathbb R^3
$$

mit der Orthonormalbasis

$$
\vec u_1
=
\frac{1}{\sqrt{2}}
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix},
\qquad
\vec u_2
=
\frac{1}{\sqrt{6}}
\begin{pmatrix}
1\\
-1\\
2
\end{pmatrix}.
$$

Weiter sei

$$
\vec x
=
\begin{pmatrix}
2\\
1\\
3
\end{pmatrix}.
$$

1. Stellen Sie die Matrix $\mathbf A$ aus den Basisvektoren auf.
2. Bestimmen Sie die Projektionsmatrix $\mathbf P_U$.
3. Berechnen Sie die orthogonale Projektion $P_U(\vec x)$.
4. Bestimmen Sie den Differenzvektor zwischen $\vec x$ und seiner Projektion.

## Lösung

Da $\vec u_1$ und $\vec u_2$ eine Orthonormalbasis von $U$ bilden, kann die Projektionsmatrix mit

$$
\mathbf P_U=\mathbf A\mathbf A^T
$$

bestimmt werden.

:::{dropdown} Matrix $\mathbf A$ aufstellen

Die Basisvektoren werden als Spalten in die Matrix $\mathbf A$ geschrieben:

$$
\mathbf A
=
\begin{pmatrix}
\vec u_1 & \vec u_2
\end{pmatrix}.
$$

Damit ergibt sich

$$
\boxed{
\mathbf A
=
\begin{pmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{6}}\\[1ex]
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{6}}\\[1ex]
0 & \frac{2}{\sqrt{6}}
\end{pmatrix}.
}
$$
:::

:::{dropdown} Projektionsmatrix bestimmen

Für eine Orthonormalbasis gilt

$$
\mathbf P_U
=
\mathbf A\mathbf A^T.
$$

Damit ist

$$
\mathbf P_U
=
\begin{pmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{6}}\\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{6}}\\
0 & \frac{2}{\sqrt{6}}
\end{pmatrix}
\begin{pmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} & 0\\
\frac{1}{\sqrt{6}} & -\frac{1}{\sqrt{6}} & \frac{2}{\sqrt{6}}
\end{pmatrix}.
$$

Durch die Matrixmultiplikation erhält man

$$
\boxed{
\mathbf P_U
=
\begin{pmatrix}
\frac{2}{3} & \frac{1}{3} & \frac{1}{3}\\[1ex]
\frac{1}{3} & \frac{2}{3} & -\frac{1}{3}\\[1ex]
\frac{1}{3} & -\frac{1}{3} & \frac{2}{3}
\end{pmatrix}.
}
$$
:::

:::{dropdown} Projektion berechnen

Die Projektion von $\vec x$ auf den Unterraum $U$ ist

$$
P_U(\vec x)
=
\mathbf P_U\vec x.
$$

Damit gilt

$$
\begin{aligned}
P_U(\vec x)
&=
\begin{pmatrix}
\frac{2}{3} & \frac{1}{3} & \frac{1}{3}\\
\frac{1}{3} & \frac{2}{3} & -\frac{1}{3}\\
\frac{1}{3} & -\frac{1}{3} & \frac{2}{3}
\end{pmatrix}
\begin{pmatrix}
2\\
1\\
3
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
\frac{8}{3}\\[1ex]
\frac{1}{3}\\[1ex]
\frac{7}{3}
\end{pmatrix}.
\end{aligned}
$$

Somit ist

$$
\boxed{
P_U(\vec x)
=
\begin{pmatrix}
\frac{8}{3}\\[1ex]
\frac{1}{3}\\[1ex]
\frac{7}{3}
\end{pmatrix}.
}
$$
:::

:::{dropdown} Aufgabe 4 – Differenzvektor bestimmen

Der Differenzvektor zwischen $\vec{x}$ und seiner Projektion auf den Unterraum $U$ ist

$$
\vec r
=
\vec{x}-P_U(\vec{x}).
$$

Mit

$$
\vec{x}
=
\begin{pmatrix}
2\\
1\\
3
\end{pmatrix}
$$

und

$$
P_U(\vec{x})
=
\begin{pmatrix}
\frac{8}{3}\\[1ex]
\frac{1}{3}\\[1ex]
\frac{7}{3}
\end{pmatrix}
$$

ergibt sich

$$
\begin{aligned}
\vec r
&=
\begin{pmatrix}
2\\
1\\
3
\end{pmatrix}
-
\begin{pmatrix}
\frac{8}{3}\\[1ex]
\frac{1}{3}\\[1ex]
\frac{7}{3}
\end{pmatrix}\\[1ex]
&=
\boxed{
\begin{pmatrix}
-\frac{2}{3}\\[1ex]
\frac{2}{3}\\[1ex]
\frac{2}{3}
\end{pmatrix}
}.
\end{aligned}
$$

Zur Überprüfung verwenden wir die beiden Basisvektoren des Unterraums

$$
\vec u_1
=
\frac{1}{\sqrt{2}}
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix},
\qquad
\vec u_2
=
\frac{1}{\sqrt{6}}
\begin{pmatrix}
1\\
-1\\
2
\end{pmatrix}.
$$

Für den ersten Basisvektor gilt

$$
\begin{aligned}
\langle \vec r,\vec u_1\rangle
&=
\begin{pmatrix}
-\frac{2}{3} &
\frac{2}{3} &
\frac{2}{3}
\end{pmatrix}
\frac{1}{\sqrt{2}}
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}\\[1ex]
&=
\frac{1}{\sqrt{2}}
\left(
-\frac{2}{3}
+
\frac{2}{3}
\right)\\
&=0.
\end{aligned}
$$

Für den zweiten Basisvektor gilt

$$
\begin{aligned}
\langle \vec r,\vec u_2\rangle
&=
\begin{pmatrix}
-\frac{2}{3} &
\frac{2}{3} &
\frac{2}{3}
\end{pmatrix}
\frac{1}{\sqrt{6}}
\begin{pmatrix}
1\\
-1\\
2
\end{pmatrix}\\[1ex]
&=
\frac{1}{\sqrt{6}}
\left(
-\frac{2}{3}
-\frac{2}{3}
+\frac{4}{3}
\right)\\
&=0.
\end{aligned}
$$

Der Differenzvektor ist somit orthogonal zu beiden Basisvektoren und damit zum gesamten Unterraum $U$:

$$
\boxed{\vec r\perp U.}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Der Normalenvektor lautet

$$
\vec n
=
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix}
$$

mit

$$
\|\vec n\|=3.
$$

Damit ist

$$
\boxed{
\vec n_0
=
\frac13
\begin{pmatrix}
1\\
2\\
2
\end{pmatrix}.
}
$$

Die Projektionsmatrix lautet

$$
\boxed{
P_H
=
\frac19
\begin{pmatrix}
8 & -2 & -2\\
-2 & 5 & -4\\
-2 & -4 & 5
\end{pmatrix}.
}
$$

Für

$$
\vec{x}
=
\begin{pmatrix}
3\\
0\\
0
\end{pmatrix}
$$

ergibt sich

$$
\boxed{
P_H\vec{x}
=
\begin{pmatrix}
\frac83\\[1ex]
-\frac23\\[1ex]
-\frac23
\end{pmatrix}.
}
$$

Da

$$
\left\langle
\vec n,
P_H\vec{x}
\right\rangle
=
0,
$$

liegt der projizierte Vektor in der Hyperebene $H$.
:::
