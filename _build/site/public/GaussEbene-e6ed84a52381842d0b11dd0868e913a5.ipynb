## Gauss-Ebene

### Die komplexe Zahlenebene (Gauss-Ebene)

- Die komplexe Ebene, auch bekannt als **Gauss-Ebene**, ist eine Möglichkeit, komplexe Zahlen grafisch darzustellen.
- Jede komplexe Zahl $z = a + bi$ kann als Punkt in der Ebene dargestellt werden, wobei:
  - $a$ der **reelle Teil** ist (x-Achse),
  - $b$ der **imaginäre Teil** ist (y-Achse).
- Der Punkt $(a,b)$ stellt die komplexe Zahl $z$ dar.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 5))

# Achsen
ax.annotate("", xy=(3, 0), xytext=(-2.5, 0),
            arrowprops=dict(arrowstyle="->", linewidth=1.8))
ax.annotate("", xy=(0, 3), xytext=(0, -1.5),
            arrowprops=dict(arrowstyle="->", linewidth=1.8))

# Punkt z = 2 + 1.5i
a, b = 2, 1.5
ax.plot(a, b, "ko")
ax.plot([a, a], [0, b], linestyle="--")
ax.plot([0, a], [b, b], linestyle="--")

# Beschriftungen
ax.text(3.05, -0.1, "Re (reelle Achse)", fontsize=11)
ax.text(0.1, 3.05, "Im (imaginäre Achse)", fontsize=11)
ax.text(a + 0.1, b + 0.1, r"$z = 2 + 1.5i$", fontsize=12)
ax.text(a, -0.2, "2", ha="center", fontsize=11)
ax.text(-0.15, b, "1.5", va="center", ha="right", fontsize=11)

ax.set_xlim(-2.5, 3.4)
ax.set_ylim(-1.5, 3.4)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

### Modul einer komplexen Zahl

Der **Modul** (oder der Betrag) einer komplexen Zahl $z = a + bi$ ist definiert als:

$$
|z| = \sqrt{a^2 + b^2}
$$

- Der Modul gibt die Entfernung der Zahl $z$ vom Ursprung der Gauss-Ebene an.

**Beispiele:**

- $z = 3 + 4i$:

$$
|z| = \sqrt{3^2 + 4^2} = 5
$$

- $z = -1 + i$:

$$
|z| = \sqrt{(-1)^2 + 1^2} = \sqrt{2}
$$

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 6))

# Achsen
ax.annotate("", xy=(4.5, 0), xytext=(-1.5, 0),
            arrowprops=dict(arrowstyle="->", linewidth=1.8))
ax.annotate("", xy=(0, 5), xytext=(0, -1),
            arrowprops=dict(arrowstyle="->", linewidth=1.8))

# Punkt z = 3 + 4i
x, y = 3, 4
ax.plot(x, y, "ko")
ax.plot([x, x], [0, y], linestyle="--")
ax.plot([0, x], [y, y], linestyle="--")

# Strecke vom Ursprung zum Punkt
ax.plot([0, x], [0, y], linewidth=2)

# Beschriftungen
ax.text(4.6, -0.1, "Re", fontsize=11)
ax.text(0.1, 5.05, "Im", fontsize=11)
ax.text(x + 0.1, y + 0.1, r"$z = 3 + 4i$", fontsize=12)
ax.text(x, -0.2, "3", ha="center", fontsize=11)
ax.text(-0.15, y, "4", va="center", ha="right", fontsize=11)

# Beschriftung |z|
ax.text(1.7, 1.8, r"$|z|$", color="blue", fontsize=13)

ax.set_xlim(-1.5, 4.8)
ax.set_ylim(-1, 5.3)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```
