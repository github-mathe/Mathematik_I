## Dimensionstheorem für Unterräume

### Das Dimensionstheorem und seine Implikationen

**Dimensionstheorem:**

Für einen Unterraum $U\subseteq\mathbb{R}^n$ und dessen orthogonales Komplement $U^\perp$ gilt:

$$
\dim(U)+\dim(U^\perp)=n.
$$

**Wichtige Fragen:**

- Welche Konsequenzen hat dies für die Struktur von $\mathbb{R}^n$?
- Wie hilft das Dimensionstheorem bei der Analyse von Unterräumen?

### Zerlegung von $\mathbb{R}^n$

**Hauptaussage:**

Jeder Vektor $\mathbf{v}\in\mathbb{R}^n$ kann eindeutig in einen Teil in $U$ und einen Teil in $U^\perp$ zerlegt werden:

$$
\mathbf{v}
=
\mathbf{u}
+
\mathbf{w},
\qquad
\mathbf{u}\in U,
\quad
\mathbf{w}\in U^\perp.
$$

**Konsequenzen:**

- Der Raum $\mathbb{R}^n$ ist die direkte Summe der beiden Unterräume:

$$
\mathbb{R}^n
=
U\oplus U^\perp.
$$

- Projektionen: Die Projektion eines Vektors auf $U$ und $U^\perp$ ist wohldefiniert.

### Erweiterung der Basis

**Hauptaussage:**

Eine Basis für $U\subset V$ kann erweitert werden, um eine Basis für $V$ zu bilden.

- Gegeben eine Basis

$$
\{\mathbf{u}_1,\ldots,\mathbf{u}_k\}
$$

von $U$, können wir

$$
\{\mathbf{v}_1,\ldots,\mathbf{v}_{n-k}\}
$$

für $U^\perp$ finden.

- Die Vereinigung

$$
\{
\mathbf{u}_1,\ldots,\mathbf{u}_k,
\mathbf{v}_1,\ldots,\mathbf{v}_{n-k}
\}
$$

ist dann eine Basis für $V$.
