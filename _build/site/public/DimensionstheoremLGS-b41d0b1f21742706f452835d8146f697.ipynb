## Dimensionstheorems für lineare Gleichungssysteme

Für eine Matrix

$$
A\in\mathbb{R}^{m\times n}
$$

gilt:

$$
\dim(\operatorname{Kern}(A))
+
\dim(\operatorname{Bild}(A))
=
n.
$$

Dabei ist

- $\operatorname{Kern}(A)$: der Lösungsraum des homogenen Systems

$$
A\mathbf{x}=\mathbf{0},
$$

- $\operatorname{Bild}(A)$: der Raum der durch $A$ erreichbaren Vektoren (Spaltenraum).

### Implikationen für lineare Systeme

1. **Homogenes System**

   Das homogene System

   $$
   A\mathbf{x}=\mathbf{0}
   $$

   hat Lösungen, wenn

   $$
   \dim(\operatorname{Kern}(A))>0.
   $$

   Die Anzahl der Freiheitsgrade der Lösung entspricht

   $$
   \dim(\operatorname{Kern}(A)).
   $$

2. **Inhomogenes System**

   Das inhomogene System

   $$
   A\mathbf{x}=\mathbf{b}
   $$

   ist lösbar, wenn

   $$
   \mathbf{b}\in\operatorname{Bild}(A).
   $$

   Die Lösungsmenge hat dieselbe Dimension wie $\operatorname{Kern}(A)$.

   Wenn

   $$
   \operatorname{Rang}(A)=n,
   $$

   gibt es genau eine Lösung; andernfalls ist die Lösungsmenge eine affine Untermenge mit der Dimension

   $$
   \dim(\operatorname{Kern}(A)).
   $$

3. **Volle Rang-Bedingung**

   Wenn

   $$
   \operatorname{Rang}(A)=n,
   $$

   ist

   $$
   \dim(\operatorname{Kern}(A))=0,
   $$

   d.h. das homogene System hat nur die triviale Lösung.

### Geometrische Interpretation

**Homogenes System:**

- Die Lösungen von

$$
A\mathbf{x}=\mathbf{0}
$$

bilden einen Unterraum, $\operatorname{Kern}(A)$, im $\mathbb{R}^n$.

- Dieser Unterraum ist orthogonal zum Raum $\operatorname{Bild}(A^\top)$, der von den Zeilen der Matrix $\mathbf{A}$ aufgespannt wird:

$$
\mathbb{R}^n
=
\operatorname{Kern}(A)
\oplus
\operatorname{Bild}(A^\top).
$$

**Inhomogenes System:**

- Für

$$
A\mathbf{x}=\mathbf{b}
$$

ist

$$
\mathbf{b}\in\operatorname{Bild}(A)
$$

eine notwendige Bedingung für die Lösbarkeit.

- Jede Lösung des inhomogenen Systems hat die Form

$$
\mathbf{x}
=
\mathbf{x}_p
+
\mathbf{x}_h,
$$

wobei $\mathbf{x}_p$ eine spezielle Lösung ist und
$\mathbf{x}_h\in\operatorname{Kern}(A)$ eine Lösung des homogenen Systems.

### Beispiel: Lösung durch Gauß-Elimination

**Gegeben:**

$$
A=
\begin{bmatrix}
1 & 2 & -1\\
2 & 4 & -2\\
1 & 0 & 1
\end{bmatrix},
\qquad
\mathbf{x}
=
\begin{bmatrix}
x_1\\
x_2\\
x_3
\end{bmatrix},
\qquad
\mathbf{b}
=
\begin{bmatrix}
1\\
2\\
1
\end{bmatrix}.
$$

Wir lösen das inhomogene System

$$
A\mathbf{x}=\mathbf{b}
$$

mit Gauß-Elimination:

$$
\begin{array}{ccc|c}
1 & 2 & -1 & 1\\
2 & 4 & -2 & 2\\
1 & 0 & 1 & 1
\end{array}
$$

$$
\begin{array}{ccc|c|l}
1 & 2 & -1 & 1 &\\
0 & 0 & 0 & 0 & \text{II'}=\text{II}-2\text{I}\\
0 & -2 & 2 & 0 & \text{III'}=\text{III}-\text{I}
\end{array}.
$$

### Lösung des Gleichungssystems

Aus der Zeilenstufenform lesen wir:

$$
\begin{aligned}
x_1+2x_2-x_3 &= 1,\\
-2x_2+2x_3 &= 0.
\end{aligned}
$$

**Parameterisierung:**

Setze

$$
x_3=t,
\qquad
t\in\mathbb{R}.
$$

Aus der zweiten Gleichung folgt

$$
x_2=x_3=t.
$$

Einsetzen in die erste Gleichung:

$$
x_1+2t-t=1
\quad\Rightarrow\quad
x_1=1-t.
$$

**Lösungsmenge:**

$$
\mathbf{x}
=
\begin{bmatrix}
1-t\\
t\\
t
\end{bmatrix}
=
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix}
+
t
\begin{bmatrix}
-1\\
1\\
1
\end{bmatrix},
\qquad
t\in\mathbb{R}.
$$

### Interpretation der Lösung

**Partikuläre Lösung:**

$$
\mathbf{x}_p
=
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix},
$$

so dass

$$
A\mathbf{x}_p=\mathbf{b}
$$

gilt.

**Kern von $A$:**

Die allgemeine Lösung des homogenen Systems

$$
A\mathbf{x}=\mathbf{0}
$$

ist

$$
\mathbf{x}_h
=
t
\begin{bmatrix}
-1\\
1\\
1
\end{bmatrix},
\qquad
t\in\mathbb{R},
$$

also

$$
\ker(A)
=
\operatorname{span}
\left\{
\begin{bmatrix}
-1\\
1\\
1
\end{bmatrix}
\right\}.
$$

**Gesamte Lösungsmenge:**

$$
\mathbf{x}
=
\mathbf{x}_p+\mathbf{x}_h
=
\mathbf{x}_p
+
t
\begin{bmatrix}
-1\\
1\\
1
\end{bmatrix},
\qquad
t\in\mathbb{R}.
$$

**Geometrische Interpretation:**

- $A$ ist eine quadratische Matrix mit nichttrivialem Kern (eine Gerade in $\mathbb{R}^3$).
- Die Lösungsmenge von $A\mathbf{x}=\mathbf{b}$ ist eine zu $\ker(A)$ parallele Gerade, die durch die partikuläre Lösung $\mathbf{x}_p$ verläuft.

```python
# Ergänzung: Lösung des linearen Gleichungssystems mit SymPy

import sympy as sp

A = sp.Matrix([
    [1, 2, -1],
    [2, 4, -2],
    [1, 0,  1]
])

b = sp.Matrix([1, 2, 1])

x1, x2, x3 = sp.symbols("x1 x2 x3")

loesung = sp.linsolve((A, b), (x1, x2, x3))

print("Lösungsmenge:")
sp.pprint(loesung)

print("\nKern von A:")
for v in A.nullspace():
    sp.pprint(v)

print("\nRang von A:", A.rank())
print("Dimension des Kerns:", len(A.nullspace()))
```
