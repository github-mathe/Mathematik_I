# Willkommen zu Mathematik I

Herzlich willkommen beim **interaktiven Vorlesungsskript Mathematik I**.

Diese Webseite begleitet Sie durch die Vorlesung und soll Ihnen helfen, die mathematischen Grundlagen schrittweise zu verstehen und sicher anzuwenden. Neben den theoretischen Inhalten finden Sie hier **anschauliche Beispiele**, **Übungsaufgaben**, **interaktive Elemente** und weiterführende **Literaturhinweise**.

Das Skript ist so aufgebaut, dass Sie die Inhalte sowohl zur Vor- und Nachbereitung der Vorlesung als auch zum selbstständigen Üben nutzen können. Zentrale Begriffe, Methoden und Zusammenhänge werden erläutert und durch Beispiele ergänzt.

Nutzen Sie die Möglichkeit, Rechnungen direkt nachzuvollziehen, eigene Lösungswege auszuprobieren und mathematische Zusammenhänge aktiv zu erkunden.

Ich wünsche Ihnen viel Erfolg und Freude beim Arbeiten mit dem interaktiven Vorlesungsskript!

## 💡 Motivation

Mathematik ist weit mehr als das Anwenden von Formeln. Sie vermittelt eine **Denkweise**, die für Ingenieurinnen und Ingenieure von zentraler Bedeutung ist.

Im Verlauf dieser Vorlesung geht es daher nicht nur darum, mathematische Methoden kennenzulernen, sondern insbesondere darum,

- **Probleme systematisch zu analysieren** und in überschaubare Teilprobleme zu zerlegen,
- **abstrakt zu denken**, Muster zu erkennen und Zusammenhänge zu verstehen,
- **kritisch zu denken** und Lösungsansätze zu bewerten,
- reale technische Fragestellungen mithilfe **mathematischer Modelle** zu beschreiben und zu untersuchen,
- sowie **Ausdauer, Kreativität und interdisziplinäres Denken** zu entwickeln.

Mathematik liefert damit grundlegende Werkzeuge, um komplexe technische Herausforderungen zu verstehen, Lösungen zu entwickeln und diese kritisch zu überprüfen.

> **Mathematik ist nicht nur Rechnen – sie ist eine Art, Probleme zu betrachten, zu verstehen und zu lösen.**
