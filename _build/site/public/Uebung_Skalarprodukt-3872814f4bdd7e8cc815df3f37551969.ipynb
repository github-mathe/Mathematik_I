## Komplexes Skalarprodukt

Gegeben seien die Vektoren

$$
\vec{x}
=
\begin{pmatrix}
1+i\\
1-i
\end{pmatrix},
\qquad
\vec{y}
=
\begin{pmatrix}
1\\
i
\end{pmatrix}
\in\mathbb{C}^2.
$$

Bearbeiten Sie die folgenden Aufgaben:

1. Berechnen Sie das komplexe Skalarprodukt

   $$
   \langle\vec{x},\vec{y}\rangle
   =
   \vec{x}^{\,*}\vec{y}.
   $$

2. Bestimmen Sie die Normen der beiden Vektoren mithilfe des Skalarprodukts:

   $$
   \|\vec{x}\|
   =
   \sqrt{\langle\vec{x},\vec{x}\rangle},
   \qquad
   \|\vec{y}\|
   =
   \sqrt{\langle\vec{y},\vec{y}\rangle}.
   $$

3. Überprüfen Sie, ob die beiden Vektoren orthogonal zueinander sind.

Verwenden Sie bei der Multiplikation komplexer Zahlen die Rechenregel

$$
(a+bi)(c+di)
=
(ac-bd)+(ad+bc)i.
$$

## Lösung

Beim komplexen Skalarprodukt wird der erste Vektor konjugiert-transponiert:

$$
\langle\vec{x},\vec{y}\rangle
=
\vec{x}^{\,*}\vec{y}.
$$

Für die Multiplikation komplexer Zahlen verwenden wir

$$
(a+bi)(c+di)
=
(ac-bd)+(ad+bc)i.
$$

:::{dropdown} Bestimmung von $\vec{x}^{\,*}$

Gegeben ist

$$
\vec{x}
=
\begin{pmatrix}
1+i\\
1-i
\end{pmatrix}.
$$

Die komplex konjugierten Einträge sind

$$
\overline{1+i}=1-i,
\qquad
\overline{1-i}=1+i.
$$

Damit ergibt sich der konjugiert-transponierte Vektor zu

$$
\boxed{
\vec{x}^{\,*}
=
\begin{pmatrix}
1-i & 1+i
\end{pmatrix}.
}
$$
:::

:::{dropdown} Berechnung des Skalarprodukts

Das komplexe Skalarprodukt lautet

$$
\langle\vec{x},\vec{y}\rangle
=
\vec{x}^{\,*}\vec{y}.
$$

Damit gilt

$$
\langle\vec{x},\vec{y}\rangle
=
\begin{pmatrix}
1-i & 1+i
\end{pmatrix}
\begin{pmatrix}
1\\
i
\end{pmatrix}.
$$

Also

$$
\langle\vec{x},\vec{y}\rangle
=
(1-i)\cdot1
+
(1+i)\cdot i.
$$

Für den ersten Term verwenden wir

$$
(1-i)\cdot1
=
(1+(-1)i)(1+0i).
$$

Mit

$$
(a+bi)(c+di)
=
(ac-bd)+(ad+bc)i
$$

ergibt sich

$$
(1-i)\cdot1
=
(1\cdot1-(-1)\cdot0)
+
(1\cdot0+(-1)\cdot1)i.
$$

Somit ist

$$
(1-i)\cdot1
=
1-i.
$$

Für den zweiten Term gilt

$$
(1+i)\cdot i
=
(1+1i)(0+1i).
$$

Damit folgt

$$
(1+i)\cdot i
=
(1\cdot0-1\cdot1)
+
(1\cdot1+1\cdot0)i.
$$

Also

$$
(1+i)\cdot i
=
-1+i.
$$

Somit ergibt sich insgesamt

$$
\begin{aligned}
\langle\vec{x},\vec{y}\rangle
&=
(1-i)+(-1+i)\\
&=
0.
\end{aligned}
$$

Damit gilt

$$
\boxed{
\langle\vec{x},\vec{y}\rangle=0.
}
$$
:::

:::{dropdown} Berechnung der Norm von $\vec{y}$

Für

$$
\vec{y}
=
\begin{pmatrix}
1\\
i
\end{pmatrix}
$$

ist

$$
\vec{y}^{\,*}
=
\begin{pmatrix}
1 & -i
\end{pmatrix}.
$$

Damit gilt

$$
\langle\vec{y},\vec{y}\rangle
=
1\cdot1
+
(-i)\cdot i.
$$

Für den zweiten Term verwenden wir

$$
(-i)\cdot i
=
(0-i)(0+i).
$$

Mit der Rechenregel ergibt sich

$$
\begin{aligned}
(-i)\cdot i
&=
(0\cdot0-(-1)\cdot1)
+
(0\cdot1+(-1)\cdot0)i\\
&=
1.
\end{aligned}
$$

Damit folgt

$$
\langle\vec{y},\vec{y}\rangle
=
1+1
=
2.
$$

Somit gilt

$$
\boxed{
\|\vec{y}\|=\sqrt{2}.
}
$$
:::

:::{dropdown} Überprüfung der Orthogonalität

Zwei komplexe Vektoren sind orthogonal, wenn

$$
\langle\vec{x},\vec{y}\rangle=0.
$$

Wir haben erhalten

$$
\langle\vec{x},\vec{y}\rangle=0.
$$

Damit gilt

$$
\boxed{
\vec{x}\perp\vec{y}.
}
$$
:::
