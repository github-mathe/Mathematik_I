## Einheitswurzeln

### Beispiel: Einheitswurzeln

Wir betrachten die Gleichung:

$$
z^6=1.
$$

Ziel ist es, alle komplexen Lösungen dieser Gleichung zu finden. Diese Lösungen nennt man die $6$-ten Einheitswurzeln.

### Darstellung der Zahl $1$ in Polarform

Die komplexe Zahl $1$ kann in Polarform dargestellt werden als:

$$
1
=
\cos(0)+i\sin(0)
=
e^{i0}.
$$

*Bemerkung:*

Aufgrund der Periodizität der trigonometrischen Funktionen können wir auch schreiben:

$$
1=e^{i2\pi k},
\qquad
k\in\mathbb Z.
$$

### Allgemeine Lösung der Gleichung $z^n=1$

Für die Gleichung

$$
z^n=1
$$

sind die Lösungen die $n$-ten Einheitswurzeln und gegeben durch:

$$
z_k
=
e^{i\left(\frac{0+2\pi k}{n}\right)},
\qquad
k=0,1,\dots,n-1.
$$

### Auflistung der Lösungen

Die sechs Lösungen sind:

$$
\begin{aligned}
z_0
&=
e^{i0}
=
1
\\[0.5em]
z_1
&=
e^{i\frac{\pi}{3}}
=
\cos\left(\frac{\pi}{3}\right)
+
i\sin\left(\frac{\pi}{3}\right)
\\[0.5em]
z_2
&=
e^{i\frac{2\pi}{3}}
=
\cos\left(\frac{2\pi}{3}\right)
+
i\sin\left(\frac{2\pi}{3}\right)
\\[0.5em]
z_3
&=
e^{i\pi}
=
\cos(\pi)+i\sin(\pi)
=
-1
\\[0.5em]
z_4
&=
e^{i\frac{4\pi}{3}}
=
\cos\left(\frac{4\pi}{3}\right)
+
i\sin\left(\frac{4\pi}{3}\right)
\\[0.5em]
z_5
&=
e^{i\frac{5\pi}{3}}
=
\cos\left(\frac{5\pi}{3}\right)
+
i\sin\left(\frac{5\pi}{3}\right).
\end{aligned}
$$

### Graphische Darstellung der Lösungen

Die Lösungen liegen gleichmäßig verteilt auf dem Einheitskreis in der komplexen Ebene.

```python
import numpy as np
import matplotlib.pyplot as plt

n = 6
k = np.arange(n)

theta = 2 * np.pi * k / n

x = np.cos(theta)
y = np.sin(theta)

fig, ax = plt.subplots(figsize=(7, 7))

# Einheitskreis
t = np.linspace(0, 2*np.pi, 500)
ax.plot(np.cos(t), np.sin(t))

# Achsen
ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

# Einheitswurzeln
ax.scatter(x, y)

for j in range(n):
    ax.plot([0, x[j]], [0, y[j]], linestyle="--")
    ax.text(1.12*x[j], 1.12*y[j], rf"$z_{j}$",
            ha="center", va="center")

ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect("equal")
ax.grid(alpha=0.3)

plt.show()
```

### Beispiel einer komplexen Gleichung

Gegeben ist die komplexe Zahl $w$ in kartesischer Form:

$$
w
=
4+4\sqrt{3}\,i.
$$

Zu lösen ist die Gleichung:

$$
z^3=w.
$$

### Umwandlung von $w$ in Polarform

**Berechnung des Betrags von $w$:**

$$
|w|
=
\sqrt{
(\operatorname{Re}(w))^2
+
(\operatorname{Im}(w))^2
}
$$

$$
=
\sqrt{
4^2+(4\sqrt{3})^2
}
=
8.
$$

**Berechnung des Arguments von $w$:**

$$
\varphi
=
\arctan\left(
\frac{\operatorname{Im}(w)}
{\operatorname{Re}(w)}
\right)
$$

$$
=
\arctan\left(
\frac{4\sqrt{3}}{4}
\right)
=
\arctan(\sqrt{3})
=
\frac{\pi}{3}.
$$

Da sowohl der Realteil $(4>0)$ als auch der Imaginärteil $(4\sqrt{3}>0)$ positiv sind, liegt $w$ im **ersten Quadranten**.

Daher ist das Argument tatsächlich

$$
\varphi=\frac{\pi}{3}
$$

und nicht ein anderer Winkel wie

$$
-\frac{5\pi}{3}.
$$

**Darstellung von $w$ in Polarform:**

$$
w
=
|w|(\cos\varphi+i\sin\varphi)
$$

$$
=
8\left(
\cos\frac{\pi}{3}
+
i\sin\frac{\pi}{3}
\right)
$$

$$
=
8e^{i\frac{\pi}{3}}.
$$

### Berechnung der Lösungen

**Betrag:**

$$
r
=
\sqrt[3]{8}
=
2.
$$

**Argument:**

$$
\begin{aligned}
\theta_k
&=
\frac{\varphi+2\pi k}{n}
\\[0.5em]
&=
\frac{\frac{\pi}{3}+2\pi k}{3}
\\[0.5em]
&=
\frac{\pi+6\pi k}{9}
\\[0.5em]
&=
\frac{\pi(1+6k)}{9}.
\end{aligned}
$$

für

$$
k=0,1,2.
$$

### Die drei Lösungen

Unter Verwendung der berechneten Winkel erhalten wir:

$$
\begin{aligned}
z_0
&=
2\left(
\cos\left(\frac{\pi}{9}\right)
+
i\sin\left(\frac{\pi}{9}\right)
\right)
\\[1em]
z_1
&=
2\left(
\cos\left(\frac{7\pi}{9}\right)
+
i\sin\left(\frac{7\pi}{9}\right)
\right)
\\[1em]
z_2
&=
2\left(
\cos\left(\frac{13\pi}{9}\right)
+
i\sin\left(\frac{13\pi}{9}\right)
\right).
\end{aligned}
$$

### Graphische Darstellung der Lösungen

| $k$ | $z_k=2e^{i\theta_k}$ |
|---:|---|
| $0$ | $z_0=2e^{i\frac{\pi}{9}}$ |
| $1$ | $z_1=2e^{i\frac{7\pi}{9}}$ |
| $2$ | $z_2=2e^{i\frac{13\pi}{9}}$ |

```python
import numpy as np
import matplotlib.pyplot as plt

k = np.arange(3)
theta = np.pi * (1 + 6*k) / 9

r = 2

x = r * np.cos(theta)
y = r * np.sin(theta)

fig, ax = plt.subplots(figsize=(7, 7))

# Kreis mit Radius 2
t = np.linspace(0, 2*np.pi, 500)
ax.plot(r*np.cos(t), r*np.sin(t))

# Achsen
ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

# Lösungen
ax.scatter(x, y)

labels = [
    r"$\theta_0=\frac{\pi}{9}$",
    r"$\theta_1=\frac{7\pi}{9}$",
    r"$\theta_2=\frac{13\pi}{9}$"
]

for j in range(3):
    ax.plot([0, x[j]], [0, y[j]], linestyle="--")
    ax.text(
        1.12*x[j],
        1.12*y[j],
        labels[j],
        ha="center",
        va="center"
    )

ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_xlim(-2.6, 2.6)
ax.set_ylim(-2.6, 2.6)
ax.set_aspect("equal")
ax.grid(alpha=0.3)

plt.show()
```

### Zusammenfassung der Lösungsmethode

Wir haben gezeigt, wie man komplexe Gleichungen der Form

$$
z^n=w
$$

löst, indem man:

- $w$ in Polarform schreibt.
- Den Betrag der Lösungen berechnet:

$$
r=|w|^{1/n}.
$$

- Die Argumente der Lösungen bestimmt:

$$
\theta_k
=
\frac{\varphi+2\pi k}{n}.
$$

- Die Lösungen in der Form

$$
z_k=re^{i\theta_k}
$$

ausdrückt.

- Die Lösungen graphisch auf dem komplexen Kreis mit Radius $r$ darstellt.

Diese Methode ermöglicht es, alle $n$ Wurzeln einer komplexen Zahl zu finden und zu visualisieren.

### $n$-te Wurzeln einer komplexen Zahl

- Die Gleichung

$$
z^n=w
$$

hat $n$ Lösungen in $\mathbb C$.

- Die Lösungen liegen auf einem Kreis mit Radius

$$
r=|w|^{1/n}.
$$

- Der erste Winkel ist

$$
\theta_0
=
\frac{\varphi}{n},
$$

wobei

$$
\varphi=\arg(w).
$$

- Die Winkel $\theta_k$ sind gleichmäßig um den Kreis verteilt:

$$
\theta_k
=
\theta_0+\frac{2\pi k}{n},
\qquad
k=0,1,\dots,n-1.
$$

- Der Winkel zwischen aufeinanderfolgenden Lösungen ist

$$
\frac{2\pi}{n}.
$$

- Die allgemeinen Lösungen sind gegeben durch:

$$
z_k
=
|w|^{1/n}e^{i\theta_k}.
$$
