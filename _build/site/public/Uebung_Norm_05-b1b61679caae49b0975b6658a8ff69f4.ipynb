## Matrixnormen mit komplexen Zahlen

Gegeben sei die komplexe Matrix

$$
\vec A=
\begin{pmatrix}
3+4i & 1-i\\
2i & -2+2i
\end{pmatrix}.
$$

Berechnen Sie für die Matrix $\vec A$

1. die maximale Spaltensummennorm

   $$
   \|\vec A\|_1
   =
   \max_{1\leq j\leq n}
   \sum_{i=1}^{m}|a_{ij}|,
   $$

2. die maximale Zeilensummennorm

   $$
   \|\vec A\|_\infty
   =
   \max_{1\leq i\leq m}
   \sum_{j=1}^{n}|a_{ij}|,
   $$

3. die Frobenius-Norm

   $$
   \|\vec A\|_F
   =
   \sqrt{
   \sum_{i=1}^{m}
   \sum_{j=1}^{n}
   |a_{ij}|^2
   }.
   $$

Bestimmen Sie zunächst die Beträge der komplexen Matrixeinträge.

## Lösung

Bei komplexen Matrixeinträgen werden für die Berechnung der Matrixnormen die Beträge der komplexen Zahlen verwendet.

Für eine komplexe Zahl

$$
z=a+bi
$$

gilt

$$
|z|=\sqrt{a^2+b^2}.
$$

:::{dropdown} Beträge der komplexen Matrixeinträge

Gegeben ist

$$
\vec A=
\begin{pmatrix}
3+4i & 1-i\\
2i & -2+2i
\end{pmatrix}.
$$

Zunächst bestimmen wir die Beträge der einzelnen Einträge.

Für den ersten Eintrag gilt

$$
|3+4i|
=
\sqrt{3^2+4^2}
=
5.
$$

Für den zweiten Eintrag gilt

$$
|1-i|
=
\sqrt{1^2+(-1)^2}
=
\sqrt{2}.
$$

Für den dritten Eintrag gilt

$$
|2i|
=
\sqrt{0^2+2^2}
=
2.
$$

Für den vierten Eintrag gilt

$$
|-2+2i|
=
\sqrt{(-2)^2+2^2}
=
\sqrt{8}
=
2\sqrt{2}.
$$

Damit erhalten wir für die Beträge

$$
\boxed{
\begin{pmatrix}
5 & \sqrt{2}\\
2 & 2\sqrt{2}
\end{pmatrix}
}
$$
:::

:::{dropdown} Berechnung der maximalen Spaltensummennorm

Die maximale Spaltensummennorm ist definiert durch

$$
\|\vec A\|_1
=
\max_{1\leq j\leq n}
\sum_{i=1}^{m}|a_{ij}|.
$$

Für die erste Spalte ergibt sich

$$
|3+4i|+|2i|
=
5+2
=
7.
$$

Für die zweite Spalte ergibt sich

$$
|1-i|+|-2+2i|
=
\sqrt{2}+2\sqrt{2}
=
3\sqrt{2}.
$$

Damit gilt

$$
\|\vec A\|_1
=
\max
\left\{
7,\,
3\sqrt{2}
\right\}.
$$

Da

$$
3\sqrt{2}\approx4{,}24,
$$

ist das Maximum gleich $7$.

Somit gilt

$$
\boxed{
\|\vec A\|_1=7
}
$$
:::

:::{dropdown} Berechnung der maximalen Zeilensummennorm

Die maximale Zeilensummennorm ist definiert durch

$$
\|\vec A\|_\infty
=
\max_{1\leq i\leq m}
\sum_{j=1}^{n}|a_{ij}|.
$$

Für die erste Zeile ergibt sich

$$
|3+4i|+|1-i|
=
5+\sqrt{2}.
$$

Für die zweite Zeile ergibt sich

$$
|2i|+|-2+2i|
=
2+2\sqrt{2}.
$$

Damit gilt

$$
\|\vec A\|_\infty
=
\max
\left\{
5+\sqrt{2},
\,
2+2\sqrt{2}
\right\}.
$$

Es ist

$$
5+\sqrt{2}\approx6{,}41
$$

und

$$
2+2\sqrt{2}\approx4{,}83.
$$

Daher gilt

$$
\boxed{
\|\vec A\|_\infty
=
5+\sqrt{2}
}
$$
:::

:::{dropdown} Berechnung der Frobenius-Norm

Die Frobenius-Norm ist definiert durch

$$
\|\vec A\|_F
=
\sqrt{
\sum_{i=1}^{m}
\sum_{j=1}^{n}
|a_{ij}|^2
}.
$$

Damit ergibt sich

$$
\begin{aligned}
\|\vec A\|_F
&=
\sqrt{
|3+4i|^2
+
|1-i|^2
+
|2i|^2
+
|-2+2i|^2
}\\
&=
\sqrt{
5^2
+
(\sqrt{2})^2
+
2^2
+
(2\sqrt{2})^2
}\\
&=
\sqrt{
25+2+4+8
}\\
&=
\sqrt{39}.
\end{aligned}
$$

Somit gilt

$$
\boxed{
\|\vec A\|_F=\sqrt{39}
}
$$
:::

:::{dropdown} Vergleich der Matrixnormen

Für die Matrix

$$
\vec A=
\begin{pmatrix}
3+4i & 1-i\\
2i & -2+2i
\end{pmatrix}
$$

haben wir erhalten:

$$
\boxed{
\|\vec A\|_1=7
}
$$

$$
\boxed{
\|\vec A\|_\infty=5+\sqrt{2}
}
$$

und

$$
\boxed{
\|\vec A\|_F=\sqrt{39}.
}
$$

Näherungsweise gilt

$$
5+\sqrt{2}\approx6{,}41
$$

und

$$
\sqrt{39}\approx6{,}24.
$$

Damit ergibt sich für diese Matrix

$$
\boxed{
\|\vec A\|_F
<
\|\vec A\|_\infty
<
\|\vec A\|_1.
}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Für

$$
\vec A=
\begin{pmatrix}
3+4i & 1-i\\
2i & -2+2i
\end{pmatrix}
$$

ergeben sich die Matrixnormen zu

$$
\boxed{
\|\vec A\|_1=7,
\qquad
\|\vec A\|_\infty=5+\sqrt{2},
\qquad
\|\vec A\|_F=\sqrt{39}.
}
$$
:::
