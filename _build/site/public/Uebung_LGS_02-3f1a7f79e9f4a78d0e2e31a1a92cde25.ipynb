## Übung: LGS mit unendlich vielen Lösungen

Gegeben ist das lineare Gleichungssystem

$$
\begin{aligned}
x_1 + 2x_2 - x_3 &= 3,\\
2x_1 + 4x_2 - 2x_3 &= 6,\\
x_1 - x_2 + 2x_3 &= 0.
\end{aligned}
$$

Stellen Sie zunächst die zugehörige erweiterte Koeffizientenmatrix auf.

Führen Sie anschließend das **Gauß-Eliminationsverfahren** durch und bringen Sie die Matrix durch elementare Zeilenumformungen in eine **reduzierte Dreiecksform**.

Verwenden Sie dabei nur die folgenden elementaren Zeilenoperationen:

- Vertauschen zweier Zeilen,
- Multiplikation einer Zeile mit einer von null verschiedenen Zahl,
- Addition eines Vielfachen einer Zeile zu einer anderen Zeile.

Dokumentieren Sie jeden Eliminationsschritt.

Bestimmen Sie anschließend die Lösung des linearen Gleichungssystems durch Rückwärtseinsetzen.

:::{dropdown} Umwandlung des LGS in Matrixschreibweise

Das lineare Gleichungssystem lautet

$$
\begin{aligned}
x_1 + 2x_2 - x_3 &= 3,\\
2x_1 + 4x_2 - 2x_3 &= 6,\\
x_1 - x_2 + 2x_3 &= 0.
\end{aligned}
$$

Die Koeffizientenmatrix ist

$$
A=
\begin{pmatrix}
1 & 2 & -1\\
2 & 4 & -2\\
1 & -1 & 2
\end{pmatrix}.
$$

Der Vektor der Unbekannten lautet

$$
\vec{x}
=
\begin{pmatrix}
x_1\\
x_2\\
x_3
\end{pmatrix},
$$

und die rechte Seite ist

$$
\vec{b}
=
\begin{pmatrix}
3\\
6\\
0
\end{pmatrix}.
$$

Damit ergibt sich

$$
A\vec{x}=\vec{b}.
$$

Für das Gauß-Eliminationsverfahren verwenden wir die erweiterte Koeffizientenmatrix

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
2 & 4 & -2 & 6\\
1 & -1 & 2 & 0
\end{array}
\right)
}
$$
:::

::::{dropdown} Durchführung der Elimination

Ausgangspunkt ist

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
2 & 4 & -2 & 6\\
1 & -1 & 2 & 0
\end{array}
\right).
$$

:::{dropdown} Schritt 1 – Eintrag unterhalb des ersten Pivotelements eliminieren

Wir verwenden

$$
Z_2 \leftarrow Z_2-2Z_1.
$$

Es ergibt sich

$$
Z_2:
\qquad
(2,\;4,\;-2\mid6)
-
2(1,\;2,\;-1\mid3).
$$

Damit erhalten wir

$$
Z_2=(0,\;0,\;0\mid0).
$$

Die Matrix lautet nun

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 0 & 0 & 0\\
1 & -1 & 2 & 0
\end{array}
\right)
}
$$
:::

:::{dropdown} Schritt 2 – Weiteren Eintrag in der ersten Spalte eliminieren

Nun verwenden wir

$$
Z_3 \leftarrow Z_3-Z_1.
$$

Es ergibt sich

$$
Z_3:
\qquad
(1,\;-1,\;2\mid0)
-
(1,\;2,\;-1\mid3).
$$

Damit ist

$$
Z_3=(0,\;-3,\;3\mid-3).
$$

Die Matrix lautet

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 0 & 0 & 0\\
0 & -3 & 3 & -3
\end{array}
\right)
}
$$
:::

:::{dropdown} Schritt 3 – Zeilen vertauschen

Für die Dreiecksform vertauschen wir die zweite und dritte Zeile:

$$
Z_2 \leftrightarrow Z_3.
$$

Damit erhalten wir

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & -3 & 3 & -3\\
0 & 0 & 0 & 0
\end{array}
\right)
}
$$
:::

:::{dropdown} Schritt 4 – Zweites Pivotelement normieren

Wir teilen die zweite Zeile durch $-3$:

$$
Z_2 \leftarrow -\frac13 Z_2.
$$

Damit ergibt sich

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & -1 & 1\\
0 & 0 & 0 & 0
\end{array}
\right)
}
$$
:::

::::

:::{dropdown} Lösbarkeit des Gleichungssystems

Nach der Gauß-Elimination erhalten wir

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & -1 & 1\\
0 & 0 & 0 & 0
\end{array}
\right).
$$

Die letzte Zeile entspricht der Gleichung

$$
0=0.
$$

Sie liefert also keine zusätzliche Bedingung.

Es gibt drei Unbekannte, aber nur zwei unabhängige Gleichungen.

Damit gilt

$$
\operatorname{rang}(A)=2
$$

und

$$
n=3.
$$

Die Anzahl der Freiheitsgrade beträgt daher

$$
n-\operatorname{rang}(A)
=
3-2
=
1.
$$

Das lineare Gleichungssystem besitzt somit **unendlich viele Lösungen**.
:::

:::{dropdown} Bestimmung der Lösungsmenge

Aus der zweiten Zeile folgt

$$
x_2-x_3=1.
$$

Da $x_3$ nicht durch ein Pivotelement bestimmt wird, ist $x_3$ eine freie Variable.

Wir setzen

$$
x_3=t,
\qquad
t\in\mathbb{R}.
$$

Damit folgt aus der zweiten Zeile

$$
x_2-t=1,
$$

also

$$
x_2=1+t.
$$

Aus der ersten Zeile

$$
x_1+2x_2-x_3=3
$$

folgt durch Einsetzen von $x_2=1+t$ und $x_3=t$

$$
x_1+2(1+t)-t=3.
$$

Damit gilt

$$
x_1+t=1
$$

und somit

$$
x_1=1-t.
$$


:::

:::{dropdown} Lösungsmenge


Die Lösungsmenge lautet

$$
\boxed{
L=
\left\{
\begin{pmatrix}
1-t\\
1+t\\
t
\end{pmatrix}
\;\middle|\;
t\in\mathbb{R}
\right\}.
}
$$

Alternativ kann sie geschrieben werden als

$$
\boxed{
L=
\left\{
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
+
t
\begin{pmatrix}
-1\\
1\\
1
\end{pmatrix}
\;\middle|\;
t\in\mathbb{R}
\right\}.
}
$$
:::
