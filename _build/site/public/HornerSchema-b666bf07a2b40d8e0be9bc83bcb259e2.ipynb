## Horner-Schema für die Polynomdivision

### Horner-Schema

- Eine effiziente Methode zur Berechnung von Polynomen und zur Polynomdivision.
- Besonders nützlich, um ein Polynom $P(x)$ durch ein lineares Polynom der Form $(x-x_0)$ zu teilen.
- Das Polynom $P(x)$ kann als

$$
P(x)=(x-x_0)\cdot Q(x)+R(x)
$$

geschrieben werden, wobei $Q(x)$ der Quotient und $R(x)$ der Rest ist.
- Spart Rechenschritte im Vergleich zur herkömmlichen Polynomdivision.

**Ziel:**

- Zeigen, wie das Horner-Schema funktioniert.
- Erklären, wie der Quotient $Q(x)$ und der Rest $R(x)$ bei der Division berechnet werden.

### Das Polynom und die Division

**Beispiel:**

- Gegeben sei das Polynom:

$$
P(x)=2x^3-6x^2+2x-1.
$$

- Wir möchten $P(x)$ durch $(x-3)$ teilen.

**Schritte:**

1. Bestimme die Koeffizienten des Polynoms:

$$
2,\;-6,\;2,\;-1.
$$

2. Wähle den Divisor

$$
c=3,
$$

da wir durch $x-3$ teilen.

### Vorbereitung des Horner-Schemas

- Schreibe die Koeffizienten des Polynoms in eine Zeile:

$$
\begin{array}{cccc}
2 & -6 & 2 & -1
\end{array}
$$

- Schreibe den Divisor-Wert $c=3$ links daneben.
- Das Horner-Schema hilft uns, den Quotienten und den Rest durch wiederholtes Multiplizieren und Addieren zu finden.

### Durchführung des Horner-Schemas

**Berechnungsschritte:**

$$
\begin{array}{c|rrrr}
  & 2 & -6 & 2 & -1 \\
3 &   & 6 & 0 & 6 \\
\hline
  & 2 & 0 & 2 & 5
\end{array}
$$

**Erklärung:**

- Bring den ersten Koeffizienten $2$ direkt nach unten.
- Multipliziere $2$ mit dem Divisor $3$ und schreibe das Ergebnis $(6)$ unter den nächsten Koeffizienten.
- Addiere:

$$
-6+6=0.
$$

Wiederhole den Vorgang.

### Ergebnisinterpretation

**Ergebnis:**

- Der untere Bereich zeigt die Koeffizienten des Quotientenpolynoms und den Rest.
- Die Koeffizienten $2,0,2$ ergeben das Polynom

$$
Q(x)=2x^2+2.
$$

- Der Rest ist

$$
5.
$$

**Polynomdivision:**

$$
P(x)=(x-3)\cdot Q(x)+R
$$

$$
P(x)
=
(x-3)\cdot(2x^2+2)+5.
$$

<a id="horner-komplex"></a>

### Horner-Schema für Polynome mit komplexen Koeffizienten

Das Horner-Schema kann auch für Polynome mit komplexen Koeffizienten verwendet werden, um Nullstellen effizient zu finden. Wenn nach Anwendung des Horner-Schemas ein quadratisches Polynom zur Faktorisierung übrig bleibt, kann man die quadratischen Formeln verwenden.

Für diese Faktorisierung können entweder die Mitternachtsformel oder die pq-Formel eingesetzt werden.

In den folgenden Beispielen zeigen wir, wie diese Formeln angewendet werden, um die Nullstellen des Polynoms

$$
x^3-(1+i)x^2+(2+i)x-2
$$

zu finden und es vollständig zu faktorisieren.

### Anwendung des Horner-Schemas

Gegebenes Polynom:

$$
f(x)
=
x^3-(1+i)x^2+(2+i)x-2.
$$

Setzen wir $x=1$ ein und verwenden das Horner-Schema:

$$
\begin{array}{r|rrrr}
  & 1 & -1-i & 2+i & -2 \\
1 &   & 1    & -i  & 2 \\
\hline
  & 1 & -i   & 2   & 0
\end{array}
$$

Da der Rest $0$ ist, ist

$$
x=1
$$

eine Nullstelle.

Das Polynom kann somit als

$$
(x-1)(x^2-ix+2)
$$

geschrieben werden.

<a id="mitternachtsformel-komplex"></a>

### Anwenden der Mitternachtsformel

Um die Nullstellen von

$$
x^2-ix+2=0
$$

zu berechnen, verwenden wir die Mitternachtsformel.

Die allgemeine Mitternachtsformel lautet:

$$
x
=
\frac{-b\pm\sqrt{b^2-4ac}}{2a}
$$

Für

$$
x^2-ix+2=0
$$

setzen wir

$$
a=1,
\qquad
b=-i,
\qquad
c=2
$$

ein:

$$
x
=
\frac{-(-i)\pm\sqrt{(-i)^2-4\cdot1\cdot2}}
     {2\cdot1}
$$

$$
=
\frac{i\pm\sqrt{-9}}{2}
$$

Damit ergeben sich:

$$
\begin{cases}
x_1=-i\\[4pt]
x_2=2i
\end{cases}
$$

Damit ist die endgültige Faktorisierung des Polynoms:

$$
x^3-(1+i)x^2+(2+i)x-2
=
(x-1)(x-2i)(x+i)
$$

<a id="pq-formel-komplex"></a>

### Alternativ: Anwenden der pq-Formel

Alternativ können wir die pq-Formel verwenden, um die Nullstellen von

$$
x^2-ix+2=0
$$

zu berechnen.

Die allgemeine pq-Formel lautet:

$$
x
=
-\frac{p}{2}
\pm
\sqrt{
\left(\frac{p}{2}\right)^2-q
}
$$

Für

$$
x^2-ix+2=0
$$

setzen wir

$$
p=-i
\qquad\text{und}\qquad
q=2
$$

ein:

$$
x
=
\frac{i}{2}
\pm
\sqrt{
\left(-\frac{i}{2}\right)^2-2
}
$$

Dies ergibt:

$$
\begin{aligned}
x
&=
\frac{i}{2}
\pm
\sqrt{-\frac14-2}\\[4pt]
&=
\frac{i}{2}
\pm
\sqrt{-\frac94}\\[4pt]
&=
\frac{i}{2}
\pm
\frac{3i}{2}
\end{aligned}
$$

Damit ergeben sich:

$$
\begin{cases}
x_1=-i\\[4pt]
x_2=2i
\end{cases}
$$

### Vorteile des Horner-Schemas

- **Effizienz:** Weniger Rechenschritte als bei der herkömmlichen Polynomdivision.
- **Einfachheit:** Leicht zu verstehen und anzuwenden, besonders für die Division durch lineare Polynome.
- **Vielseitigkeit:** Kann auch zur schnellen Auswertung von Polynomen bei gegebenen $x$-Werten verwendet werden.
