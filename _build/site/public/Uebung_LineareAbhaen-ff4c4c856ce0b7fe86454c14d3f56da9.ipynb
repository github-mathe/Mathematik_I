## Lineare Abhängigkeit von Vektoren mit dem Gauß-Eliminationsverfahren

Gegeben seien die Vektoren

$$
\vec{v}_1=
\begin{pmatrix}
1\\
1\\
2
\end{pmatrix},
\qquad
\vec{v}_2=
\begin{pmatrix}
2\\
-1\\
1
\end{pmatrix},
\qquad
\vec{v}_3=
\begin{pmatrix}
3\\
0\\
3
\end{pmatrix}.
$$

Untersuchen Sie mithilfe des **Gauß-Eliminationsverfahrens**, ob die Vektoren

$$
\vec{v}_1,\vec{v}_2,\vec{v}_3
$$

linear abhängig oder linear unabhängig sind.

Schreiben Sie die Vektoren dazu **als Zeilen einer Matrix** und bringen Sie die Matrix durch elementare Zeilenumformungen in Dreiecksform.

Bestimmen Sie anschließend den Rang der Matrix und begründen Sie damit die lineare Abhängigkeit bzw. Unabhängigkeit der Vektoren.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Aufstellen der Matrix

Die Vektoren werden als **Zeilen** in eine Matrix geschrieben:

$$
A=
\begin{pmatrix}
1 & 1 & 2\\
2 & -1 & 1\\
3 & 0 & 3
\end{pmatrix}.
$$

Die Zeilen der Matrix entsprechen also den Vektoren

$$
\vec{v}_1^\top,\qquad
\vec{v}_2^\top,\qquad
\vec{v}_3^\top.
$$

Durch das Gauß-Eliminationsverfahren bestimmen wir nun den Rang der Matrix.
:::

::::{dropdown} Durchführung der Elimination

Ausgangspunkt ist die Matrix

$$
\begin{pmatrix}
1 & 1 & 2\\
2 & -1 & 1\\
3 & 0 & 3
\end{pmatrix}.
$$


:::{dropdown} Schritt 1 – Eintrag in der ersten Spalte eliminieren

Der Eintrag $2$ in der ersten Spalte der zweiten Zeile wird eliminiert.

$$
\begin{pmatrix}
1 & 1 & 2\\
2 & -1 & 1\\
3 & 0 & 3
\end{pmatrix}
\,
\begin{array}{l}
\\
\mathrm{II}'=\mathrm{II}-2\cdot\mathrm{I}\\
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\begin{pmatrix}
1 & 1 & 2\\
0 & -3 & -3\\
3 & 0 & 3
\end{pmatrix}
}
$$

:::


:::{dropdown} Schritt 2 – Weiteren Eintrag in der ersten Spalte eliminieren

Nun wird der Eintrag $3$ in der ersten Spalte der dritten Zeile eliminiert.

$$
\begin{pmatrix}
1 & 1 & 2\\
0 & -3 & -3\\
3 & 0 & 3
\end{pmatrix}
\,
\begin{array}{l}
\\
\\
\mathrm{III}'=\mathrm{III}-3\cdot\mathrm{I}
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\begin{pmatrix}
1 & 1 & 2\\
0 & -3 & -3\\
0 & -3 & -3
\end{pmatrix}
}
$$

:::


:::{dropdown} Schritt 3 – Eintrag unterhalb des zweiten Pivotelements eliminieren

Nun wird der Eintrag $-3$ in der zweiten Spalte der dritten Zeile eliminiert.

Dabei werden bereits veränderte Zeilen mit einem Strich gekennzeichnet.

$$
\begin{pmatrix}
1 & 1 & 2\\
0 & -3 & -3\\
0 & -3 & -3
\end{pmatrix}
\,
\begin{array}{l}
\\
\\
\mathrm{III}''=\mathrm{III}'-\mathrm{II}'
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\begin{pmatrix}
1 & 1 & 2\\
0 & -3 & -3\\
0 & 0 & 0
\end{pmatrix}
}
$$

:::


:::{dropdown} Schritt 4 – Zweites Pivotelement normieren

Das zweite Pivotelement wird auf $1$ normiert.

$$
\begin{pmatrix}
1 & 1 & 2\\
0 & -3 & -3\\
0 & 0 & 0
\end{pmatrix}
\,
\begin{array}{l}
\\
\mathrm{II}''=-\frac{1}{3}\mathrm{II}'\\
\end{array}
$$

Damit erhalten wir die Dreiecksform

$$
\boxed{
\begin{pmatrix}
1 & 1 & 2\\
0 & 1 & 1\\
0 & 0 & 0
\end{pmatrix}
}
$$

:::

::::

:::{dropdown} Betrachtung des Rangs

Nach der Gauß-Elimination erhalten wir

$$
\begin{pmatrix}
1 & 1 & 2\\
0 & 1 & 1\\
0 & 0 & 0
\end{pmatrix}.
$$

Die Matrix besitzt zwei von null verschiedene Zeilen bzw. zwei Pivotelemente.

Daher gilt

$$
\boxed{
\operatorname{rang}(A)=2
}
$$

Es werden insgesamt drei Vektoren untersucht.

Damit gilt

$$
\operatorname{rang}(A)=2<3.
$$

Der Rang der Matrix ist also kleiner als die Anzahl der Vektoren.

Daher sind die Vektoren

$$
\boxed{
\vec{v}_1,\vec{v}_2,\vec{v}_3
\text{ linear abhängig.}
}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Die Gauß-Elimination liefert

$$
\begin{pmatrix}
1 & 1 & 2\\
0 & 1 & 1\\
0 & 0 & 0
\end{pmatrix}.
$$

Somit ist

$$
\operatorname{rang}(A)=2.
$$

Da drei Vektoren betrachtet werden und

$$
\operatorname{rang}(A)<3,
$$

sind die Vektoren **linear abhängig**.

Zur Kontrolle erkennt man außerdem direkt, dass

$$
\boxed{
\vec{v}_3=\vec{v}_1+\vec{v}_2
}
$$

gilt.
:::
