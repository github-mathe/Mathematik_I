## Berechnung der Inversen

Gegeben sei die Matrix

$$
\mathbf A
=
\begin{pmatrix}
1 & 1 & 0\\
0 & 1 & 1\\
1 & 0 & 1
\end{pmatrix}.
$$

1. Bestimmen Sie die inverse Matrix $\mathbf A^{-1}$ mit dem Gauß-Jordan-Algorithmus.
2. Überprüfen Sie das Ergebnis durch Berechnung von

$$
\mathbf A\mathbf A^{-1}.
$$
