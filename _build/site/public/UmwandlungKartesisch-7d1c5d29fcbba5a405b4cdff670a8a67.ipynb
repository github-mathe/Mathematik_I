## Umwandlung von kartesischer zur Polarform

- Komplexe Zahlen können in zwei Darstellungsformen beschrieben werden:
  - **Kartesische Form:** $z = x + iy$, wobei $x$ der reale Teil und $y$ der imaginäre Teil ist.
  - **Polarform:** $z = r(\cos\theta + i\sin\theta)$, wobei $r$ der Betrag (die Länge) und $\theta$ der Winkel (das Argument) ist.
- In den folgenden Beispielen lernen wir, wie man die kartesische Form in die Polarform umwandelt.
- Anstelle der Arkusfunktionen nutzen wir eine trigonometrische Wertetabelle, um die Umwandlung zu erleichtern.

### Beispiel 1 – Aufgabe

Gegeben ist die komplexe Zahl in kartesischer Form:

$$
z = \sqrt{3}+i
$$

1. Wandeln Sie $z$ in die Polarform $z=r(\cos\theta+i\sin\theta)$ um.
2. Bestimmen Sie den Betrag $r$ und den Winkel $\theta$.

### Beispiel 1 – Lösung

Berechnung des Betrags:

$$
r
=
\sqrt{(\sqrt{3})^2+1^2}
=
\sqrt{3+1}
=
\sqrt{4}
=
2
$$

Berechnung von $\cos\theta$ und $\sin\theta$:

$$
\cos\theta=\dfrac{\sqrt{3}}{2},
\qquad
\sin\theta=\dfrac{1}{2}
$$

Vergleich mit der Tabelle:

$$
\cos\frac{\pi}{6}
=
\dfrac{\sqrt{3}}{2},
\qquad
\sin\frac{\pi}{6}
=
\dfrac{1}{2}
$$

Also ist

$$
\theta=\frac{\pi}{6}.
$$

**Polarform:**

$$
z
=
2\left(
\cos\frac{\pi}{6}
+
i\sin\frac{\pi}{6}
\right)
$$

### Beispiel 2 – Aufgabe

Gegeben ist die komplexe Zahl:

$$
w=1+i
$$

1. Wandeln Sie $w$ in die Polarform um.
2. Bestimmen Sie den Betrag $r$ und den Winkel $\theta$.

### Beispiel 2 – Lösung

Berechnung des Betrags:

$$
r
=
\sqrt{1^2+1^2}
=
\sqrt{2}
$$

Berechnung von $\cos\theta$ und $\sin\theta$:

$$
\cos\theta
=
\dfrac{1}{\sqrt{2}},
\qquad
\sin\theta
=
\dfrac{1}{\sqrt{2}}
$$

Rationalisierung:

$$
\cos\theta
=
\dfrac{\sqrt{2}}{2},
\qquad
\sin\theta
=
\dfrac{\sqrt{2}}{2}
$$

Vergleich mit der Tabelle:

$$
\cos\frac{\pi}{4}
=
\dfrac{\sqrt{2}}{2},
\qquad
\sin\frac{\pi}{4}
=
\dfrac{\sqrt{2}}{2}
$$

Also ist

$$
\theta=\frac{\pi}{4}.
$$

**Polarform:**

$$
w
=
\sqrt{2}
\left(
\cos\frac{\pi}{4}
+
i\sin\frac{\pi}{4}
\right)
$$

### Beispiel 3 – Aufgabe

Gegeben ist die komplexe Zahl:

$$
u=-1+i\sqrt{3}
$$

1. Wandeln Sie $u$ in die Polarform um.
2. Bestimmen Sie den Betrag $r$ und den Winkel $\theta$.

### Beispiel 3 – Lösung

Berechnung des Betrags:

$$
r
=
\sqrt{(-1)^2+(\sqrt{3})^2}
=
\sqrt{1+3}
=
\sqrt{4}
=
2
$$

Berechnung von $\cos\theta$ und $\sin\theta$:

$$
\cos\theta
=
\dfrac{-1}{2},
\qquad
\sin\theta
=
\dfrac{\sqrt{3}}{2}
$$

Vergleich mit der Tabelle:

$$
\cos\frac{2\pi}{3}
=
-\dfrac{1}{2},
\qquad
\sin\frac{2\pi}{3}
=
\dfrac{\sqrt{3}}{2}
$$

Also ist

$$
\theta=\frac{2\pi}{3}.
$$

**Polarform:**

$$
u
=
2\left(
\cos\frac{2\pi}{3}
+
i\sin\frac{2\pi}{3}
\right)
$$

### Beispiel 4 – Aufgabe

Gegeben ist die komplexe Zahl:

$$
v=-\sqrt{2}-i\sqrt{2}
$$

1. Wandeln Sie $v$ in die Polarform um.
2. Bestimmen Sie den Betrag $r$ und den Winkel $\theta$.

### Beispiel 4 – Lösung

Berechnung des Betrags:

$$
r
=
\sqrt{(-\sqrt{2})^2+(-\sqrt{2})^2}
=
\sqrt{2+2}
=
\sqrt{4}
=
2
$$

Berechnung von $\cos\theta$ und $\sin\theta$:

$$
\cos\theta
=
\dfrac{-\sqrt{2}}{2}
=
-\dfrac{\sqrt{2}}{2},
\qquad
\sin\theta
=
\dfrac{-\sqrt{2}}{2}
=
-\dfrac{\sqrt{2}}{2}
$$

Vergleich mit der Tabelle:

$$
\cos\frac{5\pi}{4}
=
-\dfrac{\sqrt{2}}{2},
\qquad
\sin\frac{5\pi}{4}
=
-\dfrac{\sqrt{2}}{2}
$$

Also ist

$$
\theta=\frac{5\pi}{4}.
$$

**Polarform:**

$$
v
=
2\left(
\cos\frac{5\pi}{4}
+
i\sin\frac{5\pi}{4}
\right)
$$

### Umwandlung von kartesischer zu Polarform

- Der Betrag $r$ wird mit

$$
r=\sqrt{x^2+y^2}
$$

berechnet.

- Die Winkel $\theta$ werden durch Vergleich von $\cos\theta$ und $\sin\theta$ mit bekannten Werten aus der Tabelle ermittelt.
- Ohne Arkusfunktionen nutzen wir bekannte Winkel und ihre trigonometrischen Werte.
- In der Praxis kann ein Taschenrechner verwendet werden, um $\theta$ direkt mit der Arkustangensfunktion ($\arctan$) zu berechnen.
