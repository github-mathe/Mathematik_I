## Konjugierte und adjungierte Matrix

Gegeben sei die komplexe Matrix

$$
\mathbf{A}
=
\begin{pmatrix}
1+2i & -i & 3\\
2-i & 1+i & -2i\\
i & 4 & -1+3i
\end{pmatrix}.
$$

1. Bestimmen Sie die **konjugierte Matrix**

$$
\overline{\mathbf{A}}.
$$

2. Bestimmen Sie anschließend die **adjungierte Matrix**

$$
\mathbf{A}^*
=
\overline{\mathbf{A}}^{\,T}.
$$
