## Spann eines Vektorraumes

Gegeben seien die Vektoren

$$
\vec{v}_1=
\begin{pmatrix}
1\\
2\\
0
\end{pmatrix},
\qquad
\vec{v}_2=
\begin{pmatrix}
0\\
1\\
1
\end{pmatrix},
\qquad
\vec{v}_3=
\begin{pmatrix}
1\\
3\\
1
\end{pmatrix}.
$$

Untersuchen Sie den von den Vektoren aufgespannten Vektorraum

$$
\operatorname{span}
\left\{
\vec{v}_1,\vec{v}_2,\vec{v}_3
\right\}.
$$

Bearbeiten Sie dazu folgende Aufgaben:

1. Schreiben Sie die Vektoren als Spalten einer Matrix.
2. Bestimmen Sie mithilfe des **Gauß-Eliminationsverfahrens** den Rang der Matrix.
3. Bestimmen Sie die Dimension des aufgespannten Vektorraums.
4. Entscheiden Sie, ob die drei Vektoren den gesamten Raum $\mathbb{R}^3$ aufspannen.
5. Überprüfen Sie, ob der Vektor

$$
\vec{w}
=
\begin{pmatrix}
2\\
5\\
1
\end{pmatrix}
$$

im Spann der Vektoren liegt.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Aufstellen der Matrix

Die Vektoren werden als Spalten einer Matrix angeordnet:

$$
A=
\begin{pmatrix}
1 & 0 & 1\\
2 & 1 & 3\\
0 & 1 & 1
\end{pmatrix}.
$$

Die Spalten von $A$ sind also

$$
\vec{v}_1,\qquad
\vec{v}_2,\qquad
\vec{v}_3.
$$

Der von diesen Vektoren aufgespannte Raum ist

$$
\operatorname{span}
\left\{
\vec{v}_1,\vec{v}_2,\vec{v}_3
\right\}.
$$

Der Rang der Matrix gibt an, wie viele linear unabhängige Richtungen von den Vektoren aufgespannt werden.
:::

::::{dropdown} Durchführung der Elimination

Ausgangspunkt ist die Matrix

$$
\begin{pmatrix}
1 & 0 & 1\\
2 & 1 & 3\\
0 & 1 & 1
\end{pmatrix}.
$$


:::{dropdown} Schritt 1 – Eintrag in der ersten Spalte eliminieren

Die Zeilenoperation lautet

$$
\left(
\begin{array}{ccc}
1 & 0 & 1\\
2 & 1 & 3\\
0 & 1 & 1
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'=\mathrm{II}-2\cdot\mathrm{I}\\
\end{array}
$$

Danach ergibt sich

$$
\boxed{
\begin{pmatrix}
1 & 0 & 1\\
0 & 1 & 1\\
0 & 1 & 1
\end{pmatrix}
}
$$

:::


:::{dropdown} Schritt 2 – Eintrag unterhalb des zweiten Pivotelements eliminieren

Die Zeilenoperation lautet

$$
\left(
\begin{array}{ccc}
1 & 0 & 1\\
0 & 1 & 1\\
0 & 1 & 1
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}'=\mathrm{III}-\mathrm{II}'
\end{array}
$$

Danach ergibt sich

$$
\boxed{
\begin{pmatrix}
1 & 0 & 1\\
0 & 1 & 1\\
0 & 0 & 0
\end{pmatrix}
}
$$

:::

::::

:::{dropdown} Betrachtung des aufgespannten Vektorraums

Nach der Gauß-Elimination erhalten wir

$$
\begin{pmatrix}
1 & 0 & 1\\
0 & 1 & 1\\
0 & 0 & 0
\end{pmatrix}.
$$

Die Matrix besitzt zwei Pivotelemente. Daher gilt

$$
\boxed{
\operatorname{rang}(A)=2
}
$$

Somit besitzt der von den Vektoren aufgespannte Vektorraum die Dimension

$$
\boxed{
\dim\left(
\operatorname{span}
\{\vec{v}_1,\vec{v}_2,\vec{v}_3\}
\right)=2.
}
$$

Die drei Vektoren spannen daher **nicht den gesamten Raum $\mathbb{R}^3$** auf.

Sie spannen nur einen zweidimensionalen Unterraum des $\mathbb{R}^3$ auf.

Außerdem gilt

$$
\vec{v}_3=\vec{v}_1+\vec{v}_2.
$$

Der dritte Vektor erzeugt daher keine zusätzliche unabhängige Richtung.

Somit gilt

$$
\boxed{
\operatorname{span}
\{\vec{v}_1,\vec{v}_2,\vec{v}_3\}
=
\operatorname{span}
\{\vec{v}_1,\vec{v}_2\}.
}
$$

Für eine Basis des $\mathbb{R}^3$ benötigen wir drei linear unabhängige Vektoren, die den gesamten Raum $\mathbb{R}^3$ aufspannen.

Da hier

$$
\operatorname{rang}(A)=2<3
$$

gilt, sind die drei gegebenen Vektoren linear abhängig und spannen den $\mathbb{R}^3$ nicht vollständig auf.

Daher bilden

$$
\boxed{
\vec{v}_1,\vec{v}_2,\vec{v}_3
}
$$

**keine Basis des $\mathbb{R}^3$.**

Die beiden Vektoren

$$
\vec{v}_1,\vec{v}_2
$$

sind jedoch linear unabhängig und bilden eine Basis des von den drei Vektoren aufgespannten zweidimensionalen Unterraums:

$$
\boxed{
\operatorname{span}
\{\vec{v}_1,\vec{v}_2,\vec{v}_3\}
=
\operatorname{span}
\{\vec{v}_1,\vec{v}_2\}.
}
$$
:::

:::{dropdown} Liegt $\vec{w}$ im Spann der Vektoren?

Gesucht sind Skalare $\lambda_1,\lambda_2\in\mathbb{R}$ mit

$$
\lambda_1\vec{v}_1
+
\lambda_2\vec{v}_2
=
\vec{w}.
$$

Einsetzen ergibt

$$
\lambda_1
\begin{pmatrix}
1\\
2\\
0
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
0\\
1\\
1
\end{pmatrix}
=
\begin{pmatrix}
2\\
5\\
1
\end{pmatrix}.
$$

Daraus folgt das lineare Gleichungssystem

$$
\begin{aligned}
\lambda_1 &=2,\\
2\lambda_1+\lambda_2 &=5,\\
\lambda_2 &=1.
\end{aligned}
$$

Mit

$$
\lambda_1=2
\qquad\text{und}\qquad
\lambda_2=1
$$

ist auch die zweite Gleichung erfüllt:

$$
2\cdot2+1=5.
$$

Damit gilt

$$
\boxed{
\vec{w}
=
2\vec{v}_1+\vec{v}_2.
}
$$

Folglich liegt $\vec{w}$ im aufgespannten Vektorraum:

$$
\boxed{
\vec{w}
\in
\operatorname{span}
\{\vec{v}_1,\vec{v}_2,\vec{v}_3\}.
}
$$
:::

:::{dropdown} Gesamtlösung anzeigen

Für die Matrix

$$
A=
\begin{pmatrix}
1 & 0 & 1\\
2 & 1 & 3\\
0 & 1 & 1
\end{pmatrix}
$$

gilt

$$
\operatorname{rang}(A)=2.
$$

Damit ist

$$
\boxed{
\dim\left(
\operatorname{span}
\{\vec{v}_1,\vec{v}_2,\vec{v}_3\}
\right)=2.
}
$$

Die Vektoren spannen somit einen zweidimensionalen Unterraum des $\mathbb{R}^3$ auf.

Da

$$
\vec{v}_3=\vec{v}_1+\vec{v}_2,
$$

gilt

$$
\boxed{
\operatorname{span}
\{\vec{v}_1,\vec{v}_2,\vec{v}_3\}
=
\operatorname{span}
\{\vec{v}_1,\vec{v}_2\}.
}
$$

Für den gegebenen Vektor $\vec{w}$ gilt außerdem

$$
\vec{w}=2\vec{v}_1+\vec{v}_2.
$$

Daher ist

$$
\boxed{
\vec{w}
\in
\operatorname{span}
\{\vec{v}_1,\vec{v}_2,\vec{v}_3\}.
}
$$
:::
