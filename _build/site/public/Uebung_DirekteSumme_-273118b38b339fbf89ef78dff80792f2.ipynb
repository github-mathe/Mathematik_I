## Direkte Summe von Unterräumen

Gegeben seien die Unterräume

$$
U
=
\operatorname{span}
\left\{
\mathbf u_1=
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix},
\;
\mathbf u_2=
\begin{pmatrix}
0\\
1\\
1
\end{pmatrix}
\right\}
$$

und

$$
W
=
\operatorname{span}
\left\{
\mathbf w=
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}
\right\}
$$

des $\mathbb R^3$.

1. Bestimmen Sie den Schnitt $U\cap W$.
2. Prüfen Sie, ob $U+W$ eine direkte Summe ist.
3. Bestimmen Sie $U+W$.
4. Zerlegen Sie den Vektor

$$
\mathbf v=
\begin{pmatrix}
3\\
2\\
1
\end{pmatrix}
$$

in einen Vektor aus $U$ und einen Vektor aus $W$.

## Lösung

Zur Untersuchung des Schnitts stellen wir einen Vektor aus $U$ einem Vektor aus $W$ gleich.

:::{dropdown} Schnitt $U\cap W$ bestimmen

Ein Vektor aus dem Schnitt $U\cap W$ muss sowohl in $U$ als auch in $W$ liegen.

Daher muss gelten

$$
\lambda_1\mathbf u_1+\lambda_2\mathbf u_2
=
\lambda_3\mathbf w.
$$

Damit erhalten wir

$$
\lambda_1
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
0\\
1\\
1
\end{pmatrix}
-
\lambda_3
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}
=
\begin{pmatrix}
0\\
0\\
0
\end{pmatrix}.
$$

Komponentenweise folgt das Gleichungssystem

$$
\begin{aligned}
\lambda_1-\lambda_3&=0,\\
\lambda_1+\lambda_2&=0,\\
\lambda_2-\lambda_3&=0.
\end{aligned}
$$

Die zugehörige Koeffizientenmatrix lautet

$$
\begin{pmatrix}
1 & 0 & -1\\
1 & 1 & 0\\
0 & 1 & -1
\end{pmatrix}.
$$
:::

:::{dropdown} Gleichungssystem mit Gauß lösen

Wir führen die Gauß-Elimination durch:

$$
\left(
\begin{array}{ccc}
1 & 0 & -1\\
1 & 1 & 0\\
0 & 1 & -1
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'=\mathrm{II}-\mathrm{I}\\
\end{array}
$$

Damit ergibt sich

$$
\left(
\begin{array}{ccc}
1 & 0 & -1\\
0 & 1 & 1\\
0 & 1 & -1
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}'=\mathrm{III}-\mathrm{II}'
\end{array}
$$

und somit

$$
\boxed{
\begin{pmatrix}
1 & 0 & -1\\
0 & 1 & 1\\
0 & 0 & -2
\end{pmatrix}.
}
$$

Aus der dritten Zeile folgt

$$
-2\lambda_3=0
\quad\Rightarrow\quad
\lambda_3=0.
$$

Damit folgt aus der zweiten Zeile

$$
\lambda_2=0
$$

und aus der ersten Zeile

$$
\lambda_1=0.
$$

Es existiert also nur die triviale Lösung.

Damit ist

$$
\boxed{
U\cap W=\{\mathbf 0\}.
}
$$
:::

:::{dropdown} Direkte Summe untersuchen

Da

$$
U\cap W=\{\mathbf 0\},
$$

ist die Summe von $U$ und $W$ direkt.

Es gilt

$$
\dim(U)=2
$$

und

$$
\dim(W)=1.
$$

Damit ist

$$
\dim(U\oplus W)
=
\dim(U)+\dim(W)
=
2+1
=
3.
$$

Da $U\oplus W$ ein dreidimensionaler Unterraum des $\mathbb R^3$ ist, folgt

$$
\boxed{
\mathbb R^3=U\oplus W.
}
$$
:::

:::{dropdown} Vektor zerlegen

Gesucht sind Skalare $\lambda_1,\lambda_2,\lambda_3$ mit

$$
\mathbf v
=
\lambda_1\mathbf u_1
+
\lambda_2\mathbf u_2
+
\lambda_3\mathbf w.
$$

Damit gilt

$$
\lambda_1
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
0\\
1\\
1
\end{pmatrix}
+
\lambda_3
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}
=
\begin{pmatrix}
3\\
2\\
1
\end{pmatrix}.
$$

Daraus folgt

$$
\begin{aligned}
\lambda_1+\lambda_3&=3,\\
\lambda_1+\lambda_2&=2,\\
\lambda_2+\lambda_3&=1.
\end{aligned}
$$

Wir erhalten

$$
\lambda_1=2,
\qquad
\lambda_2=0,
\qquad
\lambda_3=1.
$$

Damit ist

$$
\mathbf v
=
2
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
+
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}.
$$

Die Zerlegung in einen Anteil aus $U$ und einen Anteil aus $W$ lautet daher

$$
\boxed{
\mathbf v
=
\underbrace{
\begin{pmatrix}
2\\
2\\
0
\end{pmatrix}
}_{\in U}
+
\underbrace{
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}
}_{\in W}.
}
$$

Da die Summe direkt ist, ist diese Zerlegung eindeutig.
:::

:::{dropdown} Gesamtlösung anzeigen

Für die beiden Unterräume gilt

$$
\boxed{
U\cap W=\{\mathbf 0\}.
}
$$

Damit ist die Summe direkt. Da

$$
\dim(U)=2
\qquad\text{und}\qquad
\dim(W)=1,
$$

folgt

$$
\boxed{
\mathbb R^3=U\oplus W.
}
$$

Für

$$
\mathbf v=
\begin{pmatrix}
3\\
2\\
1
\end{pmatrix}
$$

ergibt sich die eindeutige Zerlegung

$$
\boxed{
\mathbf v
=
\begin{pmatrix}
2\\
2\\
0
\end{pmatrix}
+
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix},
}
$$

wobei

$$
\begin{pmatrix}
2\\
2\\
0
\end{pmatrix}
\in U
\qquad\text{und}\qquad
\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}
\in W.
$$
:::
