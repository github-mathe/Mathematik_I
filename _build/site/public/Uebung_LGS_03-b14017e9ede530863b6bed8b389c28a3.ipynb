## LGS mit keiner Lösung

Gegeben ist das lineare Gleichungssystem

$$
\begin{aligned}
x_1 + 2x_2 - x_3 &= 3,\\
2x_1 + 4x_2 - 2x_3 &= 7,\\
x_1 - x_2 + 2x_3 &= 0.
\end{aligned}
$$

Stellen Sie zunächst die zugehörige erweiterte Koeffizientenmatrix auf.

Führen Sie anschließend das **Gauß-Eliminationsverfahren** durch und bringen Sie die Matrix durch elementare Zeilenumformungen in eine **reduzierte Dreiecksform**.

Verwenden Sie dabei nur die folgenden elementaren Zeilenoperationen:

- Vertauschen zweier Zeilen,
- Multiplikation einer Zeile mit einer von null verschiedenen Zahl,
- Addition eines Vielfachen einer Zeile zu einer anderen Zeile.

Dokumentieren Sie jeden Eliminationsschritt.

Untersuchen Sie anschließend die Lösbarkeit des linearen Gleichungssystems und geben Sie die Lösungsmenge an.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Umwandlung des LGS in Matrixschreibweise

Gegeben ist das lineare Gleichungssystem

$$
\begin{aligned}
x_1 + 2x_2 - x_3 &= 3,\\
2x_1 + 4x_2 - 2x_3 &= 7,\\
x_1 - x_2 + 2x_3 &= 0.
\end{aligned}
$$

Die Koeffizienten der Unbekannten werden in der Matrix

$$
A=
\begin{pmatrix}
1 & 2 & -1\\
2 & 4 & -2\\
1 & -1 & 2
\end{pmatrix}
$$

zusammengefasst.

Der Vektor der Unbekannten lautet

$$
\vec{x}
=
\begin{pmatrix}
x_1\\
x_2\\
x_3
\end{pmatrix},
$$

und die rechte Seite wird durch

$$
\vec{b}
=
\begin{pmatrix}
3\\
7\\
0
\end{pmatrix}
$$

beschrieben.

Damit erhält man die Matrixschreibweise

$$
\boxed{
A\vec{x}=\vec{b}
}
$$

also

$$
\begin{pmatrix}
1 & 2 & -1\\
2 & 4 & -2\\
1 & -1 & 2
\end{pmatrix}
\begin{pmatrix}
x_1\\
x_2\\
x_3
\end{pmatrix}
=
\begin{pmatrix}
3\\
7\\
0
\end{pmatrix}.
$$

Für das Gauß-Eliminationsverfahren verwenden wir die erweiterte Koeffizientenmatrix

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
2 & 4 & -2 & 7\\
1 & -1 & 2 & 0
\end{array}
\right)
}
$$
:::

::::{dropdown} Durchführung der Elimination

Ausgangspunkt ist die erweiterte Koeffizientenmatrix

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
2 & 4 & -2 & 7\\
1 & -1 & 2 & 0
\end{array}
\right).
$$


:::{dropdown} Schritt 1 – Ersten Eintrag eliminieren

Der Eintrag $2$ in der ersten Spalte der zweiten Zeile wird eliminiert.

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
2 & 4 & -2 & 7\\
1 & -1 & 2 & 0
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'=\mathrm{II}-2\cdot\mathrm{I}\\
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 0 & 0 & 1\\
1 & -1 & 2 & 0
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 2 – Weiteren Eintrag in der ersten Spalte eliminieren

Nun wird der Eintrag $1$ in der ersten Spalte der dritten Zeile eliminiert.

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 0 & 0 & 1\\
1 & -1 & 2 & 0
\end{array}
\right)
\,
\begin{array}{l}
\\
\\
\mathrm{III}'=\mathrm{III}-\mathrm{I}
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 0 & 0 & 1\\
0 & -3 & 3 & -3
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 3 – Zeilen vertauschen

Für die Dreiecksform werden die zweite und dritte Zeile vertauscht.

Die bereits veränderten Zeilen sind $\mathrm{II}'$ und $\mathrm{III}'$.

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 0 & 0 & 1\\
0 & -3 & 3 & -3
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}''=\mathrm{III}'\\
\mathrm{III}''=\mathrm{II}'
\end{array}
$$

Damit erhalten wir

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & -3 & 3 & -3\\
0 & 0 & 0 & 1
\end{array}
\right)
}
$$

:::


:::{dropdown} Schritt 4 – Zweites Pivotelement normieren

Das zweite Pivotelement wird auf $1$ normiert.

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & -3 & 3 & -3\\
0 & 0 & 0 & 1
\end{array}
\right)
\,
\begin{array}{l}
\\
\mathrm{II}'''=-\frac{1}{3}\mathrm{II}''\\
\end{array}
$$

Damit ergibt sich

$$
\boxed{
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & -1 & 1\\
0 & 0 & 0 & 1
\end{array}
\right)
}
$$

:::

::::

:::{dropdown} Betrachtung der Lösungsmenge

Nach der Gauß-Elimination erhalten wir

$$
\left(
\begin{array}{ccc|c}
1 & 2 & -1 & 3\\
0 & 1 & -1 & 1\\
0 & 0 & 0 & 1
\end{array}
\right).
$$

Die letzte Zeile entspricht der Gleichung

$$
0x_1+0x_2+0x_3=1
$$

und damit

$$
0=1.
$$

Diese Gleichung ist ein Widerspruch.

Auch über den Rang lässt sich die fehlende Lösbarkeit erkennen.

Für die Koeffizientenmatrix gilt

$$
\operatorname{rang}(A)=2.
$$

Für die erweiterte Koeffizientenmatrix gilt dagegen

$$
\operatorname{rang}(A\mid\vec b)=3.
$$

Somit gilt

$$
\boxed{
\operatorname{rang}(A)
<
\operatorname{rang}(A\mid\vec b)
}
$$

und damit ist das lineare Gleichungssystem **nicht lösbar**.

Es existiert kein Vektor $\vec{x}$, der alle drei Gleichungen gleichzeitig erfüllt.
:::

:::{dropdown} Gesamtlösung anzeigen

Das lineare Gleichungssystem enthält nach der Gauß-Elimination den Widerspruch

$$
0=1.
$$

Daher besitzt das Gleichungssystem **keine Lösung**.

Die Lösungsmenge ist

$$
\boxed{
L=\varnothing
}
$$
:::
