# Skalarprodukte

Das **Skalarprodukt** ist eine lineare Abbildung wie folgt definiert:

$$
\langle \cdot,\cdot\rangle:
\mathbb{R}^n\times\mathbb{R}^n
\rightarrow
\mathbb{R},
$$

wobei

$$
\langle\mathbf{u},\mathbf{v}\rangle
=
\mathbf{u}^T\mathbf{v}
=
\sum_{i=1}^{n}u_iv_i.
$$

<a id="eigenschaften-skalarprodukt"></a>

### Eigenschaften

- **Linearität:** Für $a,b\in\mathbb{R}$ gilt:

$$
\langle a\mathbf{u}+b\mathbf{v},\mathbf{w}\rangle
=
a\langle\mathbf{u},\mathbf{w}\rangle
+
b\langle\mathbf{v},\mathbf{w}\rangle
$$

und

$$
\langle\mathbf{u},a\mathbf{v}+b\mathbf{w}\rangle
=
a\langle\mathbf{u},\mathbf{v}\rangle
+
b\langle\mathbf{u},\mathbf{w}\rangle.
$$

- **Symmetrie:**

$$
\langle\mathbf{u},\mathbf{v}\rangle
=
\langle\mathbf{v},\mathbf{u}\rangle.
$$

- **Positivität:**

$$
\langle\mathbf{u},\mathbf{u}\rangle\geq0,
$$

und

$$
\langle\mathbf{u},\mathbf{u}\rangle=0
$$

genau dann, wenn

$$
\mathbf{u}=0.
$$

<a id="hermitesches-skalarprodukt"></a>

## Das Hermitesche Skalarprodukt

Das **Hermitesche Skalarprodukt** ist eine semilineare Abbildung wie folgt definiert:

$$
\langle\cdot,\cdot\rangle:
\mathbb{C}^n\times\mathbb{C}^n
\rightarrow
\mathbb{C},
$$

wobei

$$
\langle\mathbf{u},\mathbf{v}\rangle
=
\mathbf{u}^T\overline{\mathbf{v}}
=
\sum_{i=1}^{n}u_i\overline{v_i}.
$$

### Eigenschaften

- **Sesquilinearität:**

$$
\langle a\mathbf{u}+b\mathbf{v},\mathbf{w}\rangle
=
a\langle\mathbf{u},\mathbf{w}\rangle
+
b\langle\mathbf{v},\mathbf{w}\rangle
$$

und

$$
\langle\mathbf{u},a\mathbf{v}+b\mathbf{w}\rangle
=
\overline{a}\langle\mathbf{u},\mathbf{v}\rangle
+
\overline{b}\langle\mathbf{u},\mathbf{w}\rangle.
$$

- **Hermitesche Symmetrie:**

$$
\langle\mathbf{u},\mathbf{v}\rangle
=
\overline{\langle\mathbf{v},\mathbf{u}\rangle}.
$$

- **Positivität:**

$$
\langle\mathbf{u},\mathbf{u}\rangle\geq0,
$$

und

$$
\langle\mathbf{u},\mathbf{u}\rangle=0
$$

genau dann, wenn

$$
\mathbf{u}=0.
$$

<a id="sesquilinearitaet"></a>

## Sesquilinearität des Skalarprodukts und Konventionen

In komplexen Vektorräumen ist das Skalarprodukt **sesquilinear**. Das bedeutet, dass es:

- **linear in der ersten Komponente** ist,
- **konjugiert-linear in der zweiten Komponente** ist.

### Unterschiedliche Konventionen

- In der **Mathematik** wird das Skalarprodukt gewöhnlich so definiert, dass es **linear in der ersten Komponente** und **konjugiert-linear in der zweiten Komponente** ist.

- In der **Ingenieurwissenschaft**, z. B. in der Signalverarbeitung, ist das Skalarprodukt oft **konjugiert-linear in der ersten Komponente** und **linear in der zweiten Komponente**.

Diese Konventionen können zu unterschiedlichen Berechnungsergebnissen führen. Es ist wichtig, die jeweilige Konvention zu kennen, insbesondere beim Wechsel zwischen verschiedenen Disziplinen.

## Sesquilineare Form und das Skalarprodukt im Komplexen

Eine **sesquilineare Form** ist eine Art inneres Produkt in einem komplexen Vektorraum. Sie nimmt zwei Vektoren als Eingabe und liefert eine komplexe Zahl als Ergebnis. Doch warum ist diese Form wichtig?

- In reellen Vektorräumen ist das innere Produkt von einem Vektor mit sich selbst immer eine **nicht-negative reelle Zahl**. Diese Eigenschaft ermöglicht es uns, Längen von Vektoren zu berechnen und Projektionen durchzuführen.

- Für komplexe Vektoren ist jedoch das einfache innere Produkt unzureichend, da das Produkt eines Vektors mit sich selbst **kein reelles Ergebnis** garantiert.

### Die Rolle der Sesquilinearität

Die **sesquilineare Form** löst dieses Problem, indem sie das innere Produkt modifiziert:

- Die Komponenten des zweiten Vektors werden komplex konjugiert.
- Danach wird das Standard-Skalarprodukt (inneres Produkt) verwendet.

Damit wird sichergestellt, dass das innere Produkt eines Vektors mit sich selbst eine nicht-negative reelle Zahl ergibt.

Dies ermöglicht uns:

- die **Länge** eines Vektors zu berechnen,
- und eine zuverlässige Methode zur **Projektion** und zur Zerlegung eines Vektors in Linearkombinationen zu haben.

### Beispiel: Zerlegung eines komplexen Vektors

Angenommen, wir haben den komplexen Vektor

$$
\mathbf{v}
=
\begin{pmatrix}
1\\
1
\end{pmatrix}
\in\mathbb{C}^2,
$$

und wir wollen ihn als Linearkombination der Vektoren

$$
\mathbf{u}_1
=
\begin{pmatrix}
i\\
0
\end{pmatrix}
$$

und

$$
\mathbf{u}_2
=
\begin{pmatrix}
0\\
i
\end{pmatrix}
$$

ausdrücken.

**Frage:** Wie können wir die Koeffizienten finden?

In einem reellen Raum würden wir das innere Produkt von $\mathbf{v}$ mit den Basisvektoren nehmen. In diesem Fall ergibt sich jedoch ein **falsches Ergebnis**, da das einfache innere Produkt für komplexe Vektoren nicht funktioniert.

### Verwendung der sesquilinearen Form zur Berechnung der Projektionen

Mit der sesquilinearen Form können wir die benötigten Koeffizienten finden:

$$
\langle\mathbf{v},\mathbf{u}_1\rangle
=
\mathbf{v}^T\overline{\mathbf{u}_1}
=
(1,1)
\begin{pmatrix}
-i\\
0
\end{pmatrix}
=
-i
$$

und

$$
\langle\mathbf{v},\mathbf{u}_2\rangle
=
\mathbf{v}^T\overline{\mathbf{u}_2}
=
(1,1)
\begin{pmatrix}
0\\
-i
\end{pmatrix}
=
-i.
$$

Hier liefert die sesquilineare Form die erwarteten Koeffizienten $-i$ für beide Projektionen:

$$
\mathbf{v}
=
-i
\begin{pmatrix}
i\\
0
\end{pmatrix}
-i
\begin{pmatrix}
0\\
i
\end{pmatrix}.
$$

```python
import sympy as sp

I = sp.I

v = sp.Matrix([1, 1])
u1 = sp.Matrix([I, 0])
u2 = sp.Matrix([0, I])

skalarprodukt_v_u1 = (v.T * u1.conjugate())[0]
skalarprodukt_v_u2 = (v.T * u2.conjugate())[0]

print("<v, u1> =", skalarprodukt_v_u1)
print("<v, u2> =", skalarprodukt_v_u2)
```

<a id="induzierte-norm"></a>

## Induzierte Norm in $\mathbb{R}^n$ und $\mathbb{C}^n$

Die durch das Skalarprodukt induzierte Norm ist die **$L_2$-Norm**, definiert als:

$$
\|\mathbf{u}\|
=
\sqrt{\langle\mathbf{u},\mathbf{u}\rangle}.
$$

### Eigenschaften der Norm

- **Nichtnegativität:**

$$
\|\mathbf{u}\|\geq0,
$$

und

$$
\|\mathbf{u}\|=0
$$

genau dann, wenn

$$
\mathbf{u}=0.
$$

- **Homogenität:**

$$
\|a\mathbf{u}\|
=
|a|\,\|\mathbf{u}\|
$$

für alle $a\in\mathbb{R}$ oder $a\in\mathbb{C}$.

- **Dreiecksungleichung:**

$$
\|\mathbf{u}+\mathbf{v}\|
\leq
\|\mathbf{u}\|+\|\mathbf{v}\|.
$$

<a id="cauchy-schwarz"></a>

## Cauchy-Schwarz-Ungleichung

Die **Cauchy-Schwarz-Ungleichung** gilt für alle Vektoren $\mathbf{u},\mathbf{v}$ in einem inneren Produktraum:

$$
\boxed{
|\langle\mathbf{u},\mathbf{v}\rangle|
\leq
\|\mathbf{u}\|\|\mathbf{v}\|
}
$$

**Gleichheitsfall:** Die Gleichheit gilt genau dann, wenn $\mathbf{u}$ und $\mathbf{v}$ linear abhängig sind, d. h., wenn es ein

$$
\alpha\in\mathbb{R}
$$

für $\mathbb{R}^n$ oder

$$
\alpha\in\mathbb{C}
$$

für $\mathbb{C}^n$ gibt, sodass

$$
\mathbf{u}
=
\alpha\mathbf{v}.
$$

## Beispiele

### Beispiel: Skalarprodukt in $\mathbb{R}^2$

Gegeben seien

$$
\mathbf{u}
=
\begin{pmatrix}
1\\
2
\end{pmatrix}
$$

und

$$
\mathbf{v}
=
\begin{pmatrix}
3\\
4
\end{pmatrix}.
$$

Das Skalarprodukt ergibt sich zu:

$$
\langle\mathbf{u},\mathbf{v}\rangle
=
\mathbf{u}^T\mathbf{v}
=
\begin{pmatrix}
1&2
\end{pmatrix}
\begin{pmatrix}
3\\
4
\end{pmatrix}
$$

$$
=
1\cdot3+2\cdot4
=
11.
$$

Die induzierte Norm ist:

$$
\|\mathbf{u}\|
=
\sqrt{\langle\mathbf{u},\mathbf{u}\rangle}
=
\sqrt{1^2+2^2}
=
\sqrt{5}.
$$

```python
import sympy as sp

u = sp.Matrix([1, 2])
v = sp.Matrix([3, 4])

skalarprodukt = (u.T * v)[0]
norm_u = sp.sqrt((u.T * u)[0])

print("<u, v> =", skalarprodukt)
print("||u|| =", norm_u)
```

### Beispiel: Skalarprodukt in $\mathbb{C}^2$

Gegeben seien

$$
\mathbf{u}
=
\begin{pmatrix}
1+i\\
2-i
\end{pmatrix}
$$

und

$$
\mathbf{v}
=
\begin{pmatrix}
3\\
4+i
\end{pmatrix}.
$$

Das Hermitesche Skalarprodukt ergibt sich zu:

$$
\langle\mathbf{u},\mathbf{v}\rangle
=
\mathbf{u}^T\overline{\mathbf{v}}
=
\begin{pmatrix}
1+i&2-i
\end{pmatrix}
\begin{pmatrix}
3\\
4-i
\end{pmatrix}.
$$

Berechnung:

$$
(1+i)\cdot3
+
(2-i)\cdot(4-i)
$$

$$
=
3+3i+8-4i-2i-1
=
10-3i.
$$

Die induzierte Norm ist:

$$
\|\mathbf{u}\|
=
\sqrt{
\langle\mathbf{u},\mathbf{u}\rangle
}
$$

$$
=
\sqrt{
(1+i)(1-i)
+
(2-i)(2+i)
}
$$

$$
=
\sqrt{1+1+4+1}
=
\sqrt{7}.
$$

```python
import sympy as sp

I = sp.I

u = sp.Matrix([
    1 + I,
    2 - I
])

v = sp.Matrix([
    3,
    4 + I
])

skalarprodukt = (u.T * v.conjugate())[0]
norm_u = sp.sqrt((u.T * u.conjugate())[0])

print("<u, v> =", sp.simplify(skalarprodukt))
print("||u|| =", sp.simplify(norm_u))
```

### Geometrische Bedeutung des Skalarprodukts

Das Skalarprodukt zweier Vektoren $\mathbf{u}$ und $\mathbf{v}$ kann nicht nur algebraisch, sondern auch geometrisch interpretiert werden:

$$
\langle \mathbf{u}, \mathbf{v} \rangle
=
\|\mathbf{u}\| \, \|\mathbf{v}\| \cos(\varphi)
$$

Dabei ist $\varphi$ der Winkel zwischen den beiden Vektoren.

- Ist $\varphi < 90^\circ$, so ist $\cos(\varphi) > 0$ und damit das Skalarprodukt positiv.
- Ist $\varphi = 90^\circ$, so ist $\cos(\varphi) = 0$ und damit das Skalarprodukt gleich null.
- Ist $\varphi > 90^\circ$, so ist $\cos(\varphi) < 0$ und damit das Skalarprodukt negativ.

Die folgende Skizze zeigt zwei Vektoren und den Winkel zwischen ihnen.

```python
import numpy as np
import matplotlib.pyplot as plt

# Vektoren
u = np.array([3, 1])
v = np.array([2, 3])

# Skalarprodukt und Winkel zwischen den Vektoren
skalarprodukt = np.dot(u, v)
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)

phi = np.arccos(skalarprodukt / (norm_u * norm_v))
phi_grad = np.degrees(phi)

# Richtungswinkel der beiden Vektoren
winkel_u = np.arctan2(u[1], u[0])
winkel_v = np.arctan2(v[1], v[0])

# Plot
plt.figure(figsize=(6, 6))

plt.axhline(0)
plt.axvline(0)

# Vektoren
plt.quiver(0, 0, u[0], u[1],
           angles="xy", scale_units="xy", scale=1)

plt.quiver(0, 0, v[0], v[1],
           angles="xy", scale_units="xy", scale=1)

# Beschriftung
plt.text(u[0] + 0.1, u[1], r"$\mathbf{u}$", fontsize=12)
plt.text(v[0] + 0.1, v[1], r"$\mathbf{v}$", fontsize=12)

# Winkelbogen zwischen u und v
winkel = np.linspace(winkel_u, winkel_v, 100)

radius = 0.9
x_bogen = radius * np.cos(winkel)
y_bogen = radius * np.sin(winkel)

plt.plot(x_bogen, y_bogen)

# Winkelbeschriftung in der Mitte des Bogens
winkel_mitte = (winkel_u + winkel_v) / 2

plt.text(
    1.1 * np.cos(winkel_mitte),
    1.1 * np.sin(winkel_mitte),
    rf"$\varphi \approx {phi_grad:.1f}^\circ$",
    fontsize=12
)

plt.xlim(-1, 4)
plt.ylim(-1, 4)

plt.gca().set_aspect("equal", adjustable="box")
plt.grid()

plt.show()

print("Skalarprodukt <u,v> =", skalarprodukt)
print("Winkel phi =", round(phi_grad, 2), "Grad")
```
