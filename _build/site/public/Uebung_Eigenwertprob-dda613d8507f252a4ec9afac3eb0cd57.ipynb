## Eigenwerte und Eigenvektoren

Gegeben sei die Matrix

$$
\mathbf A
=
\begin{pmatrix}
2 & 1 & 0\\
1 & 2 & 0\\
0 & 0 & 4
\end{pmatrix}.
$$

1. Bestimmen Sie das charakteristische Polynom $p(\lambda)$.
2. Bestimmen Sie die Eigenwerte von $\mathbf A$.
3. Bestimmen Sie zu jedem Eigenwert die zugehörigen Eigenvektoren. Verwenden Sie hierzu die Gauß-Elimination.
4. Überprüfen Sie die gefundenen Eigenvektoren mit

$$
\mathbf A\mathbf v=\lambda\mathbf v.
$$
