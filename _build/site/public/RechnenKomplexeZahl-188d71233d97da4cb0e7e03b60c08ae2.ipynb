## Addition, Subtraktion, Multiplikation und Division

### Addition und Subtraktion von komplexen Zahlen

**Addition und Subtraktion:**

- Kombiniere gleichartige Terme (reell mit reell, imaginär mit imaginär):

$$
(a+bi)+(c+di)
=
(a+c)+(b+d)i
$$

$$
(a+bi)-(c+di)
=
(a-c)+(b-d)i
$$

**Beispiel:**

$$
(3+2i)+(1+4i)
=
4+6i
$$

$$
(5-3i)-(2+i)
=
3-4i
$$

### Graphische Darstellung der Addition

Die Addition komplexer Zahlen kann in der Gauss-Ebene als Vektoraddition dargestellt werden.

Für

$$
z_1=3+2i
$$

und

$$
z_2=1+4i
$$

ergibt sich

$$
z_1+z_2
=
4+6i.
$$

```python
import matplotlib.pyplot as plt

z1 = (3, 2)
z2 = (1, 4)
z_sum = (4, 6)

fig, ax = plt.subplots(figsize=(7, 7))

# Farben
farbe_z1 = "tab:blue"
farbe_z2 = "tab:green"
farbe_sum = "tab:red"

# Achsen
ax.axhline(0, linewidth=1, color="black")
ax.axvline(0, linewidth=1, color="black")

# z1
ax.arrow(
    0, 0, z1[0], z1[1],
    length_includes_head=True,
    head_width=0.15,
    color=farbe_z1
)

# z2 verschoben an die Spitze von z1
ax.arrow(
    z1[0], z1[1], z2[0], z2[1],
    length_includes_head=True,
    head_width=0.15,
    color=farbe_z2
)

# Summe
ax.arrow(
    0, 0, z_sum[0], z_sum[1],
    length_includes_head=True,
    head_width=0.15,
    color=farbe_sum
)

# z2 zusätzlich vom Ursprung gestrichelt
ax.plot(
    [0, z2[0]], [0, z2[1]],
    linestyle="--",
    color=farbe_z2
)

# Beschriftungen in passenden Farben
ax.text(2.3, 1.1, r"$3+2i$", fontsize=12, color=farbe_z1)
ax.text(3.2, 4.3, r"$1+4i$", fontsize=12, color=farbe_z2)
ax.text(2.0, 4.8, r"$4+6i$", fontsize=12, color=farbe_sum)

ax.set_xlim(-1, 5)
ax.set_ylim(-1, 7)
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_aspect("equal")
ax.grid(alpha=0.3)

plt.show()
```

### Multiplikation von komplexen Zahlen

**Multiplikation:**

Nutze das Distributivgesetz und die Tatsache, dass

$$
i^2=-1.
$$

Dann gilt:

$$
(a+bi)(c+di)
=
(ac-bd)+(ad+bc)i.
$$

**Beispiel:**

$$
(2+3i)(4-i)
=
(2\cdot4-3\cdot(-1))
+
(2\cdot(-1)+3\cdot4)i
$$

$$
=
(8+3)+(-2+12)i
=
11+10i.
$$

### Graphische Darstellung der Multiplikation

Bei der Multiplikation zweier komplexer Zahlen werden

- die Beträge (Längen) multipliziert,
- die Winkel (Argumente) addiert.

Beispiel:

$$
(2+3i)(4-i)
=
11+10i.
$$

Grafisch entspricht dies einer Skalierung und Drehung des Vektors.

### Winkelrichtung für komplexe Zahlen

**Winkelmessung:**

Der Winkel $\theta$ eines Vektors im komplexen Zahlenbereich kann auf zwei Arten gemessen werden:

- **Positiver Winkel:** Gemessen gegen den Uhrzeigersinn von der positiven x-Achse.
- **Negativer Winkel:** Gemessen im Uhrzeigersinn von der positiven x-Achse.

Diese beiden Richtungen repräsentieren denselben Vektor, da beide Winkel dieselbe Orientierung im Raum beschreiben.

**Beispiel:**

Der Vektor

$$
4-i
$$

kann mit einem Winkel von etwa

$$
\theta_2=346^\circ
$$

gegen den Uhrzeigersinn oder mit

$$
\theta_2=-14^\circ
$$

im Uhrzeigersinn beschrieben werden.

```python
import numpy as np
import matplotlib.pyplot as plt

# Vektor 4 - i
x_vec = 4
y_vec = -1

fig, ax = plt.subplots(figsize=(9, 7))

# Farben
farbe_positiv = "deeppink"
farbe_negativ = "red"

# Achsen
ax.annotate(
    "",
    xy=(4.7, 0),
    xytext=(-2.0, 0),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.8,
        color="black"
    )
)

ax.annotate(
    "",
    xy=(0, 2.2),
    xytext=(0, -2.2),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.8,
        color="black"
    )
)

ax.text(4.78, -0.08, "Re", fontsize=13)
ax.text(0.08, 2.15, "Im", fontsize=13)

# Vektor 4 - i
ax.annotate(
    "",
    xy=(x_vec, y_vec),
    xytext=(0, 0),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=2.5,
        color=farbe_negativ
    )
)

ax.text(
    3.05, -1.20,
    r"$4-i$",
    color=farbe_negativ,
    fontsize=13
)

# Positiver Winkel theta = 346°
r_pos = 1.5

theta_pos = np.linspace(
    0,
    np.deg2rad(346),
    800
)

x_pos = r_pos * np.cos(theta_pos)
y_pos = r_pos * np.sin(theta_pos)

ax.plot(
    x_pos,
    y_pos,
    linestyle="--",
    linewidth=2.2,
    color=farbe_positiv
)

# Pfeilspitze am Ende des positiven Kreisbogens
winkel_ende = np.deg2rad(346)
winkel_vorher = np.deg2rad(339)

ax.annotate(
    "",
    xy=(
        r_pos * np.cos(winkel_ende),
        r_pos * np.sin(winkel_ende)
    ),
    xytext=(
        r_pos * np.cos(winkel_vorher),
        r_pos * np.sin(winkel_vorher)
    ),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=2.2,
        color=farbe_positiv
    )
)

ax.text(
    1.65, 1.25,
    r"$\theta_2=346^\circ$",
    color=farbe_positiv,
    fontsize=13
)

# Negativer Winkel theta = -14°
r_neg = 1.15

theta_neg = np.linspace(
    0,
    np.deg2rad(-14),
    150
)

x_neg = r_neg * np.cos(theta_neg)
y_neg = r_neg * np.sin(theta_neg)

ax.plot(
    x_neg,
    y_neg,
    linewidth=2.2,
    color=farbe_negativ
)

# Pfeilspitze des negativen Winkels
ax.annotate(
    "",
    xy=(
        r_neg * np.cos(np.deg2rad(-14)),
        r_neg * np.sin(np.deg2rad(-14))
    ),
    xytext=(
        r_neg * np.cos(np.deg2rad(-8)),
        r_neg * np.sin(np.deg2rad(-8))
    ),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=2.2,
        color=farbe_negativ
    )
)

ax.text(
    1.8, -0.40,
    r"$\theta_2=-14^\circ$",
    color=farbe_negativ,
    fontsize=13
)

# Keine Zahlen und keine Teilstriche auf den Achsen
ax.set_xticks([])
ax.set_yticks([])

ax.set_xlim(-2.0, 5.0)
ax.set_ylim(-2.3, 2.3)

ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

- Der Vektor bleibt gleich, egal ob $\theta_2$ als **positiver** oder **negativer** Winkel dargestellt wird.
- Beide Winkel repräsentieren denselben Punkt im komplexen Zahlenbereich.

### Multiplikation: Winkeladdition

Für

$$
z_1=2+3i
$$

und

$$
z_2=4-i
$$

gelten näherungsweise die Winkel

$$
\theta_1=56^\circ
$$

und

$$
\theta_2=-14^\circ.
$$

Damit gilt

$$
\theta_1+\theta_2
=
42^\circ.
$$

Das Produkt

$$
(2+3i)(4-i)
=
11+10i
$$

besitzt also den Winkel

$$
42^\circ.
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# Farben
# --------------------------------------------------
farbe_z1 = "blue"
farbe_z2 = "red"
farbe_produkt = "#3f88c5"   # mittleres Blau 

# --------------------------------------------------
# Abbildung mit Tabelle links und Grafik rechts
# --------------------------------------------------
fig = plt.figure(figsize=(12, 7))

gs = fig.add_gridspec(
    1, 2,
    width_ratios=[0.35, 0.65],
    wspace=0.12
)

ax_table = fig.add_subplot(gs[0])
ax = fig.add_subplot(gs[1])

# ==================================================
# LINKE SEITE: Winkeltabelle
# ==================================================
ax_table.axis("off")

table = ax_table.table(
    cellText=[
        [r"$\theta_1$", r"$56^\circ$"],
        [r"$\theta_2$", r"$-14^\circ$"],
        [r"$\theta_1+\theta_2$", r"$42^\circ$"]
    ],
    colLabels=["Winkel", "Wert (Grad)"],
    cellLoc="center",
    loc="center"
)

table.auto_set_font_size(False)
table.set_fontsize(13)
table.scale(1.15, 1.8)

table[(0, 0)].get_text().set_weight("bold")
table[(0, 1)].get_text().set_weight("bold")

table[(1, 0)].get_text().set_color(farbe_z1)
table[(1, 1)].get_text().set_color(farbe_z1)

table[(2, 0)].get_text().set_color(farbe_z2)
table[(2, 1)].get_text().set_color(farbe_z2)

table[(3, 0)].get_text().set_color(farbe_produkt)
table[(3, 1)].get_text().set_color(farbe_produkt)

# ==================================================
# RECHTE SEITE: Gauss-Ebene
# ==================================================
ax.annotate(
    "",
    xy=(6.2, 0),
    xytext=(-2, 0),
    arrowprops=dict(arrowstyle="->", linewidth=1.2, color="black")
)

ax.annotate(
    "",
    xy=(0, 6.2),
    xytext=(0, -2),
    arrowprops=dict(arrowstyle="->", linewidth=1.2, color="black")
)

ax.text(6.25, -0.1, "Re", fontsize=13)
ax.text(0.12, 6.1, "Im", fontsize=13)

# Vektoren
Z1 = (2, 3)
Z2 = (4, -1)
Produkt = (5.5, 5)

ax.annotate(
    "",
    xy=Z1,
    xytext=(0, 0),
    arrowprops=dict(arrowstyle="->", linewidth=1.8, color=farbe_z1)
)
ax.text(1.55, 2.32, r"$2+3i$", color=farbe_z1, fontsize=13)

ax.annotate(
    "",
    xy=Z2,
    xytext=(0, 0),
    arrowprops=dict(arrowstyle="->", linewidth=1.8, color=farbe_z2)
)
ax.text(2.8, -1.1, r"$4-i$", color=farbe_z2, fontsize=13)

ax.annotate(
    "",
    xy=Produkt,
    xytext=(0, 0),
    arrowprops=dict(arrowstyle="->", linewidth=1.8, color=farbe_produkt)
)
ax.text(4.25, 4.3, r"$11+10i$", color=farbe_produkt, fontsize=13)

# ==================================================
# Winkelbögen
# ==================================================
def winkelbogen(radius, start_deg, end_deg, color, linewidth=1.5):
    winkel = np.linspace(
        np.deg2rad(start_deg),
        np.deg2rad(end_deg),
        250
    )

    x = radius * np.cos(winkel)
    y = radius * np.sin(winkel)

    ax.plot(x, y, color=color, linewidth=linewidth)

    ax.annotate(
        "",
        xy=(x[-1], y[-1]),
        xytext=(x[-8], y[-8]),
        arrowprops=dict(arrowstyle="->", linewidth=linewidth, color=color)
    )

winkelbogen(1.5, 0, 56, farbe_z1)
ax.text(1.55, 0.62, r"$\theta_1$", color=farbe_z1, fontsize=13)

winkelbogen(3.0, 0, -14, farbe_z2)
ax.text(3.15, -0.55, r"$\theta_2$", color=farbe_z2, fontsize=13)

winkelbogen(2.5, 0, 42, farbe_produkt)
ax.text(3.15, 1.35, r"$\theta_1+\theta_2=42^\circ$", color=farbe_produkt, fontsize=13)

# Achsenmarkierungen
for x in [-1, 0, 1, 2, 3, 4, 5]:
    ax.plot([x, x], [-0.05, 0.05], color="black", linewidth=0.8)

for y in [-1, 0, 1, 2, 3, 4, 5]:
    ax.plot([-0.05, 0.05], [y, y], color="black", linewidth=0.8)

# Zahlen aus der Vorlage
ax.text(2, -0.25, "2", ha="center", fontsize=10)
ax.text(4, -0.25, "4", ha="center", fontsize=10)
ax.text(-0.18, 3, "3", ha="right", va="center", fontsize=10)
ax.text(0.15, -1, "-1", ha="left", va="center", fontsize=10)
ax.text(5.5, -0.25, "11", ha="center", fontsize=10)
ax.text(-0.18, 5, "10", ha="right", va="center", fontsize=10)

ax.set_xlim(-2, 6.5)
ax.set_ylim(-2, 6.5)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```


Die Winkel können auch als positive Winkel dargestellt werden:

$$
\theta_1 = 56^\circ,
\qquad
\theta_2 = 346^\circ.
$$

Damit ergibt sich

$$
\theta_1+\theta_2
=
402^\circ.
$$

Da

$$
402^\circ = 42^\circ + 360^\circ,
$$

zeigt der Winkel auf dieselbe Richtung wie $42^\circ$.

```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# Farben
# --------------------------------------------------
farbe_z1 = "blue"
farbe_z2 = "red"
farbe_produkt = "#3f88c5"   # mittleres Blau

fig, ax = plt.subplots(figsize=(9, 8))

# --------------------------------------------------
# Achsen
# --------------------------------------------------
ax.annotate(
    "",
    xy=(6.4, 0),
    xytext=(-5.0, 0),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.2,
        color="black"
    )
)

ax.annotate(
    "",
    xy=(0, 6.3),
    xytext=(0, -4.8),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.2,
        color="black"
    )
)

ax.text(6.45, -0.12, "Re", fontsize=13)
ax.text(0.12, 6.15, "Im", fontsize=13)

# --------------------------------------------------
# Vektoren
# --------------------------------------------------
Z1 = (2, 3)
Z2 = (4, -1)

# Wie in der TikZ-Vorlage verkleinert dargestellt
Produkt = (5.5, 5)

# z1 = 2 + 3i
ax.annotate(
    "",
    xy=Z1,
    xytext=(0, 0),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.7,
        color=farbe_z1
    )
)

ax.text(
    1.55, 2.45,
    r"$2+3i$",
    color=farbe_z1,
    fontsize=13
)

# z2 = 4 - i
ax.annotate(
    "",
    xy=Z2,
    xytext=(0, 0),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.7,
        color=farbe_z2
    )
)

ax.text(
    2.9, -1.18,
    r"$4-i$",
    color=farbe_z2,
    fontsize=13
)

# Produkt 11 + 10i
ax.annotate(
    "",
    xy=Produkt,
    xytext=(0, 0),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.7,
        color=farbe_produkt
    )
)

ax.text(
    4.15, 4.45,
    r"$11+10i$",
    color=farbe_produkt,
    fontsize=13
)

# --------------------------------------------------
# Hilfsfunktion für Winkelbögen
# --------------------------------------------------
def winkelbogen(radius, start_deg, end_deg, color, linewidth=1.4):
    winkel = np.linspace(
        np.deg2rad(start_deg),
        np.deg2rad(end_deg),
        500
    )

    x = radius * np.cos(winkel)
    y = radius * np.sin(winkel)

    ax.plot(
        x,
        y,
        color=color,
        linewidth=linewidth
    )

    # Pfeilspitze
    ax.annotate(
        "",
        xy=(x[-1], y[-1]),
        xytext=(x[-8], y[-8]),
        arrowprops=dict(
            arrowstyle="->",
            linewidth=linewidth,
            color=color
        )
    )

# --------------------------------------------------
# theta_1 = 56°
# --------------------------------------------------
winkelbogen(
    radius=1.5,
    start_deg=0,
    end_deg=56,
    color=farbe_z1
)

ax.text(
    1.55, 0.60,
    r"$\theta_1$",
    color=farbe_z1,
    fontsize=13
)

# --------------------------------------------------
# theta_2 = 346°
# fast vollständiger roter Kreis
# --------------------------------------------------
winkelbogen(
    radius=2.5,
    start_deg=0,
    end_deg=346,
    color=farbe_z2
)

ax.text(
    -2.9, 1.55,
    r"$\theta_2$",
    color=farbe_z2,
    fontsize=13
)

# --------------------------------------------------
# theta_1 + theta_2 = 402°
# als nach außen laufende Spirale
# --------------------------------------------------

theta_spiral = np.linspace(
    0,
    np.deg2rad(402),
    800
)

# Radius wächst langsam von 3 nach außen
r_spiral = 3 + 0.18 * theta_spiral

x_spiral = r_spiral * np.cos(theta_spiral)
y_spiral = r_spiral * np.sin(theta_spiral)

ax.plot(
    x_spiral,
    y_spiral,
    color=farbe_produkt,
    linewidth=1.5
)

# Pfeilspitze am Ende der Spirale
ax.annotate(
    "",
    xy=(x_spiral[-1], y_spiral[-1]),
    xytext=(x_spiral[-10], y_spiral[-10]),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.5,
        color=farbe_produkt
    )
)

ax.text(
    4.1, 2.2,
    r"$\theta_1+\theta_2=402^\circ$",
    color=farbe_produkt,
    fontsize=13
)

# --------------------------------------------------
# Darstellung
# --------------------------------------------------

# Keine Zahlen auf den Achsen
ax.set_xticks([])
ax.set_yticks([])

# Grenzen so wählen, dass Kreise und Spirale vollständig sichtbar sind
ax.set_xlim(-5.0, 6.8)
ax.set_ylim(-4.8, 6.8)

ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

### Weiteres Beispiel der Multiplikation

Beispiel:

$$
(1+2i)(-1+i)
=
-3-i.
$$

Hier können beide Winkel als positiv angesehen werden:

| Zahl | Winkel |
|---|---:|
| $1+2i$ | $63.4^\circ$ |
| $-1+i$ | $135^\circ$ |
| $-3-i$ | $63.4^\circ+135^\circ=198.4^\circ$ |

Die Beträge werden multipliziert und die Winkel addiert, was zu einer Drehung und Skalierung führt.

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 7))

# --------------------------------------------------
# Farben
# --------------------------------------------------
farbe_z1 = "blue"
farbe_z2 = "red"
farbe_prod = "#4F81BD"   # mittleres Blau 

# --------------------------------------------------
# Achsen
# --------------------------------------------------
ax.annotate(
    "",
    xy=(4, 0),
    xytext=(-4, 0),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.2,
        color="black"
    )
)

ax.annotate(
    "",
    xy=(0, 4),
    xytext=(0, -3),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.2,
        color="black"
    )
)

ax.text(4.1, 0.05, "Re", fontsize=12)
ax.text(0.08, 4.05, "Im", fontsize=12)

# --------------------------------------------------
# Achsenmarkierungen
# --------------------------------------------------
for x in [-3, -2, -1, 0, 1, 2, 3]:
    ax.plot(
        [x, x],
        [-0.08, 0.08],
        color="black",
        linewidth=0.8
    )

for y in [-2, -1, 0, 1, 2, 3]:
    ax.plot(
        [-0.08, 0.08],
        [y, y],
        color="black",
        linewidth=0.8
    )

# --------------------------------------------------
# Vektoren
# --------------------------------------------------
Z1 = np.array([1, 2])      # 1 + 2i
Z2 = np.array([-1, 1])     # -1 + i
Produkt = np.array([-3, -1])   # -3 - i

# z1 = 1 + 2i
ax.annotate(
    "",
    xy=Z1,
    xytext=(0, 0),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.4,
        color=farbe_z1
    )
)

ax.text(
    1.25, 2.35,
    r"$1+2i$",
    color=farbe_z1,
    fontsize=12
)

# z2 = -1 + i
ax.annotate(
    "",
    xy=Z2,
    xytext=(0, 0),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.4,
        color=farbe_z2
    )
)

ax.text(
    -1.55, 1.25,
    r"$-1+i$",
    color=farbe_z2,
    fontsize=12
)

# Produkt = -3 - i
ax.annotate(
    "",
    xy=Produkt,
    xytext=(0, 0),
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.4,
        color=farbe_prod
    )
)

ax.text(
    -3.45, -1.35,
    r"$-3-i$",
    color=farbe_prod,
    fontsize=12
)

# --------------------------------------------------
# Funktion für Winkelbögen
# --------------------------------------------------
def winkelbogen(radius, start_deg, end_deg, color, lw=1.2):
    winkel = np.linspace(
        np.deg2rad(start_deg),
        np.deg2rad(end_deg),
        300
    )

    x = radius * np.cos(winkel)
    y = radius * np.sin(winkel)

    ax.plot(
        x,
        y,
        color=color,
        linewidth=lw
    )

    # Pfeilspitze am Ende des Winkelbogens
    ax.annotate(
        "",
        xy=(x[-1], y[-1]),
        xytext=(x[-8], y[-8]),
        arrowprops=dict(
            arrowstyle="->",
            linewidth=lw,
            color=color
        )
    )

# --------------------------------------------------
# theta_1 = 63.4°
# --------------------------------------------------
winkelbogen(
    radius=1.8,
    start_deg=0,
    end_deg=63.4,
    color=farbe_z1
)

ax.text(
    2.05, 0.40,
    r"$\theta_1$",
    color=farbe_z1,
    fontsize=12
)

# --------------------------------------------------
# theta_2 = 135°
# --------------------------------------------------
winkelbogen(
    radius=1.0,
    start_deg=0,
    end_deg=135,
    color=farbe_z2
)

ax.text(
    1.05, 0.65,
    r"$\theta_2$",
    color=farbe_z2,
    fontsize=12
)

# --------------------------------------------------
# theta_1 + theta_2 = 198.4°
# --------------------------------------------------
winkelbogen(
    radius=2.5,
    start_deg=0,
    end_deg=198.4,
    color=farbe_prod
)

ax.text(
    -2.7, 2.05,
    r"$\theta_1+\theta_2$",
    color=farbe_prod,
    fontsize=12
)

# --------------------------------------------------
# Achsenbeschriftungen wie in der Vorlage
# --------------------------------------------------
ax.text(1, -0.28, "1", ha="center", fontsize=9)
ax.text(-1, -0.28, "-1", ha="center", fontsize=9)
ax.text(-3, -0.28, "-3", ha="center", fontsize=9)

ax.text(-0.20, 2, "2", ha="right", va="center", fontsize=9)
ax.text(-0.20, 1, "1", ha="right", va="center", fontsize=9)
ax.text(-0.20, -1, "-1", ha="right", va="center", fontsize=9)

# --------------------------------------------------
# Darstellung
# --------------------------------------------------
ax.set_xlim(-4.2, 4.3)
ax.set_ylim(-3.2, 4.3)

ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

### Division von komplexen Zahlen

**Division:**

- Multipliziere Zähler und Nenner mit der konjugiert komplexen Zahl des Nenners.
- Für

$$
z=a+bi
$$

ist die konjugierte Zahl

$$
\overline z=a-bi.
$$

**Divisionsformel:**

$$
\frac{a+bi}{c+di}
=
\frac{(a+bi)(c-di)}
     {(c+di)(c-di)}
$$

$$
=
\frac{(ac+bd)+(bc-ad)i}
     {c^2+d^2}.
$$

**Beispiel:**

$$
\frac{3+4i}{1-2i}
=
\frac{(3+4i)(1+2i)}
     {(1-2i)(1+2i)}
$$

$$
=
\frac{(3-8)+(4\cdot1+3\cdot2)i}
     {1^2+2^2}
$$

$$
=
\frac{-5+10i}{5}
=
-1+2i.
$$

### Graphische Darstellung der Division

Bei der Division zweier komplexer Zahlen wird

- der Betrag des Zählers durch den Betrag des Nenners geteilt,
- der Winkel des Nenners vom Winkel des Zählers subtrahiert.

Für

$$
\frac{3+4i}{1-2i}
$$

gilt:

$$
|3+4i|
=
\sqrt{3^2+4^2}
=
5
$$

und

$$
|1-2i|
=
\sqrt{1^2+2^2}
=
\sqrt5.
$$

Damit ist der Betrag des Ergebnisses

$$
|z|
=
\frac{5}{\sqrt5}
=
\sqrt5.
$$

Für die Winkel gilt näherungsweise:

$$
\theta_1
=
53.1^\circ,
$$

$$
\theta_2
=
-63.4^\circ.
$$

Daher

$$
\theta
=
\theta_1-\theta_2
=
53.1^\circ+63.4^\circ
=
116.5^\circ.
$$

Grafisch entspricht dies einer Skalierung und einer Drehung in entgegengesetzter Richtung zur Multiplikation.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Arc, FancyArrowPatch

# -----------------------------------
# Daten
# -----------------------------------

numerator = np.array([3, 4])      # 3 + 4i
denominator = np.array([1, -2])   # 1 - 2i
quotient = np.array([-1, 2])      # -1 + 2i

# Winkel in Grad
theta1 = 53.1
theta2 = -63.4
theta_q = 116.5

# Farben
farbe_num = "blue"
farbe_den = "red"
farbe_quo = "#4F81BD"   # mittleres Blau 

# -----------------------------------
# Hilfsfunktion für Winkelbogen
# -----------------------------------

def winkelbogen(ax, radius, theta_start, theta_end, color, lw=1.4):
    arc = Arc(
        (0, 0),
        2 * radius,
        2 * radius,
        angle=0,
        theta1=min(theta_start, theta_end),
        theta2=max(theta_start, theta_end),
        linewidth=lw,
        color=color
    )
    ax.add_patch(arc)

    # Pfeilspitze am Ende des Bogens
    t_end = np.deg2rad(theta_end)
    t_prev = np.deg2rad(theta_end - 4 if theta_end >= theta_start else theta_end + 4)

    x1, y1 = radius * np.cos(t_prev), radius * np.sin(t_prev)
    x2, y2 = radius * np.cos(t_end), radius * np.sin(t_end)

    ax.annotate(
        "",
        xy=(x2, y2),
        xytext=(x1, y1),
        arrowprops=dict(arrowstyle="->", color=color, linewidth=lw)
    )

# -----------------------------------
# Abbildung
# -----------------------------------

fig, ax = plt.subplots(figsize=(8, 6))

# Achsen
ax.annotate("", xy=(6, 0), xytext=(-2, 0),
            arrowprops=dict(arrowstyle="->", linewidth=1.2, color="black"))
ax.annotate("", xy=(0, 5), xytext=(0, -2),
            arrowprops=dict(arrowstyle="->", linewidth=1.2, color="black"))

ax.text(6.08, 0, "Re", fontsize=12, va="center")
ax.text(0.12, 5.05, "Im", fontsize=12)

# kleine Striche auf den Achsen
for x in [-1, 0, 1, 2, 3, 4, 5]:
    ax.plot([x, x], [-0.08, 0.08], color="black", linewidth=0.8)

for y in [-1, 0, 1, 2, 3, 4, 5]:
    ax.plot([-0.08, 0.08], [y, y], color="black", linewidth=0.8)

# Vektoren
def zeichne_vektor(vektor, color, label, text_xy):
    arrow = FancyArrowPatch(
        (0, 0), (vektor[0], vektor[1]),
        arrowstyle="-|>",
        mutation_scale=14,
        linewidth=1.6,
        color=color
    )
    ax.add_patch(arrow)
    ax.text(text_xy[0], text_xy[1], label, color=color, fontsize=12)

zeichne_vektor(numerator,   farbe_num, r"$3+4i$",   (1.7, 2.7))
zeichne_vektor(denominator, farbe_den, r"$1-2i$",   (1.1, -1.4))
zeichne_vektor(quotient,    farbe_quo, r"$-1+2i$",  (-1.55, 2.15))

# Winkelbögen
winkelbogen(ax, 1.5, 0, theta1, farbe_num)
winkelbogen(ax, 0.6, 0, theta2, farbe_den)
winkelbogen(ax, 2.0, 0, theta_q, farbe_quo)

# Winkelbeschriftungen
ax.text(1.8, 0.55, r"$\theta_1=53.1^\circ$", color=farbe_num, fontsize=11)
ax.text(0.85, -0.55, r"$\theta_2=-63.4^\circ$", color=farbe_den, fontsize=11)
ax.text(-1.7, 1.55, r"$\theta_1-\theta_2=116.5^\circ$", color=farbe_quo, fontsize=11)

# Zahlenbeschriftungen an den Achsen
ax.text(1, -0.35, "1", ha="center", fontsize=10)
ax.text(3, -0.35, "3", ha="center", fontsize=10)
ax.text(-1, -0.35, "-1", ha="center", fontsize=10)

ax.text(-0.28, 4, "4", va="center", fontsize=10)
ax.text(-0.28, 2, "2", va="center", fontsize=10)
ax.text(-0.38, -2, "-2", va="center", fontsize=10)

# Layout
ax.set_xlim(-2, 6.3)
ax.set_ylim(-2.3, 5.3)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

### Reziproke komplexe Zahlen: Beispiel

Eine komplexe Zahl und ihre reziproke Zahl (Kehrwert) können auf der Gauss-Ebene grafisch dargestellt werden.

Betrachte

$$
z
=
\frac{1}{\sqrt2}
+
\frac{1}{\sqrt2}i.
$$

Der Betrag von $z$ ist

$$
|z|
=
\sqrt{
\left(\frac{1}{\sqrt2}\right)^2
+
\left(\frac{1}{\sqrt2}\right)^2
}
=
1.
$$

Die reziproke Zahl

$$
\frac{1}{z}
$$

hat denselben Betrag, aber der Winkel (Argument) ist das Negative des Winkels von $z$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Arc, FancyArrowPatch

# -----------------------------------
# Punkte
# -----------------------------------

z = np.array([1/np.sqrt(2), 1/np.sqrt(2)])
z_inv = np.array([1/np.sqrt(2), -1/np.sqrt(2)])

# Farben
farbe_z = "blue"
farbe_inv = "#4F81BD"   # mittleres Blau

# -----------------------------------
# Hilfsfunktion für Winkelbogen
# -----------------------------------

def winkelbogen(ax, radius, startwinkel, endwinkel, farbe, lw=1.4):
    bogen = Arc(
        (0, 0),
        2 * radius,
        2 * radius,
        angle=0,
        theta1=min(startwinkel, endwinkel),
        theta2=max(startwinkel, endwinkel),
        linewidth=lw,
        color=farbe
    )
    ax.add_patch(bogen)

    # Pfeilspitze am Ende
    t_end = np.deg2rad(endwinkel)
    t_prev = np.deg2rad(endwinkel - 4 if endwinkel >= startwinkel else endwinkel + 4)

    x1, y1 = radius * np.cos(t_prev), radius * np.sin(t_prev)
    x2, y2 = radius * np.cos(t_end), radius * np.sin(t_end)

    ax.annotate(
        "",
        xy=(x2, y2),
        xytext=(x1, y1),
        arrowprops=dict(arrowstyle="->", color=farbe, linewidth=lw)
    )

# -----------------------------------
# Abbildung
# -----------------------------------

fig, ax = plt.subplots(figsize=(6, 6))

# Achsen
ax.annotate("", xy=(1.5, 0), xytext=(-1.5, 0),
            arrowprops=dict(arrowstyle="->", linewidth=1.2, color="black"))
ax.annotate("", xy=(0, 1.5), xytext=(0, -1.5),
            arrowprops=dict(arrowstyle="->", linewidth=1.2, color="black"))

ax.text(1.55, 0, "Re", fontsize=12, va="center")
ax.text(0.08, 1.55, "Im", fontsize=12)

# kleine Achsenmarkierungen
for x in [-1, 0, 1]:
    ax.plot([x, x], [-0.04, 0.04], color="black", linewidth=0.8)

for y in [-1, 0, 1]:
    ax.plot([-0.04, 0.04], [y, y], color="black", linewidth=0.8)

# Achsenbeschriftungen
ax.text(1, -0.12, "1", ha="center", fontsize=10)
ax.text(-1, -0.12, "-1", ha="center", fontsize=10)
ax.text(-0.10, 1, "1", ha="right", va="center", fontsize=10)
ax.text(-0.12, -1, "-1", ha="right", va="center", fontsize=10)

# Vektoren
pfeil_z = FancyArrowPatch(
    (0, 0), (z[0], z[1]),
    arrowstyle="-|>",
    mutation_scale=14,
    linewidth=1.6,
    color=farbe_z
)
ax.add_patch(pfeil_z)

pfeil_inv = FancyArrowPatch(
    (0, 0), (z_inv[0], z_inv[1]),
    arrowstyle="-|>",
    mutation_scale=14,
    linewidth=1.6,
    color=farbe_inv
)
ax.add_patch(pfeil_inv)

# Beschriftungen der Vektoren
ax.text(0.38, 0.50, r"$z$", color=farbe_z, fontsize=12)
ax.text(0.34, -0.52, r"$\frac{1}{z}$", color=farbe_inv, fontsize=12)

# Winkelbögen
winkelbogen(ax, 0.4, 0, 45, farbe_z)
winkelbogen(ax, 0.4, 0, -45, farbe_inv)

# Winkelbeschriftungen
ax.text(0.53, 0.20, r"$\frac{\pi}{4}$", color=farbe_z, fontsize=11)
ax.text(0.53, -0.24, r"$-\frac{\pi}{4}$", color=farbe_inv, fontsize=11)

# Layout
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```
