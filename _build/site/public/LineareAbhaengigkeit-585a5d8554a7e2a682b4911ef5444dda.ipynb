## Lineare Abhängigkeit von Vektoren

Gegeben sei eine Menge von Vektoren

$$
\{\vec{v}_1,\vec{v}_2,\dots,\vec{v}_n\}
$$

im $\mathbb{R}^m$.

### Definition

Die Vektoren

$$
\vec{v}_1,\vec{v}_2,\dots,\vec{v}_n
$$

sind **linear abhängig**, wenn es Skalare

$$
\lambda_1,\lambda_2,\dots,\lambda_n,
$$

die nicht alle Null sind, gibt, sodass:

$$
\lambda_1\vec{v}_1
+
\lambda_2\vec{v}_2
+
\dots
+
\lambda_n\vec{v}_n
=
\vec{0}.
$$

Wenn nur die triviale Lösung

$$
\lambda_1
=
\lambda_2
=
\dots
=
\lambda_n
=
0
$$

existiert, sind die Vektoren **linear unabhängig**.

### Lineare Abhängigkeit und Gauß-Elimination

Die lineare Abhängigkeit kann mit Hilfe des **Gauß-Eliminationsverfahrens** überprüft werden:

- Man stellt die Vektoren als Zeilen einer Matrix dar.
- Dann wendet man das Gauß-Verfahren an, um die Matrix in Stufenform zu bringen.
- Wenn mindestens eine Zeile vollständig Null wird, bedeutet dies, dass die zugehörigen Vektoren linear abhängig sind.

**Beispiel:**

$$
\text{Matrix: }
\begin{pmatrix}
1 & 2 & 3\\
2 & 4 & 6\\
1 & 1 & 1
\end{pmatrix}
\quad\Rightarrow\quad
\text{Stufenform: }
\begin{pmatrix}
1 & 2 & 3\\
0 & 0 & 0\\
0 & -1 & -2
\end{pmatrix}
$$

Die zweite Zeile ist linear abhängig von der ersten.
