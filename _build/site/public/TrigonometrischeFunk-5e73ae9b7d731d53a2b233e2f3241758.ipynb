## Trigonometrische Funktionen

Zu den wichtigsten trigonometrischen Funktionen gehören die **Sinusfunktion**, die **Kosinusfunktion** und die **Tangensfunktion**.

Sie treten insbesondere bei der Beschreibung von periodischen Vorgängen auf, beispielsweise bei

- Schwingungen,
- Wellen,
- Kreisbewegungen,
- Wechselspannungen.

Die Argumente trigonometrischer Funktionen werden im Folgenden im **Bogenmaß** angegeben.

Dabei gilt beispielsweise

$$
180^\circ = \pi,
\qquad
360^\circ = 2\pi.
$$

### Periodische Funktionen

Eine Funktion $f$ heißt **periodisch**, wenn es eine Zahl $T>0$ gibt, sodass

$$
f(x+T)=f(x)
$$

für alle $x$ aus dem Definitionsbereich gilt.

Die kleinste positive Zahl $T$, für die diese Gleichung erfüllt ist, heißt **Periodenlänge** oder kurz **Periode**.

Eine periodische Funktion wiederholt ihre Funktionswerte also nach einem festen Abstand.

Bei den trigonometrischen Funktionen gilt:

$$
\sin(x+2\pi)=\sin(x),
$$

$$
\cos(x+2\pi)=\cos(x),
$$

und

$$
\tan(x+\pi)=\tan(x).
$$

Sinus und Kosinus besitzen somit die Periodenlänge

$$
T=2\pi,
$$

während die Tangensfunktion die Periodenlänge

$$
T=\pi
$$

besitzt.

<a id="sinusfunktion"></a>

### Sinusfunktion

Die Sinusfunktion ist definiert durch

$$
f(x)=\sin(x).
$$

Ihr Definitionsbereich ist

$$
D=\mathbb{R}.
$$

Der Wertebereich lautet

$$
W=[-1,1].
$$

Die Sinusfunktion besitzt die Periodenlänge

$$
T=2\pi.
$$

Wichtige Funktionswerte sind

$$
\sin(0)=0,
$$

$$
\sin\left(\frac{\pi}{2}\right)=1,
$$

$$
\sin(\pi)=0,
$$

$$
\sin\left(\frac{3\pi}{2}\right)=-1,
$$

$$
\sin(2\pi)=0.
$$

Die Nullstellen liegen bei

$$
x=k\pi,
\qquad
k\in\mathbb{Z}.
$$

Die Sinusfunktion ist punktsymmetrisch zum Ursprung:

$$
\sin(-x)=-\sin(x).
$$

Sie ist somit eine **ungerade Funktion**.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2*np.pi, 2*np.pi, 1000)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(9, 4))

ax.plot(x, y, linewidth=1.6, color="#4F81BD")

ax.axhline(0, linewidth=0.8, color="black")
ax.axvline(0, linewidth=0.8, color="black")

ax.set_xlim(-2*np.pi, 2*np.pi)
ax.set_ylim(-1.3, 1.3)

ax.set_xticks([
    -2*np.pi,
    -3*np.pi/2,
    -np.pi,
    -np.pi/2,
    0,
    np.pi/2,
    np.pi,
    3*np.pi/2,
    2*np.pi
])

ax.set_xticklabels([
    r"$-2\pi$",
    r"$-\frac{3\pi}{2}$",
    r"$-\pi$",
    r"$-\frac{\pi}{2}$",
    r"$0$",
    r"$\frac{\pi}{2}$",
    r"$\pi$",
    r"$\frac{3\pi}{2}$",
    r"$2\pi$"
])

ax.set_yticks([-1, 0, 1])

ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$\sin(x)$")

ax.grid(alpha=0.25)

plt.show()
```

### Kosinusfunktion

Die Kosinusfunktion ist definiert durch

$$
f(x)=\cos(x).
$$

Ihr Definitionsbereich ist

$$
D=\mathbb{R}.
$$

Der Wertebereich lautet

$$
W=[-1,1].
$$

Auch die Kosinusfunktion besitzt die Periodenlänge

$$
T=2\pi.
$$

Wichtige Funktionswerte sind

$$
\cos(0)=1,
$$

$$
\cos\left(\frac{\pi}{2}\right)=0,
$$

$$
\cos(\pi)=-1,
$$

$$
\cos\left(\frac{3\pi}{2}\right)=0,
$$

$$
\cos(2\pi)=1.
$$

Die Nullstellen liegen bei

$$
x=\frac{\pi}{2}+k\pi,
\qquad
k\in\mathbb{Z}.
$$

Die Kosinusfunktion ist achsensymmetrisch zur $y$-Achse:

$$
\cos(-x)=\cos(x).
$$

Sie ist somit eine **gerade Funktion**.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2*np.pi, 2*np.pi, 1000)
y = np.cos(x)

fig, ax = plt.subplots(figsize=(9, 4))

ax.plot(x, y, linewidth=1.6, color="#4F81BD")

ax.axhline(0, linewidth=0.8, color="black")
ax.axvline(0, linewidth=0.8, color="black")

ax.set_xlim(-2*np.pi, 2*np.pi)
ax.set_ylim(-1.3, 1.3)

ax.set_xticks([
    -2*np.pi,
    -3*np.pi/2,
    -np.pi,
    -np.pi/2,
    0,
    np.pi/2,
    np.pi,
    3*np.pi/2,
    2*np.pi
])

ax.set_xticklabels([
    r"$-2\pi$",
    r"$-\frac{3\pi}{2}$",
    r"$-\pi$",
    r"$-\frac{\pi}{2}$",
    r"$0$",
    r"$\frac{\pi}{2}$",
    r"$\pi$",
    r"$\frac{3\pi}{2}$",
    r"$2\pi$"
])

ax.set_yticks([-1, 0, 1])

ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$\cos(x)$")

ax.grid(alpha=0.25)

plt.show()
```

### Zusammenhang zwischen Sinus und Kosinus

Sinus- und Kosinusfunktion besitzen die gleiche Form und die gleiche Periodenlänge.

Sie unterscheiden sich lediglich durch eine Verschiebung um

$$
\frac{\pi}{2}.
$$

Es gilt

$$
\cos(x)
=
\sin\left(x+\frac{\pi}{2}\right).
$$

Die Kosinusfunktion kann also als eine um $\frac{\pi}{2}$ nach links verschobene Sinusfunktion betrachtet werden.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2*np.pi, 2*np.pi, 1000)

fig, ax = plt.subplots(figsize=(9, 4))

ax.plot(
    x,
    np.sin(x),
    linewidth=1.6,
    color="#4F81BD",
    label=r"$\sin(x)$"
)

ax.plot(
    x,
    np.cos(x),
    linewidth=1.6,
    color="orange",
    label=r"$\cos(x)$"
)

ax.axhline(0, linewidth=0.8, color="black")
ax.axvline(0, linewidth=0.8, color="black")

ax.set_xlim(-2*np.pi, 2*np.pi)
ax.set_ylim(-1.3, 1.3)

ax.set_xticks([
    -2*np.pi,
    -3*np.pi/2,
    -np.pi,
    -np.pi/2,
    0,
    np.pi/2,
    np.pi,
    3*np.pi/2,
    2*np.pi
])

ax.set_xticklabels([
    r"$-2\pi$",
    r"$-\frac{3\pi}{2}$",
    r"$-\pi$",
    r"$-\frac{\pi}{2}$",
    r"$0$",
    r"$\frac{\pi}{2}$",
    r"$\pi$",
    r"$\frac{3\pi}{2}$",
    r"$2\pi$"
])

ax.set_yticks([-1, 0, 1])

ax.set_xlabel(r"$x$")
ax.set_ylabel("Funktionswert")

ax.legend()
ax.grid(alpha=0.25)

plt.show()
```

### Tangensfunktion

Die Tangensfunktion ist definiert durch

$$
\tan(x)
=
\frac{\sin(x)}{\cos(x)}.
$$

Da eine Division durch null nicht definiert ist, darf

$$
\cos(x)=0
$$

nicht gelten.

Der Definitionsbereich lautet deshalb

$$
D
=
\mathbb{R}
\setminus
\left\{
\frac{\pi}{2}+k\pi
\;\middle|\;
k\in\mathbb{Z}
\right\}.
$$

Der Wertebereich ist

$$
W=\mathbb{R}.
$$

Die Tangensfunktion besitzt die Periodenlänge

$$
T=\pi.
$$

Es gilt also

$$
\tan(x+\pi)=\tan(x).
$$

Die Nullstellen liegen bei

$$
x=k\pi,
\qquad
k\in\mathbb{Z}.
$$

An den Stellen

$$
x=\frac{\pi}{2}+k\pi
$$

ist die Tangensfunktion nicht definiert. Dort besitzt der Graph senkrechte Asymptoten.

Außerdem gilt

$$
\tan(-x)=-\tan(x).
$$

Die Tangensfunktion ist daher ebenfalls eine **ungerade Funktion**.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2*np.pi, 2*np.pi, 4000)
y = np.tan(x)

# Funktionswerte in der Nähe der Polstellen ausblenden
y[np.abs(np.cos(x)) < 0.03] = np.nan

fig, ax = plt.subplots(figsize=(9, 4))

ax.plot(x, y, linewidth=1.4, color="#4F81BD")

# Asymptoten
for k in range(-2, 2):
    asymptote = np.pi/2 + k*np.pi
    ax.axvline(
        asymptote,
        linestyle="--",
        linewidth=0.8,
        color="gray"
    )

ax.axhline(0, linewidth=0.8, color="black")
ax.axvline(0, linewidth=0.8, color="black")

ax.set_xlim(-2*np.pi, 2*np.pi)
ax.set_ylim(-5, 5)

ax.set_xticks([
    -2*np.pi,
    -3*np.pi/2,
    -np.pi,
    -np.pi/2,
    0,
    np.pi/2,
    np.pi,
    3*np.pi/2,
    2*np.pi
])

ax.set_xticklabels([
    r"$-2\pi$",
    r"$-\frac{3\pi}{2}$",
    r"$-\pi$",
    r"$-\frac{\pi}{2}$",
    r"$0$",
    r"$\frac{\pi}{2}$",
    r"$\pi$",
    r"$\frac{3\pi}{2}$",
    r"$2\pi$"
])

ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$\tan(x)$")

ax.grid(alpha=0.25)

plt.show()
```

### Eigenschaften im Überblick

| Funktion | Definitionsbereich | Wertebereich | Periodenlänge | Symmetrie |
|---|---|---|---|---|
| $\sin(x)$ | $\mathbb{R}$ | $[-1,1]$ | $2\pi$ | ungerade |
| $\cos(x)$ | $\mathbb{R}$ | $[-1,1]$ | $2\pi$ | gerade |
| $\tan(x)$ | $\mathbb{R}\setminus\left\{\frac{\pi}{2}+k\pi\right\}$ | $\mathbb{R}$ | $\pi$ | ungerade |

Dabei gilt jeweils

$$
k\in\mathbb{Z}.
$$

### Allgemeine Sinusfunktion

Eine allgemeine Sinusfunktion kann in der Form

$$
f(x)
=
A\sin(\omega x+\varphi)+d
$$

geschrieben werden.

Die Parameter beeinflussen den Graphen:

- $A$ bestimmt die **Amplitude**,
- $\omega$ bestimmt die **Periodenlänge**,
- $\varphi$ bestimmt die **Phasenverschiebung**,
- $d$ beschreibt eine Verschiebung in $y$-Richtung.

Die Amplitude beträgt

$$
|A|.
$$

Die Periodenlänge ist

$$
T=\frac{2\pi}{|\omega|}.
$$

Der Wertebereich lautet bei $A\neq 0$

$$
[d-|A|,\;d+|A|].
$$

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2*np.pi, 2*np.pi, 1000)

f1 = np.sin(x)
f2 = 2 * np.sin(x)
f3 = np.sin(2*x)

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(
    x,
    f1,
    linewidth=1.5,
    label=r"$\sin(x)$"
)

ax.plot(
    x,
    f2,
    linewidth=1.5,
    label=r"$2\sin(x)$"
)

ax.plot(
    x,
    f3,
    linewidth=1.5,
    label=r"$\sin(2x)$"
)

ax.axhline(0, linewidth=0.8, color="black")
ax.axvline(0, linewidth=0.8, color="black")

ax.set_xlim(-2*np.pi, 2*np.pi)
ax.set_ylim(-2.5, 2.5)

ax.set_xticks([
    -2*np.pi,
    -np.pi,
    0,
    np.pi,
    2*np.pi
])

ax.set_xticklabels([
    r"$-2\pi$",
    r"$-\pi$",
    r"$0$",
    r"$\pi$",
    r"$2\pi$"
])

ax.set_xlabel(r"$x$")
ax.set_ylabel("Funktionswert")

ax.legend()
ax.grid(alpha=0.25)

plt.show()
```

### Einfluss von Amplitude und Periodenlänge

Wir betrachten die Funktion

$$
f(x)=a\sin(bx).
$$

Die Parameter $a$ und $b$ beeinflussen den Verlauf der Sinusfunktion:

- $|a|$ bestimmt die **Amplitude**.
- $b$ bestimmt die **Periodenlänge**.

Für die Periodenlänge gilt

$$
T=\frac{2\pi}{|b|}.
$$

Mit den folgenden Reglern können die Parameter $a$ und $b$ verändert werden.

```python
%pip install -q ipywidgets
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

x = np.linspace(-2*np.pi, 2*np.pi, 1000)

a_slider = widgets.FloatSlider(
    value=1.0,
    min=-3.0,
    max=3.0,
    step=0.1,
    description="a:",
    continuous_update=True
)

b_slider = widgets.FloatSlider(
    value=1.0,
    min=0.25,
    max=3.0,
    step=0.05,
    description="b:",
    continuous_update=True
)

output = widgets.Output()


def plot_sinus(a, b):
    with output:
        output.clear_output(wait=True)

        y = a * np.sin(b * x)

        fig, ax = plt.subplots(figsize=(9, 5))

        ax.plot(
            x,
            y,
            linewidth=1.6,
            color="#4F81BD"
        )

        ax.axhline(0, linewidth=0.8, color="black")
        ax.axvline(0, linewidth=0.8, color="black")

        ax.set_xlim(-2*np.pi, 2*np.pi)
        ax.set_ylim(-3.5, 3.5)

        ax.set_xticks([
            -2*np.pi,
            -np.pi,
            0,
            np.pi,
            2*np.pi
        ])

        ax.set_xticklabels([
            r"$-2\pi$",
            r"$-\pi$",
            r"$0$",
            r"$\pi$",
            r"$2\pi$"
        ])

        ax.set_xlabel(r"$x$")
        ax.set_ylabel(r"$f(x)$")

        ax.set_title(
            rf"$f(x)={a:.1f}\sin({b:.2f}x)$"
        )

        ax.grid(alpha=0.25)

        # Zusatzinformationen
        amplitude = abs(a)
        periode = 2*np.pi / abs(b)

        ax.text(
            0.02,
            0.95,
            f"Amplitude = {amplitude:.2f}\n"
            f"Periodenlänge T = {periode:.2f}",
            transform=ax.transAxes,
            verticalalignment="top"
        )

        plt.show()


def update(change=None):
    plot_sinus(a_slider.value, b_slider.value)


a_slider.observe(update, names="value")
b_slider.observe(update, names="value")

display(widgets.VBox([a_slider, b_slider]))
display(output)

update()
```
