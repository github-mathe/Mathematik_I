## Umwandlung von polarer zu kartesischer Form


**Komplexe Zahlen in Polarform:**

- Eine komplexe Zahl $z$ kann in der Polarform geschrieben werden als:

$$
z = r(\cos\theta + i\sin\theta)
$$

- Dabei ist $r$ der Betrag der Zahl und $\theta$ der Winkel (Argument).

**Ziel:**

- Erklären, wie man die Polarform in die Kartesische Form $z = x + iy$ umwandelt.
- Verwenden von bekannten trigonometrischen Werten aus der Tabelle.

### Schritte zur Umwandlung

1. Bestimme den Betrag $r$ und den Winkel $\theta$ der Polarform.

2. Verwende die Gleichungen:

$$
x = r\cos\theta
\qquad \text{und} \qquad
y = r\sin\theta
$$

3. Setze die Werte von $\cos\theta$ und $\sin\theta$ aus der trigonometrischen Tabelle ein.

4. Schreibe das Ergebnis als

$$
z = x + iy
$$

in der Kartesischen Form.

### Beispiel 1

Gegeben:

$$
z
=
4\left(
\cos\frac{\pi}{3}
+
i\sin\frac{\pi}{3}
\right)
$$

**Schritte:**

- $r=4$ und $\theta=\frac{\pi}{3}$.
- Aus der Tabelle:

$$
\cos\frac{\pi}{3}
=
\frac{1}{2}
$$

und

$$
\sin\frac{\pi}{3}
=
\frac{\sqrt{3}}{2}.
$$

**Umwandlung:**

$$
x
=
r\cos\theta
=
4\cdot\frac{1}{2}
=
2
$$

$$
y
=
r\sin\theta
=
4\cdot\frac{\sqrt{3}}{2}
=
2\sqrt{3}
$$

Damit ergibt sich:

$$
z
=
2+2\sqrt{3}i
$$

### Beispiel 2

Gegeben:

$$
z
=
3\left(
\cos\frac{\pi}{4}
+
i\sin\frac{\pi}{4}
\right)
$$

**Schritte:**

- $r=3$ und $\theta=\frac{\pi}{4}$.
- Aus der Tabelle:

$$
\cos\frac{\pi}{4}
=
\frac{\sqrt{2}}{2}
$$

und

$$
\sin\frac{\pi}{4}
=
\frac{\sqrt{2}}{2}.
$$

**Umwandlung:**

$$
x
=
r\cos\theta
=
3\cdot\frac{\sqrt{2}}{2}
=
\frac{3\sqrt{2}}{2}
$$

$$
y
=
r\sin\theta
=
3\cdot\frac{\sqrt{2}}{2}
=
\frac{3\sqrt{2}}{2}
$$

Damit ergibt sich:

$$
z
=
\frac{3\sqrt{2}}{2}
+
\frac{3\sqrt{2}}{2}i
$$

### Beispiel 3

Gegeben:

$$
z
=
5\left(
\cos\frac{2\pi}{3}
+
i\sin\frac{2\pi}{3}
\right)
$$

**Schritte:**

- $r=5$ und $\theta=\frac{2\pi}{3}$.
- Aus der Tabelle:

$$
\cos\frac{2\pi}{3}
=
-\frac{1}{2}
$$

und

$$
\sin\frac{2\pi}{3}
=
\frac{\sqrt{3}}{2}.
$$

**Umwandlung:**

$$
x
=
r\cos\theta
=
5\cdot\left(-\frac{1}{2}\right)
=
-2.5
$$

$$
y
=
r\sin\theta
=
5\cdot\frac{\sqrt{3}}{2}
=
\frac{5\sqrt{3}}{2}
$$

Damit ergibt sich:

$$
z
=
-\frac{5}{2}
+
\frac{5\sqrt{3}}{2}i
$$

### Umwandlung von polarer zu kartesischer Form

- Erkenne $r$ und $\theta$ in der Polarform.
- Verwende

$$
x = r\cos\theta
$$

und

$$
y = r\sin\theta
$$

für die Umwandlung.

- Nutze trigonometrische Werte aus bekannten Tabellen.
- Schreibe das Ergebnis in der Form

$$
z = x + iy.
$$
