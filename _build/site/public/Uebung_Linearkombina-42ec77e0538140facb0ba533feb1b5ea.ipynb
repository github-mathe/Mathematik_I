## Linearkombination von Vektoren

Gegeben seien die Vektoren

$$
\vec{v}_1=
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix},
\qquad
\vec{v}_2=
\begin{pmatrix}
2\\
-1\\
3
\end{pmatrix}
$$

und der Vektor

$$
\vec{w}=
\begin{pmatrix}
4\\
3\\
1
\end{pmatrix}.
$$

Untersuchen Sie, ob sich der Vektor $\vec{w}$ als **Linearkombination** der Vektoren $\vec{v}_1$ und $\vec{v}_2$ darstellen lässt.

Bestimmen Sie dazu die Skalare $\lambda_1,\lambda_2\in\mathbb{R}$, sodass

$$
\vec{w}
=
\lambda_1\vec{v}_1
+
\lambda_2\vec{v}_2.
$$

Geben Sie anschließend die Linearkombination explizit an.

## Lösung

Die Lösung ist in einzelne Rechenschritte unterteilt.

Bearbeiten Sie zunächst jeden Schritt selbstständig und öffnen Sie anschließend das jeweilige Zwischenergebnis zur Kontrolle.

:::{dropdown} Aufstellen der Linearkombination

Gesucht sind $\lambda_1,\lambda_2\in\mathbb{R}$ mit

$$
\lambda_1\vec{v}_1
+
\lambda_2\vec{v}_2
=
\vec{w}.
$$

Einsetzen der Vektoren ergibt

$$
\lambda_1
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
2\\
-1\\
3
\end{pmatrix}
=
\begin{pmatrix}
4\\
3\\
1
\end{pmatrix}.
$$

Durch komponentenweisen Vergleich erhalten wir das lineare Gleichungssystem

$$
\begin{aligned}
\lambda_1+2\lambda_2 &=4,\\
2\lambda_1-\lambda_2 &=3,\\
-\lambda_1+3\lambda_2 &=1.
\end{aligned}
$$
:::

:::{dropdown} Bestimmung der Koeffizienten

Aus der ersten Gleichung

$$
\lambda_1+2\lambda_2=4
$$

folgt

$$
\lambda_1=4-2\lambda_2.
$$

Einsetzen in die zweite Gleichung liefert

$$
2(4-2\lambda_2)-\lambda_2=3.
$$

Damit gilt

$$
8-5\lambda_2=3
$$

und somit

$$
\lambda_2=1.
$$

Daraus folgt

$$
\lambda_1=4-2\cdot1=2.
$$

Zur Kontrolle setzen wir die Werte in die dritte Gleichung ein:

$$
-\lambda_1+3\lambda_2
=
-2+3
=
1.
$$

Die dritte Gleichung ist ebenfalls erfüllt.
:::

:::{dropdown} Gesamtlösung anzeigen

Es gilt

$$
\lambda_1=2,
\qquad
\lambda_2=1.
$$

Damit lässt sich $\vec{w}$ als Linearkombination von $\vec{v}_1$ und $\vec{v}_2$ darstellen:

$$
\boxed{
\vec{w}
=
2\vec{v}_1+\vec{v}_2
}
$$

Zur Kontrolle:

$$
2
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix}
+
\begin{pmatrix}
2\\
-1\\
3
\end{pmatrix}
=
\begin{pmatrix}
2\\
4\\
-2
\end{pmatrix}
+
\begin{pmatrix}
2\\
-1\\
3
\end{pmatrix}
=
\begin{pmatrix}
4\\
3\\
1
\end{pmatrix}.
$$

Somit ist

$$
\boxed{
\vec{w}\in\operatorname{span}\{\vec{v}_1,\vec{v}_2\}.
}
$$
:::
