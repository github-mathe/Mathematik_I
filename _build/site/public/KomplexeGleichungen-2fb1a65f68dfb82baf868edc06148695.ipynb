## Komplexe Gleichungen

Wir betrachten die Gleichung

$$
z^n=w
$$

wobei $w$ eine gegebene komplexe Zahl in Polarform ist und $n$ eine gegebene ganze Zahl.

Ziel ist es, alle Lösungen $z$ zu bestimmen und diese graphisch in der komplexen Ebene darzustellen.

### Darstellung komplexer Zahlen in Polarform

Eine komplexe Zahl $z$ kann in Polarform geschrieben werden als:

$$
z
=
r\left(\cos\theta+i\sin\theta\right)
=
re^{i\theta}
$$

wobei:

- $r=|z|$ der Betrag von $z$ ist,
- $\theta=\arg(z)$ das Argument (Winkel) von $z$ ist.

### Gegebene Gleichung

Gegeben:

$$
z^n=w
$$

mit

$$
w=|w|e^{i\varphi}.
$$

**Ziel:**

- Bestimme alle Lösungen $z$ in der Form $z=re^{i\theta}$.
- Stelle die Lösungen graphisch dar.

### Herleitung der Lösung

Setze

$$
z=re^{i\theta}
$$

ein:

$$
z^n
=
\left(re^{i\theta}\right)^n
=
r^n e^{in\theta}.
$$

Setze gleich:

$$
r^n e^{in\theta}
=
|w|e^{i\varphi}.
$$

### Bestimmung des Betrags

Aus der Gleichung der Beträge:

$$
r^n=|w|
\quad\Longrightarrow\quad
r=|w|^{1/n}.
$$

Der Betrag $r$ ist die $n$-te Wurzel des Betrags von $w$.

### Bestimmung des Arguments

Aus der Gleichung der Argumente:

$$
n\theta
=
\varphi+2\pi k,
\qquad
k\in\mathbb Z.
$$

Daraus folgt:

$$
\theta
=
\frac{\varphi+2\pi k}{n}
$$

für

$$
k=0,1,2,\dots,n-1.
$$

### Warum $k$ von $0$ bis $n-1$ läuft

Da das Argument eines komplexen Zahlenausdrucks nur bis auf ein Vielfaches von $2\pi$ bestimmt ist, beschreiben $\theta$ und $\theta+2\pi$ denselben Punkt auf der komplexen Ebene.

Erhöht man $k$ um $n$, so ergibt sich:

$$
\theta_{k+n}
=
\frac{\varphi+2\pi(k+n)}{n}
=
\frac{\varphi+2\pi k}{n}+2\pi,
$$

also derselbe Winkel und damit dieselbe $n$-te Wurzel.

Folglich genügen $n$ verschiedene Werte von $k$, um alle unterschiedlichen Lösungen zu erhalten:

$$
k=0,1,2,\dots,n-1.
$$

Für größere $k$ wiederholen sich die Wurzeln periodisch mit der Periode $n$.

Negative Werte von $k$ liefern keine neuen Lösungen, sondern dieselben Wurzeln in umgekehrter Reihenfolge.

### Allgemeine Lösung

Die Lösungen sind gegeben durch:

$$
z_k
=
re^{i\theta_k}
=
|w|^{1/n}
e^{
i\left(
\frac{\varphi+2\pi k}{n}
\right)
}
$$

für

$$
k=0,1,2,\dots,n-1.
$$

### Graphische Darstellung

- Die Lösungen $z_k$ liegen auf dem Kreis mit Radius

$$
r=|w|^{1/n}.
$$

- Der erste Winkel ist

$$
\theta_0=\frac{\varphi}{n},
$$

wobei $\varphi=\arg(w)$.

- Die Winkel $\theta_k$ sind gleichmäßig um den Kreis verteilt:

$$
\theta_k
=
\theta_0+\frac{2\pi k}{n}.
$$

- Der Winkel zwischen aufeinanderfolgenden Lösungen ist

$$
\frac{2\pi}{n}.
$$
