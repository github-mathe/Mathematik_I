## Direkte Summe von Unterräumen

### Summe und direkte Summe von Unterräumen

**Definition (Summe von Unterräumen):**

Seien $U$ und $W$ Unterräume eines Vektorraums $V$.

Die *Summe* der Unterräume ist definiert als

$$
U+W
=
\{
\mathbf{u}+\mathbf{w}
\mid
\mathbf{u}\in U,\;
\mathbf{w}\in W
\}.
$$

**Definition (direkte Summe):**

Die Summe $U+W$ heißt *direkte Summe* und wird mit
$U\oplus W$ bezeichnet, wenn

$$
U\cap W=\{\mathbf{0}\}.
$$

In diesem Fall schreiben wir

$$
V=U\oplus W.
$$

**Eigenschaft:**

Ist

$$
V=U\oplus W,
$$

so lässt sich jeder Vektor $\mathbf{v}\in V$ eindeutig als

$$
\mathbf{v}=\mathbf{u}+\mathbf{w}
$$

mit $\mathbf{u}\in U$ und $\mathbf{w}\in W$ schreiben.

### Geometrische Intuition

**Anschaulich:**

- Die direkte Summe $U\oplus W$ ist die Kombination zweier Unterräume, die nur den Nullvektor gemeinsam haben.
- Beispiel in $\mathbb{R}^3$:
  - $U$: Eine Gerade durch den Ursprung.
  - $W$: Eine Ebene durch den Ursprung, die orthogonal zu $U$ ist.
  - $U\oplus W=\mathbb{R}^3$.

**Dimension der direkten Summe:**

$$
\dim(U\oplus W)
=
\dim(U)+\dim(W).
$$

### Dimensionssatz für Summen von Unterräumen

**Satz (Dimensionsformel):**

Seien $U$ und $W$ Unterräume eines endlichdimensionalen Vektorraums $V$. Dann gilt

$$
\dim(U+W)
=
\dim(U)+\dim(W)-\dim(U\cap W).
$$

**Folgerung (direkte Summe):**

Ist

$$
V=U\oplus W,
$$

d.h.

$$
V=U+W
\qquad\text{und}\qquad
U\cap W=\{0\},
$$

so folgt

$$
\dim(V)
=
\dim(U\oplus W)
=
\dim(U)+\dim(W).
$$

**Interpretation:**

- Die Dimension des Schnitts $U\cap W$ wird doppelt gezählt und muss daher abgezogen werden.
- Bei einer direkten Summe ist der Schnitt trivial, daher muss nichts abgezogen werden.

### Beispiel: Direkte Summe in $\mathbb{R}^3$

**Gegeben:**

- Unterraum

$$
U=\operatorname{span}(\mathbf{u}),
\qquad
\mathbf{u}=(1,0,0)^T.
$$

- Unterraum

$$
W=\operatorname{span}(\mathbf{v},\mathbf{w}),
$$

mit

$$
\mathbf{v}=(0,1,0)^T,
\qquad
\mathbf{w}=(0,0,1)^T.
$$

**Direkte Summe:**

$$
U\oplus W=\mathbb{R}^3,
\qquad
\dim(U\oplus W)
=
\dim(U)+\dim(W)
=
1+2
=
3.
$$

**Interpretation:**

Jeder Vektor in $\mathbb{R}^3$ lässt sich eindeutig schreiben als

$$
\mathbf{v}
=
c_1\mathbf{u}
+
c_2\mathbf{v}
+
c_3\mathbf{w},
\qquad
c_1,c_2,c_3\in\mathbb{R}.
$$

### Weiteres Beispiel: Direkte Summe in $\mathbb{R}^3$

**Gegeben:**

- Unterraum

$$
U=\operatorname{span}(\mathbf{u}_1,\mathbf{u}_2),
$$

mit

$$
\mathbf{u}_1=(1,1,0)^T,
\qquad
\mathbf{u}_2=(0,1,1)^T.
$$

- Unterraum

$$
W=\operatorname{span}(\mathbf{w}),
$$

mit

$$
\mathbf{w}=(1,-1,1)^T.
$$

**Direkte Summe:**

- Überprüfung:

$$
U\cap W=\{\mathbf{0}\},
$$

da kein nichttrivialer Vektor aus $U$ auch in $W$ liegt.

- Dimensionen:

$$
\dim(U)=2,
\qquad
\dim(W)=1,
\qquad
\dim(U\oplus W)
=
\dim(U)+\dim(W)
=
3.
$$

**Interpretation:**

Jeder Vektor in $\mathbb{R}^3$ lässt sich eindeutig schreiben als

$$
\mathbf{v}
=
c_1\mathbf{u}_1
+
c_2\mathbf{u}_2
+
c_3\mathbf{w},
\qquad
c_1,c_2,c_3\in\mathbb{R}.
$$

**Anschaulich:**

Der Raum $\mathbb{R}^3$ ist die direkte Summe von $U$ (einer Ebene) und $W$ (einer Geraden, die nicht in der Ebene $U$ liegt).

### Überprüfung: $U\cap W=\{\mathbf{0}\}$

**Zu prüfen:**

$$
U\cap W=\{\mathbf{0}\}.
$$

Ein Vektor $\mathbf{v}\in U\cap W$ ist sowohl Linearkombination von
$\mathbf{u}_1,\mathbf{u}_2$ als auch von $\mathbf{w}$:

$$
\mathbf{v}
=
\lambda_1\mathbf{u}_1
+
\lambda_2\mathbf{u}_2
=
\lambda_3\mathbf{w}.
$$

Komponentenweise:

$$
\lambda_1
\begin{pmatrix}
1\\
1\\
0
\end{pmatrix}
+
\lambda_2
\begin{pmatrix}
0\\
1\\
1
\end{pmatrix}
-
\lambda_3
\begin{pmatrix}
1\\
-1\\
1
\end{pmatrix}
=
\begin{pmatrix}
0\\
0\\
0
\end{pmatrix}.
$$

Daraus folgt das Gleichungssystem:

$$
\begin{aligned}
\lambda_1-\lambda_3 &= 0,\\
\lambda_1+\lambda_2+\lambda_3 &= 0,\\
\lambda_2-\lambda_3 &= 0.
\end{aligned}
$$

### Lösung des Systems mit Gauß-Elimination

Matrixdarstellung des Systems:

$$
\left[
\begin{array}{c|ccc|c}
\text{I}   & 1 & 0 & -1 & 0\\
\text{II}  & 1 & 1 &  1 & 0\\
\text{III} & 0 & 1 & -1 & 0
\end{array}
\right].
$$

Erster Eliminationsschritt:

$$
\begin{array}{r|ccc|c|l}
\text{I}    & 1 & 0 & -1 & 0 &\\
\text{II'}  & 0 & 1 & 2  & 0 & \text{II'}=\text{II}-\text{I}\\
\text{III}  & 0 & 1 & -1 & 0 &
\end{array}
$$

Danach:

$$
\begin{array}{r|ccc|c|l}
\text{I}    & 1 & 0 & -1 & 0 &\\
\text{II'}  & 0 & 1 & 2  & 0 &\\
\text{III'} & 0 & 0 & -3 & 0 & \text{III'}=\text{III}-\text{II'}
\end{array}.
$$

### Schlussfolgerung

- Aus Zeile $\text{III'}$:

$$
-3\lambda_3=0
\quad\Rightarrow\quad
\lambda_3=0.
$$

- Einsetzen in Zeile $\text{II'}$:

$$
\lambda_2+2\cdot0=0
\quad\Rightarrow\quad
\lambda_2=0.
$$

- Einsetzen in Zeile $\text{I}$:

$$
\lambda_1-0=0
\quad\Rightarrow\quad
\lambda_1=0.
$$

Der einzige Vektor, der sowohl in $U$ als auch in $W$ liegt, ist der Nullvektor.

Es folgt:

$$
U\cap W=\{\mathbf{0}\}.
$$

$U$ und $W$ haben keine nichttrivialen gemeinsamen Vektoren. Die direkte Summe $U\oplus W$ ist wohldefiniert.

```python
# Ergänzung: Überprüfung der direkten Summe

import sympy as sp

u1 = sp.Matrix([1, 1, 0])
u2 = sp.Matrix([0, 1, 1])
w  = sp.Matrix([1, -1, 1])

A = sp.Matrix.hstack(u1, u2, w)

print("Matrix aus den Basisvektoren:")
sp.pprint(A)

print("\nRang:", A.rank())
print("Determinante:", A.det())

if A.rank() == 3:
    print("\nDie drei Vektoren sind linear unabhängig.")
    print("Damit gilt U ⊕ W = R^3.")
```
