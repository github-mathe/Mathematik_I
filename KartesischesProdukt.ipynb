## Kartesisches Produkt

### Das kartesische Produkt

Das **kartesische Produkt** zweier Mengen $A$ und $B$, notiert als $A\times B$, ist die Menge aller geordneten Paare, bei denen das erste Element aus $A$ und das zweite Element aus $B$ stammt.

Formal definieren wir das kartesische Produkt als:

$$
A\times B
=
\{(a,b)\mid a\in A,\; b\in B\}
$$

**Beispiel:** Wenn

$$
A=\{1,2\}
$$

und

$$
B=\{x,y\},
$$

dann gilt:

$$
A\times B
=
\{(1,x),(1,y),(2,x),(2,y)\}.
$$

Das kartesische Produkt kann auf mehr als zwei Mengen erweitert werden.

Zum Beispiel, wenn wir drei Mengen $A$, $B$ und $C$ nehmen, dann ist

$$
A\times B\times C
$$

die Menge aller geordneten Tripel $(a,b,c)$, bei denen

$$
a\in A,\qquad b\in B,\qquad c\in C
$$

sind.

### Die Notation $\mathbb{R}^n$

Die Notation $\mathbb{R}^n$ steht für das n-dimensionale kartesische Produkt der reellen Zahlen $\mathbb{R}$ mit sich selbst und beschreibt den n-dimensionalen euklidischen Raum.

Formal lässt sich dies ausdrücken als:

$$
\mathbb{R}^n
=
\underbrace{
\mathbb{R}\times\mathbb{R}\times\cdots\times\mathbb{R}
}_{\text{n-mal}}
$$

In der Mengen-Schreibweise gilt:

$$
\mathbb{R}^n
=
\left\{
(x_1,x_2,\dotsc,x_n)
\mid
x_i\in\mathbb{R}
\text{ für alle }
i=1,2,\dotsc,n
\right\}.
$$

### Beispiele für $\mathbb{R}^n$

- $\mathbb{R}^2$ ist die Menge aller geordneten Paare $(x_1,x_2)$, wobei $x_1$ und $x_2$ reelle Zahlen sind. Dies entspricht der üblichen 2-dimensionalen Ebene.

- $\mathbb{R}^3$ ist die Menge aller geordneten Tripel $(x_1,x_2,x_3)$, was dem 3-dimensionalen Raum entspricht.

- Allgemeiner: $\mathbb{R}^n$ repräsentiert einen $n$-dimensionalen Raum, wobei jeder Punkt durch $n$ Koordinaten dargestellt wird.

### Eigenschaften von $\mathbb{R}^n$

In $\mathbb{R}^n$ sind die Vektoraddition und die skalare Multiplikation komponentenweise definiert:

- Für Vektoren

$$
\mathbf{v}
=
(v_1,v_2,\ldots,v_n)
$$

und

$$
\mathbf{w}
=
(w_1,w_2,\ldots,w_n)
$$

ist ihre Summe:

$$
\mathbf{v}+\mathbf{w}
=
(v_1+w_1,\,
v_2+w_2,\,
\ldots,\,
v_n+w_n).
$$

- Für einen Skalar $\lambda\in\mathbb{R}$ und einen Vektor

$$
\mathbf{v}
=
(v_1,v_2,\ldots,v_n)
$$

ist die skalare Multiplikation:

$$
\lambda\mathbf{v}
=
(\lambda v_1,\,
\lambda v_2,\,
\ldots,\,
\lambda v_n).
$$
