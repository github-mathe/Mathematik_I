 # Grundlagen

 In diesem Abschnitt betrachten wir grundlegende Konzepte der Mengenlehre und Algebra. Zuerst werden wir einige wichtige Symbole der Mengenlehre kennenlernen, gefolgt von einer Einführung in Gruppen, Ringe und Körper, die zentrale Objekte in der Algebra darstellen.

## Überblick über algebraische Strukturen

- **Gruppen** beschreiben Mengen mit einer Operation, die abgeschlossen, assoziativ ist und ein neutrales sowie inverses Element besitzt.
- **Ringe** erweitern Gruppen mit einer zweiten Operation (Multiplikation).
- **Körper** sind Ringe, bei denen jedes nicht-null Element ein multiplikatives Inverses hat.
