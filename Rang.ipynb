## Rang einer Matrix

### Der (Zeilen)Rang einer Matrix

Der (Zeilen)Rang einer Matrix gibt die maximale Anzahl linear unabhängiger Zeilen einer Matrix an.

### Definition

Der Rang einer Matrix $A$ ist die Anzahl der Pivot-Elemente, die nach Anwendung des Gauß-Eliminationsverfahrens in der Stufenform übrig bleiben.

**Eigenschaften:**

- Der (Zeilen)Rang gibt die Dimension des von den Zeilen aufgespannten Raums an.
- Ein (Zeilen)Rang von $r$ bedeutet, dass $r$ Zeilen linear unabhängig sind.

### Der Rang und unabhängige Zeilenvektoren

Der Rang einer Matrix gibt nicht nur die Anzahl der unabhängigen Zeilenvektoren an, sondern auch die Anzahl der unabhängigen **Spaltenvektoren**.

Dies ergibt sich aus den Eigenschaften der Zeilenumformungen und der Gauß-Elimination.

### Zeilenrang = Spaltenrang

- Der **Zeilenrang** einer Matrix ist die Anzahl der linear unabhängigen Zeilenvektoren.
- Der **Spaltenrang** ist die Anzahl der linear unabhängigen Spaltenvektoren.
- Es gilt immer:

$$
\boxed{\text{Zeilenrang}=\text{Spaltenrang}}
$$

Das ist eine fundamentale Eigenschaft jeder Matrix.

### Geometrische Bedeutung des Ranges: vom Spann zum Spalten- und Zeilenraum

Im Abschnitt zum **Spann von Vektoren** haben wir gesehen, dass Linearkombinationen von Vektoren einen Unterraum erzeugen. Wir wenden diese Idee nun auf die Spalten und Zeilen einer Matrix an und interpretieren den **Rang** geometrisch.

Wir betrachten die Matrix

$$
A=\begin{pmatrix}
1&0&1\\
0&1&1\\
0&0&0
\end{pmatrix}
$$

mit den Spaltenvektoren $\mathbf a_1=(1,0,0)^\mathsf T$, $\mathbf a_2=(0,1,0)^\mathsf T$ und $\mathbf a_3=(1,1,0)^\mathsf T$. Da

$$
\mathbf a_3=\mathbf a_1+\mathbf a_2,
$$

liefert die dritte Spalte keine zusätzliche unabhängige Richtung. Der **Spaltenraum** ist daher die von den ersten beiden Spalten aufgespannte Ebene:

$$
\operatorname{Col}(A)
=\operatorname{Span}\{\mathbf a_1,\mathbf a_2\}
=\{(x,y,0)^\mathsf T\mid x,y\in\mathbb R\}.
$$

Der Rang ist die Dimension dieses Unterraums. Für unsere Matrix gilt also

$$
\boxed{\operatorname{Rang}(A)=\dim\operatorname{Col}(A)=2.}
$$

Auch die ersten beiden **Zeilenvektoren** sind linear unabhängig; die dritte Zeile ist eine Nullzeile. Der von den Zeilen aufgespannte Raum ist eine andere Ebene:

$$
\operatorname{Row}(A)
=\operatorname{Span}\{(1,0,1)^\mathsf T,(0,1,1)^\mathsf T\}
=\{(x,y,x+y)^\mathsf T\mid x,y\in\mathbb R\}.
$$

Damit besitzen **Spaltenraum und Zeilenraum dieselbe Dimension**, obwohl sie hier unterschiedliche Ebenen im $\mathbb R^3$ sind. Dies veranschaulicht die bereits eingeführte Gleichheit von Zeilenrang und Spaltenrang.
