## Spann eines Vektorraums

### Der Spann eines Vektorraums

Der **Spann** (aufgespannte Raum) einer Menge von Vektoren

$$
\{\mathbf v_1,\mathbf v_2,\dots,\mathbf v_n\},
$$

wobei

$$
\mathbf v_1, \mathbf v_2,\dots,\mathbf v_n\in\mathbb{R}^m,
$$

ist die Menge aller Linearkombinationen dieser Vektoren:

$$
\operatorname{Span}\{\mathbf v_1,\mathbf v_2,\dots,\mathbf v_n\}
=
\left\{
\lambda_1 \mathbf v_1
+
\lambda_2 \mathbf v_2
+
\dots
+
\lambda_n \mathbf v_n
\mid
\lambda_1,\lambda_2,\dots,\lambda_n\in\mathbb{R}
\right\}.
$$

- Der Spann beschreibt den kleinsten Unterraum von $\mathbb{R}^m$, der die Vektoren

$$
\{\mathbf v_1,\mathbf v_2,\dots,\mathbf v_n\}
$$

enthält.

- Die Vektoren

$$
\{\mathbf v_1,\mathbf v_2,\dots,\mathbf v_n\}
$$

müssen nicht unbedingt linear unabhängig sein, um einen Raum aufzuspannen.

- Der aufgespannte Raum enthält alle möglichen Linearkombinationen dieser Vektoren.

Wenn die Vektoren den gesamten $\mathbb{R}^m$-Raum aufspannen, dann bilden sie eine **Basis** dieses Raumes, vorausgesetzt, sie sind linear unabhängig.

### Spaltenraum einer Matrix

Sei $A\in\mathbb{R}^{m\times n}$ eine Matrix mit den Spaltenvektoren $\mathbf a_1,\ldots,\mathbf a_n\in\mathbb{R}^m$:

$$
A=\begin{pmatrix}\vert & & \vert\\ \mathbf a_1 & \cdots & \mathbf a_n\\ \vert & & \vert\end{pmatrix}.
$$

Der **Spaltenraum** (englisch: *column space*) von $A$ ist der Spann ihrer Spaltenvektoren:

$$
\operatorname{Col}(A)=\operatorname{Span}\{\mathbf a_1,\ldots,\mathbf a_n\}
=\left\{A\mathbf x\mid\mathbf x\in\mathbb{R}^n\right\}\subseteq\mathbb{R}^m.
$$

Der Spaltenraum ist also die Menge aller Vektoren, die als Linearkombination der Spalten von $A$ dargestellt werden können. Insbesondere gilt: Das lineare Gleichungssystem $A\mathbf x=\mathbf b$ ist genau dann lösbar, wenn $\mathbf b\in\operatorname{Col}(A)$ ist.

**Beispiel (Spaltenraum).** Betrachten wir die Matrix

$$
A=\begin{pmatrix}
1&0&1\\
0&1&1\\
0&0&0
\end{pmatrix}.
$$

Ihre Spalten sind

$$
\mathbf a_1=\begin{pmatrix}1\\0\\0\end{pmatrix},\qquad
\mathbf a_2=\begin{pmatrix}0\\1\\0\end{pmatrix},\qquad
\mathbf a_3=\begin{pmatrix}1\\1\\0\end{pmatrix}
=\mathbf a_1+\mathbf a_2.
$$

Da die dritte Spalte eine Linearkombination der ersten beiden Spalten ist, gilt

$$
\begin{aligned}
\operatorname{Col}(A)
&=\operatorname{Span}\{\mathbf a_1,\mathbf a_2\}\\
&=\left\{\lambda_1\begin{pmatrix}1\\0\\0\end{pmatrix}
+\lambda_2\begin{pmatrix}0\\1\\0\end{pmatrix}
\mid\lambda_1,\lambda_2\in\mathbb{R}\right\}\\
&=\left\{\begin{pmatrix}\lambda_1\\\lambda_2\\0\end{pmatrix}
\mid\lambda_1,\lambda_2\in\mathbb{R}\right\}.
\end{aligned}
$$

Wir benennen nun die Koeffizienten $\lambda_1$ in $x$ und $\lambda_2$ in $y$ um, um die Vektoren mit den üblichen Raumkoordinaten $x$, $y$ und $z$ zu beschreiben. Damit erhalten wir

$$
\operatorname{Col}(A)
=\left\{\begin{pmatrix}x\\y\\0\end{pmatrix}
\mid x,y\in\mathbb{R}\right\}.
$$

Der Spaltenraum ist die Ebene $z=0$ im $\mathbb{R}^3$, also die $x$-$y$-Ebene.

### Zeilenraum einer Matrix

Sei $A\in\mathbb{R}^{m\times n}$ mit den Zeilenvektoren $\mathbf r_1,\ldots,\mathbf r_m\in\mathbb{R}^n$ (als Vektoren im $\mathbb{R}^n$ aufgefasst).

Der **Zeilenraum** (englisch: *row space*) von $A$ ist der Spann ihrer Zeilenvektoren:

$$
\operatorname{Row}(A)=\operatorname{Span}\{\mathbf r_1,\ldots,\mathbf r_m\}
=\left\{A^{\mathsf T}\mathbf y\mid\mathbf y\in\mathbb{R}^m\right\}\subseteq\mathbb{R}^n.
$$

Der Zeilenraum enthält alle Linearkombinationen der Zeilen von $A$. Er ist zugleich der Spaltenraum der transponierten Matrix $A^{\mathsf T}$.

**Beispiel (Zeilenraum).** Wir verwenden dieselbe Matrix

$$
A=\begin{pmatrix}
1&0&1\\
0&1&1\\
0&0&0
\end{pmatrix}.
$$

Ihre Zeilenvektoren sind

$$
\mathbf r_1=(1,0,1),\qquad
\mathbf r_2=(0,1,1),\qquad
\mathbf r_3=(0,0,0).
$$

Da die dritte Zeile der Nullvektor ist, wird der Zeilenraum von den ersten beiden Zeilen aufgespannt. Damit erhalten wir

$$
\begin{aligned}
\operatorname{Row}(A)
&=\operatorname{Span}\{\mathbf r_1,\mathbf r_2\}\\
&=\left\{\lambda_1\begin{pmatrix}1\\0\\1\end{pmatrix}
+\lambda_2\begin{pmatrix}0\\1\\1\end{pmatrix}
\mid\lambda_1,\lambda_2\in\mathbb{R}\right\}\\
&=\left\{\begin{pmatrix}\lambda_1\\\lambda_2\\\lambda_1+\lambda_2\end{pmatrix}
\mid\lambda_1,\lambda_2\in\mathbb{R}\right\}.
\end{aligned}
$$

Wir benennen auch hier $\lambda_1$ in $x$ und $\lambda_2$ in $y$ um, um die Vektoren mit den üblichen Raumkoordinaten zu beschreiben. Es folgt

$$
\operatorname{Row}(A)
=\left\{\begin{pmatrix}x\\y\\x+y\end{pmatrix}
\mid x,y\in\mathbb{R}\right\}.
$$

Der Zeilenraum ist somit die Ebene $z=x+y$ im $\mathbb{R}^3$.

**Vergleich.** In diesem Beispiel sind der Spaltenraum und der Zeilenraum verschiedene Ebenen im $\mathbb{R}^3$. Beide haben jedoch die Dimension $2$:

$$
\dim\bigl(\operatorname{Col}(A)\bigr)
=\dim\bigl(\operatorname{Row}(A)\bigr)=2.
$$

Allgemein gilt für jede Matrix $A\in\mathbb{R}^{m\times n}$:

$$
\boxed{\dim\bigl(\operatorname{Col}(A)\bigr)
=\dim\bigl(\operatorname{Row}(A)\bigr).}
$$

**Wichtig:** Der Spaltenraum liegt in $\mathbb{R}^m$, der Zeilenraum in $\mathbb{R}^n$. Bei einer nichtquadratischen Matrix sind sie daher im Allgemeinen sogar Unterräume unterschiedlicher Umgebungsräume.


Die gemeinsame Dimension des Spaltenraums und des Zeilenraums einer Matrix heißt **Rang** der Matrix. Dieser Begriff wird im nächsten Abschnitt eingeführt.
